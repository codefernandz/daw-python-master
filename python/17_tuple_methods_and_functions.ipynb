### 💡 Explicación de la Arquitectura
* **Métodos de Instancia (`tuple.method()`):**
  * **`index(value)`:** Retorna la primera posición del valor buscado. Lanza `ValueError` si el elemento no existe.
  * **`count(value)`:** Devuelve la cantidad exacta de apariciones de un valor dentro de la tupla.
* **Funciones Integradas Globales (`function(tuple)`):**
  * **`len(tuple)`:** Determina la longitud o cantidad total de elementos.
  * **`min(tuple)` / `max(tuple)`:** Encuentran el valor menor y mayor dentro de una tupla homogénea comparable.
  * **`type(tuple)`:** Confirma el tipo de estructura (`<class 'tuple'>`).

```python
# [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Métodos de Instancia y Funciones para Tuplas")
print("=" * 80)

# ================================================================================
# 🔍 1. MÉTODOS NATIVOS DE INSTANCIA (index, count)
# ================================================================================
# Búsqueda de posiciones y conteo de apariciones en tuplas inmutables.
# ================================================================================

# [1] Declaración de tupla numérica con valores repetidos
sample_tuple: tuple[int, ...] = (10, 20, 30, 20, 40, 20, 50)

print("\n--- 🚀 [PASO 1]: Métodos de Instancia (index y count) ---")
print(f"📋 Tupla de prueba: {sample_tuple}")

# [2] Búsqueda de índice con index() y manejo seguro de errores
target_value: int = 30
try:
    target_index: int = sample_tuple.index(target_value)
    print(f"✅ [1. index()]: El valor {target_value} se encuentra en el índice: {target_index}")
except ValueError:
    print(f"❌ [Error]: El valor {target_value} no existe en la tupla.")

# [3] Conteo de frecuencias con count()
search_value: int = 20
occurrence_count: int = sample_tuple.count(search_value)
print(f"✅ [2. count()]: El valor {search_value} aparece {occurrence_count} veces en la tupla.")
```

## 📊 Métricas e Inspección General con Funciones Integradas (`min`, `max`, `len`, `type`)

### 💡 Explicación de Funciones Globales:
1. **`min()` / `max()`:** Operan sobre colecciones donde los elementos son comparables entre sí (ej. enteros con flotantes).
2. **`len()`:** Ejecución de complejidad $O(1)$ que consulta directamente la cabecera del objeto en CPython.
3. **`type()`:** Permite validar dinámicamente la clase a la que pertenece la colección.

```python
# ================================================================================
# 📊 2. FUNCIONES GLOBALES DE INSPECCIÓN (min, max, len, type)
# ================================================================================
print("\n--- 📊 [PASO 2]: Inspección General con Funciones Globales ---")

# Cálculo de valores extremos y métricas
min_value: int = min(sample_tuple)
max_value: int = max(sample_tuple)
tuple_length: int = len(sample_tuple)
data_type: type = type(sample_tuple)

print(f"✅ [3. min()]:  Valor mínimo de la tupla: {min_value}")
print(f"✅ [4. max()]:  Valor máximo de la tupla: {max_value}")
print(f"✅ [5. len()]:  Longitud de la tupla:      {tuple_length} elementos")
print(f"✅ [6. type()]: Tipo de objeto:           {data_type}")

# --------------------------------------------------------------------------------
# 🏁 [PASO 3]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Módulo de métodos de tuplas finalizado con éxito.")
print("=" * 80)
```
