| Propiedad | Detalle |
| :--- | :--- |

| 🎯 **Objetivo** | Explorar el funcionamiento de los operadores matemáticos nativos (`+`, `-`, `*`, `/`, `//`, `%`, `**`) y el uso sobrecargado de `+` para la concatenación de cadenas de texto. |
| 🛠️ **Conceptos** | `Addition (+)`, `Subtraction (-)`, `Multiplication (*)`, `Division (/)`, `Floor Division (//)`, `Modulus (%)`, `Exponentiation (**)`, `String Concatenation` |

---

### 💡 Explicación de la Arquitectura
* **Suma (`+`):** Suma dos números o concatena dos cadenas de texto (`str`).
* **Resta (`-`):** Resta el operando derecho del izquierdo.
* **Multiplicación (`*`):** Multiplica dos números (o repite cadenas).
* **División (`/`):** Divide dos números y **siempre** retorna un número flotante (`float`).
* **División Entera (`//`):** Divide dos números y descarta la parte decimal (retorna la parte entera).
* **Módulo (`%`):** Retorna el residuo/resto de una división entera.
* **Potencia (`**`):** Eleva la base a la potencia del exponente.

```python
# [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Operadores Aritméticos y Concatenación")
print("=" * 80)

# ================================================================================
# 🔢 1. OPERACIONES MATEMÁTICAS BÁSICAS (+, -, *, /)
# ================================================================================
# Declaración de variables numéricas operando en Python.
# ================================================================================

# [1] Operandos numéricos iniciales
first_number: int = 15
second_number: int = 4

print("\n--- 🚀 [PASO 1]: Operaciones Matemáticas Básicas ---")

# Operaciones básicas
addition_result: int = first_number + second_number
subtraction_result: int = first_number - second_number
multiplication_result: int = first_number * second_number
division_result: float = first_number / second_number  # Siempre retorna float

print(f"✅ [SUMA (+)]:          {first_number} + {second_number} = {addition_result}")
print(f"✅ [RESTA (-)]:         {first_number} - {second_number} = {subtraction_result}")
print(f"✅ [MULTIPLICACIÓN (*)]: {first_number} * {second_number} = {multiplication_result}")
print(f"✅ [DIVISIÓN (/)]:       {first_number} / {second_number} = {division_result} (Tipo: {type(division_result).__name__})")
```

## 📐 Operaciones Avanzadas y Concatenación de Cadenas

### 💡 Explicación de las Operaciones:
1. **División Entera (`//`) vs Módulo (`%`):** La división entera extrae únicamente el cociente sin decimales, mientras que el módulo obtiene el residuo restante de la división.
2. **Potencia (`**`):** Permite calcular exponentes de forma nativa sin requerir librerías externas.
3. **Sobrecarga del Operador Suma (`+`):** Al aplicar el operador `+` entre dos cadenas de texto (`str`), este actúa concatenando ambos valores en una sola secuencia.

```python
# ================================================================================
# ⚙️ 2. OPERACIONES AVANZADAS (//, %, **)
# ================================================================================
print("\n--- ⚙️ [PASO 2]: División Entera, Módulo y Potencias ---")

# Operaciones avanzadas
floor_division_result: int = first_number // second_number
modulus_result: int = first_number % second_number
exponentiation_result: int = first_number ** second_number

print(f"✅ [DIVISIÓN ENTERA (//)]: {first_number} // {second_number} = {floor_division_result}")
print(f"✅ [MÓDULO (%)]:           {first_number} % {second_number}  = {modulus_result}")
print(f"✅ [POTENCIA (**)]:         {first_number} ** {second_number} = {exponentiation_result}")

# ================================================================================
# 🔤 3. CONCATENACIÓN DE CADENAS DE TEXTO
# ================================================================================
print("\n--- 🔤 [PASO 3]: Sobrecarga de Operador (+) para Concatenación ---")

# [1] Declaración de componentes de texto en inglés
greeting_prefix: str = "Hola, "
user_name: str = "Ana"

# [2] Concatenación utilizando el operador +
full_message: str = greeting_prefix + user_name

print(f"✅ [TEXTO CONCATENADO]: '{full_message}'")
print(f"📋 [TIPO RESULTANTE]:   {type(full_message)}")

# --------------------------------------------------------------------------------
# 🏁 [PASO 4]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Módulo de operadores aritméticos finalizado con éxito.")
print("=" * 80)
```
