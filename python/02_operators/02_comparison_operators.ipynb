# ⚙️ Fundamentos de Python: Operadores de Comparación (`Comparison Operators`)

| Propiedad | Detalle |
| :--- | :--- |
| 📍 **Módulo** | Fundamentos de Python |
| 🎯 **Objetivo** | Comprender la evaluación de relaciones entre valores mediante los operadores de comparación (`==`, `!=`, `>`, `<`, `>=`, `<=`) y su resultado booleano (`True` / `False`). |
| 🛠️ **Conceptos** | `Equality (==)`, `Inequality (!=)`, `Greater Than (>)`, `Less Than (<)`, `Greater or Equal (>=)`, `Less or Equal (<=)` |

---

### 💡 Explicación de la Arquitectura
* **Igualdad (`==`):** Evalúa si ambos operandos son idénticos en valor.
* **Desigualdad (`!=`):** Evalúa si los operandos poseen valores distintos.
* **Mayor que (`>`) / Menor que (`<`):** Comparan magnitud entre valores numéricos o lexicográfica en cadenas.
* **Mayor o igual (`>=`) / Menor o igual (`<=`):** Incluyen la posibilidad de igualdad dentro de la condición.

```python
# [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Operadores de Comparación")
print("=" * 80)

# ================================================================================
# 🔢 1. COMPARACIONES NUMÉRICAS DIRECTAS
# ================================================================================
# Evaluación de relaciones numéricas obteniendo resultados de tipo 'bool'.
# ================================================================================

# [1] Operandos numéricos iniciales
left_value: int = 10
right_value: int = 20

print("\n--- 🚀 [PASO 1]: Evaluación de Operandos Numéricos (10 vs 20) ---")

# Evaluaciones relacionales
is_equal: bool = left_value == right_value
is_not_equal: bool = left_value != right_value
is_greater_than: bool = left_value > right_value
is_less_than: bool = left_value < right_value
is_greater_or_equal: bool = left_value >= right_value
is_less_or_equal: bool = left_value <= right_value

print(f"✅ [IGUALDAD (==)]:           {left_value} == {right_value}  -> {is_equal}")
print(f"✅ [DESIGUALDAD (!=)]:        {left_value} != {right_value}  -> {is_not_equal}")
print(f"✅ [MAYOR QUE (>)]:           {left_value} >  {right_value}  -> {is_greater_than}")
print(f"✅ [MENOR QUE (<)]:           {left_value} <  {right_value}  -> {is_less_than}")
print(f"✅ [MAYOR O IGUAL QUE (>=)]: {left_value} >= {right_value}  -> {is_greater_or_equal}")
print(f"✅ [MENOR O IGUAL QUE (<=)]: {left_value} <= {right_value}  -> {is_less_or_equal}")
```

## 📝 Comparación de Cadenas de Texto y Evaluación en Estructuras de Control

### 💡 Explicación de los Casos de Uso:
1. **Comparación Lexicográfica (Textos):** Python evalúa la igualdad entre cadenas respetando mayúsculas, minúsculas y orden alfabético según la tabla ASCII/Unicode.
2. **Evaluación de Límites:** Los operadores `>=` y `<=` son esenciales para verificar límites de edad, inventarios o rangos numéricos.

```python
# ================================================================================
# 🔤 2. COMPARACIÓN DE CADENAS DE TEXTO
# ================================================================================
print("\n--- 🔤 [PASO 2]: Comparación Lexicográfica entre Cadenas ---")

first_str: str = "Python"
second_str: str = "python"  # En minúscula para verificar case-sensitivity

print(f"✅ [TEXTO 1]: '{first_str}' | [TEXTO 2]: '{second_str}'")
print(f"📋 ¿Son idénticas? ('{first_str}' == '{second_str}'): {first_str == second_str}")
print(f"📋 ¿Son distintas? ('{first_str}' != '{second_str}'): {first_str != second_str}")

# ================================================================================
# 🎯 3. EVALUACIÓN PRÁCTICA DE LÍMITES
# ================================================================================
print("\n--- 🎯 [PASO 3]: Validación de Rangos y Edades Minimales ---")

user_age: int = 18
minimum_required_age: int = 18

is_eligible_to_vote: bool = user_age >= minimum_required_age

print(f"📋 Edad del usuario:        {user_age} años")
print(f"📋 Edad mínima requerida:   {minimum_required_age} años")
print(f"✅ ¿Cumple la condición?    {is_eligible_to_vote}")

# --------------------------------------------------------------------------------
# 🏁 [PASO 4]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Módulo de operadores de comparación finalizado con éxito.")
print("=" * 80)
```
