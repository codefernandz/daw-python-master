| 📍 **Módulo** | Fundamentos de Python |
| 🎯 **Objetivo** | Comprender la combinación y negación de condiciones booleanas utilizando los operadores lógicos `and`, `or` y `not`. |
| 🛠️ **Conceptos** | `Conjunction (and)`, `Disjunction (or)`, `Negation (not)`, `Short-circuit Evaluation` |

---

### 💡 Explicación de la Arquitectura
* **Operador `and`:** Devuelve `True` **únicamente si ambas** condiciones son verdaderas. Si al menos una es `False`, el resultado es `False`.
* **Operador `or`:** Devuelve `True` si **al menos una** de las condiciones es verdadera. Solo devuelve `False` si todas las condiciones son falsas.
* **Operador `not`:** Invierte el valor booleano actual (`True` se convierte en `False` y viceversa).

```python
# [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Operadores Lógicos (and, or, not)")
print("=" * 80)

# ================================================================================
# ⚙️ 1. DECLARACIÓN DE VARIABLES Y EVALUACIÓN CON 'AND' Y 'OR'
# ================================================================================
# Evaluación de condiciones compuestas combinando valores de entrada.
# ================================================================================

# [1] Variables de estado en inglés
user_age: int = 20
is_student: bool = True
has_license: bool = False

print("\n--- 🚀 [PASO 1]: Evaluación de Condiciones con 'and' y 'or' ---")

# Condición 1: Puede conducir si tiene más de 18 años Y (and) posee licencia de conducir
can_drive: bool = (user_age > 18) and has_license

# Condición 2: Es considerado "joven" si tiene menos de 25 años O (or) es estudiante
is_young: bool = (user_age < 25) or is_student

print(f"📋 Edad del usuario:      {user_age} años")
print(f"📋 ¿Es estudiante?:       {is_student}")
print(f"📋 ¿Tiene licencia?:      {has_license}")
print("-" * 50)
print(f"✅ ¿Puede conducir? (edad > 18 and licencia): {can_drive}")
print(f"✅ ¿Es joven? (edad < 25 or estudiante):     {is_young}")
```

## 🧪 Operador de Negación (`not`) y Evaluación en Cortocircuito (*Short-circuiting*)

### 💡 Explicación de Conceptos Avanzados:
1. **Negación Lógica (`not`):** Permite verificar condiciones inversas (ej. no estar registrado, no ser estudiante).
2. **Evaluación de Cortocircuito (*Short-circuiting*):**
   - En un `and`, si el primer término es `False`, Python **no evalúa** el segundo término porque ya sabe que el resultado final será `False`.
   - En un `or`, si el primer término es `True`, Python **no evalúa** el segundo término porque ya sabe que el resultado final será `True`.

```python
# ================================================================================
# 🔄 2. OPERADOR DE NEGACIÓN (NOT) Y COMBINACIONES
# ================================================================================
print("\n--- 🔄 [PASO 2]: Negación Lógica con 'not' ---")

# Invertimos los estados con 'not'
is_not_student: bool = not is_student
does_not_have_license: bool = not has_license

print(f"✅ Original 'is_student': {is_student} | Negado ('not is_student'): {is_not_student}")
print(f"✅ Original 'has_license': {has_license} | Negado ('not has_license'): {does_not_have_license}")

# ================================================================================
# 🎯 3. CASO DE USO INTEGRADO (REGLAS DE NEGOCIO)
# ================================================================================
print("\n--- 🎯 [PASO 3]: Validación Compuesta de Descuento ---")

# Un usuario recibe un descuento si es estudiante O si no tiene licencia Y es mayor de 18 años
is_eligible_for_discount: bool = is_student or (not has_license and user_age > 18)

print(f"✅ ¿Elegible para descuento especial?: {is_eligible_for_discount}")

# --------------------------------------------------------------------------------
# 🏁 [PASO 4]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Módulo de operadores lógicos finalizado con éxito.")
print("=" * 80)
```
