# ⚙️ Fundamentos de Python: Operadores de Asignación Combinados (`Augmented Assignment Operators`)

| Propiedad | Detalle |
| :--- | :--- |
| 📍 **Módulo** | Fundamentos de Python |
| 🎯 **Objetivo** | Comprender la sintaxis reducida para realizar operaciones aritméticas sobre una variable y reasignar su valor resultante en el mismo paso (`+=`, `-=`, `*=`, `/=`, `//=`, `%=`, `**=`). |
| 🛠️ **Conceptos** | `In-place Operations`, `Augmented Assignment`, `+=`, `-=`, `*=`, `/=`, `//=`, `%=`, `**=` |

---

### 💡 Explicación de la Arquitectura
* **`x += y`**: Equivalente a `x = x + y` (Suma e in-place assignment).
* **`x -= y`**: Equivalente a `x = x - y` (Resta e in-place assignment).
* **`x *= y`**: Equivalente a `x = x * y` (Multiplicación e in-place assignment).
* **`x /= y`**: Equivalente a `x = x / y` (División e in-place assignment, convierte la variable a `float`).
* **`x //= y`**: Equivalente a `x = x // y` (División entera e in-place assignment).
* **`x %= y`**: Equivalente a `x = x % y` (Cálculo del módulo e in-place assignment).
* **`x **= y`**: Equivalente a `x = x ** y` (Potenciación e in-place assignment).

```python
# [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Operadores de Asignación Combinados")
print("=" * 80)

# ================================================================================
# 🔢 1. OPERADORES BÁSICOS DE ASIGNACIÓN AUMENTADA (+=, -=, *=, /=)
# ================================================================================
# Modificación del valor de una variable conservando su identificador.
# ================================================================================

# [1] Variable de estado inicial (Puntaje de un juego o contador)
current_score: float = 100.0

print("\n--- 🚀 [PASO 1]: Operaciones Básicas In-place (Base inicial: 100.0) ---")

# Asignación combinada: Suma (+-)
current_score += 50.0
print(f"✅ [+= 50.0]:  Puntaje tras suma acumulada:          {current_score}")

# Asignación combinada: Resta (-=)
current_score -= 20.0
print(f"✅ [-= 20.0]:  Puntaje tras resta acumulada:         {current_score}")

# Asignación combinada: Multiplicación (*=)
current_score *= 2.0
print(f"✅ [*= 2.0]:   Puntaje tras multiplicación doble:     {current_score}")

# Asignación combinada: División (/=)
current_score /= 4.0
print(f"✅ [/= 4.0]:   Puntaje tras división entre 4:         {current_score}")
```

## 📐 Asignaciones Combinadas Avanzadas (`//=`, `%=`, `**=`)

### 💡 Explicación de los Operadores Avanzados:
1. **División Entera In-place (`//=`):** Realiza la división descartando la parte decimal y actualiza la variable.
2. **Módulo In-place (`%=`):** Calcula el residuo de la división y actualiza la variable.
3. **Potencia In-place (`**=`):** Eleva la variable a la potencia especificada y actualiza su valor.

```python
# ================================================================================
# ⚙️ 2. OPERADORES AVANZADOS DE ASIGNACIÓN AUMENTADA (//=, %=, **=)
# ================================================================================
print("\n--- ⚙️ [PASO 2]: Operaciones Avanzadas In-place ---")

# Variable numérica para pruebas avanzadas
total_units: int = 27

print(f"📋 Valor base inicial de 'total_units': {total_units}")

# Asignación combinada: División entera (//=)
total_units //= 4
print(f"✅ [//= 4]:    División entera de 27 // 4:            {total_units}")

# Asignación combinada: Potencia (**=)
total_units **= 2
print(f"✅ [**= 2]:    Elevado al cuadrado (6 ** 2):          {total_units}")

# Asignación combinada: Módulo (%=)
total_units %= 5
print(f"✅ [%= 5]:     Residuo del módulo 36 % 5:              {total_units}")

# --------------------------------------------------------------------------------
# 🏁 [PASO 3]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Módulo de operadores de asignación combinados completado.")
print("=" * 80)
```
