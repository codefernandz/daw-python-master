---

### 💡 ¿Qué es la Programación Orientada a Objetos?

La **Programación Orientada a Objetos** (*Object-Oriented Programming - OOP*) es un paradigma de programación que nos permite estructurar el código agrupando datos (**atributos**) y comportamientos (**métodos**) en entidades individuales llamadas **objetos**.

Los objetos en software modelan elementos del mundo real (por ejemplo: un usuario, una transacción financiera, un mueble o un empleado), dotándolos de un estado interno y acciones que pueden realizar.

---

### 🧩 Conceptos Clave de POO:

1. 🏗️ **Clase (*Class*):** Es el plano, plantilla o esquema a partir del cual se construyen los objetos. Define la estructura común que compartirán sus instancias (sus atributos y métodos disponibles).
2. 🏷️ **Atributos (*Attributes*):** Son las variables o características asociadas a un objeto (ej. `name`, `position`, `salary`).
3. ⚙️ **Métodos (*Methods*):** Son las funciones definidas dentro de una clase que representan las acciones o comportamientos que un objeto puede realizar sobre sus datos o el entorno.
4. 🧱 **Objeto e Instancia (*Object / Instance*):** Es el ejemplar concreto creado a partir de una clase, con sus propios valores específicos asignados a los atributos.

> 📌 **Convención de Nombres:** Por convención estándar (**PEP 8**), el nombre de las clases debe escribirse en formato **PascalCase** (primera letra de cada palabra en mayúscula, ej: `Employee`, `DatabaseConnection`).

---

### 📊 Comparativa Conceptual:

| Concepto | Metáfora | Ejemplo en Código |
| :--- | :--- | :--- |
| **Clase** | Plano de construcción de una casa | `class Employee:` |
| **Atributo** | Color de paredes, número de habitaciones | `self.name`, `self.salary` |
| **Método** | Encender luces, abrir puertas | `def get_details(self):` |
| **Objeto / Instancia** | La casa física construida en la calle X | `employee_1 = Employee(...)` |

```python
from typing import Any

# 🛰️ [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 PROGRAMACIÓN ORIENTADA A OBJETOS | Introducción a Clases e Instancias")
print("=" * 80)

# ================================================================================
# 🧪 1. DEFINICIÓN Y MODELADO DE UNA CLASE
# ================================================================================
# Definición de la clase Employee representando un empleado del mundo real.
# ================================================================================

print("\n--- 🚀 [PASO 1]: Definición de la Clase Employee ---")


class Employee:
    """Representa a un empleado dentro del sistema de la empresa.

    Attributes:
        name (str): Nombre completo del empleado.
        position (str): Cargo o puesto de trabajo.
        salary (float): Salario mensual en USD.
    """

    # ⚙️ [1] Método constructor (__init__): Inicializa los atributos del objeto
    def __init__(self, name: str, position: str, salary: float) -> None:
        """Inicializa una nueva instancia de la clase Employee.

        Args:
            name (str): Nombre completo del empleado.
            position (str): Cargo o puesto de trabajo.
            salary (float): Salario mensual asignado.
        """
        self.name: str = name
        self.position: str = position
        self.salary: float = salary

    # ⚙️ [2] Método de instancia: Comportamiento para mostrar detalles del empleado
    def get_details(self) -> str:
        """Devuelve una representación formateada con la información del empleado.

        Returns:
            str: Ficha descriptiva del empleado.
        """
        return f"Empleado: {self.name} | Puesto: {self.position} | Salario: ${self.salary:,.2f}"

    # ⚙️ [3] Método de instancia: Modifica el estado interno (Incremento salarial)
    def apply_raise(self, percentage: float) -> None:
        """Aplica un incremento porcentual al salario del empleado.

        Args:
            percentage (float): Porcentaje de aumento (ej. 0.10 para un 10%).
        """
        if percentage > 0:
            increase_amount: float = self.salary * percentage
            self.salary += increase_amount
            print(f"  ✅ Salario de {self.name} incrementado en un {percentage*100:.0f}%. Nuevo salario: ${self.salary:,.2f}")
        else:
            print(f"  ❌ Porcentaje inválido para el aumento de {self.name}.")


print("✅ Clase 'Employee' definida correctamente.")
```

## 🧱 Instanciación y Creación de Objetos

### 💡 Crear Instancias a Partir de una Clase:
Para instanciar un objeto invocamos el nombre de la clase como si fuera una función, pasando los argumentos requeridos por el constructor `__init__()`.

```python
# ================================================================================
# 📊 2. INSTANCIACIÓN DE OBJETOS Y ACCESO A ATRIBUTOS Y MÉTODOS
# ================================================================================
print("\n--- 📊 [PASO 2]: Creación de Instancias (Objetos) ---")

# 🔌 [1] Instanciación del primer objeto (Ana)
employee_1: Employee = Employee(name="Ana López", position="Software Engineer", salary=3500.0)

# 🔌 [2] Instanciación del segundo objeto (Juan)
employee_2: Employee = Employee(name="Juan Pérez", position="Data Analyst", salary=2800.0)

print(f"🔍 Tipo del objeto 'employee_1': {type(employee_1)}")
print(f"🔍 Identificador en memoria:    {hex(id(employee_1))}")
print("-" * 50)

# 🔌 [3] Acceso directo a atributos de instancia
print(f"📋 Atributos directos de employee_1 -> Nombre: {employee_1.name}, Salario: ${employee_1.salary}")

# 🔌 [4] Invocación de métodos de instancia
print(f"  🔹 Ficha 1: {employee_1.get_details()}")
print(f"  🔹 Ficha 2: {employee_2.get_details()}")
```

## ⚙️ Modificación de Estado e Interacción entre Objetos

### 💡 Comportamientos Dinámicos:
Los métodos de instancia no solo leen datos; también pueden alterar el estado de los atributos internos (`self.salary`) aplicando reglas de negocio.

```python
# ================================================================================
# 🧪 3. MODIFICACIÓN DEL ESTADO DE UN OBJETO
# ================================================================================
print("\n--- 🧪 [PASO 3]: Modificación de Estado Mediante Métodos ---")

print(f"📋 Salario inicial de {employee_1.name}: ${employee_1.salary:,.2f}")
print("-" * 50)

# 🔌 Aumento salarial del 10%
employee_1.apply_raise(percentage=0.10)

# Verificación de la actualización del estado
print(f"  ✅ Estado actualizado: {employee_1.get_details()}")
```

## 🛒 Ejercicios Prácticos Integrados

### 🧪 Ejercicio 1: Modelado de Entidades del Mundo Real (Muebles / Producto)
Creación de una clase `Furniture` para representar piezas de mobiliario con características físicas y acciones como abrir o cerrar componentes.

```python
# ================================================================================
# 🧪 4. EJERCICIO PRÁCTICO: CLASE FURNITURE (MOBILIARIO)
# ================================================================================
print("\n--- 🧪 [PASO 4]: Ejercicio Práctico - Clase Furniture ---")


class Furniture:
    """Representa una pieza de mobiliario con propiedades físicas y puertas.

    Attributes:
        furniture_type (str): Tipo de mueble (ej. 'Armario', 'Escritorio').
        color (str): Color principal.
        door_count (int): Número de puertas que posee.
        is_open (bool): Estado actual de las puertas (Abiertas/Cerradas).
    """

    def __init__(self, furniture_type: str, color: str, door_count: int) -> None:
        """Inicializa una instancia de Furniture."""
        self.furniture_type: str = furniture_type
        self.color: str = color
        self.door_count: int = door_count
        self.is_open: bool = False  # Las puertas inician cerradas por defecto

    def open_doors(self) -> None:
        """Abre las puertas del mueble si posee alguna."""
        if self.door_count == 0:
            print(f"  ❌ El mueble de tipo '{self.furniture_type}' no tiene puertas para abrir.")
        elif self.is_open:
            print(f"  ⚠️ Las puertas del mueble '{self.furniture_type}' ya están abiertas.")
        else:
            self.is_open = True
            print(f"  ✅ Se han abierto las {self.door_count} puertas del mueble '{self.furniture_type}'.")

    def close_doors(self) -> None:
        """Cierra las puertas del mueble."""
        if self.is_open:
            self.is_open = False
            print(f"  ✅ Se han cerrado las puertas del mueble '{self.furniture_type}'.")
        else:
            print(f"  ⚠️ Las puertas del mueble '{self.furniture_type}' ya estaban cerradas.")


# 🔌 Instanciación y prueba del objeto Furniture
wardrobe: Furniture = Furniture(furniture_type="Armario ropero", color="Madera Roble", door_count=3)

print(f"📋 Mueble creado: {wardrobe.furniture_type} | Color: {wardrobe.color} | Puertas: {wardrobe.door_count}")
print("-" * 50)

# Interacción con el objeto
wardrobe.open_doors()
wardrobe.open_doors()  # Reintento para verificar control de estado
wardrobe.close_doors()

# --------------------------------------------------------------------------------
# 🏁 [PASO 5]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Cuaderno de Introducción a la POO finalizado con éxito.")
print("=" * 80)
```
