---

### 💡 Instanciación de Objetos

La **instanciación** es el proceso de crear un objeto concreto (*ejemplar o instancia*) en memoria a partir de la plantilla definida por una clase.

#### 📐 Sintaxis Básica:
```python
object_name = ClassName(attribute_1, attribute_2, attribute_3)
```

- **`ClassName`**: Nombre de la clase escrita en convención PascalCase.
- **`attribute_N`**: Argumentos requeridos por el método constructor `__init__()`.
- **`object_name`**: Variable que almacena la referencia en memoria del objeto instanciado.

---

### 🔍 Métodos de Acceso a los Atributos de un Objeto

Una vez creado el objeto, disponemos de dos formas principales de acceder a su estado interno:

- **📌 Notación de Punto (`object.attribute`)**: Acceso directo e individual al valor de un atributo específico de la instancia.
- **🗂️ Atributo Especial `__dict__` (`object.__dict__`)**: Devuelve un diccionario nativo de Python (`dict`) que contiene todos los atributos de instancia asociados a sus respectivos valores actuales (`{'attribute_name': value}`).

---

### 📊 Comparativa de Métodos de Inspección

| Método | Sintaxis | Retorno | Uso Recomendado |
| :--- | :--- | :--- | :--- |
| Notación de Punto | `employee.position` | Valor del atributo (str, int, etc.) | Lectura/escritura puntual en la lógica de negocio. |
| Atributo `__dict__` | `employee.__dict__` | `dict[str, Any]` | Serialización, depuración y registro en logs. |

````python
from typing import Any

# 🛰️ [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 PROGRAMACIÓN ORIENTADA A OBJETOS | Instanciación e Inspección con __dict__")
print("=" * 80)

# ================================================================================
# 🧪 1. DEFINICIÓN DE CLASE E INSTANCIACIÓN DE UN OBJETO
# ================================================================================
# Creación del objeto 'juan' a partir de la clase Employee.
# ================================================================================

print("\n--- 🚀 [PASO 1]: Instanciación del Objeto 'juan' ---")


class Employee:
    """Representa a un empleado dentro de una organización.

    Attributes:
        name (str): Nombre del empleado.
        age (int): Edad del empleado.
        position (str): Cargo o posición laboral.
        salary (float): Salario asignado.
    """

    def __init__(self, name: str, age: int, position: str, salary: float) -> None:
        """Inicializa los atributos de instancia del empleado."""
        self.name: str = name
        self.age: int = age
        self.position: str = position
        self.salary: float = salary


# 🔌 [1] Instanciación del nuevo objeto 'juan'
juan: Employee = Employee(name="Juan", age=30, position="Gerente", salary=50000.0)

print(f"✅ Objeto creado exitosamente: {juan}")
print(f"🔍 Referencia de memoria:      {hex(id(juan))}")
````

## 📊 Lectura e Inspección de Atributos

### 💡 Acceso Individual vs. Inspección Completa (`__dict__`):
Demostración de lectura puntual con notación de punto y extracción de todos los atributos en formato `dict`.

````python
# ================================================================================
# 📊 2. LECTURA DE ATRIBUTOS DIRECTA Y VÍA __DICT__
# ================================================================================
print("\n--- 📊 [PASO 2]: Acceso a Atributos de la Instancia ---")

# 🔌 [1] Acceso a atributos individuales usando notación de punto
print(f"📋 Nombre:   {juan.name}")
print(f"📋 Posición: {juan.position}")
print(f"📋 Salario:  ${juan.salary:,.2f}")

print("-" * 50)

# 🔌 [2] Inspección de la totalidad de atributos mediante __dict__
juan_attributes_dict: dict[str, Any] = juan.__dict__

print(f"✅ Todos los atributos de 'juan' (__dict__):")
print(f"  🔍 {juan_attributes_dict}")
print(f"  🔍 Tipo devuelto por __dict__: {type(juan_attributes_dict)}")
````

## 🛒 Ejercicio Práctico Integrado

### 🧪 Ejercicio: Modificación de Atributos e Inspección Dinámica
Demostración de cómo la notación de punto permite actualizar el estado interno y cómo dichos cambios se reflejan inmediatamente en el diccionario `__dict__`.

````python
# ================================================================================
# 🧪 3. EJERCICIO PRÁCTICO: DUALIDAD ENTRE ATRIBUTOS Y DICCIONARIO
# ================================================================================
print("\n--- 🧪 [PASO 3]: Modificación de Estado y Reflejo en __dict__ ---")

print("📋 Estado inicial de 'juan':")
print(f"  • {juan.__dict__}")

print("-" * 50)

# ⚙️ Modificación de atributos mediante notación de punto
juan.position = "Director General"
juan.salary = 65000.0

print("✅ Estado actualizado tras promoción laboral:")
print(f"  🔹 Posición actualizada (punto): {juan.position}")
print(f"  🔹 Salario actualizado (punto):  ${juan.salary:,.2f}")

print("-" * 50)

# 🔌 Comprobación de la actualización automática dentro de __dict__
print("📋 Nuevo estado reflejado en __dict__:")
for attr_name, attr_value in juan.__dict__.items():
    print(f"  • {attr_name}: {attr_value}")

# --------------------------------------------------------------------------------
# 🏁 [PASO 4]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Cuaderno de Instanciación e Inspección finalizado con éxito.")
print("=" * 80)
````
