### 💡 Métodos de Instancia en Python

Los **métodos de instancia** son funciones definidas dentro del cuerpo de una clase que representan los comportamientos o acciones que un objeto puede realizar.

#### 📐 Características Principales:
1. **Acceso al Estado:** Tienen acceso directo a los atributos del objeto a través del primer parámetro implícito **`self`**.
2. **Ubicación:** Se definen a continuación del método constructor `__init__()`.
3. **Modificación de Estado:** Permiten alterar los valores de los atributos de la instancia activa de forma controlada.

#### 📐 Sintaxis de Invocación:
```python
object_name.method_name(argument_1, argument_2)
```

---

### 📊 Comparativa: Atributos vs. Métodos

| Componente | Definición | Función Principal | Ejemplo en Código |
| :--- | :--- | :--- | :--- |
| Atributo | Variable asociada a la instancia | Almacenar datos/estado | `self.salary` |
| Método | Función asociada a la clase | Ejecutar acciones / Alterar estado | `juan.raise_salary(10.0)` |
```

````python
from typing import Any

# 🛰️ [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 PROGRAMACIÓN ORIENTADA A OBJETOS | Métodos de Instancia")
print("=" * 80)

# ================================================================================
# 🧪 1. DEFINICIÓN DE LA CLASE CON MÉTODOS DE INSTANCIA
# ================================================================================
# Implementación de la clase Employee extendida con métodos de negocio.
# ================================================================================

print("\n--- 🚀 [PASO 1]: Definición de la Clase Employee y sus Métodos ---")


class Employee:
    """Representa a un empleado dentro de la organización.

    Attributes:
        name (str): Nombre del empleado.
        age (int): Edad del empleado.
        position (str): Cargo o puesto ocupado.
        salary (float): Salario actual en USD.
    """

    def __init__(self, name: str, age: int, position: str, salary: float) -> None:
        """Inicializa una nueva instancia de Employee."""
        self.name: str = name
        self.age: int = age
        self.position: str = position
        self.salary: float = salary

    # ⚙️ [1] Método de instancia para incrementar el salario según un porcentaje
    def raise_salary(self, percentage: float) -> None:
        """Aumenta el salario del empleado en un determinado porcentaje.

        Args:
            percentage (float): Porcentaje de aumento (ej. 10.0 para un 10%).
        """
        # ⚙️ [2] Calcula el nuevo salario aumentado por el porcentaje dado
        self.salary *= (1 + percentage / 100)

        # ⚙️ [3] Redondea el salario resultante a dos decimales
        self.salary = round(self.salary, 2)

        # ⚙️ [4] Emisión de traza visual de éxito
        print(f"  ✅ ¡El salario de {self.name} ha subido! Nuevo salario: ${self.salary:,.2f}")


print("✅ Clase 'Employee' con método 'raise_salary()' definida correctamente.")
````

## 🧱 Instanciación e Invocación del Método de Instancia

### 💡 Caso Práctico: Aumento Salarial
Instanciación del objeto `juan` y ejecución del método `raise_salary()` para verificar la actualización de su salario.

````python
# ================================================================================
# 📊 2. INSTANCIACIÓN Y EJECUCIÓN DE MÉTODOS
# ================================================================================
print("\n--- 📊 [PASO 2]: Instanciación de 'juan' e Invocación de raise_salary() ---")

# 🔌 [1] Instanciación del objeto 'juan'
juan: Employee = Employee(name="Juan", age=30, position="Gerente", salary=50000.0)

print(f"📋 Estado inicial -> Nombre: {juan.name} | Salario: ${juan.salary:,.2f}")
print("-" * 50)

# 🔌 [2] Invocación del método para incrementar el salario en un 10%
print("🚀 Aplicando incremento salarial del 10%...")
juan.raise_salary(percentage=10.0)

print("-" * 50)
print(f"📋 Estado final verificado -> Salario actualizado: ${juan.salary:,.2f}")
````

## 🛒 Ejercicio Práctico Integrado

### 🧪 Ejercicio: Métodos con Validaciones e Interacción Completa
Extensión de métodos para incluir validaciones de parámetros (ej. evitar incrementos negativos o nulos) y garantizar la integridad de los datos.

````python
# ================================================================================
# 🧪 3. EJERCICIO PRÁCTICO: VALIDACIÓN DE REGLAS DE NEGOCIO EN MÉTODOS
# ================================================================================
print("\n--- 🧪 [PASO 3]: Validación de Parámetros dentro de Métodos ---")


class RobustEmployee(Employee):
    """Extensión de Employee con validación estricta de parámetros en métodos."""

    def raise_salary(self, percentage: float) -> None:
        """Aumenta el salario del empleado validando que el porcentaje sea positivo.

        Args:
            percentage (float): Porcentaje de aumento.
        """
        if percentage <= 0:
            print(f"  ❌ Error: El porcentaje de aumento ({percentage}%) debe ser mayor a cero.")
            return

        super().raise_salary(percentage)


# 🔌 Instanciación de objeto para pruebas de validación
ana: RobustEmployee = RobustEmployee(name="Ana López", age=28, position="Líder Técnica", salary=60000.0)

print(f"📋 Salario inicial de {ana.name}: ${ana.salary:,.2f}")
print("-" * 50)

# 🔌 [1] Prueba con porcentaje inválido (negativo)
ana.raise_salary(percentage=-5.0)

# 🔌 [2] Prueba con porcentaje válido (15%)
ana.raise_salary(percentage=15.0)

# --------------------------------------------------------------------------------
# 🏁 [PASO 4]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Cuaderno de Métodos de Instancia finalizado con éxito.")
print("=" * 80)
````
