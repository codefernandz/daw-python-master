### 💡 ¿Qué es la Herencia?

La **herencia** es uno de los pilares fundamentales de la Programación Orientada a Objetos. Permite crear nuevas clases (**subclases o clases hijas**) que heredan automáticamente los atributos y métodos de una clase existente (**superclase o clase madre/padre**).

#### 📐 Principales Ventajas:
1. ♻️ **Reutilización de Código:** Evita la duplicación de lógica común entre múltiples clases relacionadas.
2. 🏗️ **Extensibilidad:** Permite añadir atributos y métodos exclusivos en la clase hija sin modificar la superclase.
3. 🔄 **Especialización:** Permite sobreescribir o adaptar comportamientos heredados según el contexto específico de la subclase.

---

### 📊 Relación Conceptual (Superclase vs. Subclase)

```markdown
            ┌────────────────────────────────────────┐
            │        Superclase: Product             │
            ├────────────────────────────────────────┤
            │ • name: str                            │
            │ • price: float                         │
            │ • sales_platform: str                  │
            │ • vat_rate: float                      │
            │ ────────────────────────────────────── │
            │ + calculate_tax(): float               │
            └──────────────────┬─────────────────────┘
                               │
                               │ (Hereda atributos y métodos)
                               ▼
            ┌────────────────────────────────────────┐
            │       Subclase: Appliance              │
            ├────────────────────────────────────────┤
            │ • energy_certificate: str (Exclusivo)  │
            │ ────────────────────────────────────── │
            │ + evaluate_certificate(): None (Propio)│
            └────────────────────────────────────────┘
```

### 📐 Sintaxis Básica de Herencia:

```python
class ChildClass(ParentClass):
    def __init__(self, parent_attr: str, child_attr: str) -> None:
        # Invoca el constructor de la superclase
        super().__init__(parent_attr)
        # Inicializa atributos exclusivos de la subclase
        self.child_attr = child_attr
```
```

````python
from typing import Any

# 🛰️ [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 PROGRAMACIÓN ORIENTADA A OBJETOS | Herencia y Extensión de Clases")
print("=" * 80)

# ================================================================================
# 🧪 1. DEFINICIÓN DE LA SUPERCLASE (PRODUCT)
# ================================================================================
# Clase base para representar cualquier producto a la venta.
# ================================================================================

print("\n--- 🚀 [PASO 1]: Definición de la Superclase 'Product' ---")


class Product:
    """Representa un producto genérico para la venta.

    Attributes:
        name (str): Nombre comercial del producto.
        price (float): Precio base sin impuestos.
        sales_platform (str): Plataforma de venta (ej. 'online', 'física').
        vat_rate (float): Tasa de IVA aplicada (ej. 0.21 para el 21%).
    """

    def __init__(self, name: str, price: float, sales_platform: str, vat_rate: float) -> None:
        """Inicializa los atributos genéricos del producto."""
        self.name: str = name
        self.price: float = price
        self.sales_platform: str = sales_platform
        self.vat_rate: float = vat_rate

    def calculate_tax(self) -> float:
        """Calcula el monto de impuestos sobre el precio del producto.

        Returns:
            float: Cantidad de impuestos a pagar en USD.
        """
        return self.price * self.vat_rate


print("✅ Superclase 'Product' definida correctamente.")
````

## 🧱 Definición de la Subclase y Uso de `super()`

### 💡 Extensión de Funcionalidad:
La clase `Appliance` (Electrodoméstico) hereda de `Product`, invoca a `super().__init__()` para inicializar los atributos genéricos y añade el atributo `energy_certificate` con su método específico `evaluate_certificate()`.

````python
# ================================================================================
# 🧪 2. DEFINICIÓN DE LA SUBCLASE (APPLIANCE) CON SUPER()
# ================================================================================
print("\n--- 🚀 [PASO 2]: Definición de Subclase 'Appliance' derivando de 'Product' ---")


class Appliance(Product):
    """Representa un electrodoméstico como un tipo específico de producto.

    Attributes:
        energy_certificate (str): Clasificación de eficiencia energética ('A', 'B', ..., 'F').
    """

    def __init__(
        self, 
        name: str, 
        price: float, 
        sales_platform: str, 
        vat_rate: float, 
        energy_certificate: str
    ) -> None:
        """Inicializa un electrodoméstico delegando los datos comunes a la superclase.

        Args:
            name (str): Nombre del electrodoméstico.
            price (float): Precio base.
            sales_platform (str): Canal de venta.
            vat_rate (float): Tasa de IVA aplicada.
            energy_certificate (str): Calificación de certificado energético.
        """
        # ⚙️ [1] Llama al constructor de la clase madre (Product)
        super().__init__(name=name, price=price, sales_platform=sales_platform, vat_rate=vat_rate)

        # ⚙️ [2] Añadimos los atributos específicos de la clase hija
        self.energy_certificate: str = energy_certificate

    def evaluate_certificate(self) -> None:
        """Evalúa el nivel de eficiencia energética del electrodoméstico."""
        if self.energy_certificate == "F":
            print(f"  ❌ El producto '{self.name}' tiene un certificado energético deficiente (Clase F).")
        elif self.energy_certificate in ("A", "A+", "A++", "A+++"):
            print(f"  ✅ El producto '{self.name}' cuenta con una excelente eficiencia energética (Clase {self.energy_certificate}).")
        else:
            print(f"  ℹ️ El producto '{self.name}' tiene una calificación energética estándar (Clase {self.energy_certificate}).")


print("✅ Subclase 'Appliance' definida correctamente.")
````

## 📊 Instanciación y Verificación de Herencia

### 💡 Ejecución Práctica:
Creación de la instancia `washing_machine` para verificar el acceso a atributos/métodos heredados (`calculate_tax()`) y específicos (`evaluate_certificate()`).

````python
# ================================================================================
# 📊 3. INSTANCIACIÓN DE SUBCLASE Y EJECUCIÓN DE MÉTODOS HEREDADOS Y PROPIOS
# ================================================================================
print("\n--- 📊 [PASO 3]: Instanciación y Demostración de Herencia ---")

# 🔌 [1] Instanciamos un objeto de la clase hija (Electrodoméstico)
washing_machine: Appliance = Appliance(
    name="Lavadora Siemens", 
    price=678.0, 
    sales_platform="online", 
    vat_rate=0.21, 
    energy_certificate="F"
)

# 🔌 [2] Acceso a atributos heredados y propios
print(f"📋 Nombre del producto (heredado):     {washing_machine.name}")
print(f"📋 Precio base (heredado):              ${washing_machine.price:,.2f}")
print(f"📋 Certificado energético (exclusivo): {washing_machine.energy_certificate}")

print("-" * 50)

# 🔌 [3] Ejecución del método heredado de la superclase Product
tax_amount: float = washing_machine.calculate_tax()
print(f"  🔹 Impuesto calculado (método heredado 'calculate_tax'): ${tax_amount:,.2f}")

# 🔌 [4] Invocación del método propio de la clase hija Appliance
print("🚀 Evaluación del certificado energético (método propio):")
washing_machine.evaluate_certificate()
````

## 🛒 Ejercicio Práctico Integrado

### 🧪 Ejercicio: Extensión con Múltiples Subclases e Inspección con `isinstance()`
Modelado de una segunda subclase `ElectronicDevice` e inspección de jerarquía mediante la función integrada `isinstance()`.

````python
# ================================================================================
# 🧪 4. EJERCICIO PRÁCTICO: JERARQUÍA E INSPECCIÓN CON ISINSTANCE
# ================================================================================
print("\n--- 🧪 [PASO 4]: Verificación de Jerarquía con isinstance() ---")


class ElectronicDevice(Product):
    """Representa un dispositivo electrónico con garantía extendida."""

    def __init__(self, name: str, price: float, sales_platform: str, vat_rate: float, warranty_months: int) -> None:
        super().__init__(name, price, sales_platform, vat_rate)
        self.warranty_months: int = warranty_months


# 🔌 Instanciación de un dispositivo electrónico
laptop: ElectronicDevice = ElectronicDevice(
    name="Laptop Dell XPS", price=1200.0, sales_platform="física", vat_rate=0.21, warranty_months=24
)

print("📋 Verificación de Tipos e Hitos de Herencia:")
print(f"  • ¿laptop es de tipo ElectronicDevice?: {isinstance(laptop, ElectronicDevice)}")
print(f"  • ¿laptop es de tipo Product?:          {isinstance(laptop, Product)}")
print(f"  • ¿laptop es de tipo Appliance?:        {isinstance(laptop, Appliance)}")

# --------------------------------------------------------------------------------
# 🏁 [PASO 5]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Cuaderno de Herencia finalizado con éxito.")
print("=" * 80)
````
