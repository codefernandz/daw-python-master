---

### 💡 ¿Qué es el Polimorfismo?

El **Polimorfismo** (del griego *"muchas formas"*) es el principio fundamental de la POO que permite a objetos de distintas clases responder a una **misma llamada o mensaje de método** ejecutando comportamientos específicos para cada tipo de objeto.

#### 📐 Sobrescritura de Métodos (*Method Overriding*):
Para lograr el polimorfismo mediante herencia, una **subclase** redefine (*sobrescribe*) un método que ya fue declarado en su **superclase**.

---

### 📊 Comparativa de Comportamiento Polimórfico:

| Clase | Método | Lógica de Cálculo | Resultado con Precio = $1,000, IVA = 21% |
| :--- | :--- | :--- | :--- |
| **`Product`** (Base) | `calculate_price_with_vat()` | `price * (1 + vat_rate)` | **$1,210.00** |
| **`Appliance`** (Subclase) | `calculate_price_with_vat()` | `price * (1 + vat_rate) - 100` (Descuento de $100) | **$1,110.00** |

```markdown
                     Llamada unificada: object.calculate_price_with_vat()
                                           │
                    ┌──────────────────────┴──────────────────────┐
                    │                                             │
                    ▼                                             ▼
       ┌──────────────────────────┐                  ┌──────────────────────────┐
       │     Objeto: Product      │                  │    Objeto: Appliance     │
       ├──────────────────────────┤                  ├──────────────────────────┤
       │ Retorna:                 │                  │ Retorna:                 │
       │ Precio + IVA Estándar    │                  │ Precio + IVA - $100      │
       └──────────────────────────┘                  └──────────────────────────┘

````python
from typing import Any

# 🛰️ [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 PROGRAMACIÓN ORIENTADA A OBJETOS | Polimorfismo y Sobrescritura")
print("=" * 80)

# ================================================================================
# 🧪 1. DEFINICIÓN DE LA SUPERCLASE PRODUCT CON MÉTODOS BASE
# ================================================================================
# Definición de la superclase Product y su cálculo estándar de IVA.
# ================================================================================

print("\n--- 🚀 [PASO 1]: Definición de la Superclase 'Product' ---")


class Product:
    """Representa un producto genérico para la venta.

    Attributes:
        name (str): Nombre comercial del producto.
        price (float): Precio base sin impuestos.
        sales_platform (str): Canal de venta.
        vat_rate (float): Tasa de IVA (ej. 0.21 para el 21%).
    """

    def __init__(self, name: str, price: float, sales_platform: str, vat_rate: float) -> None:
        """Inicializa los atributos base del producto."""
        self.name: str = name
        self.price: float = price
        self.sales_platform: str = sales_platform
        self.vat_rate: float = vat_rate

    def calculate_price_with_vat(self) -> float:
        """Calcula el precio final incluyendo la tasa de IVA estándar.

        Returns:
            float: Precio final con impuesto aplicado.
        """
        return round(self.price * (1 + self.vat_rate), 2)


print("✅ Superclase 'Product' definida con el método base 'calculate_price_with_vat()'.")
````

## 🧱 Definición de Subclase y Sobrescritura de Método

### 💡 Comportamiento Especializado en `Appliance`:
Sobrescribimos el método `calculate_price_with_vat()` en la clase `Appliance` para aplicar automáticamente un descuento promocional fijo de **$100.00**.

````python
# ================================================================================
# 🧪 2. SOBRESCRITURA DEL MÉTODO EN LA SUBCLASE APPLIANCE
# ================================================================================
print("\n--- 🚀 [PASO 2]: Definición de la Subclase 'Appliance' con Sobrescritura ---")


class Appliance(Product):
    """Representa un electrodoméstico con un cálculo de IVA con descuento promocional."""

    def __init__(
        self, 
        name: str, 
        price: float, 
        sales_platform: str, 
        vat_rate: float, 
        energy_certificate: str
    ) -> None:
        """Inicializa la instancia delegando a la superclase mediante super()."""
        super().__init__(name=name, price=price, sales_platform=sales_platform, vat_rate=vat_rate)
        self.energy_certificate: str = energy_certificate

    # ⚙️ [1] Sobrescritura Polimórfica: Redefinición del método calculate_price_with_vat
    def calculate_price_with_vat(self) -> float:
        """Calcula el precio con IVA restándole una bonificación especial de $100.

        Returns:
            float: Precio con IVA e incentivo promocional aplicado.
        """
        # ⚙️ Opción A: Reutilizar el cálculo de la superclase mediante super() y aplicar el descuento
        base_price_with_vat: float = super().calculate_price_with_vat()
        final_price: float = base_price_with_vat - 100.0

        return round(final_price, 2)


print("✅ Subclase 'Appliance' definida sobrescribiendo 'calculate_price_with_vat()'.")
````

## 📊 Demostración Práctica de Polimorfismo

### 💡 Procesamiento Homogéneo de Colecciones Polimórficas:
La verdadera potencia del polimorfismo se evidencia cuando iteramos sobre una lista de objetos de distintas clases invocando el **mismo método** sin requerir condicionales de tipo.

````python
# ================================================================================
# 📊 3. EJECUCIÓN Y VERIFICACIÓN DEL COMPORTAMIENTO POLIMÓRFICO
# ================================================================================
print("\n--- 📊 [PASO 3]: Iteración Polimórfica sobre Múltiples Productos ---")

# 🔌 [1] Instanciación de un producto estándar y un electrodoméstico
generic_product: Product = Product(
    name="Escritorio de Oficina", 
    price=1000.0, 
    sales_platform="física", 
    vat_rate=0.21
)

appliance_product: Appliance = Appliance(
    name="Lavadora Siemens", 
    price=1000.0, 
    sales_platform="online", 
    vat_rate=0.21, 
    energy_certificate="F"
)

# 🔌 [2] Colección heterogénea de objetos
catalog: list[Product] = [generic_product, appliance_product]

print("📋 Comparativa de Cálculos de Precio Final (Polimorfismo en Acción):")
print("-" * 65)

# 🔌 [3] Iteración unificada invocando el mismo método en cada elemento
for item in catalog:
    final_price: float = item.calculate_price_with_vat()
    print(f"  • Producto: {item.name:<25} | Clase: {item.__class__.__name__:<10} | Precio Final: ${final_price:,.2f}")
````

## 🛒 Ejercicio Práctico Integrado

### 🧪 Ejercicio: Extensión del Catálogo con Polimorfismo Adicional
Creación de una subclase `DigitalBook` con **0% de IVA** para reforzar el concepto de sobrescritura de comportamientos según la lógica de negocio.

````python
# ================================================================================
# 🧪 4. EJERCICIO PRÁCTICO: SUBCLASE DIGITALBOOK
# ================================================================================
print("\n--- 🧪 [PASO 4]: Extensión con la Clase 'DigitalBook' ---")


class DigitalBook(Product):
    """Representa un libro en formato digital exento de IVA."""

    def __init__(self, name: str, price: float, sales_platform: str, file_format: str) -> None:
        # Se pasa un IVA de 0.0 al constructor base
        super().__init__(name=name, price=price, sales_platform=sales_platform, vat_rate=0.0)
        self.file_format: str = file_format

    def calculate_price_with_vat(self) -> float:
        """Los libros digitales no devengan IVA."""
        print(f"  ℹ️ Aplicando exención fiscal para el libro digital '{self.name}'.")
        return self.price


# 🔌 Instanciación e inclusión en el catálogo general
ebook: DigitalBook = DigitalBook(
    name="Python Clean Code", 
    price=35.0, 
    sales_platform="online", 
    file_format="EPUB"
)

catalog.append(ebook)

print("\n📋 Recorrido completo del Catálogo Polimórfico:")
print("-" * 65)

for item in catalog:
    price: float = item.calculate_price_with_vat()
    print(f"  ✅ [{item.__class__.__name__}] {item.name}: ${price:,.2f}")

# --------------------------------------------------------------------------------
# 🏁 [PASO 5]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Cuaderno de Polimorfismo finalizado con éxito.")
print("=" * 80)
````
