### 💡 El Método Constructor `__init__()` y el Parámetro `self`

En Python, el método constructor es una función especial dunder (*Double Underscore*) llamada **`__init__()`**. Su propósito principal es inicializar el estado del objeto en el momento exacto de la instanciación.

#### 📐 Sintaxis Básica:
```python
class ClassName:
    def __init__(self, attribute_1: type, attribute_2: type) -> None:
        self.attribute_1 = attribute_1
        self.attribute_2 = attribute_2
```

- **`__init__()`**: Método invocado automáticamente cuando se ejecuta `ClassName(...)`.
- **`self`**: Referencia explícita a la instancia de la clase que se está creando o manipulando. Permite asociar los parámetros recibidos a la estructura interna del objeto mediante `self.attribute_name`.
- **Atributos de Instancia (`self.variable`)**: Variables asociadas individualmente a cada objeto creado a partir de la clase.

---

### 🔍 ¿Por qué es Necesario `self`?

Cuando creas múltiples objetos a partir de la misma clase (ej. `employee_1` y `employee_2`), Python necesita saber a qué objeto en memoria pertenece cada atributo. `self` actúa como el enlace dinámico a la dirección de memoria de la instancia activa.

```python
Employee Class Definition
           │
           ├───> Instancia 1 (self = 0x7f8a) ───> self.name = "Ana"
           │
           └───> Instancia 2 (self = 0x7f8b) ───> self.name = "Juan"
```

````python
from typing import Any

# 🛰️ [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 PROGRAMACIÓN ORIENTADA A OBJETOS | Método Constructor (__init__) y self")
print("=" * 80)

# ================================================================================
# 🧪 1. DEFINICIÓN DEL CONSTRUCTOR Y ATRIBUTOS DE INSTANCIA
# ================================================================================
# Implementación de la clase Employee con docstring norma Google y Type Hints.
# ================================================================================

print("\n--- 🚀 [PASO 1]: Definición de la Clase Employee con Constructor ---")


class Employee:
    """Representa a un empleado dentro de la organización.

    Attributes:
        name (str): Nombre completo del empleado.
        age (int): Edad del empleado en años.
        position (str): Puesto o cargo asignado.
        salary (float): Salario base en USD.
    """

    def __init__(self, name: str, age: int, position: str, salary: float) -> None:
        """Inicializa los atributos de instancia de un nuevo empleado.

        Args:
            name (str): Nombre completo del empleado.
            age (int): Edad del empleado.
            position (str): Cargo dentro de la empresa.
            salary (float): Salario asignado.
        """
        # ⚙️ Vinculación de parámetros a atributos de instancia mediante self
        self.name: str = name
        self.age: int = age
        self.position: str = position
        self.salary: float = salary


print("✅ Clase 'Employee' definida correctamente con su método constructor.")
````

## 🧱 Instanciación y Verificación de Atributos Independientes

### 💡 Asignación de Estado Interno:
Al invocar `Employee(...)`, Python ejecuta `__init__()` internamente pasando la nueva instancia en el primer argumento (`self`).

````python
# ================================================================================
# 📊 2. INSTANCIACIÓN Y ACCESO A ATRIBUTOS
# ================================================================================
print("\n--- 📊 [PASO 2]: Instanciación de Objetos y Lectura de Atributos ---")

# 🔌 [1] Creación de dos instancias con datos independientes
employee_1: Employee = Employee(name="Ana López", age=28, position="Software Engineer", salary=3500.0)
employee_2: Employee = Employee(name="Juan Pérez", age=35, position="Data Analyst", salary=2800.0)

print("📋 Ficha del Empleado 1:")
print(f"  • Nombre:   {employee_1.name}")
print(f"  • Edad:     {employee_1.age} años")
print(f"  • Posición: {employee_1.position}")
print(f"  • Salario:  ${employee_1.salary:,.2f}")

print("-" * 50)

print("📋 Ficha del Empleado 2:")
print(f"  • Nombre:   {employee_2.name}")
print(f"  • Edad:     {employee_2.age} años")
print(f"  • Posición: {employee_2.position}")
print(f"  • Salario:  ${employee_2.salary:,.2f}")
````

## 🛒 Ejercicio Práctico Integrado

### 🧪 Ejercicio: Inicialización con Parámetros por Defecto y Validación de Estado
Modelado de una clase `BankAccount` que demuestra cómo el constructor `__init__()` puede recibir **valores por defecto** e inicializar atributos calculados o protegidos.

````python
# ================================================================================
# 🧪 3. EJERCICIO PRÁCTICO: CLASE BANKACCOUNT CON PARÁMETROS OPCIONALES
# ================================================================================
print("\n--- 🧪 [PASO 3]: Ejercicio Práctico - Constructor con Valores Por Defecto ---")


class BankAccount:
    """Representa una cuenta bancaria con saldo e historial de cuenta.

    Attributes:
        account_holder (str): Nombre del titular de la cuenta.
        balance (float): Salario/saldo disponible en la cuenta.
        currency (str): Moneda de la cuenta ('USD', 'EUR', etc.).
        is_active (bool): Estado operativo de la cuenta.
    """

    def __init__(
        self, 
        account_holder: str, 
        initial_balance: float = 0.0, 
        currency: str = "USD"
    ) -> None:
        """Inicializa una cuenta bancaria con saldo y moneda configurable.

        Args:
            account_holder (str): Nombre del titular.
            initial_balance (float, optional): Saldo inicial. Por defecto 0.0.
            currency (str, optional): Código de moneda. Por defecto 'USD'.
        """
        self.account_holder: str = account_holder
        self.balance: float = initial_balance
        self.currency: str = currency
        self.is_active: bool = True  # Atributo con valor inicial fijo (no pasado por parámetro)


# 🔌 [1] Creación de cuenta con saldo inicial
account_1: BankAccount = BankAccount(account_holder="María Gómez", initial_balance=1500.0)

# 🔌 [2] Creación de cuenta usando parámetros por defecto
account_2: BankAccount = BankAccount(account_holder="Carlos Ruiz")

print("📋 Detalles de Cuentas Bancarias Creadas:")
print(f"  🔹 Titular: {account_1.account_holder} | Saldo: ${account_1.balance:,.2f} {account_1.currency} | Activa: {account_1.is_active}")
print(f"  🔹 Titular: {account_2.account_holder} | Saldo: ${account_2.balance:,.2f} {account_2.currency} | Activa: {account_2.is_active}")

# --------------------------------------------------------------------------------
# 🏁 [PASO 4]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Cuaderno sobre Atributos y Constructor finalizado con éxito.")
print("=" * 80)
````
