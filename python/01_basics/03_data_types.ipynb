# ⚙️ Fundamentos de Python: Tipos de Datos Primitivos (`Data Types`)

| Propiedad | Detalle |
| :--- | :--- |
| 📍 **Módulo** | Fundamentos de Python |
| 🎯 **Objetivo** | Explorar la clasificación de los tipos de datos nativos en Python: numéricos (`int`, `float`, `complex`), texto (`str`), booleanos (`bool`) y tipo nulo (`NoneType`). |
| 🛠️ **Conceptos** | `int`, `float`, `complex`, `str`, `bool`, `NoneType`, `Type Hints`, `type()` |

---

### 💡 Explicación de la Arquitectura
* **Números:** `int` (enteros sin decimales), `float` (de punto flotante), `complex` (números complejos con componente real e imaginaria `j`).
* **Cadenas de Texto (`str`):** Cadenas simples, con comillas dobles y multilinea (`"""` o `'''`).
* **Booleanos (`bool`):** Representaciones lógicas de verdad (`True`) o falsedad (`False`).
* **Tipo Nulo (`NoneType`):** Representación explícita de la ausencia de valor (`None`).

```python
# [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Clasificación de Tipos de Datos Nativos")
print("=" * 80)

# ================================================================================
# 🔢 1. TIPOS NUMÉRICOS (NUMERIC TYPES)
# ================================================================================
# Declaración de variables numéricas con Type Hints explícitos.
# ================================================================================

# [1] Entero (int)
negative_number: int = -10

# [2] Decimales de punto flotante (float)
pi_decimal: float = 3.14
euler_decimal: float = 2.718

# [3] Número Complejo (complex) -> 2 real + 3 imaginario
complex_number: complex = 2 + 3j

print("\n--- 🚀 [PASO 1]: Tipos Numéricos (int, float, complex) ---")
print(f"✅ [ENTERO]:   {negative_number:<10} | 📋 Tipo: {type(negative_number)}")
print(f"✅ [FLOAT 1]:  {pi_decimal:<10} | 📋 Tipo: {type(pi_decimal)}")
print(f"✅ [FLOAT 2]:  {euler_decimal:<10} | 📋 Tipo: {type(euler_decimal)}")
print(f"✅ [COMPLEJO]: {str(complex_number):<10} | 📋 Tipo: {type(complex_number)}")
```

## 📝 Cadenas de Texto (`str`), Booleanos (`bool`) y Tipo Nulo (`NoneType`)

### 💡 Explicación de las Estructuras:
1. **Cadenas Multilinea:** Se definen mediante triples comillas (`"""` o `'''`) y preservan los saltos de línea internos.
2. **Objeto `None`:** Representa la ausencia explícita de un valor. Es un singleton de la clase `NoneType`.

```python
# ================================================================================
# 🔤 2. CADENAS DE TEXTO (STRINGS)
# ================================================================================
print("\n--- 🔤 [PASO 2]: Cadenas de Texto (str) ---")

# [1] Cadenas simples y multilinea en inglés
single_line_text: str = "¡Hola, mundo!"
framework_phrase: str = "Python es genial!"
multiline_double: str = """Esta es una cadena de texto
que ocupa varias líneas."""
multiline_single: str = '''También se pueden usar
comillas simples para crear
cadenas de texto multilínea.'''

print(f"✅ [CADENA 1]: {single_line_text:<20} | 📋 Tipo: {type(single_line_text)}")
print(f"✅ [CADENA 2]: {framework_phrase:<20} | 📋 Tipo: {type(framework_phrase)}")
print(f"📋 Tipo de 'multiline_double': {type(multiline_double)}")
print(f"📋 Tipo de 'multiline_single': {type(multiline_single)}")

# ================================================================================
# 🔘 3. BOOLEANOS (BOOLEANS)
# ================================================================================
print("\n--- 🔘 [PASO 3]: Valores Booleanos (bool) ---")

is_active: bool = True
is_completed: bool = False

print(f"✅ [VERDADERO]: {is_active:<10} | 📋 Tipo: {type(is_active)}")
print(f"✅ [FALSO]:     {is_completed:<10} | 📋 Tipo: {type(is_completed)}")

# ================================================================================
# 🚫 4. TIPO NULO (NONETYPE) Y COMPARACIONES
# ================================================================================
print("\n--- 🚫 [PASO 4]: Tipo Nulo (NoneType) e Identidad ---")

null_value: None = None

print(f"✅ [VALOR NULO]: {null_value} | 📋 Tipo: {type(null_value)}")

# 📌 Pruebas de Identidad: Demostración de que None no es equivalente a 0, False o ""
print("\n--- 🔍 Verificación de Identidad de 'None' ---")
print(f"📋 ¿None == 0?           {null_value == 0}")
print(f"📋 ¿None == False?       {null_value == False}")
print(f"📋 ¿None == '' (vacío)?  {null_value == ''}")

# --------------------------------------------------------------------------------
# 🏁 [PASO 5]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Módulo de tipos de datos finalizado correctamente.")
print("=" * 80)
```
