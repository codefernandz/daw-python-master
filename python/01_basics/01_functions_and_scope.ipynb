---

### 💡 Explicación de la Arquitectura
* `def greet() -> None`: Define la función. El *Type Hint* `-> None` indica explícitamente que la función no retorna un valor.
* `message: str`: Variable interna en inglés restringida al ámbito local (*Local Scope*).
* `print(...)`: Emisión de la respuesta formateada hacia la consola.

```python
# [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Definición de Funciones y Ámbito de Variables")
print("=" * 80)

def greet() -> None:
    """
    Imprime un mensaje de saludo inicial en la consola.

    Demuestra la creación de una variable local 'message' de tipo string (str)
    que existe estrictamente dentro del ámbito de ejecución de esta función.
    """
    # [1] Asignación de variable local con tipado explícito
    message: str = "¡Hola, mundo!"

    # [2] Emisión del resultado por pantalla en español
    print(f"✅ [GREET] {message}")
```

## 🧪 Flujo de Ejecución y Pruebas de Ámbito (*Scope Integration*)

### 💡 Explicación de las Pruebas:
1. **Paso 1:** Ejecuta la función `greet()` dentro de su flujo permitido.
2. **Paso 2:** Provoca un error de lectura `NameError` intencional para validar que las variables locales no se fugan hacia el ámbito global (*Global Scope*).

```python
# --------------------------------------------------------------------------------
# 🚀 [PASO 1]: Invocación Estándar de la Función
# --------------------------------------------------------------------------------
print("\n--- 🚀 [PASO 1]: Invocación de la función greet() ---")

greet()

# --------------------------------------------------------------------------------
# 🔍 [PASO 2]: Prueba de Aislamiento de Variables (Global Scope Check)
# --------------------------------------------------------------------------------
print("\n--- 🔍 [PASO 2]: Verificación del Ámbito de la Variable (Scope) ---")

try:
    # ⚠️ Intentamos acceder a 'message' desde fuera (provocará NameError intencional)
    print(message)  # type: ignore
except NameError as error:
    print(f"❌ [ERROR DE ÁMBITO] Intento de acceso a la variable local 'message' desde el ámbito global:")
    print(f"   ↳ Detalle técnico del sistema: {error}")

# --------------------------------------------------------------------------------
# 🏁 [PASO 3]: Cierre y Resumen de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Módulo de prueba finalizado con éxito.")
print("=" * 80)
```
