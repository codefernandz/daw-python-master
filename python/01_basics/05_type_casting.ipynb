# ⚙️ Fundamentos de Python: Conversión de Tipos de Datos (`Type Casting`)

| Propiedad | Detalle |
| :--- | :--- |
| 📍 **Módulo** | Fundamentos de Python |
| 🎯 **Objetivo** | Comprender la conversión explícita (*Casting*) e implícita de tipos de datos utilizando las funciones nativas `int()`, `float()`, `str()` y `bool()`, evaluando el comportamiento de valores veraces (*Truthy*) y falsos (*Falsy*). |
| 🛠️ **Conceptos** | `Explicit Casting`, `Implicit Casting`, `int()`, `float()`, `str()`, `bool()`, `Truthy & Falsy Values` |

---

### 💡 Explicación de la Arquitectura
* **`int(x)`:** Convierte cadenas numéricas enteras o flotantes a tipo entero (`int`). Descarta los decimales sin redondear.
* **`float(x)`:** Convierte cadenas numéricas o enteros a números de punto flotante (`float`).
* **`str(x)`:** Transforma cualquier objeto o tipo de dato en su representación de cadena de texto (`str`).
* **`bool(x)`:** Evalúa el valor de verdad. Devuelve `False` para `0`, `0.0`, cadenas vacías `""` o `None` (*Falsy*). Cualquier otro valor devuelve `True` (*Truthy*).

```python
# [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Conversión de Tipos de Datos (Type Casting)")
print("=" * 80)

# ================================================================================
# 🔢 1. CONVERSIÓN A ENTEROS Y FLOTANTES (int, float)
# ================================================================================
# Transformación de entradas de texto y valores numéricos cruzados.
# ================================================================================

# [1] Cadenas de texto numéricas de entrada
string_number: str = "42"
string_decimal: str = "3.14159"

print("\n--- 🚀 [PASO 1]: Conversión Explícita con int() y float() ---")

# Conversiones explícitas
parsed_integer: int = int(string_number)
parsed_float: float = float(string_decimal)
truncated_integer: int = int(parsed_float)  # Trunca la parte decimal

print(f"✅ [STR -> INT]:   '{string_number}' ({type(string_number).__name__}) -> {parsed_integer} ({type(parsed_integer).__name__})")
print(f"✅ [STR -> FLOAT]: '{string_decimal}' ({type(string_decimal).__name__}) -> {parsed_float} ({type(parsed_float).__name__})")
print(f"✅ [FLOAT -> INT]: {parsed_float} ({type(parsed_float).__name__}) -> {truncated_integer} ({type(truncated_integer).__name__}) [Truncado]")
```

## 🔤 Conversión a Cadena (`str`) y Evaluación Booleana (`bool`)

### 💡 Explicación de los Casos de Uso:
1. **Conversión a Cadenas (`str()`):** Imprescindible para concatenar o imprimir variables numéricas en formatos de texto legibles.
2. **Evaluación Booleana (`bool()`):** Determina la veracidad de una variable. En Python, los valores "vacíos" o nulos se evalúan como `False` (*Falsy*), mientras que aquellos con contenido o números distintos de `0` se evalúan como `True` (*Truthy*).

```python
# ================================================================================
# 🔤 2. CONVERSIÓN A CADENAS DE TEXTO (str)
# ================================================================================
print("\n--- 🔤 [PASO 2]: Conversión de Objetos a Texto con str() ---")

numeric_code: int = 2026
is_active_status: bool = True

# Conversión a texto
code_as_string: str = str(numeric_code)
status_as_string: str = str(is_active_status)

print(f"✅ [INT -> STR]:  {numeric_code} ({type(numeric_code).__name__}) -> '{code_as_string}' ({type(code_as_string).__name__})")
print(f"✅ [BOOL -> STR]: {is_active_status} ({type(is_active_status).__name__}) -> '{status_as_string}' ({type(status_as_string).__name__})")

# ================================================================================
# 🧪 3. EVALUACIÓN BOOLEANA (bool) - TRUTHY Y FALSY
# ================================================================================
print("\n--- 🧪 [PASO 3]: Evaluación Booleana (bool) ---")

# Casos Falsy (Evalúan a False)
zero_number: int = 0
empty_text: str = ""

# Casos Truthy (Evalúan a True)
positive_number: int = 100
negative_number: int = -5
non_empty_text: str = "Hola Python"

print(f"❌ [FALSY - Entero 0]:           bool({zero_number}) -> {bool(zero_number)}")
print(f"❌ [FALSY - Cadena Vacía]:      bool('{empty_text}') -> {bool(empty_text)}")
print(f"✅ [TRUTHY - Entero Positivo]:  bool({positive_number}) -> {bool(positive_number)}")
print(f"✅ [TRUTHY - Entero Negativo]:  bool({negative_number}) -> {bool(negative_number)}")
print(f"✅ [TRUTHY - Cadena con Texto]: bool('{non_empty_text}') -> {bool(non_empty_text)}")

# --------------------------------------------------------------------------------
# 🏁 [PASO 4]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Módulo de conversión de tipos de datos finalizado.")
print("=" * 80)
```
