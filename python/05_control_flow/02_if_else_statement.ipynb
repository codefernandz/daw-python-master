# ⚙️ Fundamentos de Python: Estructura Condicional `if ... else` (`Control Flow - If/Else Bifurcation`)

| Propiedad | Detalle |
| :--- | :--- |
| 📍 **Módulo** | Fundamentos de Python |
| 🎯 **Objetivo** | Dominar la bifurcación lógica bidireccional mediante la estructura `if ... else`, asegurando la ejecución de un bloque alternativo cuando la condición principal evalúa como `False`. |
| 🛠️ **Conceptos** | `Binary Decision Making`, `Fallback Execution`, `Exclusive Code Blocks`, `State Evaluation` |

---

### 💡 Explicación de la Arquitectura
* **Bifurcación Bidireccional (`if ... else`):**
  * Si la `condición` del `if` es `True`, se ejecuta su bloque de código y el bloque `else` **se ignora por completo**.
  * Si la `condición` del `if` es `False`, el bloque `if` se salta y se ejecuta **obligatoriamente** el bloque de código indentado bajo `else`.
* **Reglas de Sintaxis:**
  1. La palabra clave `else` requiere sus propios dos puntos (`:`) al final.
  2. No admite una condición directa en la sentencia `else` (para múltiples condiciones se evalúa `elif`).
  3. Indentación estricta de 4 espacios tanto en el bloque principal como en el bloque de alternativa.

```python
from typing import Any

# [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Flujo de Control - Estructura 'if ... else'")
print("=" * 80)

# ================================================================================
# 🧪 1. EVALUACIÓN Y BIFURCACIÓN DE FLUJO (Caso Positivo / Bloque if)
# ================================================================================
# Escenario donde la condición del if resulta ser verdadera (True).
# ================================================================================

print("\n--- 🚀 [PASO 1]: Evaluación con Cumplimiento de Condición (if = True) ---")

# [1] Declaración de variables iniciales con Type Hints
completed_lessons: int = 3
daily_lessons: int = 5

print(f"📋 Lecciones previa acumuladas: {completed_lessons}")
print(f"📋 Lecciones realizadas hoy:    {daily_lessons}")
print("-" * 50)

# [2] Evaluación del condicional con bifurcación
if daily_lessons > completed_lessons:
    print("🎉 ¡Felicidades! Has completado más lecciones hoy que las que tenías registradas previamente.")
else:
    print("⏳ ¡Sigue adelante! Aún te faltan lecciones por completar para superar el acumulado.")

# [3] Verificación del estado
print(f"📋 Estado total de lecciones completadas: {completed_lessons}")
```

## 🔄 Manejo de Casos de Fallback (`else` Execution)

### 💡 Comportamiento del Bloque Alternativo:
Cuando la evaluación de la condición `if` da como resultado `False`, la instrucción `else` actúa como una red de seguridad que ejecuta la lógica o los mensajes secundarios del programa sin detener el flujo de ejecución.

```python
# ================================================================================
# 🔄 2. EVALUACIÓN DE CASO NEGATIVO (Bloque else)
# ================================================================================
print("\n--- 🔄 [PASO 2]: Evaluación sin Cumplimiento de Condición (if = False) ---")

# [1] Reinicialización de variables para simular un día sin actividad
completed_lessons = 3
daily_lessons = 0

print(f"📋 Lecciones previa acumuladas: {completed_lessons}")
print(f"📋 Lecciones realizadas hoy:    {daily_lessons}")
print("-" * 50)

# [2] Evaluación relacional con mutación condicional
if daily_lessons > 0:
    completed_lessons += daily_lessons
    print("🎉 ¡Felicidades! Has registrado avance en el día de hoy.")
else:
    print("⏳ ¡Sigue adelante! Aún no has completado lecciones el día de hoy.")

# [3] Verificación del estado final
print("-" * 50)
print(f"📋 Estado final de lecciones completadas: {completed_lessons}")

# --------------------------------------------------------------------------------
# 🏁 [PASO 3]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Módulo de estructura 'if ... else' finalizado con éxito.")
print("=" * 80)
```
