# ⚙️ Fundamentos de Python: Sentencias de Control de Bucles (`break`, `continue`, `pass`)

| Propiedad | Detalle |
| :--- | :--- |
| 📍 **Módulo** | Fundamentos de Python |
| 🎯 **Objetivo** | Dominar el uso de las sentencias `break`, `continue` y `pass` para alterar el flujo normal de iteración en bucles `for` y `while` mediante evaluaciones condicionales. |
| 🛠️ **Conceptos** | `Loop Control`, `Early Exit (break)`, `Iteration Skip (continue)`, `Null Statement (pass)`, `Syntactic Placeholder` |

---

### 💡 Explicación de la Arquitectura

Las sentencias de control de flujo permiten modificar el comportamiento habitual de repetición de un bucle:

* **`break`:** Interrumpe e interrumpe **de inmediato** la ejecución del bucle, transfiriendo el flujo de control a la primera instrucción fuera de él.
* **`continue`:** Omite el resto de instrucciones del cuerpo del bucle para la iteración actual y salta **inmediatamente** al siguiente ciclo del bucle.
* **`pass`:** Funciona como un **marcador de posición nulo** (*null operation*). No altera el flujo de ejecución ni salta iteraciones; simplemente evita errores de sintaxis en bloques de código aún no implementados.

```python
from typing import Any

# 🛰️ [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Flujo de Control - break, continue y pass")
print("=" * 80)

# ================================================================================
# 🧪 1. SENTENCIA BREAK: INTERRUPCIÓN INMEDIATA DEL BUCLE
# ================================================================================
# Suma de números pares hasta encontrar el primer número impar.
# ================================================================================

print("\n--- 🚀 [PASO 1]: Uso de 'break' (Salida Anticipada) ---")

# 📋 Declaración de lista de números enteros
numbers_list: list[int] = [2, 4, 6, 8, 10, 3, 12, 14]
even_sum: int = 0

print(f"📋 Lista analizada: {numbers_list}")
print("-" * 50)

# 🔌 [1] Iteración con interrupción break al detectar impar
for number in numbers_list:
    if number % 2 != 0:
        print(f"⚠️ [BREAK]: Se detectó el número impar {number}. Interrumpiendo el bucle...")
        break
    even_sum += number

print(f"✅ La suma de los números pares previos al primer impar es: {even_sum}")
```

## ⏭️ Sentencia `continue`: Omisión de Iteración Actual

### 💡 Filtrado de Ciclos:
La sentencia `continue` descarta cualquier línea de código que se encuentre por debajo de ella **únicamente** en el ciclo en curso, pasando de inmediato al siguiente elemento del iterable.

```python
# ================================================================================
# 📊 2. SENTENCIA CONTINUE: SALTO A LA SIGUIENTE ITERACIÓN
# ================================================================================
print("\n--- 📊 [PASO 2]: Uso de 'continue' (Omisión de Elementos) ---")

# 📋 Declaración de lista para acumular impares
numbers_list = [1, 3, 5, 2, 7, 9, 4, 11]
odd_sum: int = 0

print(f"📋 Lista analizada: {numbers_list}")
print("-" * 50)

# 🔌 [1] Iteración omitiendo números pares mediante continue
for number in numbers_list:
    if number % 2 == 0:
        print(f"⏩ [CONTINUE]: Omitiendo el número par {number}")
        continue
    odd_sum += number

print(f"✅ La suma total de los números impares es: {odd_sum}")
```

## 🧱 Sentencia `pass`: Marcador de Posición Sintáctico

### 💡 Operación Nula:
`pass` se utiliza cuando la sintaxis de Python requiere un bloque de código pero no se desea ejecutar ninguna instrucción. A diferencia de `continue`, **no salta** a la siguiente iteración.

```python
# ================================================================================
# 🔣 3. SENTENCIA PASS: OPERACIÓN NULA / PLACEHOLDER
# ================================================================================
print("\n--- 🔣 [PASO 3]: Uso de 'pass' (Marcador de Posición) ---")

# 📋 Declaración de secuencia numérica de prueba
numbers_list = [1, 2, 3, 4, 5, 6, 7, 8, 9]

print("📋 [Ejemplo]: Filtrado explícito utilizando bloque pass para pares")
print("-" * 50)

# 🔌 [1] Iteración con sentencia pass
for number in numbers_list:
    if number % 2 == 0:
        # ⚙️ Operación nula: Reservado para lógica futura si fuera necesario
        pass
    else:
        print(f"🔢 Número impar: {number}")

# --------------------------------------------------------------------------------
# 🏁 [PASO 4]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Módulo de sentencias de control finalizado con éxito.")
print("=" * 80)
```
