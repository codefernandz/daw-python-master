# ⚙️ Fundamentos de Python: Operadores Lógicos en Condicionales (`and`, `or`, `not`)

| Propiedad | Detalle |
| :--- | :--- |
| 📍 **Módulo** | Fundamentos de Python |
| 🎯 **Objetivo** | Combinar múltiples expresiones booleanas mediante los operadores lógicos `and`, `or` y `not` para tomar decisiones complejas sin recurrir al anidamiento excesivo. |
| 🛠️ **Conceptos** | `Logical AND (Conjunction)`, `Logical OR (Disjunction)`, `Short-Circuit Logic`, `Inequality Checking (!= / not in)` |

---

### 💡 Explicación de la Arquitectura
* **Operador `and` (Conjunción):** Requiere que **todas** las condiciones evaluadas resulten en `True` para que el bloque de código se ejecute.
* **Operador `or` (Disyunción):** Requiere que **al menos una** de las condiciones resulte en `True`.
* **Evaluación de Cortocircuito (*Short-Circuiting*):**
  * En `and`, si la primera condición es `False`, Python ignora la segunda condición (ya no puede ser `True`).
  * En `or`, si la primera condición es `True`, Python ignora la segunda (ya se garantiza `True`).
* **Optimización Idiomática (*Pythonic Refactoring*):**
  * En lugar de encadenar múltiples desigualdades: `dia != "viernes" and dia != "sábado" and dia != "domingo"`
  * Se recomienda simplificar utilizando membresía: `dia not in ("viernes", "sábado", "domingo")`

```python
from typing import Any

# [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Flujo de Control - Operadores Lógicos (and / or)")
print("=" * 80)

# ================================================================================
# 🧪 1. EVALUACIÓN DE ESCENARIO 1: VIERNES POR LA TARDE
# ================================================================================
# Conjunción (and) de día de la semana y rango de hora.
# ================================================================================

print("\n--- 🚀 [PASO 1]: Caso 1 (Viernes 20:00 h) ---")

# [1] Declaración de variables con Type Hints
day_of_week: str = "viernes"
current_hour: int = 20

print(f"📋 Día evaluado: {day_of_week} | Hora: {current_hour}:00 h")
print("-" * 50)

# [2] Evaluación condicional compuesta
if day_of_week == "viernes" and current_hour >= 15:
    print("🎉 Es viernes y es después de las 3 PM. ¡Hora de relajarse!")

elif day_of_week == "sábado" or day_of_week == "domingo":
    print("🥳 Es fin de semana. ¡Disfruta tu tiempo libre!")

elif day_of_week not in ("viernes", "sábado", "domingo") and current_hour >= 18:
    print("🌙 Es un día laborable y es después de las 6 PM. ¡Hora de descansar!")

else:
    print("💼 Es un día laborable y es antes de las 6 PM. ¡Sigue trabajando!")
```

## 💼 Evaluación de Escenario 2: Lunes por la Tarde (`Fallthrough to Else`)

### 💡 Análisis del Flujo de Control:
1. `day_of_week == "viernes" and current_hour >= 18` $\rightarrow$ `False`
2. `day_of_week == "sábado" or day_of_week == "domingo"` $\rightarrow$ `False`
3. `day_of_week not in (...) and current_hour >= 18` $\rightarrow$ `True` and `False` ($17 < 18$) $\rightarrow$ `False`
4. **Fallback (`else`):** Se ejecuta el bloque por defecto al no cumplirse ninguna de las expresiones anteriores.

```python
# ================================================================================
# 📊 2. EVALUACIÓN DE ESCENARIO 2: LUNES EN HORARIO LABORAL
# ================================================================================
print("\n--- 📊 [PASO 2]: Caso 2 (Lunes 17:00 h) ---")

# [1] Reasignación de estado
day_of_week = "lunes"
current_hour = 17

print(f"📋 Día evaluado: {day_of_week} | Hora: {current_hour}:00 h")
print("-" * 50)

# [2] Evaluación lógica compuesta
if day_of_week == "viernes" and current_hour >= 18:
    print("🎉 Es viernes y es después de las 6 PM. ¡Hora de relajarse!")

elif day_of_week == "sábado" or day_of_week == "domingo":
    print("🥳 Es fin de semana. ¡Disfruta tu tiempo libre!")

elif day_of_week not in ("viernes", "sábado", "domingo") and current_hour >= 18:
    print("🌙 Es un día laborable y es después de las 6 PM. ¡Hora de descansar!")

else:
    print("💼 Es un día laborable y es antes de las 6 PM. ¡Sigue trabajando!")
```

## 🌙 Evaluación de Escenario 3: Martes por la Noche (`Laborable fuera de horario`)

### 💡 Análisis del Flujo de Control:
La condición `day_of_week not in ("viernes", "sábado", "domingo")` evalúa como `True` para `"martes"`, y como `current_hour >= 18` ($19 \ge 18$) también es `True`, el operador `and` habilita la ejecución de la tercera rama (`elif`).

```python
# ================================================================================
# 🔣 3. EVALUACIÓN DE ESCENARIO 3: MARTES FUERA DE HORARIO
# ================================================================================
print("\n--- 🔣 [PASO 3]: Caso 3 (Martes 19:00 h) ---")

# [1] Reasignación de estado
day_of_week = "martes"
current_hour = 19

print(f"📋 Día evaluado: {day_of_week} | Hora: {current_hour}:00 h")
print("-" * 50)

# [2] Evaluación lógica compuesta
if day_of_week == "viernes" and current_hour >= 18:
    print("🎉 Es viernes y es después de las 6 PM. ¡Hora de relajarse!")

elif day_of_week == "sábado" or day_of_week == "domingo":
    print("🥳 Es fin de semana. ¡Disfruta tu tiempo libre!")

elif day_of_week not in ("viernes", "sábado", "domingo") and current_hour >= 18:
    print("🌙 Es un día laborable y es después de las 6 PM. ¡Hora de descansar!")

else:
    print("💼 Es un día laborable y es antes de las 6 PM. ¡Sigue trabajando!")

# --------------------------------------------------------------------------------
# 🏁 [PASO 4]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Módulo de operadores lógicos 'and/or' finalizado con éxito.")
print("=" * 80)
```
