# ⚙️ Fundamentos de Python: Iteración con Índices (`enumerate Function`)

| Propiedad | Detalle |
| :--- | :--- |
| 📍 **Módulo** | Fundamentos de Python |
| 🎯 **Objetivo** | Utilizar la función incorporada `enumerate()` para iterar sobre secuencias obteniendo de forma simultánea el índice de posición y el valor correspondiente a cada elemento. |
| 🛠️ **Conceptos** | `Built-in Functions`, `Sequence Indexing`, `Tuple Unpacking`, `Indexed Traversal`, `enumerate(iterable, start=0)` |

---

### 💡 Explicación de la Arquitectura
* **Función `enumerate()`:** Recibe como argumento un objeto iterable (`list`, `tuple`, `str`, etc.) y devuelve un generador que produce tuplas de la forma `(índice, elemento)` en cada paso de la iteración.
* **Sintaxis Idiomática (*Pythonic*):**
  * `for index, item in enumerate(iterable):`
  * Evita tener que gestionar contadores manuales (`i += 1`) o indexaciones secundarias basadas en rango (`iterable[i]`).
* **Parámetro `start` (Opcional):** Permite cambiar el número inicial del contador de índice (por defecto `start = 0`).

```python
from typing import Any

# [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Flujo de Control - Función enumerate()")
print("=" * 80)

# ================================================================================
# 🧪 1. RECORRIDO DE LISTAS CON ÍNDICES (Indexación Base 0)
# ================================================================================
# Desempaquetado directo de (index, item) sobre una lista de cadenas.
# ================================================================================

print("\n--- 🚀 [PASO 1]: Iteración sobre Lista con enumerate() ---")

# [1] Declaración de lista con Type Hints
fruits_list: list[str] = ["apple", "banana", "cherry", "strawberry", "kiwi"]

print(f"📋 Lista de frutas analizada | Total de elementos: {len(fruits_list)}")
print("-" * 50)

# [2] Iteración desempaquetando índice y valor
print("📋 [Ejemplo 1]: Lectura indexada de lista")
for index, fruit in enumerate(fruits_list):
    print(f"📍 Índice: {index} ➔ 🍎 Fruta: {fruit}")
```

## 🔒 Recorrido de Tuplas Inmutables con Índices

### 💡 Conservación de Estructura:
Incluso en estructuras inmutables como las tuplas (`tuple`), `enumerate()` provee un rastreo de posición seguro sin alterar el contenido ni requerir conversiones intermedias.

```python
# ================================================================================
# 📊 2. RECORRIDO DE TUPLAS
# ================================================================================
print("\n--- 📊 [PASO 2]: Iteración sobre Tupla inmutable ---")

# [1] Declaración de tupla con Type Hints
fruits_tuple: tuple[str, ...] = ("apple", "banana", "cherry", "strawberry", "kiwi")

print(f"🔒 Tupla de frutas analizada | Longitud: {len(fruits_tuple)}")
print("-" * 50)

# [2] Iteración con enumerate sobre tupla
print("📋 [Ejemplo 2]: Lectura indexada de tupla")
for index, fruit in enumerate(fruits_tuple):
    print(f"📍 Índice: {index} ➔ 🍒 Fruta: {fruit}")
```

## 🔤 Iteración Carácter a Carácter sobre Cadenas (`str`)

### 💡 Análisis Posicional de Cadenas:
Al aplicar `enumerate()` sobre un objeto de tipo `str`, la función trata la cadena como una secuencia de caracteres, asignando a cada carácter su índice numérico dentro del texto.

```python
# ================================================================================
# 🔣 3. ITERACIÓN SOBRE CADENAS DE CARACTERES
# ================================================================================
print("\n--- 🔣 [PASO 3]: Iteración sobre Cadena de Texto ---")

# [1] Declaración de cadena de texto
text_characters: str = "Python"

print(f"🔤 Cadena de texto a procesar: '{text_characters}'")
print("-" * 50)

# [2] Iteración carácter a carácter con índice
print("📋 [Ejemplo 3]: Posición de cada carácter en el texto")
for index, character in enumerate(text_characters):
    print(f"📍 Posición {index}: '{character}'")

# --------------------------------------------------------------------------------
# 🏁 [PASO 4]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Módulo de la función 'enumerate()' finalizado con éxito.")
print("=" * 80)
```
