| 🛠️ **Conceptos** | `Dictionary Iteration`, `Keys Traversal (.keys())`, `Values Traversal (.values())`, `Items Unpacking (.items())`, `Direct Iteration` |

---

### 💡 Explicación de la Arquitectura
* **Iterabilidad Nativa:** En Python, los diccionarios son **objetos iterables**. Al iterar directamente sobre una variable de tipo `dict`, la variable de control toma el valor de cada **clave** (`key`) en cada paso.
* **Mecanismos de Recorrido:**
  1. **Directo / `.keys()`:** Accede a las claves del diccionario. Permitido el acceso indirecto al valor mediante la sintaxis de corchetes `dict[key]`.
  2. **`.values()`:** Recorre únicamente los valores almacenados, omitiendo la referencia a las claves.
  3. **`.items()`:** Devuelve tuplas de dos elementos `(clave, valor)` que se pueden desempaquetar directamente en dos variables (`key, value`). Es la forma más idiomática (*pythonic*) cuando se necesitan ambos datos.

```python
from typing import Any

# [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Flujo de Control - Iteración en Diccionarios")
print("=" * 80)

# ================================================================================
# 🧪 1. DECLARACIÓN Y RECORRIDO DIRECTO POR CLAVES (Comportamiento Por Defecto)
# ================================================================================
# Al hacer 'for item in dict', item toma el valor de cada clave.
# ================================================================================

print("\n--- 🚀 [PASO 1]: Iteración Directa sobre Claves ---")

# [1] Declaración de diccionario con Type Hints
months_dict: dict[int, str] = {
    1: "Enero",
    2: "Febrero",
    3: "Marzo",
    4: "Abril",
    5: "Mayo",
    6: "Junio",
    7: "Julio",
    8: "Agosto",
    9: "Septiembre",
    10: "Octubre",
    11: "Noviembre",
    12: "Diciembre"
}

print(f"📋 Diccionario de meses registrado | Total de elementos: {len(months_dict)}")
print("-" * 50)

# [2] Iteración directa por claves
print("📋 [Ejemplo 1]: Iteración directa por claves (for key in dict)")
for month_key in months_dict:
    # Acceso al valor mediante indexación directa months_dict[month_key]
    print(f"🗝️ Clave: {month_key:<2} | 📄 Valor: {months_dict[month_key]}")
```

## 📖 Recorrido Explícito de Claves y Valores (`.keys()` & `.values()`)

### 💡 Uso de Vistas de Diccionario (*Dictionary Views*):
Los métodos `.keys()` y `.values()` devuelven vistas iterables del diccionario. Usar `.keys()` hace el código más explícito en intencionalidad, mientras que `.values()` es ideal cuando no se necesita rastrear la clave de origen.

```python
# ================================================================================
# 📊 2. RECORRIDO CON .keys() Y .values()
# ================================================================================
print("\n--- 📊 [PASO 2]: Iteración Explícita con .keys() y .values() ---")

# [1] Ejemplo 2: Iteración por el método explícito .keys()
print("📋 [Ejemplo 2]: Uso del método explícito .keys()")
for month_key in months_dict.keys():
    print(f"🗝️ Clave: {month_key:<2} | 📄 Valor: {months_dict[month_key]}")

print("-" * 50)

# [2] Ejemplo 3: Iteración únicamente sobre valores con .values()
print("📋 [Ejemplo 3]: Uso del método .values()")
for month_name in months_dict.values():
    print(f"📄 Valor (Nombre del mes): {month_name}")
```

## 🔑 Desempaquetado de Pares Clave-Valor (`.items()`)

### 💡 Idioma Pythonic por Excelencia:
El método `.items()` es la técnica recomendada y más eficiente cuando se requiere manipular simultáneamente tanto la clave como el valor dentro del cuerpo del bucle, evitando lecturas adicionales a la memoria.

```python
# ================================================================================
# 🔣 3. DESEMPAQUETADO SIMULTÁNEO CON .items()
# ================================================================================
print("\n--- 🔣 [PASO 3]: Desempaquetado Idiomático con .items() ---")

# [1] Ejemplo 4: Iteración y desempaquetado de tuplas (key, value)
print("📋 [Ejemplo 4]: Desempaquetado de Clave y Valor con .items()")
for month_key, month_name in months_dict.items():
    print(f"🔑 Clave: {month_key:<2} ➔ 📄 Valor: {month_name}")

# --------------------------------------------------------------------------------
# 🏁 [PASO 4]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Módulo de iteración en diccionarios finalizado con éxito.")
print("=" * 80)
```
