### 💡 Explicación de la Arquitectura
* **Función `range()`:** Genera un objeto de tipo `range` que representa una secuencia inmutable de números enteros. Es un objeto **evaluado de forma diferida** (*Lazy Evaluation*), lo que significa que no almacena todos los números en memoria al mismo tiempo, sino que los genera a medida que se iteran.
* **Sintaxis Completa:** `range(start, stop, step)`
  * **`start` (Opcional):** Valor inicial inclusivo de la secuencia (por defecto `0`).
  * **`stop` (Obligatorio):** Valor final **exclusivo** de la secuencia (la secuencia se detiene en `stop - 1`).
  * **`step` (Opcional):** Salto o incremento entre cada número de la secuencia (por defecto `1`).

```python
from typing import Any

# [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Flujo de Control - Función range()")
print("=" * 80)

# ================================================================================
# 🧪 1. SECUENCIA SIMPLE (Inicio por defecto en 0)
# ================================================================================
# range(10) genera números del 0 al 9 (10 elementos en total).
# ================================================================================

print("\n--- 🚀 [PASO 1]: Generación de Secuencia Básica (0 a N-1) ---")

print("📋 [Ejemplo 1]: Iteración con range(10)")
print("-" * 50)

# [1] Iteración directa con range(stop)
for number in range(10):
    print(f"🔢 Valor actual: {number}")
```

## 🔢 Secuencias Personalizadas con Inicio y Paso (`start`, `stop`, `step`)

### 💡 Parámetros Avanzados:
Al especificar `range(10, 51, 2)`:
* **`start = 10`**: La secuencia comienza en $10$.
* **`stop = 51`**: La secuencia se detiene antes de $51$ (el último valor evaluable es $50$).
* **`step = 2`**: Incrementa de $2$ en $2$, generando únicamente números pares.

```python
# ================================================================================
# 📊 2. SECUENCIA CON INICIO, PARADA E INCREMENTO PERSONALIZADO
# ================================================================================
print("\n--- 📊 [PASO 2]: Secuencia con Pasos Incrementales ---")

print("📋 [Ejemplo 2]: Generación de números del 10 al 50 con salto de 2 en 2")
print("-" * 50)

# [1] Iteración con range(start, stop, step)
for even_number in range(10, 51, 2):
    print(f"🔢 Número par en la secuencia: {even_number}")
```

## 📋 Construcción y Poblamiento de Listas con `range()`

### 💡 Integración de `range()` con Colecciones:
Es común utilizar la secuencia generada por `range()` para poblar estructuras de datos como listas mediante el método `.append()`, o convirtiendo el objeto `range` directamente mediante la función `list()`.

```python
# ================================================================================
# 🔣 3. CONSTRUCCIÓN DE LISTAS A PARTIR DE RANGE
# ================================================================================
print("\n--- 🔣 [PASO 3]: Acumulación en Lista ---")

# [1] Declaración de lista vacía con Type Hints
generated_numbers: list[int] = []

print("📋 [Ejemplo 3]: Poblamiento de lista mediante range(10)")
print("-" * 50)

# [2] Bucle for para agregar elementos a la lista
for number in range(10):
    generated_numbers.append(number)

print(f"✅ Lista final generada: {generated_numbers}")
print(f"📊 Longitud de la lista: {len(generated_numbers)} elementos")

# --------------------------------------------------------------------------------
# 🏁 [PASO 4]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Módulo de la función 'range()' finalizado con éxito.")
print("=" * 80)
```
