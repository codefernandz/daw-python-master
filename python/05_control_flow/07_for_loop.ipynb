### 💡 Explicación de la Arquitectura
* **Iteración Determinada (`for`):** A diferencia del bucle `while`, el bucle `for` se ejecuta un número **conocido y finito** de veces, correspondiente a la cantidad de elementos presentes en el objeto iterable.
* **Comportamiento según la Estructura de Datos:**
  * **Listas y Tuplas (`list`, `tuple`):** Mantienen el orden de inserción y permiten la iteración indexada/secuencial.
  * **Conjuntos (`set`):** Son colecciones no ordenadas de elementos únicos; el orden de iteración no está garantizado.
  * **Diccionarios (`dict`):** Permiten iterar directamente sobre sus claves (`.keys()`), valores (`.values()`) o pares clave-valor (`.items()`).

```python
from typing import Any

# [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Flujo de Control - Bucles 'for' en Colecciones")
print("=" * 80)

# ================================================================================
# 🧪 1. ITERACIÓN SOBRE LISTAS (Recorrido y Transformación de Elementos)
# ================================================================================
# Recorrido elemento a elemento y construcción de nuevas listas acumulativas.
# ================================================================================

print("\n--- 🚀 [PASO 1]: Iteración y Transformación sobre Listas ---")

# [1] Ejemplo 1: Recorrido simple sobre lista de días
days_of_week: list[str] = ["Lunes", "Martes", "Miércoles", "Jueves", "Viernes", "Sábado", "Domingo"]

print("📋 [Ejemplo 1]: Recorrido de Días de la Semana")
for day in days_of_week:
    print(f"🗓️ El día de la semana es: {day}")

print("-" * 50)

# [2] Ejemplo 2: Transformación de elementos (A mayúsculas mediante .append())
fruits: list[str] = ["manzana", "banana", "cereza", "durazno", "kiwi"]
fruits_upper: list[str] = []

for fruit in fruits:
    # Transformación del elemento actual
    transformed_fruit: str = fruit.upper()
    fruits_upper.append(transformed_fruit)

print("📋 [Ejemplo 2]: Transformación de Lista de Frutas")
print(f"🍎 Lista original: {fruits}")
print(f"🔤 Lista en mayúsculas: {fruits_upper}")
```

## 📦 Iteración sobre Tuplas y Conjuntos (`tuple` & `set`)

### 💡 Diferencias Clave:
* **Tuplas (`tuple`):** Son inmutables. La iteración permite extraer y pasar sus elementos a otras estructuras (como listas) sin modificar la tupla original.
* **Conjuntos (`set`):** Son colecciones de elementos únicos no ordenados. Son ideales para operaciones matemáticas de acumulación y filtrado de duplicados.

```python
# ================================================================================
# 📊 2. ITERACIÓN EN TUPLAS Y CONJUNTOS (Conversión y Acumulación)
# ================================================================================
print("\n--- 📊 [PASO 2]: Conversión de Tuplas y Sumatoria en Sets ---")

# [1] Ejemplo 3: Conversión de Tupla a Lista usando 'for'
programming_languages: tuple[str, ...] = ("Python", "Java", "C++", "JavaScript", "Ruby")
list_from_tuple: list[str] = []

for language in programming_languages:
    list_from_tuple.append(language)

print("📋 [Ejemplo 3]: Extracción de Tupla a Lista")
print(f"🔒 Tupla original: {programming_languages}")
print(f"📋 Lista resultante: {list_from_tuple}")

print("-" * 50)

# [2] Ejemplo 4: Sumatoria acumulativa sobre un Conjunto (Set)
numbers_set: set[int] = {1, 2, 3, 4, 5}
total_sum: int = 0

for number in numbers_set:
    total_sum += number

print("📋 [Ejemplo 4]: Acumulación Matemática sobre Set")
print(f"🔢 Set original: {numbers_set}")
print(f"➕ Suma total acumulada: {total_sum}")
```

## 📖 Iteración sobre Diccionarios (`dict`)

### 💡 Métodos de Desempaquetado:
* **`.items()`:** Devuelve un par `(clave, valor)` en cada iteración, permitiendo desempaquetar directamente dos variables de control (`for key, value in dict.items()`).
* **`.keys()`:** Devuelve únicamente las claves del diccionario.
* **`.values()`:** Devuelve únicamente los valores asociados.

```python
# ================================================================================
# 🔣 3. ITERACIÓN SOBRE DICCIONARIOS (Items y Keys)
# ================================================================================
print("\n--- 🔣 [PASO 3]: Exploración de Diccionarios ---")

# [1] Declaración de diccionario estructurado
person_info: dict[str, Any] = {
    "nombre": "Juan",
    "edad": 30,
    "ciudad": "Madrid"
}

# [2] Ejemplo 5: Desempaquetado de Clave-Valor con .items()
print("📋 [Ejemplo 5]: Lectura Completa con .items()")
for key, value in person_info.items():
    print(f"🔑 {key.capitalize()}: {value}")

print("-" * 50)

# [3] Ejemplo 6: Extracción de Claves con .keys()
keys_list: list[str] = []

for key in person_info.keys():
    keys_list.append(key)

print("📋 [Ejemplo 6]: Extracción de Claves con .keys()")
print(f"📖 Diccionario fuente: {person_info}")
print(f"🗝️ Lista de claves extraídas: {keys_list}")

# --------------------------------------------------------------------------------
# 🏁 [PASO 4]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Módulo de bucles 'for' en colecciones finalizado con éxito.")
print("=" * 80)
```
