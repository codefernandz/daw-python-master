# ⚙️ Fundamentos de Python: Condicionales Múltiples (`if ... elif ... else`)

| Propiedad | Detalle |
| :--- | :--- |
| 📍 **Módulo** | Fundamentos de Python |
| 🎯 **Objetivo** | Dominar la evaluación encadenada y mutuamente excluyente de múltiples condiciones mediante `if`, `elif` y `else`. |
| 🛠️ **Conceptos** | `Multi-Way Branching`, `Short-Circuit Evaluation`, `Range Classification`, `Modulo Operator (%)`, `Membership Checking (in)` |

---

### 💡 Explicación de la Arquitectura
* **Flujo Multicamino (`Multi-Way Branching`):**
  * `if`: Evalúa la primera condición obligatoria.
  * `elif` (*else if*): Se evalúa **únicamente si todas las condiciones anteriores resultaron `False`**. Pueden encadenarse múltiples bloques `elif`.
  * `else`: Bloque opcional de captura global (*fallback*) que se ejecuta cuando **ninguna** de las condiciones anteriores fue `True`.
* **Evaluación de Cortocircuito:** Tan pronto como una condición resulta `True`, Python ejecuta únicamente el bloque indentado correspondiente e **ignora completamente el resto de la cadena condicional**.

```python
from typing import Any

# [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Flujo de Control - Condicionales Múltiples (if-elif-else)")
print("=" * 80)

# ================================================================================
# 🧪 1. EVALUACIÓN LÓGICA GENERAL Y CLASIFICACIÓN DE RANGOS NUMÉRICOS
# ================================================================================
# Demostración del encadenamiento de condiciones y rangos jerárquicos.
# ================================================================================

print("\n--- 🚀 [PASO 1]: Evaluación General y Categorización por Edad/Notas ---")

# [1] Ejemplo 1: Estructura base de demostración booleana
is_first_condition: bool = False
is_second_condition: bool = True

print("📋 [Ejemplo 1]: Evaluación de Banderas Booleanas")
if is_first_condition:
    print("✅ La condición 1 es verdadera.")
elif is_second_condition:
    print("✅ La condición 2 es verdadera.")
else:
    print("❌ Ninguna de las condiciones es verdadera.")

print("-" * 50)

# [2] Ejemplo 2: Clasificación de rangos por edad
age: int = 25
print(f"📋 [Ejemplo 2]: Clasificación por Edad ({age} años)")

if age < 13:
    print("👦 Categoría: Eres un niño.")
elif age < 20:
    print("🧑 Categoría: Eres un adolescente.")
elif age < 65:
    print("👨 Categoría: Eres un adulto.")
else:
    print("👴 Categoría: Eres un adulto mayor.")

print("-" * 50)

# [3] Ejemplo 3: Calificación académica
academic_score: float = 85.0
print(f"📋 [Ejemplo 3]: Clasificación Académica (Puntaje: {academic_score})")

if academic_score >= 90.0:
    print("🏆 Desempeño: Excelente")
elif academic_score >= 80.0:
    print("🎖️ Desempeño: Muy bien")
elif academic_score >= 70.0:
    print("👍 Desempeño: Bien")
else:
    print("⚠️ Desempeño: Necesitas mejorar")
```

## 🔢 Signo de Números, Días de la Semana y Clima

### 💡 Inspección de Tipos y Evaluaciones Secuenciales:
* **Invariante de Positividad/Negatividad:** Evaluación contra cero ($0$).
* **Mapeo de Cadenas (`str`):** Comparación directa mediante el operador de igualdad `==`.

```python
# ================================================================================
# 📊 2. CLASIFICACIÓN NUMÉRICA, TEMPORAL Y CLIMÁTICA
# ================================================================================
print("\n--- 📊 [PASO 2]: Evaluación de Signo, Días y Temperatura ---")

# [1] Ejemplo 4: Signo del número
target_number: int = -5
print(f"📋 [Ejemplo 4]: Signo de Número ({target_number})")

if target_number > 0:
    print("➕ El número es positivo.")
elif target_number < 0:
    print("➖ El número es negativo.")
else:
    print("⚪ El número es cero.")

print("-" * 50)

# [2] Ejemplo 5: Día de la semana
day_of_week: str = "miércoles"
print(f"📋 [Ejemplo 5]: Categorización de Día ('{day_of_week}')")

if day_of_week == "lunes":
    print("📅 Hoy es lunes.")
elif day_of_week == "martes":
    print("📅 Hoy es martes.")
elif day_of_week == "miércoles":
    print("📅 Hoy es miércoles.")
elif day_of_week == "jueves":
    print("📅 Hoy es jueves.")
elif day_of_week == "viernes":
    print("📅 Hoy es viernes.")
else:
    print("🎉 Es fin de semana.")

print("-" * 50)

# [3] Ejemplo 6: Sensación térmica
current_temperature: float = 30.0
print(f"📋 [Ejemplo 6]: Clasificación de Temperatura ({current_temperature}°C)")

if current_temperature < 0:
    print("❄️ Hace mucho frío.")
elif current_temperature < 15:
    print("🧥 Hace frío.")
elif current_temperature < 25:
    print("🌤️ Hace una temperatura agradable.")
else:
    print("☀️ Hace calor.")
```

## 🔣 Evaluación Aritmética y Pertenencia (`Modulus %` & `Membership in`)

### 💡 Evaluaciones Avanzadas:
* **Paridad (`% 2`):** Permite determinar la divisibilidad exacta de un número. Si `numero % 2 == 0`, el residuo es cero (par).
* **Pertenencia (`in`):** Verifica de forma eficiente si un carácter se encuentra contenido dentro de una colección o secuencia de texto.

```python
# ================================================================================
# 🔣 3. OPERADORES ARITMÉTICOS Y PERTENENCIA EN CADENAS
# ================================================================================
print("\n--- 🔣 [PASO 3]: Paridad y Verificación de Vocales ---")

# [1] Ejemplo 7: Evaluación de paridad
eval_number: int = 7
print(f"📋 [Ejemplo 7]: Evaluación de Paridad ({eval_number})")

if eval_number % 2 == 0:
    print("🔢 El número es PAR.")
else:
    print("🔢 El número es IMPAR.")

print("-" * 50)

# [2] Ejemplo 8: Evaluación de vocales/consonantes
character: str = "a"
vowels: str = "aeiouAEIOU"

print(f"📋 [Ejemplo 8]: Verificación de Pertenencia ('{character}')")

if character in vowels:
    print(f"🔤 La letra '{character}' es una VOCAL.")
else:
    print(f"🔤 La letra '{character}' es una CONSONANTE.")

# --------------------------------------------------------------------------------
# 🏁 [PASO 4]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Módulo de condicionales múltiples 'if-elif-else' finalizado con éxito.")
print("=" * 80)
```
