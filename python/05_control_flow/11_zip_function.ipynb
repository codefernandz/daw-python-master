# ⚙️ Fundamentos de Python: Iteración en Paralelo (`zip Function`)

| Propiedad | Detalle |
| :--- | :--- |
| 📍 **Módulo** | Fundamentos de Python |
| 🎯 **Objetivo** | Dominar el uso de la función incorporada `zip()` para combinar y recorrer simultáneamente múltiples iterables en tuplas indexadas por posición. |
| 🛠️ **Conceptos** | `Built-in Functions`, `Parallel Iteration`, `Tuple Aggregation`, `Truncation Mechanics`, `zip(iter1, iter2, ...)` |

---

### 💡 Explicación de la Arquitectura
* **Función `zip()`:** Toma como argumento dos o más objetos iterables (`list`, `tuple`, `str`, etc.) y devuelve un objeto iterador que empareja los elementos del mismo índice en tuplas: `(iter1[i], iter2[i], ...)`.
* **Evaluación perezosa (*Lazy Evaluation*):** Devuelve un iterador, lo que significa que evalúa los elementos bajo demanda sin cargar todo el dataset en memoria. Puede convertirse explícitamente a estructura de lista mediante `list(zip(...))`.
* **Comportamiento con Longitudes Desiguales:** Por defecto, `zip()` detiene la agregación en el momento en que se agota el **iterable más corto** (trincado/truncamiento).

```python
from typing import Any

# 🛰️ [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Flujo de Control - Función zip()")
print("=" * 80)

# ================================================================================
# 🧪 1. COMBINACIÓN BÁSICA DE DOS ITERABLES (Lista + Tupla)
# ================================================================================
# Unión de estudiantes con sus respectivas calificaciones numéricas.
# ================================================================================

print("\n--- 🚀 [PASO 1]: Agregación Básica de 2 Iterables ---")

# 📋 Declaración de datos con Type Hints
students_list: list[str] = ["Juan", "María", "Pedro", "Ana"]
grades_tuple: tuple[float, ...] = (8.5, 9.0, 7.5, 10.0)

print(f"📋 Estudiantes: {students_list}")
print(f"🔒 Calificaciones: {grades_tuple}")
print("-" * 50)

# 🔌 [1] Combinación y conversión explícita a lista de tuplas
students_grades: list[tuple[str, float]] = list(zip(students_list, grades_tuple))
print(f"⚙️ Lista combinada generada: {students_grades}")

# 📋 [2] Iteración desempaquetada mediante bucle for
print("\n📋 [Ejemplo 1]: Recorrido en paralelo con desempaquetado de tuplas")
for student, grade in zip(students_list, grades_tuple):
    print(f"🎓 El alumno {student} tiene una nota de {grade}")
```

## 🌐 Agregación de Múltiples Colecciones (N Iterables)

### 💡 Extensibilidad de `zip()`:
La función acepta cualquier número de colecciones como argumentos, emparejando $N$ elementos por tupla en cada ciclo de la iteración.

```python
# ================================================================================
# 📊 2. COMBINACIÓN DE TRES O MÁS ITERABLES
# ================================================================================
print("\n--- 📊 [PASO 2]: Combinación de Múltiples Fuentes de Datos ---")

# 📋 Declaración de colecciones paralelas
names_list: list[str] = ["Juan", "María", "Pedro", "Ana"]
ages_list: list[int] = [20, 22, 19, 21]
cities_list: list[str] = ["Madrid", "Barcelona", "Valencia", "Sevilla"]

print("📋 [Ejemplo 2]: Generación de tuplas de 3 dimensiones (Nombre, Edad, Ciudad)")
print("-" * 50)

# 🔌 [1] Combinación de 3 iterables
personal_info: list[tuple[str, int, str]] = list(zip(names_list, ages_list, cities_list))
print(f"⚙️ Colección agregada: {personal_info}")

# 📋 [2] Iteración en paralelo con desempaquetado de 3 variables
print("\n📋 Imprimiendo registros combinados:")
for name, age, city in zip(names_list, ages_list, cities_list):
    print(f"👤 {name} tiene {age} años y vive en {city}")
```

## ⚠️ Manejo de Iterables con Diferente Longitud

### 💡 Mecánica de Truncamiento (*Shortest-First*):
Cuando se ingresan colecciones con distinto número de elementos, `zip()` omite los elementos sobrantes de las listas más largas y se detiene automáticamente al agotar la colección más corta.

```python
# ================================================================================
# 🔣 3. EVALUACIÓN DE ITERABLES DESIGUALES
# ================================================================================
print("\n--- 🔣 [PASO 3]: Truncamiento con Colecciones Desiguales ---")

# 📋 Declaración de iterables de longitud 3 y 2 respectivamente
fruits_list: list[str] = ["manzana", "banana", "naranja"]
colors_list: list[str] = ["rojo", "amarillo"]

print(f"🍎 Frutas ({len(fruits_list)} elementos): {fruits_list}")
print(f"🎨 Colores ({len(colors_list)} elementos): {colors_list}")
print("-" * 50)

# 🔌 [1] Agregación de colecciones desiguales
fruits_colors_list: list[tuple[str, str]] = list(zip(fruits_list, colors_list))

print("📋 [Ejemplo 3]: Resultado de la combinación (Truncado automático):")
print(f"⚙️ Lista de tuplas resultante: {fruits_colors_list}")
print(f"📊 Total de pares formados: {len(fruits_colors_list)} (se ignoró 'naranja')")

# --------------------------------------------------------------------------------
# 🏁 [PASO 4]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Módulo de la función 'zip()' finalizado con éxito.")
print("=" * 80)
```
