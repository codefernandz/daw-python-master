### 💡 Explicación de la Arquitectura
* **Flujo de Control (`Control Flow`):** Refiere al orden secuencial en el que el intérprete de Python ejecuta las instrucciones de un programa.
* **Evaluación Condicional (`if`):**
  * Si la expresión lógica se evalúa como `True`, se ejecuta el bloque de código indentado (4 espacios).
  * Si la expresión se evalúa como `False`, Python omite el bloque indentado y continúa la ejecución en la siguiente instrucción al mismo nivel del `if`.
* **Sintaxis Crucial:**
  1. Dos puntos (`:`) al final de la sentencia condicional.
  2. Indentación constante (4 espacios o 1 tabulación) para definir el ámbito (`scope`) de ejecución.

```python
from typing import Any

# [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Flujo de Control - Sentencia 'if'")
print("=" * 80)

# ================================================================================
# 🧪 1. EVALUACIÓN DE CONDICIONES BOOLEANAS BÁSICAS
# ================================================================================
# Comparaciones directas para observar la ejecución y omisión de bloques.
# ================================================================================

print("\n--- 🚀 [PASO 1]: Evaluación de Expresiones Relacionales ---")

# [1] Condición verdadera (5 > 2) -> Se ejecuta el bloque indentado
if 5 > 2:
    print("✅ [Condición 1 (5 > 2)]: La condición se cumple (True). Bloque ejecutado.")

# [2] Condición falsa (5 < 2) -> Se omite el bloque indentado sin generar error
if 5 < 2:
    print("❌ [Condición 2 (5 < 2)]: Este texto NO se imprimirá porque es False.")

print("📋 La ejecución del programa continúa secuencialmente sin interrupciones.")
```

## 📊 Aplicación Práctica: Control de Estado del Programa

### 💡 Modificación Condicional de Variables:
La sentencia `if` permite ejecutar mutaciones de estado (como acumulaciones o felicitaciones) únicamente cuando se cumplen criterios específicos de negocio.

```python
# ================================================================================
# 📊 2. LÓGICA DE NEGOCIO Y ACUMULACIÓN CONDICIONAL
# ================================================================================
print("\n--- 📊 [PASO 2]: Seguimiento de Lecciones y Mutación Condicional ---")

# [1] Estado inicial del usuario con Type Hints
completed_lessons: int = 5
daily_lessons: int = 0

print(f"📋 Estado inicial de lecciones completadas: {completed_lessons}")

# [2] Verificación de logro por objetivo diario/acumulado
if completed_lessons >= 5:
    print("🎉 ¡Felicidades! Has alcanzado o superado la meta de 5 lecciones completadas.")

# [3] Evaluación e incremento condicional basado en la actividad del día
if daily_lessons > 0:
    completed_lessons += daily_lessons
    print(f"✅ Se sumaron {daily_lessons} lecciones realizadas hoy.")

# [4] Imprimir estado final (como daily_lessons es 0, completed_lessons no cambia)
print("-" * 50)
print(f"📋 Estado final de lecciones completadas:   {completed_lessons}")

# --------------------------------------------------------------------------------
# 🏁 [PASO 3]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Módulo de sentencia 'if' finalizado con éxito.")
print("=" * 80)
```
