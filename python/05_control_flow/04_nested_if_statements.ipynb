### 💡 Explicación de la Arquitectura
* **Anidamiento Condicional (`Nested If`):** Consiste en colocar una estructura `if ... else` dentro del bloque indentado de otra sentencia condicional.
* **Evaluación Jerárquica:** El nivel interno de evaluación **solo se ejecuta si la condición del nivel externo resulta `True`**.
* **⚠️ Regla de Refactorización (Buenas Prácticas):**
  * El uso excesivo de `if` anidados genera código en "pirámide" (*Pyramid of Doom*), incrementando la complejidad ciclomática y dificultando la lectura.
  * Se recomienda limitar el anidamiento a un máximo de 2 niveles y considerar combinaciones con operadores lógicos (`and`, `or`) o cláusulas de guarda (*Guard Clauses*).

```python
from typing import Any

# [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Flujo de Control - Sentencias Condicionales Anidadas")
print("=" * 80)

# ================================================================================
# 🧪 1. EVALUACIÓN JERÁRQUICA DE PROGRESO (2 Niveles de Anidamiento)
# ================================================================================
# Condicional interno evaluado tras verificar la condición primaria externa.
# ================================================================================

print("\n--- 🚀 [PASO 1]: Anidamiento de Progreso de Lecciones ---")

# [1] Declaración de variables con Type Hints
lessons_done_today: int = 2
total_topic_lessons: int = 10

print(f"📋 Lecciones completadas hoy: {lessons_done_today} / {total_topic_lessons}")
print("-" * 50)

# [2] Nivel 1 (Externo): Comprobar si se hizo más de 1 lección
if lessons_done_today > 1:
    print("✅ [Nivel 1]: ¡Has hecho más de una lección hoy!")
    
    # [3] Nivel 2 (Anidado): Comprobar si se completó el tema entero
    if lessons_done_today == total_topic_lessons:
        print("🎉 [Nivel 2]: ¡Felicidades! Has completado el tema en su totalidad.")
    else:
        print(f"⏳ [Nivel 2]: Aún te faltan {total_topic_lessons - lessons_done_today} lecciones para completar el tema.")
else:
    print("ℹ️ [Nivel 1]: Has hecho 1 o ninguna lección hoy.")
```

## 🔢 Clasificación Multicamino y Evaluación de Paridad

### 💡 Análisis de Propiedades Numéricas:
Mediante condicionales anidados podemos evaluar primero el **signo** de un número (positivo, negativo o cero) y, posteriormente, determinar su **paridad** (par o impar) usando el operador módulo (`%`).

```python
# ================================================================================
# 📊 2. CLASIFICACIÓN DUAL DE NÚMEROS (Signo y Paridad)
# ================================================================================
print("\n--- 📊 [PASO 2]: Clasificación por Signo y Paridad ---")

# [1] Declaración del número a evaluar
target_number: int = -3

print(f"📋 Número a analizar: {target_number}")
print("-" * 50)

# [2] Evaluación primaria por signo (if-elif-else)
if target_number > 0:
    print("➕ [Signo]: El número es positivo.")
    
    # Anidamiento para paridad en números positivos
    if target_number % 2 == 0:
        print("🔢 [Paridad]: El número es PAR.")
    else:
        print("🔢 [Paridad]: El número es IMPAR.")

elif target_number < 0:
    print("➖ [Signo]: El número es negativo.")
    
    # Anidamiento para paridad en números negativos
    if target_number % 2 == 0:
        print("🔢 [Paridad]: El número es PAR.")
    else:
        print("🔢 [Paridad]: El número es IMPAR.")

else:
    print("⚪ [Signo]: El número es cero (neutro).")
```

## 🔣 Evaluación Corta y Resumen de Anidamiento

### 💡 Ejemplo de Anidamiento Simple:
Demostración de verificación de signo y paridad sobre una variable `base_number`.

```python
# ================================================================================
# 🔣 3. EVALUACIÓN DE NÚMERO BASE
# ================================================================================
print("\n--- 🔣 [PASO 3]: Evaluación de Número Base ---")

# [1] Declaración de variable base
base_number: int = 5

print(f"📋 Valor base a evaluar: {base_number}")
print("-" * 50)

# [2] Anidamiento directo
if base_number > 0:
    print("✅ [Externo]: La base es positiva.")
    if base_number % 2 == 0:
        print("🔢 [Anidado]: La base es PAR.")
    else:
        print("🔢 [Anidado]: La base es IMPAR.")

# --------------------------------------------------------------------------------
# 🏁 [PASO 4]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Módulo de 'if' anidados finalizado con éxito.")
print("=" * 80)
```
