### 💡 Explicación de la Arquitectura
* **Iteración Indeterminada (`while`):** Se utiliza cuando **no se conoce con antelación** el número exacto de iteraciones que requerirá el programa, dependiendo de la evaluación continua de una expresión booleana.
* **Mecanismos de Control de Ejecución:**
  1. **Actualización de Condición:** El contador o la variable de control debe modificarse dentro del bucle para garantizar que la condición evalúe como `False` en algún momento.
  2. **Sentencia `break`:** Interrumpe la ejecución del bucle de forma inmediata, transfiriendo el control a la línea de código posterior.
  3. **Banderas Booleanas (*State Flags*):** Variables del tipo `bool` (`should_continue = True`) utilizadas para controlar de forma legible la permanencia en el bucle.

```python
from typing import Any

# [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Flujo de Control - Bucles Condicionales (while)")
print("=" * 80)

# ================================================================================
# 🧪 1. CONTADORES Y RECORRIDOS INDEXADOS (Incremento, Decremento e Índices)
# ================================================================================
# Iteraciones deterministas controladas por condición de rango numérico.
# ================================================================================

print("\n--- 🚀 [PASO 1]: Contadores Incrementales y Decrementales ---")

# [1] Ejemplo 1: Conteo ascendente del 1 al 5
counter: int = 1
print("📋 [Ejemplo 1]: Conteo Ascendente (1 a 5)")

while counter <= 5:
    print(f"🔢 Contador actual: {counter}")
    counter += 1  # Incremento de control

print("-" * 50)

# [2] Ejemplo 2: Conteo descendente del 10 al 1
countdown: int = 10
print("📋 [Ejemplo 2]: Conteo Descendente (10 a 1)")

while countdown > 0:
    print(f"⏳ Cuenta regresiva: {countdown}")
    countdown -= 1  # Decremento de control

print("-" * 50)

# [3] Ejemplo 3: Recorrido indexado sobre una lista
fruits: list[str] = ["manzana", "banana", "cereza"]
index: int = 0

print("📋 [Ejemplo 3]: Recorrido de Colección por Índice")
while index < len(fruits):
    print(f"🍎 Fruta en índice {index}: {fruits[index]}")
    index += 1  # Incremento de posición
```

## 🛑 Acumulación de Datos y Cláusulas de Interrupción (`Infinite Loop & break`)

### 💡 Patrón Bucle Infinito + `break`:
Es una práctica habitual utilizar `while True:` cuando el punto de salida depende de la lectura de un valor centinela o comando específico del usuario (`0`, `"salir"`), interrumpiendo el flujo con la palabra clave `break`.

```python
# ================================================================================
# 📊 2. BUCLES INTERACTIVOS Y MANTENIMIENTO DE ESTADO (Acumuladores y Comandos)
# ================================================================================
print("\n--- 📊 [PASO 2]: Acumulación y Manejo de Comandos con break ---")

# [1] Ejemplo 4: Acumulación interactiva con valor centinela (Simulación de Entradas)
# Nota: Se utiliza una lista simulada de entradas para evitar bloquear entornos no interactivos.
simulated_numbers: list[int] = [10, 25, 5, 0]  # El '0' marca la salida
simulated_index: int = 0
total_sum: int = 0

print("📋 [Ejemplo 4]: Acumulación mediante valor centinela (0 para salir)")

while True:
    # Simulación de entrada de usuario: int(input("Introduce un número (0 para salir): "))
    current_number: int = simulated_numbers[simulated_index]
    simulated_index += 1
    
    if current_number == 0:
        print("🛑 Valor centinela '0' detectado. Interrumpiendo bucle con 'break'...")
        break  # Salida directa del bucle
    
    total_sum += current_number
    print(f"➕ Número ingresado: {current_number} | Suma acumulada parcial: {total_sum}")

print(f"✅ Suma total final: {total_sum}")

print("-" * 50)

# [2] Ejemplo 5: Bucle infinito controlado por comando de salida
simulated_commands: list[str] = ["status", "help", "salir"]
command_index: int = 0

print("📋 [Ejemplo 5]: Procesador de Comandos")

while True:
    # Simulación de comando: input("Introduce un comando (escribe 'salir' para terminar): ")
    command: str = simulated_commands[command_index]
    command_index += 1
    
    if command.lower() == "salir":
        print("🚪 Saliendo del programa...")
        break
    else:
        print(f"⚙️ Procesando comando: '{command}'")
```

## 🔄 Validaciones y Banderas Booleanas (`State Flags & Validation`)

### 💡 Banderas de Estado (*State Flags*):
Permiten abstraer la condición de permanencia dentro de una variable booleana (`should_continue = True`), mejorando la legibilidad del código cuando la condición depende de lógica compleja.

```python
# ================================================================================
# 🔣 3. VALIDACIÓN DE ENTRADAS Y CONTROL CON BANDERAS BOOLEANAS
# ================================================================================
print("\n--- 🔣 [PASO 3]: Validaciones y Banderas Booleanas ---")

# [1] Ejemplo 6: Validación de respuesta requerida ("sí")
simulated_answers: list[str] = ["tal vez", "no", "SÍ"]
answer_index: int = 0
user_response: str = ""

print("📋 [Ejemplo 6]: Validación Estricta de Respuesta")

while user_response.lower() != "sí":
    # Simulación: user_response = input("¿Quieres continuar? (sí/no): ")
    user_response = simulated_answers[answer_index]
    answer_index += 1
    print(f"🔍 Respuesta recibida: '{user_response}'")

print("🎉 ¡Gracias por continuar! Validación superada con éxito.")

print("-" * 50)

# [2] Ejemplo 7: Control del bucle mediante variable booleana de estado
simulated_flags: list[str] = ["sí", "sí", "no"]
flag_index: int = 0
should_continue: bool = True

print("📋 [Ejemplo 7]: Control de Estado mediante Bandera Booleana")

while should_continue:
    # Simulación: response_flag = input("¿Deseas continuar? (sí/no): ")
    response_flag: str = simulated_flags[flag_index]
    flag_index += 1
    
    print(f"🔍 Estado evaluado: '{response_flag}'")
    
    if response_flag.lower() == "no":
        print("🛑 Cambiando la variable booleana 'should_continue' a False...")
        should_continue = False  # Alteración de la condición del bucle

print("✅ Bucle finalizado mediante cambio de bandera.")

# --------------------------------------------------------------------------------
# 🏁 [PASO 4]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Módulo de bucles 'while' finalizado con éxito.")
print("=" * 80)
```
