### 💡 Explicación de la Arquitectura

La **comprensión de listas** es una sintaxis idiomática (*Pythonic*) que permite construir listas a partir de otros iterables en una sola línea de código. Además de reducir la verbosidad del código, suele ser más eficiente en tiempo de ejecución que un bucle `for` tradicional con `.append()`.

#### 📐 Variaciones Sintácticas Principales:

1. **Transformación Básica:**
   $$\text{[ \textit{expression} \textbf{ for } \textit{item} \textbf{ in } \textit{iterable} ]}$$
2. **Filtrado (`if` al final):**
   $$\text{[ \textit{expression} \textbf{ for } \textit{item} \textbf{ in } \textit{iterable} \textbf{ if } \textit{condition} ]}$$
3. **Mapeo Condicional (`if-else` al inicio):**
   $$\text{[ \textit{expr\_if\_true} \textbf{ if } \textit{condition} \textbf{ else } \textit{expr\_if\_false} \textbf{ for } \textit{item} \textbf{ in } \textit{iterable} ]}$$

> ⚠️ **Nota:** No se admite la cláusula `elif` dentro de una comprensión de listas. Si se requiere una lógica condicional multirrama compleja, se recomienda el uso de un bucle `for` tradicional para mantener la legibilidad.

```python
from typing import Any

# 🛰️ [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Flujo de Control - List Comprehension")
print("=" * 80)

# ================================================================================
# 🧪 1. COMPAÑÍA DE SINTAXIS: BUCLE TRADICIONAL VS LIST COMPREHENSION
# ================================================================================
# Creación de una secuencia de 1000 elementos mediante dos enfoques.
# ================================================================================

print("\n--- 🚀 [PASO 1]: Transformación Básica (1000 Elementos) ---")

# 📋 1. Enfoque Tradicional con Bucle for y .append()
built_list: list[int] = []
for number in range(1000):
    built_list.append(number)

print(f"📋 [Bucle Tradicional]: Longitud de la lista generada = {len(built_list)}")

# 🔌 2. Enfoque Pythonic con List Comprehension
comprehension_list: list[int] = [number for number in range(1000)]

print(f"⚡ [List Comprehension]: Longitud de la lista generada = {len(comprehension_list)}")
print(f"✅ ¿Ambas listas son idénticas en contenido? {built_list == comprehension_list}")
```

## 🔍 Filtrado de Elementos mediante Cláusula `if`

### 💡 Sintaxis de Filtrado:
Cuando la cláusula `if` se ubica al **final** de la expresión, actúa como un filtro (*filter*). El elemento solo se evalúa e incluye en la lista resultante si la condición devuelve `True`.

```python
# ================================================================================
# 📊 2. FILTRADO CON CONDICIONAL IF
# ================================================================================
print("\n--- 📊 [PASO 2]: Filtrado de Elementos (number > 50) ---")

# 📋 1. Enfoque Tradicional con Bucle for e if
filtered_built_list: list[int] = []
for number in range(1000):
    if number > 50:
        filtered_built_list.append(number)

print(f"📋 [Bucle Tradicional con filtro]: Elementos resultantes = {len(filtered_built_list)}")

# 🔌 2. Enfoque con List Comprehension y filtro if
filtered_comprehension_list: list[int] = [
    number for number in range(1000) if number > 50
]

print(f"⚡ [List Comprehension con filtro]: Elementos resultantes = {len(filtered_comprehension_list)}")
print(f"✅ Muestra de los primeros 5 elementos filtrados: {filtered_comprehension_list[:5]}")
```

## 🔀 Mapeo Condicional mediante `if...else`

### 💡 Sintaxis Ternaria:
Cuando se requiere asignar un valor alternativo si la condición no se cumple, la expresión condicional `expr_a if cond else expr_b` debe colocarse **antes** de la cláusula `for`.

```python
# ================================================================================
# 🔣 3. MAPEO CONDICIONAL CON IF-ELSE (Etiquetado de Paridad)
# ================================================================================
print("\n--- 🔣 [PASO 3]: Transformación Condicional (Par / Impar) ---")

# 📋 1. Enfoque Tradicional con Bucle for e if-else
parity_labels_built: list[str] = []
for number in range(10):
    if number % 2 == 0:
        parity_labels_built.append("par")
    else:
        parity_labels_built.append("impar")

print("📋 [Bucle Tradicional]:")
print(f"   {parity_labels_built}")

# 🔌 2. Enfoque con List Comprehension y Expresión Ternaria
parity_labels_comprehension: list[str] = [
    "par" if number % 2 == 0 else "impar" for number in range(10)
]

print("\n⚡ [List Comprehension con if-else]:")
print(f"   {parity_labels_comprehension}")

# --------------------------------------------------------------------------------
# 🏁 [PASO 4]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Módulo de 'List Comprehension' finalizado con éxito.")
print("=" * 80)
```
