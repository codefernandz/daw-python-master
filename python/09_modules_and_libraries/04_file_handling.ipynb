### 💡 El Manejo de Ficheros en Python

El trabajo con archivos de texto y binarios es una tarea fundamental en el desarrollo de software para garantizar la persistencia de datos. Python ofrece la función nativa **`open()`**, la cual retorna un objeto de tipo archivo (*file object*) que habilita operaciones de lectura, escritura y anexado de información.

---

### 🗂️ Modos Principales de Apertura de Ficheros

| Modo | Nombre | Descripción | Ejemplo de Uso |
| :--- | :--- | :--- | :--- |
| 📖 **`'r'`** | *Read* (Lectura) | Abre el archivo para lectura (modo por defecto). Lanza `FileNotFoundError` si no existe. | `open('data.txt', 'r')` |
| ✍️ **`'w'`** | *Write* (Escritura) | Abre el archivo para escritura. **Sobrescribe** completamente el contenido previo o crea el archivo si no existe. | `open('data.txt', 'w')` |
| ➕ **`'a'`** | *Append* (Anexar) | Abre el archivo para escribir al final del documento. Preserva el contenido existente y crea el archivo si no existe. | `open('data.txt', 'a')` |
| 🔄 **`'r+'`** | *Read/Write* | Abre el archivo permitiendo tanto operaciones de lectura como de escritura sobre el mismo puntero. | `open('data.txt', 'r+')` |

```python
from typing import Any

# 🛰️ [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 MANEJO DE FICHEROS | Lectura, Escritura y Administradores de Contexto")
print("=" * 80)

# ================================================================================
# 🧪 1. APERTURA, ESCRITURA Y CIERRE MANUAL DE FICHEROS (OPEN, WRITE, CLOSE)
# ================================================================================
# Uso explícito del ciclo de vida de un archivo mediante open() y close().
# ================================================================================

print("\n--- 🚀 [PASO 1]: Escritura y Cierre Manual de Ficheros ---")

sample_filename: str = "demo_file.txt"

# 🔌 [1] Apertura de un fichero en modo escritura ('w')
file_handle = open(sample_filename, "w", encoding="utf-8")

try:
    # 🔌 [2] Escritura de contenido textual en el fichero
    file_handle.write("Hola, mundo!\n")
    file_handle.write("Esta es una prueba de manejo nativo de ficheros en Python.\n")
    print(f"📋 Contenido escrito exitosamente en el archivo '{sample_filename}'.")
finally:
    # 🔌 [3] Cierre explícito del fichero para liberar recursos del sistema
    file_handle.close()
    print("  ✅ Archivo cerrado de forma explícita mediante close().")
```

## 🧱 Métodos de Lectura (`read`, `readline`, `readlines`) y Administradores de Contexto (`with`)

### 💡 Métodos de Procesamiento de Contenido:
- **`read(size)`:** Lee la totalidad del fichero como una sola cadena de texto (o $n$ caracteres si se especifica `size`).
- **`readline()`:** Lee una sola línea por llamada manteniendo el puntero de lectura posicionado.
- **`readlines()`:** Lee todas las líneas y devuelve una lista conteniendo cada renglón como una cadena individual.
- **`with open(...) as file:` (*Context Manager*):** Estructura recomendada que **garantiza el cierre automático** del fichero al finalizar el bloque, previniendo fugas de memoria o bloqueos de archivos.

```python
# ================================================================================
# 🧪 2. LECTURA Y USO DE CONTEXT MANAGERS (WITH STATEMENT)
# ================================================================================
print("\n--- 🚀 [PASO 2]: Lectura de Ficheros con Context Manager (with) ---")

# ⚙️ [1] Lectura completa mediante read() con sentencia 'with'
with open(sample_filename, "r", encoding="utf-8") as file_reader:
    full_content: str = file_reader.read()
    print("📋 Lectura completa del contenido (read()):")
    print(f"  🔹 Raw Content:\n{full_content}")

print("-" * 65)

# ⚙️ [2] Lectura línea por línea utilizando readline()
with open(sample_filename, "r", encoding="utf-8") as line_reader:
    first_line: str = line_reader.readline()
    second_line: str = line_reader.readline()
    print("📋 Lectura mediante líneas individuales (readline()):")
    print(f"  🔹 Línea 1: {first_line.strip()}")
    print(f"  🔹 Línea 2: {second_line.strip()}")

print("-" * 65)

# ⚙️ [3] Anexar datos sin sobrescribir utilizando el modo append ('a')
with open(sample_filename, "a", encoding="utf-8") as file_appender:
    file_appender.write("Tercera línea agregada con modo de anexado ('a').\n")
    print("  ✅ Nuevos datos anexados al final del archivo.")
```

### 🧪 Ejercicio 1: Procesador de Listas de Datos y Generación de Reportes

**Escenario Empresarial:** Una aplicación necesita procesar una lista de transacciones registradas en un archivo `.txt`, calcular métricas agregadas (total acumulado y promedio) y escribir un informe de resumen estructurado dentro de un nuevo archivo de reporte.

```python
import os

# ================================================================================
# 🧪 3. EJERCICIO PRÁCTICO 1: PROCESAMIENTO Y REPORTING DE DATOS EN FICHEROS
# ================================================================================
print("\n--- 🧪 [PASO 3.1]: Ejercicio 1 - Procesador de Transacciones Financieras ---")


def process_transaction_file(input_path: str, report_output_path: str) -> dict[str, Any]:
    """Lee un archivo de transacciones numéricas, calcula acumulados y genera un reporte formateado.

    Args:
        input_path (str): Ruta del archivo fuente con registros de importes.
        report_output_path (str): Ruta donde se escribirá el archivo de reporte.

    Returns:
        dict[str, Any]: Métricas calculadas procesadas.
    """
    metrics: dict[str, Any] = {"total_amount": 0.0, "transaction_count": 0, "average": 0.0}

    # [1] Validar existencia del fichero de entrada
    if not os.path.exists(input_path):
        print(f"❌ Error: El archivo fuente '{input_path}' no existe.")
        return metrics

    # [2] Leer y procesar las líneas del archivo
    with open(input_path, "r", encoding="utf-8") as input_file:
        for line in input_file:
            stripped_line: str = line.strip()
            if stripped_line:
                try:
                    amount: float = float(stripped_line)
                    metrics["total_amount"] += amount
                    metrics["transaction_count"] += 1
                except ValueError:
                    print(f"  ⚠️ Advertencia: Omitiendo valor no numérico '{stripped_line}'")

    # [3] Calcular promedio
    if metrics["transaction_count"] > 0:
        metrics["average"] = metrics["total_amount"] / metrics["transaction_count"]

    # [4] Escribir reporte en el archivo de salida
    with open(report_output_path, "w", encoding="utf-8") as report_file:
        report_file.write("=========================================\n")
        report_file.write("📊 REPORTE RESUMEN DE TRANSACCIONES\n")
        report_file.write("=========================================\n")
        report_file.write(f"Total de Transacciones: {metrics['transaction_count']}\n")
        report_file.write(f"Monto Acumulado: ${metrics['total_amount']:.2f}\n")
        report_file.write(f"Promedio por Transacción: ${metrics['average']:.2f}\n")

    return metrics


# Creación de archivo temporal con datos de prueba
transactions_file: str = "raw_transactions.txt"
report_file_path: str = "transactions_summary_report.txt"

with open(transactions_file, "w", encoding="utf-8") as source_writer:
    source_writer.write("150.50\n")
    source_writer.write("200.00\n")
    source_writer.write("50.25\n")
    source_writer.write("300.00\n")

print(f"📋 Procesando archivo de entradas: '{transactions_file}'...")
results: dict[str, Any] = process_transaction_file(transactions_file, report_file_path)

print(f"  ✅ Transacciones procesadas: {results['transaction_count']}")
print(f"  ✅ Monto total calculado:     ${results['total_amount']:.2f}")

# Mostrar contenido del reporte generado
if os.path.exists(report_file_path):
    print("\n📋 Contenido del reporte generado:")
    with open(report_file_path, "r", encoding="utf-8") as reader:
        print(reader.read())

# Limpieza de archivos del ejercicio
for temp_path in [transactions_file, report_file_path]:
    if os.path.exists(temp_path):
        os.remove(temp_path)
```

### 🧪 Ejercicio 2: Gestor y Rotador de Logs de Errores del Sistema

**Escenario Empresarial:** Un servicio backend genera archivos de registro (*logs*). Se requiere implementar una función que añada eventos de error con marcas de tiempo formateadas, inspeccione el tamaño o cantidad de líneas acumuladas y mueva las entradas a un archivo de historial comprimido o rotado cuando se supere el umbral máximo de almacenamiento.

```python
from datetime import datetime

# ================================================================================
# 🧪 4. EJERCICIO PRÁCTICO 2: ROTACIÓN Y AUDITORÍA DE ARCHIVOS DE LOG
# ================================================================================
print("\n--- 🧪 [PASO 3.2]: Ejercicio 2 - Rotador y Auditor de Logs ---")


def append_system_log(log_path: str, log_level: str, message: str, max_lines_threshold: int = 5) -> bool:
    """Anexa entradas a un archivo de log y realiza rotación si supera el límite de líneas.

    Args:
        log_path (str): Ruta del archivo activo de log.
        log_level (str): Nivel de evento (ej. 'INFO', 'ERROR', 'WARN').
        message (str): Mensaje descriptivo del evento.
        max_lines_threshold (int): Límite de líneas antes de rotar.

    Returns:
        bool: True si se realizó rotación de archivo, False en caso contrario.
    """
    timestamp: str = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    formatted_entry: str = f"[{timestamp}] [{log_level.upper()}] {message}\n"

    # [1] Anexar entrada al log activo
    with open(log_path, "a", encoding="utf-8") as log_file:
        log_file.write(formatted_entry)
        print(f"  📝 Registrado: {formatted_entry.strip()}")

    # [2] Contar líneas actuales para verificar umbral de rotación
    lines_count: int = 0
    with open(log_path, "r", encoding="utf-8") as log_file:
        lines_count = len(log_file.readlines())

    # [3] Rotar archivo si supera el umbral
    if lines_count >= max_lines_threshold:
        rotated_filename: str = f"rotated_{datetime.now().strftime('%Y%m%d_%H%M%S')}_{log_path}"
        os.rename(log_path, rotated_filename)
        print(f"  ⚠️ Umbral de {max_lines_threshold} líneas alcanzado. Archivo rotado a: '{rotated_filename}'")

        # Limpieza del archivo rotado para mantener el entorno limpio
        if os.path.exists(rotated_filename):
            os.remove(rotated_filename)

        return True

    return False


# Simulación de eventos en el sistema
active_log_file: str = "app_system.log"

print(f"📋 Escribiendo eventos en log activo: '{active_log_file}'")
append_system_log(active_log_file, "INFO", "Inicio del servicio web.")
append_system_log(active_log_file, "WARN", "Uso de memoria elevado (>80%).")
append_system_log(active_log_file, "ERROR", "Fallo de conexión con la base de datos.")
append_system_log(active_log_file, "INFO", "Reconexión exitosa.")
was_rotated: bool = append_system_log(active_log_file, "CRITICAL", "Límite de peticiones alcanzado.", max_lines_threshold=5)

print(f"  🔹 Estado final de rotación: {was_rotated}")

# Limpieza final si aún existe el log activo
if os.path.exists(active_log_file):
    os.remove(active_log_file)
if os.path.exists(sample_filename):
    os.remove(sample_filename)

# --------------------------------------------------------------------------------
# 🏁 [PASO 4]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Cuaderno de Manejo de Ficheros finalizado con éxito.")
print("=" * 80)
```
