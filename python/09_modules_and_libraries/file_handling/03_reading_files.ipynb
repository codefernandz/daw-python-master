### 💡 Métodos de Lectura de Archivos

Una vez abierto un fichero en modo lectura (`'r'`), Python proporciona diferentes métodos para extraer su contenido según la cantidad de datos y la gestión de memoria requerida.

---

### 🗂️ Comparativa de Métodos de Lectura

| Método | Descripción | Retorno | Casos de Uso Recomendados |
| :--- | :--- | :--- | :--- |
| 📖 **`read(size)`** | Lee todo el contenido del archivo de golpe (o $n$ caracteres/bytes si se pasa `size`). | `str` (o `bytes`) | Archivos pequeños cuyo contenido completo se necesita procesar junto. |
| 📑 **`readline()`** | Lee únicamente una línea del archivo hasta encontrar un salto de línea (`\n`). | `str` | Procesamiento paso a paso o análisis de archivos con líneas de encabezado. |
| 📋 **`readlines()`** | Lee todas las líneas y las empaqueta en una lista. | `list[str]` | Archivos medianos donde se requiere acceso por índice a líneas específicas. |

```python
import os
from typing import Any

# 🛰️ [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 MANEJO DE FICHEROS | Lectura con read(), readline() y readlines()")
print("=" * 80)

# ================================================================================
# 🧪 1. DEMOSTRACIÓN DE MÉTODOS DE LECTURA (READ Y READLINE)
# ================================================================================
# Métodos para consumir contenido completo o fragmentado por renglones.
# ================================================================================

print("\n--- 🚀 [PASO 1]: Métodos read() y readline() ---")

sample_file_name: str = "demo_reading_example.txt"

# Creación previa de un archivo multitexto de prueba
with open(sample_file_name, "w", encoding="utf-8") as setup_file:
    setup_file.write("Línea 1: Configuración inicial del servidor.\n")
    setup_file.write("Línea 2: Conexión establecida a la base de datos.\n")
    setup_file.write("Línea 3: Procesamiento de transacciones iniciado.\n")

try:
    # 🔌 [1] Lectura de una sola línea mediante readline()
    with open(sample_file_name, "r", encoding="utf-8") as line_file:
        first_line: str = line_file.readline()
        second_line: str = line_file.readline()
        
        print("📋 Lectura renglón por renglón (readline()):")
        print(f"  🔹 Renglón 1: {first_line.strip()}")
        print(f"  🔹 Renglón 2: {second_line.strip()}")

    print("-" * 65)

    # 🔌 [2] Lectura completa mediante read()
    with open(sample_file_name, "r", encoding="utf-8") as full_file:
        entire_content: str = full_file.read()
        
        print("📋 Lectura completa del archivo (read()):")
        print(f"  🔹 Contenido extraído:\n{entire_content}")

finally:
    # Limpieza del archivo temporal
    if os.path.exists(sample_file_name):
        os.remove(sample_file_name)
```

## 🧱 Procesamiento Eficiente de Archivos Grandes mediante Iteración Directa

### 💡 Buenas Prácticas de Memoria:
Cargar archivos de gran tamaño (GBs) mediante `read()` o `readlines()` puede agotar la memoria RAM disponible. La forma más **eficiente y recomendada en Python** para procesar archivos línea por línea es iterar directamente sobre el objeto de archivo (*file stream*), ya que carga cada renglón bajo demanda.

```python
# ================================================================================
# 🧪 2. ITERACIÓN EFICIENTE Y USO DE READLINES()
# ================================================================================
print("\n--- 🚀 [PASO 2]: Iteración Eficiente y readlines() ---")

large_file_demo: str = "large_log_simulation.txt"

# Generación de archivo con múltiples líneas
with open(large_file_demo, "w", encoding="utf-8") as setup_file:
    for index in range(1, 6):
        setup_file.write(f"LOG_ENTRY_00{index}: Evento de sistema procesado.\n")

# ⚙️ [1] Uso de readlines() para obtener una lista de renglones
with open(large_file_demo, "r", encoding="utf-8") as list_file:
    lines_list: list[str] = list_file.readlines()
    print(f"📋 Contenido cargado con readlines() ({len(lines_list)} elementos):")
    print(f"  🔹 Primer elemento de la lista: {lines_list[0].strip()}")

print("-" * 65)

# ⚙️ [2] Iteración directa sobre el file stream (Mejor práctica para memoria)
print("📋 Iteración directa sobre el objeto de archivo:")
with open(large_file_demo, "r", encoding="utf-8") as stream_file:
    for line_number, line_text in enumerate(stream_file, start=1):
        print(f"  🔹 [Línea {line_number}]: {line_text.strip()}")

# Limpieza del archivo de demostración
if os.path.exists(large_file_demo):
    os.remove(large_file_demo)
```

## 🛒 Ejercicios Prácticos Avanzados Integrados

---

### 🧪 Ejercicio 1: Extractor de Encabezados y Filtrado de Logs Críticos

**Escenario Empresarial:** Se requiere analizar un archivo de registros del sistema para extraer la línea de encabezado (*header*) utilizando `readline()`, e inspeccionar el resto del documento para identificar únicamente las líneas que contengan eventos clasificados como `[ERROR]`.

```python
# ================================================================================
# 🧪 3. EJERCICIO PRÁCTICO 1: PARSER Y FILTRADO DE EVENTOS CRÍTICOS
# ================================================================================
print("\n--- 🧪 [PASO 3.1]: Ejercicio 1 - Extractor de Errores de Logs ---")


def parse_and_filter_log_file(log_path: str) -> dict[str, Any]:
    """Extrae el encabezado de un archivo de log y filtra los eventos de error.

    Args:
        log_path (str): Ruta del archivo de registros a procesar.

    Returns:
        dict[str, Any]: Estructura con el encabezado y la lista de errores encontrados.
    """
    analysis_results: dict[str, Any] = {
        "header": "",
        "error_entries": [],
        "total_lines_processed": 0
    }

    if not os.path.exists(log_path):
        print(f"❌ Error: El archivo '{log_path}' no existe.")
        return analysis_results

    with open(log_path, "r", encoding="utf-8") as log_file:
        # [1] Extraer únicamente el encabezado con readline()
        analysis_results["header"] = log_file.readline().strip()

        # [2] Iterar eficientemente sobre el resto de las líneas
        for line in log_file:
            analysis_results["total_lines_processed"] += 1
            if "[ERROR]" in line:
                analysis_results["error_entries"].append(line.strip())

    return analysis_results


# Creación de archivo ficticio de log
system_log_path: str = "app_server.log"
with open(system_log_path, "w", encoding="utf-8") as log_writer:
    log_writer.write("SYSTEM LOG V2.1 - FECHA: 2026-08-14\n")
    log_writer.write("[INFO] Servidor iniciado correctamente.\n")
    log_writer.write("[ERROR] Fallo al conectar a la base de datos MySQL.\n")
    log_writer.write("[WARN] Latencia de red alta en la petición REQ-104.\n")
    log_writer.write("[ERROR] Timeout excedido en la pasarela de pagos.\n")

print(f"📋 Procesando registros en: '{system_log_path}'")
log_report: dict[str, Any] = parse_and_filter_log_file(system_log_path)

print(f"  🔹 Encabezado extraído:  '{log_report['header']}'")
print(f"  🔹 Líneas procesadas:   {log_report['total_lines_processed']}")
print(f"  🚨 Errores detectados ({len(log_report['error_entries'])}):")
for error in log_report["error_entries"]:
    print(f"     ❌ {error}")

# Limpieza del archivo de prueba
if os.path.exists(system_log_path):
    os.remove(system_log_path)

# --------------------------------------------------------------------------------
# 🏁 [PASO 4]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Lección de Lectura de Ficheros finalizada.")
print("=" * 80)
```
