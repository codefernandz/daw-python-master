| 🛠️ **Conceptos** | `write()`, `writelines()`, `Modo Escritura ('w')`, `Modo Anexar ('a')`, `Buffer de Escritura (flush)` |

---

### 💡 Escritura de Ficheros en Python

Para guardar información en un archivo mediante Python, es imprescindible abrir el fichero en un modo que permita operaciones de salida. El método **`write()`** recibe una cadena de texto (*string*) y la escribe en el fichero, mientras que **`writelines()`** permite escribir una lista iterable de cadenas.

---

### 🗂️ Comportamiento según el Modo de Apertura

| Modo | Nombre | Efecto sobre Ficheros Existentes | Efecto si el Fichero NO Existe |
| :--- | :--- | :--- | :--- |
| ✍️ **`'w'`** | *Write* | **Borra todo el contenido** previo y escribe desde el inicio. | Crea un archivo nuevo automáticamente. |
| ➕ **`'a'`** | *Append* | Preserva los datos existentes y escribe al **final** del documento. | Crea un archivo nuevo automáticamente. |

```python
import os
from typing import Any

# 🛰️ [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 MANEJO DE FICHEROS | Escritura con write(), writelines() y modos 'w' / 'a'")
print("=" * 80)

# ================================================================================
# 🧪 1. DEMOSTRACIÓN DE ESCRITURA Y ANEXADO (WRITE, MODE 'w' vs MODE 'a')
# ================================================================================
# Diferencias prácticas entre sobrescribir y anexar información en ficheros.
# ================================================================================

print("\n--- 🚀 [PASO 1]: Escritura Básica y Anexado de Contenido ---")

target_filename: str = "demo_writing_example.txt"

# 🔌 [1] Escritura inicial en modo 'w' (Sobrescribe / Crea el archivo)
with open(target_filename, "w", encoding="utf-8") as file_writer:
    bytes_written_1: int = file_writer.write("Hola, mundo!\n")
    bytes_written_2: int = file_writer.write("Primera línea guardada en el archivo.\n")
    print(f"📋 Contenido escrito en modo 'w' ({bytes_written_1 + bytes_written_2} caracteres).")

# Leer el contenido para verificar
with open(target_filename, "r", encoding="utf-8") as file_reader:
    print(f"  🔹 Contenido actual:\n{file_reader.read().strip()}")

print("-" * 65)

# 🔌 [2] Anexado de información en modo 'a' (Append)
with open(target_filename, "a", encoding="utf-8") as file_appender:
    file_appender.write("Segunda línea añadida usando el modo anexar ('a').\n")
    print("  ✅ Nuevos datos anexados sin borrar el contenido anterior.")

# Leer nuevamente para verificar la preservación de datos
with open(target_filename, "r", encoding="utf-8") as file_reader:
    print(f"  🔹 Contenido tras anexar:\n{file_reader.read().strip()}")

# Limpieza del archivo temporal
if os.path.exists(target_filename):
    os.remove(target_filename)
```

## 🧱 Escritura Múltiple con `writelines()`

### 💡 El Método `writelines()`:
Permite escribir una secuencia de cadenas (como una lista de `str`) directamente en el fichero. 

> ⚠️ **Nota:** `writelines()` **no agrega saltos de línea (`\n`) automáticamente** entre cada elemento de la lista; deben ser incluidos dentro de los textos antes de llamar al método.

```python
# ================================================================================
# 🧪 2. ESCRITURA ESTRUCTURADA DE LISTAS CON WRITELINES()
# ================================================================================
print("\n--- 🚀 [PASO 2]: Escritura Múltiple con writelines() ---")

list_filename: str = "demo_writelines_example.txt"

# ⚙️ [1] Definición de lista de renglones formateados previamente
data_lines: list[str] = [
    "Servidor Alpha: Activo\n",
    "Servidor Beta: En Mantenimiento\n",
    "Servidor Gamma: Activo\n"
]

# ⚙️ [2] Escritura de toda la lista en una sola operación
with open(list_filename, "w", encoding="utf-8") as multi_writer:
    multi_writer.writelines(data_lines)
    print(f"📋 {len(data_lines)} renglones escritos exitosamente mediante writelines().")

# Lectura de verificación
with open(list_filename, "r", encoding="utf-8") as multi_reader:
    print(f"  🔹 Contenido guardado:\n{multi_reader.read().strip()}")

# Limpieza del archivo de prueba
if os.path.exists(list_filename):
    os.remove(list_filename)
```

### 🧪 Ejercicio 1: Generador y Acumulador de Reportes de Diagnóstico

**Escenario Empresarial:** Una herramienta de monitorización necesita registrar diagnósticos continuos. Debe permitir inicializar un archivo de informe diario sobrescribiendo datos viejos (`'w'`) si es un nuevo día, y añadir nuevos registros de auditoría (`'a'`) sin perder el histórico acumulado durante la jornada.

```python
from datetime import datetime

# ================================================================================
# 🧪 3. EJERCICIO PRÁCTICO 1: ACUMULADOR Y REGISTRO DE AUDITORÍA
# ================================================================================
print("\n--- 🧪 [PASO 3.1]: Ejercicio 1 - Generador de Registros de Auditoría ---")


def append_audit_event(report_path: str, component: str, status: str) -> bool:
    """Anexa un evento de diagnóstico formateado a un informe existente.

    Args:
        report_path (str): Ruta del archivo donde se registra el evento.
        component (str): Nombre del componente evaluado.
        status (str): Estado del componente ('SUCCESS', 'FAILURE').

    Returns:
        bool: True si la operación de escritura fue exitosa.
    """
    timestamp: str = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    formatted_record: str = f"[{timestamp}] | COMPONENTE: {component:<12} | ESTADO: {status}\n"

    try:
        # Modo 'a' para conservar registros anteriores
        with open(report_path, "a", encoding="utf-8") as audit_file:
            audit_file.write(formatted_record)
            print(f"  📝 Evento registrado: {component} -> {status}")
        return True
    except Exception as error:
        print(f"❌ Error al escribir en el registro: {error}")
        return False


# [1] Inicialización del informe del día (modo 'w')
daily_report_file: str = "daily_audit_report.txt"

with open(daily_report_file, "w", encoding="utf-8") as header_writer:
    header_writer.write("=== INFORME DE AUDITORÍA DE SISTEMAS ===\n")
    print(f"📋 Informe inicializado en: '{daily_report_file}'")

# [2] Registro secuencial de eventos de auditoría (modo 'a')
append_audit_event(daily_report_file, "Database", "SUCCESS")
append_audit_event(daily_report_file, "AuthService", "SUCCESS")
append_audit_event(daily_report_file, "PaymentGateway", "FAILURE")

print("-" * 65)

# [3] Verificación del contenido completo escrito
print("📋 Contenido final del informe acumulado:")
with open(daily_report_file, "r", encoding="utf-8") as final_reader:
    print(final_reader.read())

# Limpieza del archivo de prueba
if os.path.exists(daily_report_file):
    os.remove(daily_report_file)

# --------------------------------------------------------------------------------
# 🏁 [PASO 4]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Lección de Escritura en Ficheros finalizada.")
print("=" * 80)
```
