### 💡 La Función `open()` en Python

La función integrada **`open()`** es el punto de entrada para interactuar con cualquier archivo en el sistema de archivos desde Python. Retorna un objeto de tipo *file stream* (o *file handle*) que actúa como un puente para ejecutar operaciones posteriores como leer, escribir o modificar datos.

---

### 🗂️ Modos Principales de Apertura

| Modo | Descripción | Comportamiento si el archivo NO existe | Comportamiento si el archivo SI existe |
| :--- | :--- | :--- | :--- |
| 📖 **`'r'`** | Lectura (Default) | Lanza un error `FileNotFoundError`. | Abre el archivo posicionado al inicio. |
| ✍️ **`'w'`** | Escritura | Crea un archivo nuevo. | **Sobrescribe** borrando todo el contenido previo. |
| ➕ **`'a'`** | Anexar (*Append*) | Crea un archivo nuevo. | Abre el archivo posicionado al **final** para preservar datos. |
| 🔄 **`'x'`** | Creación Exclusiva | Crea un archivo nuevo. | Lanza un error `FileExistsError`. |
| 🔢 **`'b'`** | Modo Binario | Se combina con otros modos (`'rb'`, `'wb'`) para imágenes o ejecutables. | Operaciones basadas en `bytes` en lugar de `str`. |

```python
import os
from typing import Any

# 🛰️ [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 MANEJO DE FICHEROS | Apertura de Archivos con open()")
print("=" * 80)

# ================================================================================
# 🧪 1. DEMOSTRACIÓN DE APERTURA EN MODO LECTURA ('r')
# ================================================================================
# Instanciación de un file stream usando open() de forma segura.
# ================================================================================

print("\n--- 🚀 [PASO 1]: Apertura Básica de Ficheros ---")

sample_file_name: str = "demo_open_example.txt"

# Creación previa de un archivo de prueba
with open(sample_file_name, "w", encoding="utf-8") as setup_file:
    setup_file.write("Contenido de prueba para la lección de apertura de archivos.")

try:
    # 🔌 [1] Apertura del fichero en modo lectura ('r') con codificación explícita
    file_stream = open(sample_file_name, "r", encoding="utf-8")
    
    print(f"📋 Objeto de archivo instanciado: {file_stream}")
    print(f"  🔹 Nombre del archivo: {file_stream.name}")
    print(f"  🔹 Modo de apertura:  {file_stream.mode}")
    print(f"  🔹 Codificación:      {file_stream.encoding}")

finally:
    # 🔌 [2] Liberación manual del recurso
    if 'file_stream' in locals() and not file_stream.closed:
        file_stream.close()
        print("  ✅ Archivo cerrado de manera explícita.")

    # Limpieza del archivo temporal
    if os.path.exists(sample_file_name):
        os.remove(sample_file_name)
```

## 🧱 Verificación Preventiva de Errores al Abrir Archivos

### 💡 Manejo de Excepciones con `FileNotFoundError`:
Al abrir un fichero en modo lectura (`'r'`), si el archivo no existe en el directorio especificado, Python detendrá la ejecución lanzando un error `FileNotFoundError`. Es una buena práctica envolver la apertura en un bloque `try-except` o verificar previamente su existencia con `os.path.exists()`.

```python
# ================================================================================
# 🧪 2. CAPTURA DE ERRORES AL ABRIR ARCHIVOS INEXISTENTES
# ================================================================================
print("\n--- 🚀 [PASO 2]: Manejo Seguro de Apertura con Excepciones ---")

non_existent_file: str = "archivo_inexistente.txt"

# ⚙️ [1] Intento de apertura controlado con bloque try-except
try:
    print(f"📋 Intentando abrir el archivo '{non_existent_file}'...")
    file_stream = open(non_existent_file, "r", encoding="utf-8")
except FileNotFoundError as error:
    print(f"  ❌ Excepción capturada exitosamente: No se encontró el archivo.")
    print(f"  🔍 Detalle técnico: {error}")
```

### 🧪 Ejercicio 1: Apertura y Validación de Archivos de Configuración del Sistema

**Escenario Empresarial:** Una aplicación necesita abrir un archivo de configuración crítico (`config.json` o `.env`). Debe validar que la ruta existe, intentar instanciar el flujo de apertura en modo lectura y retornar un reporte con los metadatos del objeto de archivo creado sin provocar caídas en el sistema.

```python
# ================================================================================
# 🧪 3. EJERCICIO PRÁCTICO 1: VALIDADOR Y APERTURA SEGURA DE ARCHIVOS
# ================================================================================
print("\n--- 🧪 [PASO 3.1]: Ejercicio 1 - Gestor Seguro de Apertura de Configuración ---")


def safe_open_configuration_file(file_path: str, mode: str = "r") -> dict[str, Any]:
    """Valida la existencia de un archivo y gestiona su apertura de forma segura.

    Args:
        file_path (str): Ruta completa o relativa del archivo a abrir.
        mode (str): Modo de apertura deseado ('r', 'w', 'a').

    Returns:
        dict[str, Any]: Reporte de estado con el objeto de archivo si fue exitoso o el error.
    """
    execution_report: dict[str, Any] = {
        "success": False,
        "file_object": None,
        "error_message": None
    }

    try:
        # [1] Intentar la apertura del fichero
        file_stream = open(file_path, mode, encoding="utf-8")
        execution_report["success"] = True
        execution_report["file_object"] = file_stream
        print(f"  ✅ Archivo '{file_path}' abierto exitosamente en modo '{mode}'.")
    except FileNotFoundError:
        execution_report["error_message"] = f"El archivo '{file_path}' no fue encontrado en el sistema."
        print(f"  ❌ Error: {execution_report['error_message']}")
    except PermissionError:
        execution_report["error_message"] = f"Permisos insuficientes para abrir '{file_path}'."
        print(f"  ❌ Error: {execution_report['error_message']}")
    except Exception as general_error:
        execution_report["error_message"] = f"Error insospechado: {general_error}"
        print(f"  ❌ Error: {execution_report['error_message']}")

    return execution_report


# Caso de Prueba 1: Archivo inexistente
print("📋 Prueba 1: Apertura de archivo inexistente")
report_failed = safe_open_configuration_file("settings_missing.conf")

print("-" * 65)

# Caso de Prueba 2: Archivo válido
valid_config_path: str = "app_settings.conf"
with open(valid_config_path, "w", encoding="utf-8") as temp_file:
    temp_file.write("ENV=PRODUCTION")

print("📋 Prueba 2: Apertura de archivo existente")
report_success = safe_open_configuration_file(valid_config_path)

# [2] Garantizar el cierre del archivo abierto si la prueba fue exitosa
if report_success["success"] and report_success["file_object"]:
    report_success["file_object"].close()
    print("  🔒 Recurso de archivo cerrado correctamente.")

# Limpieza del archivo de prueba
if os.path.exists(valid_config_path):
    os.remove(valid_config_path)

# --------------------------------------------------------------------------------
# 🏁 [PASO 4]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Lección de Apertura de Ficheros finalizada.")
print("=" * 80)
```
