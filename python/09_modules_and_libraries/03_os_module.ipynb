### 💡 El Módulo `os` en Python

El módulo estándar **`os`** proporciona una interfaz portable para interactuar con las funcionalidades dependientes del sistema operativo en el que se ejecuta Python. Permite manipular rutas de forma multiplataforma, listar el contenido de carpetas, crear/eliminar recursos y operar sobre archivos a bajo nivel mediante descriptores numéricos (*file descriptors*).

---

### 🗂️ Principales Métodos del Módulo

| Método / Función | Descripción | Ejemplo de Uso |
| :--- | :--- | :--- |
| 📄 **`os.open()` / `os.close()`** | Abre/cierra un archivo devolviendo/liberando un descriptor de bajo nivel. | `file_descriptor = os.open('file.txt', os.O_RDONLY)` |
| 📖 **`os.read()`** | Lee $n$ bytes de un archivo usando su descriptor numérico. | `data_bytes = os.read(file_descriptor, 100)` |
| 📂 **`os.getcwd()` / `os.chdir()`** | Obtiene o cambia el directorio de trabajo actual del proceso. | `current_dir = os.getcwd()` |
| 📋 **`os.listdir()`** | Lista todos los archivos y subdirectorios dentro de una ruta. | `file_list = os.listdir('./my_folder')` |
| 🔨 **`os.mkdir()` / `os.rmdir()`** | Crea o elimina carpetas (esta última debe estar vacía). | `os.mkdir('new_directory')` |
| 🗑️ **`os.remove()`** | Elimina permanentemente un archivo específico del sistema. | `os.remove('temp_file.txt')` |
| 🛠️ **`os.path.join()`** | Concatena componentes de ruta con los separadores correctos (`/` o `\`). | `full_path = os.path.join(base_dir, filename)` |
| 🔍 **`os.path.exists()`** | Evalúa si una ruta de archivo o directorio realmente existe. | `path_exists = os.path.exists('/path/to/check')` |

```python
import os
from typing import Any

# 🛰️ [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 MÓDULO OS | Interacción con el Sistema Operativo y Rutas")
print("=" * 80)

# ================================================================================
# 🧪 1. LECTURA DE BAJO NIVEL CON DESCRIPTORES DE ARCHIVO (OPEN, READ, CLOSE)
# ================================================================================
# Operaciones sobre descriptores numéricos de bajo nivel.
# ================================================================================

print("\n--- 🚀 [PASO 1]: Operaciones de Bajo Nivel sobre Archivos ---")

# Preparamos un archivo temporal para la demostración
sample_filename: str = "temp_demo_file.txt"
with open(sample_filename, "w", encoding="utf-8") as file_handle:
    file_handle.write("Python OS Module - Demostración de lectura a bajo nivel.")

try:
    # 🔌 [1] Apertura del archivo en modo lectura a bajo nivel (Obtiene el File Descriptor)
    file_descriptor: int = os.open(sample_filename, os.O_RDONLY)
    print(f"📋 Descriptor de archivo (File Descriptor ID): {file_descriptor}")

    # 🔌 [2] Lectura de los primeros 100 bytes utilizando el descriptor numérico
    raw_bytes_data: bytes = os.read(file_descriptor, 100)
    decoded_text_content: str = raw_bytes_data.decode("utf-8")
    print(f"  🔹 Contenido leído y decodificado: '{decoded_text_content}'")

finally:
    # 🔌 [3] Cierre obligatorio del descriptor de archivo
    if 'file_descriptor' in locals():
        os.close(file_descriptor)
        print("  ✅ Descriptor de archivo cerrado correctamente.")

    # Limpieza del archivo de demostración
    if os.path.exists(sample_filename):
        os.remove(sample_filename)
```

## 🧱 Navegación y Gestión del Sistema de Archivos

### 💡 Métodos de Directorios y Verificación de Rutas:
- **`getcwd()` (*Get Current Working Directory*):** Muestra el directorio de trabajo en ejecución.
- **`chdir(path)` (*Change Directory*):** Cambia el contexto de ejecución a una nueva ruta.
- **`listdir(path)`:** Obtiene un arreglo con los nombres de todos los elementos presentes en una carpeta.
- **`os.path.join()`:** Permite construir rutas multiplataforma evitando errores de diagonales (`/` vs `\`).

```python
# ================================================================================
# 🧪 2. NAVEGACIÓN Y OPERACIONES SOBRE DIRECTORIOS
# ================================================================================
print("\n--- 🚀 [PASO 2]: Navegación y Construcción de Rutas ---")

# ⚙️ [1] Obtención del directorio de trabajo actual
current_working_directory: str = os.getcwd()
print(f"📋 Directorio de trabajo actual (getcwd): {current_working_directory}")

print("-" * 65)

# ⚙️ [2] Construcción segura y multiplataforma de rutas con os.path.join
target_folder_name: str = "temp_workspace_dir"
full_directory_path: str = os.path.join(current_working_directory, target_folder_name)
print(f"📋 Ruta completa construida (path.join): {full_directory_path}")

# ⚙️ [3] Verificación de existencia y creación de carpeta
if not os.path.exists(full_directory_path):
    os.mkdir(full_directory_path)
    print(f"  ✅ Directorio creado exitosamente: {target_folder_name}")
else:
    print(f"  ⚠️ El directorio ya existe: {target_folder_name}")

print("-" * 65)

# ⚙️ [4] Listado de contenido del directorio actual
directory_contents: list[str] = os.listdir(current_working_directory)
print(f"📋 Archivos y carpetas detectadas en la ruta ({len(directory_contents)} elementos):")
for item_name in directory_contents[:5]:  # Muestra los primeros 5 elementos
    print(f"  🔹 Elemento: {item_name}")

print("-" * 65)

# ⚙️ [5] Demostración de cambio de directorio temporal
os.chdir(full_directory_path)
print(f"  🔄 Nuevo directorio de trabajo tras chdir(): {os.getcwd()}")

# Restauración del directorio original
os.chdir(current_working_directory)
print(f"  🔄 Restauración al directorio base: {os.getcwd()}")

# Limpieza del directorio de pruebas
if os.path.exists(full_directory_path):
    os.rmdir(full_directory_path)
    print(f"  🗑️ Directorio temporal eliminado correctamente: {target_folder_name}")
```

### 🧪 Ejercicio 1: Automatizador de Auditoría y Organización de Archivos

**Escenario Empresarial:** Un servidor recibe continuamente archivos de logs y reportes. Se requiere construir una función automatizada que escanee un directorio de entrada, identifique los archivos con extensión `.log` o `.txt`, cree carpetas organizadas por extensión y mueva/clasifique los archivos existentes verificando la presencia de las rutas destino.

```python
# ================================================================================
# 🧪 3. EJERCICIO PRÁCTICO 1: ORGANIZADOR AUTOMÁTICO DE ARCHIVOS DE AUDITORÍA
# ================================================================================
print("\n--- 🧪 [PASO 3.1]: Ejercicio 1 - Clasificación Autómata de Archivos ---")


def organize_directory_files(target_dir: str) -> dict[str, int]:
    """Escanea un directorio y organiza archivos creando carpetas según su extensión.

    Args:
        target_dir (str): Ruta base del directorio a procesar.

    Returns:
        dict[str, int]: Métricas con la cantidad de archivos organizados por categoría.
    """
    processing_metrics: dict[str, int] = {"processed": 0, "directories_created": 0}

    # [1] Verificar si el directorio objetivo existe
    if not os.path.exists(target_dir):
        print(f"❌ Error: El directorio objetivo '{target_dir}' no existe.")
        return processing_metrics

    # [2] Listar elementos presentes en la ruta
    dir_items: list[str] = os.listdir(target_dir)

    for item_name in dir_items:
        item_full_path: str = os.path.join(target_dir, item_name)

        # Filtrar para procesar únicamente archivos regulares
        if not os.path.isdir(item_full_path):
            # Extraer extensión de archivo
            file_extension: str = item_name.split(".")[-1].lower() if "." in item_name else "no_extension"

            # Definir carpeta destino
            category_dir_name: str = f"folder_{file_extension}"
            category_dir_path: str = os.path.join(target_dir, category_dir_name)

            # [3] Crear la carpeta de categoría si no existe
            if not os.path.exists(category_dir_path):
                os.mkdir(category_dir_path)
                processing_metrics["directories_created"] += 1

            # [4] Simular/Ejecutar reubicación del archivo
            new_file_path: str = os.path.join(category_dir_path, item_name)
            # Reubicación utilizando renombrado de bajo nivel (os.rename)
            os.rename(item_full_path, new_file_path)
            processing_metrics["processed"] += 1

    return processing_metrics


# Entorno de prueba aislado
sandbox_dir: str = os.path.join(os.getcwd(), "sandbox_audit_environment")
if not os.path.exists(sandbox_dir):
    os.mkdir(sandbox_dir)

# Creación de archivos ficticios de prueba
test_files: list[str] = ["system_audit.log", "data_report.txt", "network_trace.log", "backup.csv"]
for single_file in test_files:
    file_path: str = os.path.join(sandbox_dir, single_file)
    with open(file_path, "w", encoding="utf-8") as fh:
        fh.write("Contenido de prueba de auditoría.")

print(f"📋 Ejecutando organizador autómata en: '{sandbox_dir}'")
results: dict[str, int] = organize_directory_files(sandbox_dir)
print(f"  ✅ Archivos clasificados y movidos: {results['processed']}")
print(f"  📁 Nuevas carpetas estructuradas:   {results['directories_created']}")

# Limpieza total del entorno de pruebas tras el ejercicio
for root_path, dirs_list, files_list in os.walk(sandbox_dir, topdown=False):
    for f in files_list:
        os.remove(os.path.join(root_path, f))
    for d in dirs_list:
        os.rmdir(os.path.join(root_path, d))
os.rmdir(sandbox_dir)
print("  🗑️ Entorno de prueba aislado de la auditoría eliminado con éxito.")
```

### 🧪 Ejercicio 2: Gestor Seguro de Limpieza de Archivos Temporales y Descargas

**Escenario Empresarial:** Los servidores web acumulan constantemente logs e imágenes temporales. Se necesita un sistema de purga automatizado que inspeccione un directorio de almacenamiento temporal, valide la existencia de archivos que coincidan con un sufijo o extensión dada, elimine de forma segura cada archivo mediante `os.remove()` e intente remover la carpeta contenedor si esta queda completamente vacía.

```python
# ================================================================================
# 🧪 4. EJERCICIO PRÁCTICO 2: PURGA SEGURA DE ALMACENAMIENTO TEMPORAL
# ================================================================================
print("\n--- 🧪 [PASO 3.2]: Ejercicio 2 - Limpieza de Archivos Temporales ---")


def purge_temporary_storage(target_directory: str, file_extension: str) -> dict[str, Any]:
    """Elimina de forma segura archivos temporales con una extensión determinada y limpia carpetas vacías.

    Args:
        target_directory (str): Ruta base donde se realizará la purga.
        file_extension (str): Extensión de los archivos a eliminar (ej. '.tmp').

    Returns:
        dict[str, Any]: Reporte de auditoría de archivos eliminados y estado de limpieza.
    """
    purge_summary: dict[str, Any] = {
        "deleted_files": [],
        "errors": [],
        "directory_removed": False
    }

    # [1] Validar que la ruta especificada existe
    if not os.path.exists(target_directory):
        purge_summary["errors"].append(f"La ruta '{target_directory}' no existe.")
        return purge_summary

    # [2] Iterar sobre los archivos del directorio objetivo
    for item_filename in os.listdir(target_directory):
        full_item_path: str = os.path.join(target_directory, item_filename)

        # [3] Verificar si es un archivo que coincide con la extensión requerida
        if os.path.isfile(full_item_path) and item_filename.endswith(file_extension):
            try:
                os.remove(full_item_path)
                purge_summary["deleted_files"].append(item_filename)
                print(f"  🗑️ Archivo eliminado: {item_filename}")
            except OSError as error:
                purge_summary["errors"].append(f"No se pudo eliminar {item_filename}: {error}")

    # [4] Evaluar si el directorio quedó completamente vacío para eliminarlo
    if len(os.listdir(target_directory)) == 0:
        try:
            os.rmdir(target_directory)
            purge_summary["directory_removed"] = True
            print(f"  ✅ Directorio vacío eliminado exitosamente: {target_directory}")
        except OSError as error:
            purge_summary["errors"].append(f"Error eliminando directorio: {error}")

    return purge_summary


# Creación de directorio temporal de prueba
temp_storage_dir: str = os.path.join(os.getcwd(), "temp_cache_data")
if not os.path.exists(temp_storage_dir):
    os.mkdir(temp_storage_dir)

# Creación de archivos temporales ficticios
dummy_cache_files: list[str] = ["session_01.tmp", "session_02.tmp", "important_data.csv"]
for cache_file in dummy_cache_files:
    cache_path: str = os.path.join(temp_storage_dir, cache_file)
    with open(cache_path, "w", encoding="utf-8") as fh:
        fh.write("Datos temporales en caché.")

print(f"📋 Iniciando purga de archivos '.tmp' en: '{temp_storage_dir}'")
audit_report: dict[str, Any] = purge_temporary_storage(temp_storage_dir, ".tmp")

print(f"  🔹 Total de archivos borrados: {len(audit_report['deleted_files'])}")
print(f"  🔹 ¿Se eliminó la carpeta raíz?: {audit_report['directory_removed']}")

# Limpieza final de seguridad para archivos no purgados (.csv)
if os.path.exists(temp_storage_dir):
    for remaining_file in os.listdir(temp_storage_dir):
        os.remove(os.path.join(temp_storage_dir, remaining_file))
    os.rmdir(temp_storage_dir)

# --------------------------------------------------------------------------------
# 🏁 [PASO 4]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Cuaderno del Módulo OS finalizado con éxito.")
print("=" * 80)
```
