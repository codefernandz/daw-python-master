### 💡 ¿Qué es una Librería en Python?

Una **librería** o **módulo** en Python es un conjunto de variables, funciones, clases y herramientas predefinidas almacenadas en archivos estructurados (`.py`). Nos permite reutilizar código existente e integrarlo en nuestras aplicaciones sin necesidad de reescribirlo desde cero.

---

### 🗂️ Clasificación de Librerías

1. 📦 **Librerías Integradas (*Standard Library*):**
   Vienen preinstaladas con la distribución estándar de Python y no requieren instalación adicional.
   - 🧮 `math`: Operaciones matemáticas avanzadas y constantes.
   - 💻 `os` / `sys`: Interacción con el sistema operativo y entorno de ejecución.
   - ⏱️ `datetime`: Manipulación de fechas y tiempos.

2. 🌐 **Librerías Externas (*Third-Party Libraries*):**
   Desarrolladas por la comunidad y alojadas en [PyPI (Python Package Index)](https://pypi.org/). Se instalan manualmente utilizando el gestor de paquetes `pip`.
   - 📊 `numpy`: Cálculo numérico vectorizado y álgebra lineal.
   - 📈 `pandas`: Análisis, manipulación y limpieza de datos estructurados (DataFrames).

```bash
# Comando de instalación desde la terminal / CLI
pip install numpy pandas
```

---

### 📐 Sintaxis de Importación

Existen dos estrategias principales para importar y utilizar librerías en Python:

#### 1️⃣ Importación Completa de la Librería (`import library_name`)
Importa el espacio de nombres completo. Para acceder a una función o clase se utiliza la notación de punto (`library_name.function_name()`).

#### 2️⃣ Importación de Funcionalidades Concretas (`from library_name import function_name`)
Importa únicamente los miembros específicos indicados. Permite invocarlos directamente sin prefijar el nombre del módulo.

---

### 📊 Comparativa de Sintaxis de Importación

| Sintaxis | Ejemplo de Código | Ventaja | Riesgo / Desventaja |
| :--- | :--- | :--- | :--- |
| `import math` | `math.sqrt(36)` | Evita colisiones de nombres (namespace clean). | Código ligeramente más extenso al escribir. |
| `from math import sqrt` | `sqrt(36)` | Código más conciso e intencional. | Riesgo de sobreescribir funciones locales con el mismo nombre. |
| `import pandas as pd` | `pd.DataFrame(...)` | Alias estandarizados y breves. | Requiere conocer las convenciones de la comunidad. |
```

````python
from typing import Any

# 🛰️ [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 MÓDULOS Y LIBRERÍAS | Introducción e Importaciones Básicas")
print("=" * 80)

# ================================================================================
# 🧪 1. IMPORTACIÓN COMPLETA DE MÓDULOS (ESTRATEGIA 1)
# ================================================================================
# Importamos el módulo 'math' completo de la Standard Library.
# ================================================================================

print("\n--- 🚀 [PASO 1]: Importación Completa (import math) ---")

# ⚙️ Importación del módulo estándar de matemáticas
import math

# 🔌 [1] Definición de variable numérica
number_1: float = 36.0

# ⚙️ [2] Cálculo de la raíz cuadrada invocando math.sqrt() mediante notación de punto
square_root_1: float = math.sqrt(number_1)

print(f"📋 Número analizado: {number_1}")
print(f"  ✅ La raíz cuadrada calculada con math.sqrt() es: {square_root_1:.2f}")

# 🔌 [3] Uso de constantes integradas en la librería
print(f"  ℹ️ Valor de la constante Pi (math.pi): {math.pi:.5f}")
````

## 🧱 Importación de Funcionalidades Específicas

### 💡 Uso de `from ... import ...`:
Al importar únicamente los miembros requeridos (`sqrt`, `pow`), estos se incorporan al espacio de nombres local y pueden ejecutarse directamente.

````python
# ================================================================================
# 🧪 2. IMPORTACIÓN DE FUNCIONALIDADES ESPECÍFICAS (ESTRATEGIA 2)
# ================================================================================
print("\n--- 🚀 [PASO 2]: Importación Específica (from math import sqrt, pow) ---")

# ⚙️ Importación selectiva de funciones requeridas
from math import pow, sqrt

# 🔌 [1] Definición de variable numérica
number_2: float = 49.0

# ⚙️ [2] Invocación directa de la función sqrt() sin prefijo de librería
square_root_2: float = sqrt(number_2)

# ⚙️ [3] Uso de pow() importado de math
power_result: float = pow(square_root_2, 2)

print(f"📋 Número analizado: {number_2}")
print(f"  ✅ La raíz cuadrada calculada directamente con sqrt() es: {square_root_2:.2f}")
print(f"  ✅ Potencia calculada con pow({square_root_2}, 2):         {power_result:.2f}")
````

## 📊 Uso de Alias (`as`) e Inspección de Módulos

### 💡 Asignación de Sobrenombres o Alias:
Es una convención estándar en Python asignar alias cortos a librerías para agilizar la escritura del código (ej. `import os as operating_system`).

````python
# ================================================================================
# 📊 3. USO DE ALIAS E INSPECCIÓN DEL SISTEMA OPERATIVO WITH OS
# ================================================================================
print("\n--- 📊 [PASO 3]: Uso de Alias con 'import os as op_sys' ---")

# ⚙️ Importación con alias para evitar ambigüedades
import os as op_sys

# 🔌 [1] Obtención del directorio de trabajo actual
current_directory: str = op_sys.getcwd()

# 🔌 [2] Obtención del nombre del sistema operativo
os_name: str = op_sys.name

print(f"📋 Nombre del sistema operativo (os.name): {os_name}")
print(f"  📂 Ruta del directorio de trabajo actual:  {current_directory}")
````

## 🛒 Ventajas de la Utilización de Librerías

### 💡 Principales Beneficios en el Desarrollo de Software:

1. ♻️ **Reutilización de Código:** Evita reinventar la rueda al ofrecer soluciones probadas y optimizadas para tareas comunes y complejas.

2. 🚀 **Facilidad y Velocidad de Desarrollo:** Encapsula algoritmos complejos tras interfaces y funciones simples de consumir.

3. ⚡ **Eficiencia y Rendimiento:** Muchas librerías populares (como `numpy` o `math`) incluyen núcleos optimizados en C/C++, ofreciendo mayor velocidad de ejecución.

4. 🛠️ **Mantenimiento y Corrección de Errores:** Las librerías de la comunidad reciben constantes actualizaciones, parches de seguridad y mejoras de rendimiento.

5. 📚 **Documentación y Soporte:** Disponibilidad de guías, ejemplos y foros de soporte ampliamente extendidos en el ecosistema.

````python
# ================================================================================
# 🧪 4. EJERCICIO PRÁCTICO INTEGRADO: COMBINACIÓN DE LIBRERÍAS ESTÁNDAR
# ================================================================================
print("\n--- 🧪 [PASO 4]: Ejercicio Integrado - Módulos 'random' y 'math' ---")

from math import ceil, floor
import random

# ⚙️ [1] Generación de un valor flotante aleatorio entre 10 y 50
random_value: float = random.uniform(10.5, 49.9)

# ⚙️ [2] Aplicación de funciones de redondeo superior e inferior de math
rounded_up: int = ceil(random_value)
rounded_down: int = floor(random_value)

print(f"📋 Valor flotante aleatorio generado: {random_value:.4f}")
print(f"  🔹 Redondeo hacia arriba (math.ceil):   {rounded_up}")
print(f"  🔹 Redondeo hacia abajo (math.floor):   {rounded_down}")

# --------------------------------------------------------------------------------
# 🏁 [PASO 5]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Cuaderno de Módulos y Librerías finalizado con éxito.")
print("=" * 80)
````
