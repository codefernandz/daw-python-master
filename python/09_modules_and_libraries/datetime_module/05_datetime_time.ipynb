---

### 💡 La Clase `datetime.time`

La clase **`datetime.time`** representa una hora del día (local o con información de huso horario), totalmente abstraída de un día, mes o año específico. Es ideal para representar horarios fijos (p. ej., horarios de apertura, turnos laborales o ejecuciones de tareas programadas).

---

### 🗂️ Obtención e Instanciación de la Hora

```text
                  ┌── datetime.datetime.now().time() ──> [ Hora actual del sistema ]
[ datetime.time ]
                  └── datetime.time(hour, min, sec) ──> [ Hora específica ]
```

| Método / Atributo | Ejemplo / Firma | Descripción |
| :--- | :--- | :--- |
| ⏰ `datetime.time(h, m, s, ms)` | `datetime.time(23, 14, 56)` | Instancia una hora concreta (Hora, Minuto, Segundo, Microsegundo) |
| ⌚ `now().time()` | `datetime.datetime.now().time()` | Extrae el objeto `time` del instante presente capturado por `datetime` |
| 🔹 `.hour` / `.minute` / `.second` | `hora.hour`, `hora.minute` | Atributos de acceso directo a las unidades de tiempo |
```

````python
import datetime
from typing import Any

# 🛰️ [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 MÓDULO DATETIME | Manipulación de Horas con datetime.time")
print("=" * 80)

# ================================================================================
# 🧪 1. CREACIÓN Y EXTRACCIÓN DE OBJETOS TIME
# ================================================================================
# Extraer la hora actual del sistema e instanciar un objeto time específico.
# ================================================================================

print("\n--- 🚀 [PASO 1]: Instanciación y Extracción de Objetos datetime.time ---")

# 🔌 [1] Creamos un objeto time con la hora actual extraída de datetime.now()
current_time: datetime.time = datetime.datetime.now().time()
print("La hora es:          ", current_time)

# 🔌 [2] Creamos un objeto time con una hora concreta (Hora, Minuto, Segundo)
specific_time: datetime.time = datetime.time(23, 14, 56)
print("La hora concreta es: ", specific_time)

# 🔌 [3] Extracción de componentes individuales
print("\n📋 Componentes extraídos de la hora concreta:")
print(f"  🔹 Hora (hour):        {specific_time.hour}")
print(f"  🔹 Minuto (minute):    {specific_time.minute}")
print(f"  🔹 Segundo (second):   {specific_time.second}")
print(f"  🔹 Microsegundo (ms):  {specific_time.microsecond}")
````

## 🧱 Comparación y Formateo de Objetos `time`

### 💡 Características Clave:
1. **Atributos Opcionales:** Si no se especifican minutos, segundos o microsegundos, sus valores predeterminados serán `0`.
2. **Comparación Directa:** Los objetos `datetime.time` admiten operadores de comparación (`<`, `>`, `==`) para validar secuencias cronológicas dentro de un mismo día.

````python
# ================================================================================
# 🧪 2. COMPARACIÓN Y FORMATEO DE HORAS
# ================================================================================
print("\n--- 🚀 [PASO 2]: Formateo y Comparaciones Cronológicas ---")

# ⚙️ [1] Definición de dos horarios para comparación
morning_shift_start: datetime.time = datetime.time(8, 30, 0)
night_shift_start: datetime.time = datetime.time(22, 0, 0)

print(f"📋 Hora Turno Mañana: {morning_shift_start.strftime('%I:%M %p')}")
print(f"📋 Hora Turno Noche:  {night_shift_start.strftime('%I:%M %p')}")

# ⚙️ [2] Operación lógica de comparación entre horas
is_earlier: bool = morning_shift_start < night_shift_start
print(f"\n📋 ¿El turno de la mañana es previo al de la noche?: {is_earlier}")
````

## 🛒 Ejercicios Prácticos Avanzados Integrados

---

### 🧪 Ejercicio 1: Validador de Rango de Horario Laboral y Ventana de Mantenimiento

**Escenario Empresarial:** Un backend bancario necesita verificar si una transacción solicitada por un cliente se realiza dentro del horario de atención permitido (08:00 a 20:00) o si cae dentro de la ventana de mantenimiento nocturno.

````python
# ================================================================================
# 🧪 3. EJERCICIO PRÁCTICO 1: AUDITOR DE HORARIO DE OPERACIONES
# ================================================================================
print("\n--- 🧪 [PASO 3.1]: Ejercicio 1 - Validador de Horario Operativo ---")


def validate_operating_window(
    request_time: datetime.time,
    window_start: datetime.time = datetime.time(8, 0, 0),
    window_end: datetime.time = datetime.time(20, 0, 0)
) -> dict[str, Any]:
    """Valida si un horario dado se encuentra dentro del rango operativo permitido.

    Args:
        request_time (datetime.time): Hora enviada por el cliente.
        window_start (datetime.time): Hora de inicio de operaciones.
        window_end (datetime.time): Hora de cierre de operaciones.

    Returns:
        dict[str, Any]: Diagnóstico con el estado de autorización y detalle.
    """
    # [1] Verificación del rango usando operadores relacionales
    is_within_window: bool = window_start <= request_time <= window_end

    result: dict[str, Any] = {
        "request_time": request_time.strftime("%H:%M:%S"),
        "window_start": window_start.strftime("%H:%M:%S"),
        "window_end": window_end.strftime("%H:%M:%S"),
        "is_authorized": is_within_window,
        "status_code": "ACCEPTED" if is_within_window else "REJECTED_OUT_OF_HOURS"
    }

    print(f"  ✅ Solicitud recibida a las: {result['request_time']}")
    print(f"  🔹 Rango permitido:        {result['window_start']} - {result['window_end']}")
    print(f"  🔹 Resultado de validación: {result['status_code']}")

    return result


# Prueba del ejercicio
custom_request_time: datetime.time = datetime.time(23, 14, 56)

print(f"📋 Evaluando transacción realizada a las {custom_request_time}...")
validation_info: dict[str, Any] = validate_operating_window(custom_request_time)

print("\n📋 Resumen del Análisis Operativo:")
print(f"  🔹 Hora Solicitada:  {validation_info['request_time']}")
print(f"  🔹 ¿Autorizada?:    {validation_info['is_authorized']}")
print(f"  🔹 Código Estado:   {validation_info['status_code']}")

# --------------------------------------------------------------------------------
# 🏁 [PASO 4]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Lección del método datetime.time finalizada.")
print("=" * 80)
````
