| Propiedad | Detalle |
| :--- | :--- |
| 📍 **Módulo** | Módulos, Paquetes y Librerías / Módulo Datetime |

| 🛠️ **Conceptos** | `import datetime`, `datetime.datetime()`, `Argumentos Obligatorios/Opcionales`, `Extracción de Atributos` |

---

### 💡 La Clase `datetime.datetime`

En Python, la sintaxis **`datetime.datetime(...)`** puede parecer redundante, pero obedece a una jerarquía clara:
1. El **primer `datetime`** hace referencia al **módulo** importado.
2. El **segundo `datetime`** hace referencia a la **clase** dentro de dicho módulo.

---

### 🗂️ Argumentos del Constructor `datetime.datetime()`

```text
datetime.datetime( year, month, day [, hour [, minute [, second [, microsecond [, tzinfo ]]]]] )
                   └───────────┬───────────┘ └──────────────────────────────┬──────────────────────────────┘
                        OBLIGATORIOS                                   OPCIONALES (Default: 0)
```

| Parámetro | Rango Válido | Estado | Descripción |
| :--- | :--- | :--- | :--- |
| 📅 `year` | `MINYEAR` (1) a `MAXYEAR` (9999) | Obligatorio | Año de la fecha |
| 🗓️ `month` | 1 a 12 | Obligatorio | Mes del año |
| 📆 `day` | 1 a 31 (según el mes) | Obligatorio | Día del mes |
| ⏰ `hour` | 0 a 23 | Opcional (Default: 0) | Hora del día |
| ⏱️ `minute` | 0 a 59 | Opcional (Default: 0) | Minuto |
| ⏱️ `second` | 0 a 59 | Opcional (Default: 0) | Segundo |
| 🔬 `microsecond` | 0 a 999999 | Opcional (Default: 0) | Microsegundo |
```

````python
import datetime
from typing import Any

# 🛰️ [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 MÓDULO DATETIME | Instanciación de la Clase datetime.datetime")
print("=" * 80)

# ================================================================================
# 🧪 1. CREACIÓN DE OBJETOS DATETIME ESPECÍFICOS
# ================================================================================
# Instanciación con año, mes, día, hora, minuto y segundo explícitos.
# ================================================================================

print("\n--- 🚀 [PASO 1]: Instanciación con Argumentos Posicionales ---")

# 🔌 [1] Creamos un objeto datetime con una fecha y hora específicas (Año, mes, día, hora, min, segundo)
specific_date_time: datetime.datetime = datetime.datetime(1495, 2, 14, 12, 30, 0)

print(f"📋 Fecha y hora específica: {specific_date_time}")

# 🔌 [2] Inspección individual de los componentes del objeto
print("\n📋 Componentes extraídos del objeto instanciado:")
print(f"  🔹 Año (year):     {specific_date_time.year}")
print(f"  🔹 Mes (month):    {specific_date_time.month}")
print(f"  🔹 Día (day):      {specific_date_time.day}")
print(f"  🔹 Hora (hour):    {specific_date_time.hour}")
print(f"  🔹 Minuto (minute): {specific_date_time.minute}")
print(f"  🔹 Segundo (second): {specific_date_time.second}")
````

## 🧱 Parámetros Opcionales y Métodos Alternativos de Importación

### 💡 Formas de Importación:
Para evitar la redundancia `datetime.datetime`, es habitual en proyectos Python realizar una importación directa de la clase:
```python
from datetime import datetime

# Permite instanciar directamente:
fecha = datetime(2026, 8, 15)

````python
# ================================================================================
# 🧪 2. USO DE VALORES POR DEFECTO Y MÉTODOS DE CONSTRUCCIÓN
# ================================================================================
print("\n--- 🚀 [PASO 2]: Argumentos Opcionales y Atributos Por Defecto ---")

# ⚙️ [1] Omisión de hora, minuto y segundo (asume 00:00:00 por defecto)
date_only_time: datetime.datetime = datetime.datetime(2026, 8, 15)
print(f"📋 Objeto instanciado solo con año, mes y día: {date_only_time}")

# ⚙️ [2] Instanciación usando argumentos por nombre (Keyword Arguments)
named_date_time: datetime.datetime = datetime.datetime(
    year=1789, month=7, day=14, hour=17, minute=15
)
print(f"📋 Objeto instanciado mediante Keyword Arguments: {named_date_time}")
````

### 🧪 Ejercicio 1: Registrador e Inspector de Eventos Históricos

**Escenario Empresarial:** Un sistema de archivo digital para museos e historiadores requiere registrar eventos históricos precisos, validar si la fecha cae en fin de semana y calcular los años transcurridos desde el acontecimiento hasta la actualidad.

````python
# ================================================================================
# 🧪 3. EJERCICIO PRÁCTICO 1: INSPECTOR Y VALIDADOR DE EVENTOS HISTÓRICOS
# ================================================================================
print("\n--- 🧪 [PASO 3.1]: Ejercicio 1 - Registrador de Eventos Históricos ---")


def analyze_historical_event(
    event_name: str, year: int, month: int, day: int, hour: int = 0, minute: int = 0
) -> dict[str, Any]:
    """Crea un objeto datetime para un evento histórico y genera un diagnóstico temporal.

    Args:
        event_name (str): Nombre del evento histórico.
        year (int): Año de realización.
        month (int): Mes de realización (1-12).
        day (int): Día de realización (1-31).
        hour (int): Hora del evento (0-23).
        minute (int): Minuto del evento (0-59).

    Returns:
        dict[str, Any]: Diagnóstico detallado del evento procesado.
    """
    # [1] Instanciación explícita de datetime.datetime
    event_datetime: datetime.datetime = datetime.datetime(
        year, month, day, hour, minute
    )
    current_year: int = datetime.datetime.now().year

    # [2] Determinación del día de la semana (0 = Lunes, 6 = Domingo)
    weekday_index: int = event_datetime.weekday()
    is_weekend: bool = weekday_index >= 5

    days_map: list[str] = [
        "Lunes", "Martes", "Miércoles", "Jueves", "Viernes", "Sábado", "Domingo"
    ]

    event_summary: dict[str, Any] = {
        "event_name": event_name,
        "formatted_datetime": event_datetime.strftime("%Y-%m-%d %H:%M:%S"),
        "weekday": days_map[weekday_index],
        "is_weekend": is_weekend,
        "years_elapsed": current_year - event_datetime.year
    }

    print(f"  ✅ Evento registrado: '{event_name}'")
    print(f"  🔹 Fecha/Hora: {event_summary['formatted_datetime']}")
    print(f"  🔹 Día de la semana: {event_summary['weekday']}")
    print(f"  🔹 Años transcurridos: {event_summary['years_elapsed']} años")

    return event_summary


# Prueba del ejercicio
print("📋 Registrando evento histórico de prueba...")
historical_info: dict[str, Any] = analyze_historical_event(
    event_name="Firma de Documento Histórico",
    year=1495,
    month=2,
    day=14,
    hour=12,
    minute=30
)

print("\n📋 Resumen del Análisis de Evento:")
print(f"  🔹 Nombre:            {historical_info['event_name']}")
print(f"  🔹 Representación ISO: {historical_info['formatted_datetime']}")
print(f"  🔹 Es fin de semana:   {historical_info['is_weekend']}")

# --------------------------------------------------------------------------------
# 🏁 [PASO 4]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Lección de instanciación datetime.datetime finalizada.")
print("=" * 80)
````
