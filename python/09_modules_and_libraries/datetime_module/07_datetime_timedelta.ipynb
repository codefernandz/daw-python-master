| 🛠️ **Conceptos** | `datetime.timedelta`, `Aritmética Temporal`, `Desplazamiento de Fechas`, `Duración de Intervalos` |

---

### 💡 La Clase `datetime.timedelta`

Un objeto **`timedelta`** representa una duración o la diferencia entre dos fechas u horas. Permite realizar operaciones matemáticas de suma (`+`) y resta (`-`) directamente sobre objetos `datetime.date` y `datetime.datetime`.

---

### 🗂️ Parámetros del Constructor `datetime.timedelta()`

```text
[ Objeto Date / DateTime ]  ±  [ datetime.timedelta(days=N, weeks=N, ...) ]  ──> [ Nueva Fecha / Hora ]
```

| Parámetro | Valor Predeterminado | Descripción |
| :--- | :--- | :--- |
| `days` | 0 | Desplazamiento en número de días |
| `seconds` | 0 | Desplazamiento en segundos (0 a 86399) |
| `microseconds` | 0 | Desplazamiento en microsegundos |
| `milliseconds` | 0 | Desplazamiento en milisegundos |
| `minutes` | 0 | Desplazamiento en minutos |
| `hours` | 0 | Desplazamiento en horas |
| `weeks` | 0 | Desplazamiento en semanas (convertido internamente a días) |

---

### 📚 Recurso Adicional:
Para explorar todas las funcionalidades avanzadas sobre aritmética de fechas y zonas horarias, puedes consultar la [Documentación Oficial del Módulo Datetime en Python](https://docs.python.org/3/library/datetime.html).
```

````python
import datetime
from typing import Any

# 🛰️ [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 MÓDULO DATETIME | Aritmética de Fechas con timedelta")
print("=" * 80)

# ================================================================================
# 🧪 1. CÁLCULO DE FECHAS FUTURAS Y PASADAS
# ================================================================================
# Obtenemos la fecha actual y aplicamos sumas y restas con datetime.timedelta().
# ================================================================================

print("\n--- 🚀 [PASO 1]: Operaciones Matemáticas sobre Objetos date ---")

# 🔌 [1] Obtenemos la fecha actual
current_date: datetime.date = datetime.date.today()
print("Fecha actual:", current_date)

# 🔌 [2] Calculamos la fecha de hace una semana (resta de 7 días)
one_week_ago: datetime.date = current_date - datetime.timedelta(days=7)
print("Fecha de hace una semana:", one_week_ago)

# 🔌 [3] Calculamos la fecha de dentro de tres meses (aproximadamente 90 días)
in_three_months: datetime.date = current_date + datetime.timedelta(days=90)
print("Fecha de dentro de tres meses:", in_three_months)
````

## 🧱 Aritmética Combinada de Horas, Minutos y Semanas

### 💡 Flexibilidad de `timedelta`:
Se pueden combinar múltiples unidades de tiempo en una sola instancia de `timedelta` para realizar ajustes complejos en marcas de tiempo `datetime`.

````python
# ================================================================================
# 🧪 2. OPERACIONES AVANZADAS CON FECHA Y HORA (datetime.datetime)
# ================================================================================
print("\n--- 🚀 [PASO 2]: Desplazamiento de Precisión con Horas y Semanas ---")

current_datetime: datetime.datetime = datetime.datetime.now()
print(f"📋 Timestamp actual: {current_datetime.strftime('%Y-%m-%d %H:%M:%S')}")

# ⚙️ [1] Desplazamiento combinado: 2 semanas, 3 días, 5 horas y 30 minutos al futuro
complex_offset: datetime.timedelta = datetime.timedelta(
    weeks=2, days=3, hours=5, minutes=30
)
future_event: datetime.datetime = current_datetime + complex_offset

print(f"📋 Objeto timedelta aplicado: {complex_offset}")
print(f"  ✅ Fecha/Hora futura calculada: {future_event.strftime('%Y-%m-%d %H:%M:%S')}")

# ⚙️ [2] Diferencia entre dos objetos datetime (devuelve un objeto timedelta)
time_difference: datetime.timedelta = future_event - current_datetime
print(f"\n📋 Diferencia calculada (Días): {time_difference.days} días")
print(f"📋 Diferencia en total de segundos: {time_difference.total_seconds()} segundos")
````

## 🛒 Ejercicios Prácticos Avanzados Integrados

---

### 🧪 Ejercicio 1: Gestor de Fechas de Vencimiento de Facturas y Penalizaciones

**Escenario Empresarial:** Un sistema de contabilidad necesita calcular la fecha de vencimiento de una factura (30 días tras la emisión), determinar si está vencida respecto a la fecha actual y calcular los días de mora junto con el interés acumulado.

````python
# ================================================================================
# 🧪 3. EJERCICIO PRÁCTICO 1: CALCULADOR DE VENCIMIENTO Y INTERESES DE MORA
# ================================================================================
print("\n--- 🧪 [PASO 3.1]: Ejercicio 1 - Gestor de Vencimiento de Facturas ---")


def calculate_invoice_status(
    issue_date: datetime.date,
    credit_days: int = 30,
    daily_penalty_rate: float = 0.001
) -> dict[str, Any]:
    """Calcula la fecha de vencimiento y el estado de mora de una factura.

    Args:
        issue_date (datetime.date): Fecha de emisión de la factura.
        credit_days (int): Días de crédito otorgados.
        daily_penalty_rate (float): Tasa de penalización diaria por mora.

    Returns:
        dict[str, Any]: Resumen con fecha de vencimiento, días de mora y estado.
    """
    today: datetime.date = datetime.date.today()

    # [1] Cálculo de la fecha de vencimiento mediante timedelta
    due_date: datetime.date = issue_date + datetime.timedelta(days=credit_days)

    # [2] Determinación de estado y días de retraso
    is_overdue: bool = today > due_date
    overdue_days: int = (today - due_date).days if is_overdue else 0

    invoice_summary: dict[str, Any] = {
        "issue_date": issue_date.isoformat(),
        "due_date": due_date.isoformat(),
        "is_overdue": is_overdue,
        "overdue_days": overdue_days,
        "penalty_factor": round(overdue_days * daily_penalty_rate, 4)
    }

    print(f"  ✅ Factura emitida el: {invoice_summary['issue_date']}")
    print(f"  🔹 Fecha de vencimiento: {invoice_summary['due_date']}")
    print(f"  🔹 ¿Está vencida?:        {invoice_summary['is_overdue']}")
    if is_overdue:
        print(f"  ⚠️ Días de retraso:     {invoice_summary['overdue_days']} días")

    return invoice_summary


# Prueba del ejercicio
sample_issue_date: datetime.date = datetime.date.today() - datetime.timedelta(days=45)

print("📋 Evaluando factura emitida hace 45 días...")
invoice_report: dict[str, Any] = calculate_invoice_status(
    issue_date=sample_issue_date, credit_days=30
)

print("\n📋 Resumen del Estado de la Factura:")
print(f"  🔹 Emisión:        {invoice_report['issue_date']}")
print(f"  🔹 Vencimiento:    {invoice_report['due_date']}")
print(f"  🔹 Estado Mora:    {'VENCIDA' if invoice_report['is_overdue'] else 'AL DÍA'}")
print(f"  🔹 Días en Mora:   {invoice_report['overdue_days']} días")

# --------------------------------------------------------------------------------
# 🏁 [PASO 4]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Lección del método datetime.timedelta finalizada.")
print("=" * 80)
````
