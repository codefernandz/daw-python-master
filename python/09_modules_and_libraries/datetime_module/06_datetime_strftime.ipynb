| 🛠️ **Conceptos** | `strftime()`, `Directivas de Formato (%Y, %m, %d, %H, %M, %S)`, `Formateo de Sub-objetos (.date(), .time())` |

---

### 💡 El Método `strftime()`

El método **`strftime(format)`** (String Format Time) permite representar objetos temporales como cadenas de texto formateadas. Utiliza códigos de formato precedidos por `%` para representar componentes específicos de la fecha o la hora.

---

### 🗂️ Tabla de Directivas de Formato Frecuentes

```text
[ Objeto temporal ] ──( .strftime("%Y-%m-%d %H:%M:%S") )──> "2026-08-15 23:05:01"
```

| Directiva | Descripción | Ejemplo de Salida |
| :--- | :--- | :--- |
| `%Y` | Año completo de 4 dígitos | 2026 |
| `%m` | Mes formateado con cero inicial (01-12) | 08 |
| `%d` | Día del mes con cero inicial (01-31) | 15 |
| `%H` | Hora en formato 24 horas (00-23) | 23 |
| `%I` | Hora en formato 12 horas (01-12) | 11 |
| `%p` | Indicador AM / PM | PM |
| `%M` | Minuto con cero inicial (00-59) | 05 |
| `%S` | Segundo con cero inicial (00-59) | 01 |
| `%A` / `%B` | Nombre completo del día / mes | Saturday / August |
```

````python
import datetime
from typing import Any

# 🛰️ [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 MÓDULO DATETIME | Formateo de Fechas y Horas con strftime()")
print("=" * 80)

# ================================================================================
# 🧪 1. FORMATEO BÁSICO DE FECHA Y HORA
# ================================================================================
# Obtenemos la fecha/hora actual y aplicamos distintos patrones de strftime().
# ================================================================================

print("\n--- 🚀 [PASO 1]: Aplicación de Máscaras de Formato con strftime() ---")

# 🔌 [1] Obtenemos la fecha y hora actual
current_date_time: datetime.datetime = datetime.datetime.now()
print("La fecha sin formatear es :", current_date_time)

# 🔌 [2] Formateamos la fecha y hora en un formato específico (ISO Estándar)
formatted_date_time: str = current_date_time.strftime("%Y-%m-%d %H:%M:%S")
print("Fecha y hora formateada:", formatted_date_time)

# 🔌 [3] Formateamos solo la fecha en otro formato (Día/Mes/Año)
formatted_date_only: str = current_date_time.date().strftime("%d/%m/%Y")
print("Fecha formateada:", formatted_date_only)

# 🔌 [4] Formateamos solo la hora en otro formato (Hora:Minuto:Segundo)
formatted_time_only: str = current_date_time.time().strftime("%H:%M:%S")
print("Hora formateada:", formatted_time_only)
````

## 🧱 Patrones Alternativos y Localización

### 💡 Formateo Avanzado (12 Horas vs 24 Horas):
`strftime()` se adapta a necesidades específicas de presentación, permitiendo generar reportes corporativos o adaptar la salida según el formato regional requerido.

````python
# ================================================================================
# 🧪 2. PATRONES REGIONALES Y DE 12 HORAS
# ================================================================================
print("\n--- 🚀 [PASO 2]: Variaciones de Representación ---")

# ⚙️ [1] Formato de 12 horas con indicador AM/PM
formatted_12h: str = current_date_time.strftime("%I:%M:%S %p")
print(f"📋 Formato 12 horas (AM/PM): {formatted_12h}")

# ⚙️ [2] Formato extenso para documentos o informes
formatted_textual: str = current_date_time.strftime("%A, %d de %B de %Y - %H:%M hrs")
print(f"📋 Formato textual extenso: {formatted_textual}")
````

### 🧪 Ejercicio 1: Generador de Nombres de Archivo para Backups con Timestamps

**Escenario Empresarial:** Un sistema de copias de seguridad de base de datos requiere generar automáticamente nombres de archivo estandarizados que incluyan la fecha, hora y entorno para evitar sobrescribir respaldos anteriores (`backup_PROD_YYYYMMDD_HHMMSS.sql`).

````python
# ================================================================================
# 🧪 3. EJERCICIO PRÁCTICO 1: GENERADOR DE NOMBRES DE ARCHIVO CON TIMESTAMPS
# ================================================================================
print("\n--- 🧪 [PASO 3.1]: Ejercicio 1 - Generador de Nombres de Backup ---")


def generate_backup_filename(
    database_name: str, environment: str = "PROD", extension: str = "sql"
) -> dict[str, Any]:
    """Genera un nombre de archivo único para copias de seguridad usando strftime().

    Args:
        database_name (str): Nombre de la base de datos.
        environment (str): Entorno (PROD, STG, DEV).
        extension (str): Extensión del archivo resultante.

    Returns:
        dict[str, Any]: Detalle de la marca de tiempo y ruta generada.
    """
    now: datetime.datetime = datetime.datetime.now()

    # [1] Formateo compacto para nombres de archivo
    timestamp_pattern: str = now.strftime("%Y%m%d_%H%M%S")
    file_name: str = f"backup_{database_name}_{environment}_{timestamp_pattern}.{extension}"

    result: dict[str, Any] = {
        "database_name": database_name,
        "environment": environment,
        "timestamp_used": timestamp_pattern,
        "generated_filename": file_name,
        "human_readable_time": now.strftime("%Y-%m-%d a las %H:%M:%S")
    }

    print(f"  ✅ Copia de seguridad programada para: {result['human_readable_time']}")
    print(f"  🔹 Nombre de archivo generado:        '{result['generated_filename']}'")

    return result


# Prueba del ejercicio
print("📋 Generando nombre de archivo para respaldo de base de datos...")
backup_info: dict[str, Any] = generate_backup_filename(
    database_name="finance_db", environment="PROD"
)

print("\n📋 Resumen del Generador de Respaldos:")
print(f"  🔹 Base de Datos:    {backup_info['database_name']}")
print(f"  🔹 Entorno:          {backup_info['environment']}")
print(f"  🔹 Archivo Creado:   {backup_info['generated_filename']}")

# --------------------------------------------------------------------------------
# 🏁 [PASO 4]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Lección del método strftime() finalizada.")
print("=" * 80)
````
