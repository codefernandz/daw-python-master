### 💡 El Método `datetime.datetime.now()`

El método de clase **`datetime.datetime.now()`** obtiene la fecha y hora locales del sistema en el instante preciso de ejecución.

---

### 🗂️ Estructura y Sintaxis de Invocación

```text
import datetime

fecha_hora_actual = datetime.datetime.now()
│                   └──┬───┘ └──┬───┘ └──┬┘
│                      │        │        └── Método: Captura el instante presente del sistema
│                      │        └────────── Class: Clase dentro del módulo
│                      └────────────────── Module: Nombre de la librería importada
```

| Atributo | Ejemplo de Valor | Descripción |
| :--- | :--- | :--- |
| 📅 `.year` / `.month` / `.day` | 2026, 8, 15 | Componentes de la fecha local |
| ⏰ `.hour` / `.minute` / `.second` | 22, 55, 32 | Componentes del tiempo local |
| 🔬 `.microsecond` | 482910 | Precisión fraccionaria de segundo (0 a 999,999) |
```

````python
import datetime
from typing import Any

# 🛰️ [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 MÓDULO DATETIME | Captura del Instante Actual con now()")
print("=" * 80)

# ================================================================================
# 🧪 1. CAPTURA Y EXPLORACIÓN DE LA FECHA Y HORA ACTUAL
# ================================================================================
# Invocación de datetime.datetime.now() para obtener el sello temporal del sistema.
# ================================================================================

print("\n--- 🚀 [PASO 1]: Invocación de datetime.datetime.now() ---")

# 🔌 [1] Creamos un objeto datetime con la fecha y hora actual del ordenador
current_date_time: datetime.datetime = datetime.datetime.now()

print(f"📋 Fecha y hora actual: {current_date_time}")

# 🔌 [2] Desglose individual de todos sus atributos temporales
print("\n📋 Componentes extraídos del instante actual:")
print(f"  🔹 Año:         {current_date_time.year}")
print(f"  🔹 Mes:         {current_date_time.month}")
print(f"  🔹 Día:         {current_date_time.day}")
print(f"  🔹 Hora:        {current_date_time.hour}")
print(f"  🔹 Minuto:      {current_date_time.minute}")
print(f"  🔹 Segundo:     {current_date_time.second}")
print(f"  🔹 Microsegundo: {current_date_time.microsecond}")
````

## 🧱 Comparativa entre `now()` y `today()`

### 💡 Diferencias clave:
* **`datetime.datetime.now()`**: Permite opcionalmente recibir una zona horaria (`tzinfo`) para obtener el instante actual en cualquier parte del mundo.
* **`datetime.datetime.today()`**: Retorna la fecha y hora local sin posibilidad de asignarle directamente un parámetro de zona horaria.

```python
# Equivalencia en hora local:
ahora = datetime.datetime.now()
hoy = datetime.datetime.today()
```

````python
# ================================================================================
# 🧪 2. COMPARACIÓN Y MARCAS DE TIEMPO (TIMESTAMPS)
# ================================================================================
print("\n--- 🚀 [PASO 2]: Comparativa de Métodos y Obtención de POSIX Timestamp ---")

# ⚙️ [1] Comparación visual entre now() y today()
now_instance: datetime.datetime = datetime.datetime.now()
today_instance: datetime.datetime = datetime.datetime.today()

print(f"📋 Resultado con now():   {now_instance}")
print(f"📋 Resultado con today(): {today_instance}")

# ⚙️ [2] Conversión del instante actual a timestamp de UNIX / POSIX (segundos desde 1970)
unix_timestamp: float = current_date_time.timestamp()
print(f"\n📋 Marca de tiempo UNIX (segundos transcurridos): {unix_timestamp}")
````

### 🧪 Ejercicio 1: Benchmark de Rendimiento y Benchmark de Procesos

**Escenario Empresarial:** Un equipo de ingeniería de datos necesita medir el tiempo de ejecución exacto de un algoritmo intensivo registrando la marca de tiempo inicial con `now()`, la marca final y calculando la latencia en milisegundos.

````python
# ================================================================================
# 🧪 3. EJERCICIO PRÁCTICO 1: CRONÓMETRO DE RENDIMIENTO DE ALGORITMOS
# ================================================================================
print("\n--- 🧪 [PASO 3.1]: Ejercicio 1 - Medidor de Tiempo de Ejecución ---")


def measure_execution_time(iterations: int = 1_000_000) -> dict[str, Any]:
    """Mide la latencia de un bucle de cómputo calculando la diferencia con datetime.now().

    Args:
        iterations (int): Número de ciclos a simular.

    Returns:
        dict[str, Any]: Reporte del tiempo de inicio, fin y diferencia calculada.
    """
    # [1] Captura del instante inicial
    start_time: datetime.datetime = datetime.datetime.now()

    # [2] Simulación de procesamiento intensivo
    dummy_counter: int = sum(i * 2 for i in range(iterations))

    # [3] Captura del instante final
    end_time: datetime.datetime = datetime.datetime.now()

    # [4] Cálculo del intervalo transcurrido
    elapsed_time: datetime.timedelta = end_time - start_time
    elapsed_milliseconds: float = elapsed_time.total_seconds() * 1000

    metrics: dict[str, Any] = {
        "iterations_processed": iterations,
        "start_time": start_time.strftime("%H:%M:%S.%f"),
        "end_time": end_time.strftime("%H:%M:%S.%f"),
        "elapsed_seconds": elapsed_time.total_seconds(),
        "elapsed_milliseconds": round(elapsed_milliseconds, 2)
    }

    print(f"  ✅ Proceso finalizado ({iterations:,} iteraciones)")
    print(f"  🔹 Hora inicio: {metrics['start_time']}")
    print(f"  🔹 Hora fin:    {metrics['end_time']}")
    print(f"  🔹 Duración:    {metrics['elapsed_milliseconds']} ms")

    return metrics


# Prueba del ejercicio
print("📋 Iniciando benchmark de rendimiento...")
performance_results: dict[str, Any] = measure_execution_time(iterations=2_000_000)

print("\n📋 Resumen del Benchmark de Ejecución:")
print(f"  🔹 Iteraciones:           {performance_results['iterations_processed']:,}")
print(f"  🔹 Tiempo Total (seg):    {performance_results['elapsed_seconds']} s")
print(f"  🔹 Tiempo Total (msec):   {performance_results['elapsed_milliseconds']} ms")

# --------------------------------------------------------------------------------
# 🏁 [PASO 4]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Lección del método datetime.datetime.now() finalizada.")
print("=" * 80)
````
