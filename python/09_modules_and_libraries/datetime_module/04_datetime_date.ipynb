---

### 💡 La Clase `datetime.date`

La clase **`datetime.date`** representa una fecha idealizada en el calendario gregoriano proléptico. A diferencia de `datetime.datetime`, un objeto `date` contiene **únicamente** los atributos de año, mes y día, ignorando horas, minutos y segundos.

---

### 🗂️ Constructor y Métodos Principales

```text
               ┌── datetime.date.today() ────> [ Fecha actual del sistema ]
[ datetime.date ]
               └── datetime.date(year, month, day) ──> [ Fecha específica ]
```

| Método / Atributo | Ejemplo / Firma | Descripción |
| :--- | :--- | :--- |
| 📅 `datetime.date.today()` | `datetime.date.today()` | Retorna la fecha local actual |
| 🗓️ `datetime.date(y, m, d)` | `datetime.date(1495, 12, 31)` | Instancia una fecha concreta (Año, Mes, Día) |
| 🔹 `.year` / `.month` / `.day` | `fecha.year`, `fecha.month` | Atributos de acceso directo a los componentes |
```

````python
import datetime
from typing import Any

# 🛰️ [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 MÓDULO DATETIME | Manipulación de Fechas con datetime.date")
print("=" * 80)

# ================================================================================
# 🧪 1. CREACIÓN Y CONSULTA DE OBJETOS DATE
# ================================================================================
# Obtención de fecha actual con today() e instanciación de fecha específica.
# ================================================================================

print("\n--- 🚀 [PASO 1]: Instanciación de Objetos datetime.date ---")

# 🔌 [1] Creamos un objeto date con la fecha actual del sistema
today_date: datetime.date = datetime.date.today()
print("Fecha hoy:        ", today_date)

# 🔌 [2] Creamos un objeto date con una fecha concreta (Año, Mes, Día)
specific_date: datetime.date = datetime.date(1495, 12, 31)
print("Fecha específica: ", specific_date)

# 🔌 [3] Extracción de componentes individuales
print("\n📋 Componentes extraídos de la fecha específica:")
print(f"  🔹 Año (year):  {specific_date.year}")
print(f"  🔹 Mes (month): {specific_date.month}")
print(f"  🔹 Día (day):   {specific_date.day}")
````

## 🧱 Métodos Útiles de Conversión e Inspección en `date`

### 💡 Métodos Recomendados:
1. **`isoformat()`**: Devuelve la fecha representada en formato ISO 8601 (`YYYY-MM-DD`).
2. **`strftime(format)`**: Permite formatear la fecha según patrones de texto personalizados.
3. **`weekday()` / `isoweekday()`**: Indican el día de la semana (`0-6` o `1-7`).

````python
# ================================================================================
# 🧪 2. FORMATEO Y CONVERSIÓN DE OBJETOS DATE
# ================================================================================
print("\n--- 🚀 [PASO 2]: Representación e Inspección de Fechas ---")

# ⚙️ [1] Conversión a formato Estándar ISO 8601
iso_date_str: str = specific_date.isoformat()
print(f"📋 Formato ISO 8601: {iso_date_str}")

# ⚙️ [2] Formateo personalizado mediante strftime
formatted_custom: str = specific_date.strftime("%d de %B de %Y")
print(f"📋 Formato Lectura: {formatted_custom}")

# ⚙️ [3] Obtención del día de la semana (0 = Lunes, 6 = Domingo)
weekday_num: int = specific_date.weekday()
print(f"📋 Índice de día de la semana (0=Lunes): {weekday_num}")
````

### 🧪 Ejercicio 1: Calcular Edad e Intervalos de Días Festivos/Aniversarios

**Escenario Empresarial:** Un departamento de recursos humanos necesita calcular con precisión la edad exacta en años de un empleado basándose en su fecha de nacimiento y determinar cuántos días faltan para su próximo cumpleaños respecto a la fecha actual.

````python
# ================================================================================
# 🧪 3. EJERCICIO PRÁCTICO 1: CALCULADOR DE EDAD Y PRÓXIMO CUMPLEAÑOS
# ================================================================================
print("\n--- 🧪 [PASO 3.1]: Ejercicio 1 - Gestor de Fechas de Nacimiento ---")


def calculate_age_and_next_birthday(birth_date: datetime.date) -> dict[str, Any]:
    """Calcula la edad actual en años y determina los días restantes para el próximo cumpleaños.

    Args:
        birth_date (datetime.date): Fecha de nacimiento.

    Returns:
        dict[str, Any]: Resumen con edad actual, próximo cumpleaños y días faltantes.
    """
    today: datetime.date = datetime.date.today()

    # [1] Cálculo de edad en años
    age: int = today.year - birth_date.year
    # Ajuste si aún no ha cumplido años en el año actual
    if (today.month, today.day) < (birth_date.month, birth_date.day):
        age -= 1

    # [2] Proyección del próximo cumpleaños
    next_birthday_year: int = today.year
    if (today.month, today.day) >= (birth_date.month, birth_date.day):
        next_birthday_year += 1

    next_birthday: datetime.date = datetime.date(
        next_birthday_year, birth_date.month, birth_date.day
    )

    # [3] Diferencia de días
    days_until_next_birthday: int = (next_birthday - today).days

    results: dict[str, Any] = {
        "birth_date": birth_date.isoformat(),
        "current_age": age,
        "next_birthday": next_birthday.isoformat(),
        "days_remaining": days_until_next_birthday
    }

    print(f"  ✅ Fecha analizada: {results['birth_date']}")
    print(f"  🔹 Edad actual:     {results['current_age']} años")
    print(f"  🔹 Próximo cumple:  {results['next_birthday']} (Faltan {results['days_remaining']} días)")

    return results


# Prueba del ejercicio
sample_birth_date: datetime.date = datetime.date(1995, 12, 31)

print(f"📋 Procesando expediente de fecha de nacimiento ({sample_birth_date})...")
employee_age_info: dict[str, Any] = calculate_age_and_next_birthday(sample_birth_date)

print("\n📋 Resumen del Análisis de Edad:")
print(f"  🔹 Edad Calculada:         {employee_age_info['current_age']} años")
print(f"  🔹 Días para Cumpleaños:  {employee_age_info['days_remaining']} días")

# --------------------------------------------------------------------------------
# 🏁 [PASO 4]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Lección del método datetime.date finalizada.")
print("=" * 80)
````
