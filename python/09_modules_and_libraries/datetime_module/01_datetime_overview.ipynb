### 💡 El Módulo `datetime`

El módulo estándar **`datetime`** de Python ofrece clases para manipular fechas y horas de forma orientada a objetos. Aunque soporta aritmética temporal precisa, su diseño está optimizado para la **extracción eficiente de atributos**, la **formateación** y la **transformación entre cadenas y objetos de tiempo**.

---

### 🗂️ Clases Principales del Módulo

```text
[ modulo datetime ] ──┬──> date       (Solo Fecha: Año, Mes, Día)
                     ├──> time       (Solo Hora: Hora, Minuto, Segundo, Microsegundo)
                     ├──> datetime   (Combinación de Fecha y Hora)
                     ├──> timedelta  (Diferencia / Duración entre dos momentos)
                     └──> tzinfo     (Base para gestión de Zonas Horarias)
```

| Clase | Descripción | Ejemplo de Uso |
| :--- | :--- | :--- |
| 📅 `date` | Representa una fecha idealizada (año, mes, día) asumiendo el calendario gregoriano. | `date(2026, 8, 15)` |
| ⏰ `time` | Representa una hora del día independiente de cualquier fecha. | `time(14, 30, 0)` |
| 🗓️ `datetime` | Combina todos los atributos de `date` y `time`. | `datetime(2026, 8, 15, 14, 30)` |
| ⏳ `timedelta` | Expresa la duración o diferencia entre dos fechas o horas. | `timedelta(days=7)` |
| 🌍 `tzinfo` | Clase abstracta para manejo e información de zonas horarias. | Usada con `zoneinfo` o `pytz` |
```

````python
import datetime
from typing import Any

# 🛰️ [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 MÓDULO DATETIME | Visión General y Clases Fundamentales")
print("=" * 80)

# ================================================================================
# 🧪 1. INSPECCIÓN Y CREACIÓN DE OBJETOS BÁSICOS DE DATETIME
# ================================================================================
# Exploración de date, time y datetime para la fecha y hora actuales.
# ================================================================================

print("\n--- 🚀 [PASO 1]: Instanciación de Clases Principales ---")

# 🔌 [1] Captura del momento actual (Fecha y Hora)
current_now: datetime.datetime = datetime.datetime.now()

# 🔌 [2] Captura de la fecha actual
current_date: datetime.date = datetime.date.today()

# 🔌 [3] Creación de una hora específica
custom_time: datetime.time = datetime.time(hour=14, minute=30, second=0)

print(f"📋 Objeto datetime actual (now): {current_now}")
print(f"  🔹 Año: {current_now.year} | Mes: {current_now.month} | Día: {current_now.day}")
print(f"  🔹 Hora: {current_now.hour} | Min: {current_now.minute} | Seg: {current_now.second}")

print(f"\n📋 Objeto date (today):           {current_date}")
print(f"📋 Objeto time (personalizado):   {custom_time}")
````

## 🧱 Operaciones Aritméticas Básicas y `timedelta`

### 💡 Aritmética de Fechas:
El módulo `datetime` permite realizar operaciones matemáticas como sumar o restar días, horas o semanas utilizando la clase **`timedelta`**, así como calcular la diferencia exacta entre dos instantes de tiempo.

````python
# ================================================================================
# 🧪 2. OPERACIONES Y ARITMÉTICA TEMPORAL CON TIMEDELTA
# ================================================================================
print("\n--- 🚀 [PASO 2]: Cánhulo de Intervalos y Desplazamientos Temporales ---")

# ⚙️ [1] Definición del momento actual y un intervalo de 10 días
today: datetime.date = datetime.date.today()
ten_days: datetime.timedelta = datetime.timedelta(days=10)

# ⚙️ [2] Operaciones de suma y resta de tiempo
future_date: datetime.date = today + ten_days
past_date: datetime.date = today - ten_days

print(f"📋 Fecha base hoy:           {today}")
print(f"  ✅ Fecha (+10 días futuros): {future_date}")
print(f"  ✅ Fecha (-10 días pasados): {past_date}")

# ⚙️ [3] Cálculo de diferencia entre dos fechas (Retorna un timedelta)
target_event: datetime.date = datetime.date(2026, 12, 31)
difference: datetime.timedelta = target_event - today

print(f"\n📋 Días faltantes para fin de año ({target_event}): {difference.days} días")
````

### 🧪 Ejercicio 1: Generador de Reportes de Auditoría de Sesión

**Escenario Empresarial:** Un sistema web necesita registrar el momento de inicio de sesión de un usuario, calcular el tiempo proyectado de expiración del token (30 minutos) y generar un reporte detallado con los atributos temporales extraídos.

````python
# ================================================================================
# 🧪 3. EJERCICIO PRÁCTICO 1: GESTOR DE EXPIRACIÓN DE SESIONES
# ================================================================================
print("\n--- 🧪 [PASO 3.1]: Ejercicio 1 - Calculador de Expiración de Sesión ---")


def create_session_audit(
    user_id: str, session_duration_minutes: int = 30
) -> dict[str, Any]:
    """Genera un reporte de auditoría de inicio de sesión y proyecta su tiempo de expiración.

    Args:
        user_id (str): Identificador único del usuario.
        session_duration_minutes (int): Minutos de validez de la sesión.

    Returns:
        dict[str, Any]: Reporte detallado con marcas de tiempo y desglose.
    """
    login_time: datetime.datetime = datetime.datetime.now()
    duration: datetime.timedelta = datetime.timedelta(minutes=session_duration_minutes)
    expiration_time: datetime.datetime = login_time + duration

    audit_summary: dict[str, Any] = {
        "user_id": user_id,
        "is_active": True,
        "login_timestamp": login_time.isoformat(),
        "expiration_timestamp": expiration_time.isoformat(),
        "details": {
            "day_of_week": login_time.strftime("%A"),
            "time_formatted": login_time.strftime("%H:%M:%S"),
            "minutes_valid": session_duration_minutes
        }
    }

    print(f"  ✅ Sesión iniciada para usuario: '{user_id}'")
    print(f"  🔹 Inicio:     {audit_summary['details']['time_formatted']}")
    print(f"  🔹 Expiración: {expiration_time.strftime('%H:%M:%S')}")

    return audit_summary


# Prueba del ejercicio
demo_user: str = "USR-9842"

print(f"📋 Generando sesión de auditoría para '{demo_user}'...")
session_report: dict[str, Any] = create_session_audit(demo_user, session_duration_minutes=45)

print("\n📋 Resumen del Informe de Auditoría de Sesión:")
print(f"  🔹 ID Usuario:           {session_report['user_id']}")
print(f"  🔹 Timestamp de Inicio: {session_report['login_timestamp']}")
print(f"  🔹 Timestamp Expiración: {session_report['expiration_timestamp']}")
print(f"  🔹 Día de la semana:     {session_report['details']['day_of_week']}")

# --------------------------------------------------------------------------------
# 🏁 [PASO 4]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Lección de introducción al Módulo datetime finalizada.")
print("=" * 80)
````
