# 📅 El Módulo `datetime` en Python

La librería estándar `datetime` permite manipular fechas y horas de manera eficiente y flexible. Aunque soporta aritmética de fechas, su foco principal es facilitar la creación, extracción y formateo de componentes temporales.

### Clases Principales
- **`datetime.date`**: Representa una fecha idealizada (año, mes, día).
- **`datetime.time`**: Representa una hora del día independiente de la fecha (hora, minuto, segundo, microsegundo).
- **`datetime.datetime`**: Combinación de fecha y hora.
- **`datetime.timedelta`**: Duración o diferencia entre dos fechas o horas.
- **`datetime.tzinfo`**: Soporte para zonas horarias (zonas UTC, offset).

--- 
## 🛠️ Métodos y Uso Fundamental

### 1. Creación y Lectura de Fechas (`datetime` y `now`)

```python
import datetime

# 🛰️ MÓDULO DATETIME: Fechas específicas y actuales

# [1] Instanciar una fecha y hora específica: (año, mes, día, hora, minuto, segundo)
specific_datetime = datetime.datetime(1495, 2, 14, 12, 30, 0)
print("Fecha y hora específica:", specific_datetime)
# ✅ Output: 1495-02-14 12:30:00

# [2] Obtener fecha y hora actuales del sistema
current_datetime = datetime.datetime.now()
print("Fecha y hora actual:", current_datetime)
```

### 2. Trabajo Solo con Fechas (`date`) o Solo con Horas (`time`)

```python
import datetime

# 🔌 MÓDULO DATETIME: Objetos date y time

# [1] Trabajar solo con fechas
today_date = datetime.date.today()
specific_date = datetime.date(1495, 12, 31)

print("Hoy es:", today_date)
print("Fecha histórica:", specific_date)

# [2] Trabajar solo con horas
# NOTA: Extraemos la hora actual llamando .time() sobre datetime.now()
current_time = datetime.datetime.now().time()
specific_time = datetime.time(23, 14, 56)

print("Hora actual:", current_time)
print("Hora fija:", specific_time)
```

### 3. Formateo de Fechas (`strftime` y `strptime`)

- **`strftime` (String Format Time)**: Convierte un objeto de fecha a un `string` legible.
- **`strptime` (String Parse Time)**: Parsea una cadena de texto y la transforma en objeto `datetime`.

| Código | Significado | Ejemplo |
| :--- | :--- | :--- |
| `%Y` / `%y` | Año (4 dígitos / 2 dígitos) | `2026` / `26` |
| `%m` | Mes en número (01-12) | `08` |
| `%d` | Día del mes (01-31) | `14` |
| `%H` / `%I` | Hora (00-23 / 01-12) | `15` / `03` |
| `%M` / `%S` | Minutos / Segundos | `30` / `05` |
| `%p` | AM o PM | `PM` |

```python
import datetime

# ⚙️ MÓDULO DATETIME: Formateo con strftime y strptime

now = datetime.datetime.now()

# [1] Convertir datetime -> string
custom_format = now.strftime("%d/%m/%Y %H:%M:%S")
print("Fecha formateada:", custom_format)

# [2] Convertir string -> datetime
text_date = "2026-12-25 08:30:00"
parsed_date = datetime.datetime.strptime(text_date, "%Y-%m-%d %H:%M:%S")
print("Objeto parseado:", type(parsed_date), parsed_date)
```

### 4. Aritmética Temporal (`timedelta`)

Permite sumar o restar días, horas, minutos o semanas a una fecha.

```python
import datetime

# ⚙️ MÓDULO DATETIME: Aritmética de fechas con timedelta

today = datetime.date.today()

# [1] Restar o sumar días/semanas
one_week_ago = today - datetime.timedelta(days=7)
in_30_days = today + datetime.timedelta(days=30)

print("Hoy:", today)
print("Hace 7 días:", one_week_ago)
print("En 30 días:", in_30_days)

# [2] Diferencia entre dos fechas
event_date = datetime.date(2026, 12, 31)
days_remaining = (event_date - today).days
print(f"Faltan {days_remaining} días para fin de año.")
```

### Ejercicio 1.1 (Nivel Básico)
Calculadora de días vividos hasta la fecha actual.

```python
import datetime

# 📋 EJERCICIO 1.1: Calculadora de días vividos

def calculate_days_lived(year: int, month: int, day: int) -> int:
    try:
        # [1] Instanciar fecha de nacimiento y fecha de hoy
        birth_date = datetime.date(year, month, day)
        today = datetime.date.today()
        # [2] Calcular diferencia total en días
        days = (today - birth_date).days
        return days
    except ValueError as e:
        print(f"❌ Fecha inválida: {e}")
        return 0

# Prueba de la función:
days = calculate_days_lived(2000, 5, 15)
print(f"✅ Has vivido {days} días.")
```

### Ejercicio 2.1 (Nivel Intermedio)
Comprobar si un año es bisiesto evaluando la validez de la fecha `29 de febrero` con `datetime`.

```python
import datetime

# 📋 EJERCICIO 2.1: Comprobar año bisiesto con datetime

def is_leap_year(year: int) -> bool:
    try:
        # [1] Intentar instanciar el 29 de febrero para el año dado
        datetime.date(year, 2, 29)
        return True
    except ValueError:
        # [2] Si lanza ValueError, el año no es bisiesto
        return False

print(f"¿2024 es bisiesto?: {is_leap_year(2024)}")  # ✅ True
print(f"¿2026 es bisiesto?: {is_leap_year(2026)}")  # ❌ False
```

### Ejercicio 2.2 (Nivel Intermedio)
Generador de vencimiento de facturas (30 días de plazo) formateado en texto legible.

```python
import datetime

# 📋 EJERCICIO 2.2: Generador de vencimiento de facturas

def calculate_invoice_due_date(issue_date: datetime.date = None) -> str:
    # [1] Asignar fecha actual si no se provee argumento
    if issue_date is None:
        issue_date = datetime.date.today()
        
    # [2] Sumar 30 días de plazo
    due_date = issue_date + datetime.timedelta(days=30)
    
    # [3] Formatear a texto legible
    readable_format = due_date.strftime("%A, %d de %B de %Y")
    return readable_format

print("📋 Vencimiento de factura emitida hoy:", calculate_invoice_due_date())
```

### Ejercicio 3.1 (Nivel Avanzado)
Calculador de duración de logs de sesión a partir de cadenas ISO.

```python
import datetime

# 📋 EJERCICIO 3.1: Calculador de duración de logs de sesión

def calculate_session_duration(start_str: str, end_str: str) -> str:
    date_format = "%Y-%m-%d %H:%M:%S"
    
    try:
        # [1] Convertir strings de log a objetos datetime
        start_time = datetime.datetime.strptime(start_str, date_format)
        end_time = datetime.datetime.strptime(end_str, date_format)
        
        # [2] Validar coherencia temporal
        if end_time < start_time:
            return "❌ Error: La fecha de fin no puede ser anterior al inicio."
            
        # [3] Calcular minutos y segundos transcurridos
        difference = end_time - start_time
        total_seconds = int(difference.total_seconds())
        minutes = total_seconds // 60
        seconds = total_seconds % 60
        
        return f"✅ Duración de sesión: {minutes} min {seconds} seg"
        
    except ValueError as e:
        return f"❌ Error de formato en la fecha: {e}"

# Ejemplo de uso:
log_start = "2026-08-14 14:00:00"
log_end = "2026-08-14 14:45:32"
print(calculate_session_duration(log_start, log_end))
```
