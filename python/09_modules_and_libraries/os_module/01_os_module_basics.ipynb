### 💡 El Módulo `os` en Python

El módulo estándar **`os`** proporciona una interfaz unificada y portable para acceder a las funcionalidades del sistema operativo subyacente (Linux, macOS, Windows). Permite realizar operaciones de administración de infraestructura como la consulta de variables de entorno, la navegación por directorios y la manipulación del sistema de archivos.

---

### 🗂️ Principales Ámbitos de Aplicación de `os`

| Área de Trabajo | Descripción de Funcionalidades | Métodos / Atributos Clave |
| :--- | :--- | :--- |
| 💻 **Información del Sistema** | Identificación de la plataforma y configuración del entorno del SO. | `os.name`, `os.environ`, `os.uname()` |
| 📁 **Sistema de Archivos** | Creación, borrado, renombrado y navegación por directorios. | `os.getcwd()`, `os.chdir()`, `os.mkdir()`, `os.listdir()` |
| 🛣️ **Gestión de Rutas (`os.path`)** | Manipulación y validación de componentes de rutas multiplataforma. | `os.path.join()`, `os.path.exists()`, `os.path.isabs()` |

---

### ⚠️ Regla Importante de Importación

> 📌 **Buenas Prácticas:** La declaración **`import os`** debe situarse siempre en la **parte superior (encabezado)** del archivo o notebook de Python, ejecutándose una única vez al inicio del programa para garantizar la disponibilidad global de sus funciones.

```python
# 🛰️ [1] Importación estandarizada al inicio del archivo (PEP 8)
import os
from typing import Any

# Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 MÓDULO OS | Fundamentos e Interacción con el Sistema Operativo")
print("=" * 80)

# ================================================================================
# 🧪 1. INSPECCIÓN DEL ENTORNO DEL SISTEMA OPERATIVO
# ================================================================================
# Consultas iniciales de plataforma y variables de entorno del SO.
# ================================================================================

print("\n--- 🚀 [PASO 1]: Inspección de Datos del Sistema Operativo ---")

# 🔌 [1] Identificación del sistema operativo subyacente
operating_system_type: str = os.name
print(f"📋 Tipo de Sistema Operativo (os.name): {operating_system_type}")
print(f"  🔹 'posix' = Linux/macOS | 'nt' = Windows")

print("-" * 65)

# 🔌 [2] Consulta de variables de entorno del sistema
user_home_dir: str = os.environ.get("HOME") or os.environ.get("USERPROFILE") or "No Detectado"
current_path_env: str = os.environ.get("PATH", "No Configurado")

print("📋 Variables de Entorno Detectadas:")
print(f"  🔹 Carpeta Home del Usuario: {user_home_dir}")
print(f"  🔹 Tamaño de variable PATH: {len(current_path_env)} caracteres")
```

## 🧱 Operaciones de Validación Inicial del Sistema de Archivos

### 💡 Interacción Segura con el Entorno:
Antes de ejecutar cualquier lectura o escritura de datos, el módulo `os` nos permite comprobar el estado actual del directorio de trabajo mediante `os.getcwd()` y validar recursos antes de modificarlos.

```python
# ================================================================================
# 🧪 2. VERIFICACIÓN DE RUTAS Y DIRECTORIO DE TRABAJO
# ================================================================================
print("\n--- 🚀 [PASO 2]: Verificación del Directorio Actual ---")

# ⚙️ [1] Obtención del directorio de trabajo actual
current_execution_directory: str = os.getcwd()
print(f"📋 Directorio actual de ejecución (getcwd): {current_execution_directory}")

# ⚙️ [2] Construcción de ruta segura usando os.path
target_filename: str = "environment_audit.log"
absolute_file_path: str = os.path.join(current_execution_directory, target_filename)

print(f"📋 Ruta absoluta construida: {absolute_file_path}")
print(f"  🔹 ¿La ruta construida existe actualmente?: {os.path.exists(absolute_file_path)}")
```

### 🧪 Ejercicio 1: Inspector de Compatibilidad y Entorno del Servidor

**Escenario Empresarial:** Una aplicación multiplataforma debe ejecutarse en servidores Linux y Windows. Se requiere construir una función de diagnóstico que inspeccione las propiedades del sistema operativo utilizando `os`, verifique la presencia de variables de entorno clave y genere un informe preliminar de compatibilidad.

```python
# ================================================================================
# 🧪 3. EJERCICIO PRÁCTICO 1: INSPECTOR Y AUDITOR DE ENTORNO
# ================================================================================
print("\n--- 🧪 [PASO 3.1]: Ejercicio 1 - Diagnóstico del Entorno del SO ---")


def diagnose_system_environment() -> dict[str, Any]:
    """Analiza las características del sistema operativo y sus variables de entorno.

    Returns:
        dict[str, Any]: Reporte de auditoría del sistema con datos de plataforma.
    """
    system_report: dict[str, Any] = {
        "os_family": os.name,
        "is_compatible": os.name in ["posix", "nt"],
        "working_directory": os.getcwd(),
        "has_home_var": False,
        "system_user": "Desconocido"
    }

    # [1] Inspeccionar variable de entorno del usuario
    user_name: str | None = os.environ.get("USER") or os.environ.get("USERNAME")
    if user_name:
        system_report["system_user"] = user_name

    # [2] Evaluar presencia de carpeta HOME
    if "HOME" in os.environ or "USERPROFILE" in os.environ:
        system_report["has_home_var"] = True

    return system_report


# Ejecución del diagnóstico
print("📋 Iniciando análisis del entorno del sistema...")
environment_audit: dict[str, Any] = diagnose_system_environment()

print(f"  ✅ Familia de SO detectada:   {environment_audit['os_family']}")
print(f"  ✅ ¿Plataforma compatible?:   {environment_audit['is_compatible']}")
print(f"  👤 Usuario del sistema:       {environment_audit['system_user']}")
print(f"  📁 Ruta de trabajo activa:    {environment_audit['working_directory']}")

# --------------------------------------------------------------------------------
# 🏁 [PASO 4]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Lección de Fundamentos del Módulo OS finalizada.")
print("=" * 80)
```
