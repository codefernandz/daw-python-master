---

### 💡 El Método `os.path.join()`

La función **`os.path.join(path, *paths)`** une uno o más componentes de ruta de manera inteligente. Inserta automáticamente el separador de archivos adecuado según el sistema operativo en el que se ejecute el script (`/` en sistemas Linux/macOS y `\` en Windows).

---

### 🗂️ Portabilidad y Comportamiento de `os.path.join()`

```text
[ Base: '/ruta/base' ] + [ Archivo: 'archivo.txt' ] ──( os.path.join )──> '/ruta/base/archivo.txt' (POSIX)
                                                                    └──> '\ruta\base\archivo.txt' (Windows)
```

| Sistema Operativo | Separador (`os.sep`) | Ejemplo de Resultado |
| :--- | :--- | :--- |
| 🐧 POSIX (Linux / macOS) | `/` | `/ruta/base/archivo.txt` |
| 🪟 Windows | `\` | `\ruta\base\archivo.txt` |

---

### 📚 Recurso Adicional:
Para explorar todas las funcionalidades adicionales de manipulación del sistema de archivos y rutas, puedes consultar la [Documentación Oficial del Módulo OS en Python](https://docs.python.org/3/library/os.html).
```

````python
import os
from typing import Any

# 🛰️ [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 MÓDULO OS | Construcción de Rutas con os.path.join()")
print("=" * 80)

# ================================================================================
# 🧪 1. CONCATENACIÓN BÁSICA DE RUTAS MULTIPLATAFORMA
# ================================================================================
# Unión de ruta base y nombre de archivo mediante os.path.join().
# ================================================================================

print("\n--- 🚀 [PASO 1]: Concatenación Segura de Segmentos de Ruta ---")

base_path: str = "/ruta/base"
file_name: str = "archivo.txt"

# 🔌 [1] Unir partes de rutas de archivos para crear una ruta completa
full_path: str = os.path.join(base_path, file_name)

print(f"📋 Ruta base:    '{base_path}'")
print(f"📋 Archivo:      '{file_name}'")
print(f"📋 Ruta completa: '{full_path}'")

# 🔌 [2] Mostrar el separador de sistema detectado en el entorno actual
print(f"  🔹 Separador activo del sistema (os.sep): '{os.sep}'")
````

## 🧱 Unión Multi-Segmento y Comportamiento con Rutas Absolutas

### 💡 Reglas Importantes de `os.path.join()`:
1. **Múltiples Segmentos:** Se pueden pasar N argumentos para construir estructuras de carpetas profundas.
2. **Reinicio por Ruta Absoluta:** Si uno de los componentes unidos es una ruta absoluta (empieza con `/` o letra de unidad `C:\`), `os.path.join()` descartará todos los componentes anteriores y comenzará desde dicha ruta absoluta.

````python
# ================================================================================
# 🧪 2. CASOS AVANZADOS: MÚLTIPLES NIVELES Y RUTAS ABSOLUTAS INTERMEDIAS
# ================================================================================
print("\n--- 🚀 [PASO 2]: Múltiples Argumentos y Anulaciones por Ruta Absoluta ---")

# ⚙️ [1] Concatenación de múltiples subdirectorios en una sola llamada
multi_level_path: str = os.path.join("proyectos", "python", "logs", "app.log")
print(f"📋 Ruta multinivel construida: '{multi_level_path}'")

# ⚙️ [2] Demostración de anulación al encontrar un segmento absoluto intermedio
root_override_path: str = os.path.join("carpeta_ignorada", "/sistema/root", "config.ini")
print(f"\n📋 Ejemplo con segmento absoluto intermedio ('/sistema/root'):")
print(f"  ⚠️ Ruta final (se descarta 'carpeta_ignorada'): '{root_override_path}'")
````

## 🛒 Ejercicios Prácticos Avanzados Integrados

---

### 🧪 Ejercicio 1: Generador Dinámico de Rutas de Almacenamiento Jerárquico

**Escenario Empresarial:** Un sistema de gestión documental requiere organizar archivos adjuntos guardándolos en una estructura de carpetas basada en el año, mes y categoría del documento (`almacen/2026/08/facturas/factura_001.pdf`), garantizando la portabilidad del código.

````python
# ================================================================================
# 🧪 3. EJERCICIO PRÁCTICO 1: CONSTRUCTOR DE ESTRUCTURAS DE ALMACENAMIENTO
# ================================================================================
print("\n--- 🧪 [PASO 3.1]: Ejercicio 1 - Generador de Rutas de Almacenamiento ---")


def build_storage_path(
    root_dir: str, year: str, month: str, category: str, file_name: str
) -> dict[str, Any]:
    """Construye dinámicamente una ruta jerárquica segura y crea los directorios si no existen.

    Args:
        root_dir (str): Directorió raíz de almacenamiento.
        year (str): Año de clasificación.
        month (str): Mes de clasificación.
        category (str): Categoría del archivo.
        file_name (str): Nombre del archivo final.

    Returns:
        dict[str, Any]: Resumen con el directorio contenedor y la ruta absoluta final.
    """
    # [1] Construcción jerárquica del directorio destino
    target_directory: str = os.path.join(root_dir, year, month, category)

    # [2] Construcción de la ruta completa del archivo
    file_full_path: str = os.path.join(target_directory, file_name)

    # [3] Crear la estructura de carpetas de forma segura
    os.makedirs(target_directory, exist_ok=True)

    return {
        "target_directory": target_directory,
        "file_full_path": file_full_path,
        "is_absolute": os.path.isabs(file_full_path)
    }


# Prueba del ejercicio
root_storage: str = "media_vault"
doc_year: str = "2026"
doc_month: str = "08"
doc_category: str = "invoices"
doc_file: str = "invoice_1024.pdf"

print("📋 Generando ruta jerárquica para almacenamiento de documento...")
storage_info: dict[str, Any] = build_storage_path(
    root_storage, doc_year, doc_month, doc_category, doc_file
)

print("\n📋 Resultado de la Construcción de Ruta:")
print(f"  📁 Directorio objetivo: {storage_info['target_directory']}")
print(f"  📄 Ruta completa:      {storage_info['file_full_path']}")

# Crear el archivo simbólico dentro de la ruta para validar la creación
with open(storage_info["file_full_path"], "w", encoding="utf-8") as f:
    f.write("Simulación de contenido PDF.")

print(f"  ✅ Confirmación: Archivo creado en '{os.path.exists(storage_info['file_full_path'])}'")

# Limpieza final de la estructura creada para el ejercicio
if os.path.exists(root_storage):
    for root, dirs, files in os.walk(root_storage, topdown=False):
        for name in files:
            os.remove(os.path.join(root, name))
        for name in dirs:
            os.rmdir(os.path.join(root, name))
    os.rmdir(root_storage)
    print("  🔒 Estructura de prueba eliminada correctamente.")

# --------------------------------------------------------------------------------
# 🏁 [PASO 4]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Lección de os.path.join() finalizada.")
print("=" * 80)
````
