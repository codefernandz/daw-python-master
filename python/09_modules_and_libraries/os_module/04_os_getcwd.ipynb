### 💡 El Método `os.getcwd()`

La función **`os.getcwd()`** (*Get Current Working Directory*) devuelve una cadena de texto (`str`) con la **ruta absoluta** del directorio desde el cual se está ejecutando el script o la sesión activa de Python.

---

### 🗂️ Importancia del Directorio de Trabajo

```text
[ Terminal / IDE ] ──( Ejecuta script desde /app )──> os.getcwd() ──> '/app'
```

| Aspecto | Descripción |
| :--- | :--- |
| 📍 Resolución de Rutas Relativas | Cuando abres un archivo usando una ruta relativa como `'mi_fichero.txt'`, Python lo busca dentro de la ruta devuelta por `os.getcwd()`. |
| 🖥️ Independencia del Entorno | Permite verificar en qué punto del sistema de archivos se encuentra el proceso antes de realizar operaciones de lectura o escritura. |
```

````python
import os
from typing import Any

# 🛰️ [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 MÓDULO OS | Directorio de Trabajo Actual con os.getcwd()")
print("=" * 80)

# ================================================================================
# 🧪 1. OBTENCIÓN Y CONSULTA DEL CWD (CURRENT WORKING DIRECTORY)
# ================================================================================
# Consulta de la ruta absoluta actual de ejecución.
# ================================================================================

print("\n--- 🚀 [PASO 1]: Consulta de os.getcwd() ---")

# 🔌 [1] Acceso al directorio de trabajo actual
current_directory: str = os.getcwd()

print(f"📋 Directorio de trabajo actual: {current_directory}")
print(f"  🔹 Tipo de dato retornado: {type(current_directory)}")
print(f"  🔹 ¿Es una ruta absoluta?: {os.path.isabs(current_directory)}")
````

## 🧱 Inspección y Resolución de Rutas Absolutas

### 💡 Construcción Segura de Rutas a Partir del CWD:
Saber el directorio de trabajo actual permite construir rutas absolutas a archivos usando `os.path.join()`, evitando fallos por ejecutar el código desde carpetas distintas.

````python
# ================================================================================
# 🧪 2. RESOLUCIÓN DE RUTAS ABSOLUTAS BASADAS EN EL CWD
# ================================================================================
print("\n--- 🚀 [PASO 2]: Construcción de Rutas Basadas en getcwd() ---")

# ⚙️ [1] Obtener el CWD activo
working_dir: str = os.getcwd()

# ⚙️ [2] Unir el CWD con el nombre de un archivo
target_file_name: str = "app_data.json"
absolute_file_path: str = os.path.join(working_dir, target_file_name)

print(f"📋 Ruta construida dinámicamente:")
print(f"  🔹 Directorio base: {working_dir}")
print(f"  🔹 Archivo objetivo: {target_file_name}")
print(f"  🔹 Ruta absoluta:   {absolute_file_path}")
````

## 🛒 Ejercicios Prácticos Avanzados Integrados

---

### 🧪 Ejercicio 1: Validador y Auditor del Entorno de Ejecución

**Escenario Empresarial:** Una tarea programada (*Cron job* / *Task Scheduler*) requiere verificar que se está ejecutando en el directorio de trabajo correcto antes de generar archivos de salida, evitando así la creación no deseada de ficheros en carpetas del sistema.

````python
# ================================================================================
# 🧪 3. EJERCICIO PRÁCTICO 1: AUDITOR DE DIRECTORIO DE EJECUCIÓN
# ================================================================================
print("\n--- 🧪 [PASO 3.1]: Ejercicio 1 - Auditor del CWD ---")


def validate_execution_directory(expected_folder_name: str | None = None) -> dict[str, Any]:
    """Audita el directorio de trabajo actual y valida si coincide con la carpeta esperada.

    Args:
        expected_folder_name (str | None): Nombre de la carpeta requerida (opcional).

    Returns:
        dict[str, Any]: Reporte del CWD actual y estado de validación.
    """
    current_cwd: str = os.getcwd()
    folder_basename: str = os.path.basename(current_cwd)

    audit_report: dict[str, Any] = {
        "current_cwd": current_cwd,
        "folder_basename": folder_basename,
        "is_valid": True,
        "warning": None
    }

    if expected_folder_name and folder_basename.lower() != expected_folder_name.lower():
        audit_report["is_valid"] = False
        audit_report["warning"] = (
            f"El script se está ejecutando en '{folder_basename}', "
            f"se esperaba '{expected_folder_name}'."
        )
        print(f"  ❌ Advertencia de directorio: {audit_report['warning']}")
    else:
        print(f"  ✅ Ejecución en directorio válido: '{folder_basename}'")

    return audit_report


# Prueba del ejercicio
print("📋 Ejecutando auditoría del CWD actual...")
cwd_audit: dict[str, Any] = validate_execution_directory()

print("\n📋 Resumen del Informe de Auditoría:")
print(f"  🔹 Ruta absoluta actual:  {cwd_audit['current_cwd']}")
print(f"  🔹 Carpeta de ejecución: {cwd_audit['folder_basename']}")

# --------------------------------------------------------------------------------
# 🏁 [PASO 4]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Lección de os.getcwd() finalizada.")
print("=" * 80)
````
