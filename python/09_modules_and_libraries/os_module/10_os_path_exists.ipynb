| 🛠️ **Conceptos** | `os.path.exists()`, `Validación Preventiva`, `Control de Flujo Guardián`, `Rutas Relativas y Absolutas` |

---

### 💡 El Método `os.path.exists()`

La función **`os.path.exists(path)`** devuelve un valor booleano (`True` o `False`) indicando si la ruta especificada existe en el sistema de archivos. Funciona indistintamente para **archivos**, **directorios** y **enlaces simbólicos**.

---

### 🗂️ Comportamiento de `os.path.exists()`

```text
[ Ruta Objetivo: '/ruta/al/elemento' ] ──( os.path.exists )──> ¿Existe? ──┬──> True  ──> Ejecutar lectura/modificación
                                                                        └──> False ──> Crear o notificar ausencia
```

| Resultado | Causa | Acción Sugerida |
| :--- | :--- | :--- |
| `True` | La ruta apunta a un archivo, carpeta o enlace válido y accesible. | Proceder con la operación (lectura, borrado, etc.). |
| `False` | La ruta no existe, la ruta es inválida o no hay permisos de lectura en la ruta base. | Prevenir operaciones que requieran la existencia del elemento. |
```

````python
import os
from typing import Any

# 🛰️ [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 MÓDULO OS | Verificación de Existencia con os.path.exists()")
print("=" * 80)

# ================================================================================
# 🧪 1. COMPROBACIÓN BÁSICA DE EXISTENCIA CON BLOQUE IF-ELSE
# ================================================================================
# Verificación condicional de archivos y directorios preexistentes o ausentes.
# ================================================================================

print("\n--- 🚀 [PASO 1]: Evaluación Condicional con os.path.exists() ---")

target_path: str = "archivo_o_directorio.txt"

# ⚙️ [1] Crear un archivo de prueba temporal
with open(target_path, "w", encoding="utf-8") as file:
    file.write("Registro de prueba para verificación de existencia.")

try:
    # 🔌 [1] Verificar existencia de un elemento existente
    if os.path.exists(target_path):
        print(f"  ✅ El elemento '{target_path}' existe en el sistema de archivos.")
    else:
        print(f"  ❌ El elemento '{target_path}' no existe.")

    # ⚙️ [2] Probar con una ruta que sabemos que no existe
    missing_path: str = "directorio_ficticio/no_existente.dat"
    
    # 🔌 [2] Verificar existencia de un elemento ausente
    if os.path.exists(missing_path):
        print(f"  ✅ El elemento '{missing_path}' existe en el sistema de archivos.")
    else:
        print(f"  ❌ El elemento '{missing_path}' no existe.")

finally:
    # Limpieza del entorno de prueba
    if os.path.exists(target_path):
        os.remove(target_path)
        print("  🔒 Entorno de prueba limpiado correctamente.")
````

## 🧱 Patrón Guardián y Distinción de Tipos de Rutas

### 💡 Buenas Prácticas al Usar `os.path.exists()`:
Aunque `os.path.exists()` valida la presencia de un elemento, no especifica si es un archivo o una carpeta. Es común combinarlo con `os.path.isfile()` o `os.path.isdir()` cuando se requiere realizar operaciones específicas.

````python
# ================================================================================
# 🧪 2. PATRÓN GUARDIÁN Y CLASIFICACIÓN DE ELEMENTOS EXISTENTES
# ================================================================================
print("\n--- 🚀 [PASO 2]: Inspección Combinada de Existencia y Tipo ---")

demo_file: str = "config_sample.json"
demo_folder: str = "folder_sample"

# Crear estructura de prueba
with open(demo_file, "w", encoding="utf-8") as f:
    f.write("{}")
os.makedirs(demo_folder, exist_ok=True)

paths_to_inspect: list[str] = [demo_file, demo_folder, "invalid_item.txt"]

try:
    for current_path in paths_to_inspect:
        print(f"\n📋 Inspeccionando: '{current_path}'...")

        # ⚙️ [1] Aplicación del patrón guardián con os.path.exists()
        if not os.path.exists(current_path):
            print(f"  ❌ Error Guardián: La ruta '{current_path}' no fue encontrada.")
            continue

        # ⚙️ [2] Clasificación del tipo de elemento una vez confirmada su existencia
        if os.path.isfile(current_path):
            print(f"  📄 El elemento es un ARCHIVO válido.")
        elif os.path.isdir(current_path):
            print(f"  📁 El elemento es un DIRECTORIO válido.")

finally:
    if os.path.exists(demo_file):
        os.remove(demo_file)
    if os.path.exists(demo_folder):
        os.rmdir(demo_folder)
    print("\n  🔒 Archivos de prueba eliminados correctamente.")
````

## 🛒 Ejercicios Prácticos Avanzados Integrados

---

### 🧪 Ejercicio 1: Verificador Guardián de Dependencias de Archivos

**Escenario Empresarial:** Se necesita una función de precarga para una canalización de datos (*Data Pipeline*) que verifique si existen todos los archivos de configuración y datos requeridos antes de iniciar el procesamiento, cancelando la ejecución de forma segura si falta algún componente esencial.

````python
# ================================================================================
# 🧪 3. EJERCICIO PRÁCTICO 1: AUDITOR PREVENTIVO DE DEPENDENCIAS
# ================================================================================
print("\n--- 🧪 [PASO 3.1]: Ejercicio 1 - Verificador Guardián de Pipeline ---")


def verify_pipeline_dependencies(required_paths: list[str]) -> dict[str, Any]:
    """Verifica si todas las dependencias de rutas de un pipeline existen.

    Args:
        required_paths (list[str]): Lista de rutas requeridas para el proceso.

    Returns:
        dict[str, Any]: Resumen con el estado general de validación y elementos ausentes.
    """
    verification_result: dict[str, Any] = {
        "ready_to_execute": True,
        "found_paths": [],
        "missing_paths": []
    }

    for path in required_paths:
        # [1] Evaluación con os.path.exists()
        if os.path.exists(path):
            verification_result["found_paths"].append(path)
            print(f"  ✅ [OK] Encontrado: '{path}'")
        else:
            verification_result["missing_paths"].append(path)
            verification_result["ready_to_execute"] = False
            print(f"  ❌ [FALTA] No encontrado: '{path}'")

    return verification_result


# Crear algunos archivos simulando dependencias del pipeline
required_manifest: list[str] = ["dataset_part1.csv", "settings.env", "schema.json"]

with open("dataset_part1.csv", "w", encoding="utf-8") as f:
    f.write("id,value\n1,100")
with open("settings.env", "w", encoding="utf-8") as f:
    f.write("ENV=PROD")

print("📋 Verificando dependencias necesarias para iniciar el pipeline...")
pipeline_status: dict[str, Any] = verify_pipeline_dependencies(required_manifest)

print("\n📋 Reporte Final de Readiness del Pipeline:")
print(f"  🔹 ¿Listo para ejecutar?: {pipeline_status['ready_to_execute']}")
print(f"  🔹 Elementos confirmados ({len(pipeline_status['found_paths'])}): {pipeline_status['found_paths']}")
print(f"  🔹 Elementos faltantes   ({len(pipeline_status['missing_paths'])}): {pipeline_status['missing_paths']}")

if not pipeline_status["ready_to_execute"]:
    print("\n  ⚠️ ALERTA: La ejecución del pipeline se detiene debido a dependencias faltantes.")

# Limpieza final de dependencias de prueba
for created_file in pipeline_status["found_paths"]:
    if os.path.exists(created_file):
        os.remove(created_file)

# --------------------------------------------------------------------------------
# 🏁 [PASO 4]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Lección de os.path.exists() finalizada.")
print("=" * 80)
````
