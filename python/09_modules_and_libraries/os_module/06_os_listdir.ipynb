# ⚙️ Módulos y Librerías: Listado de Archivos y Directorios (`os.listdir`)

| Propiedad | Detalle |
| :--- | :--- |
| 📍 **Módulo** | Módulos, Paquetes y Librerías / Módulo OS |
| 🎯 **Objetivo** | Entender el uso de la función `os.listdir()` para obtener el contenido de un directorio específico, filtrando elementos entre archivos y carpetas, y gestionando excepciones comunes como `FileNotFoundError`. |
| 🛠️ **Conceptos** | `os.listdir()`, `Filtrado de Elementos`, `os.path.isfile()`, `os.path.isdir()`, `FileNotFoundError` |

---

### 💡 El Método `os.listdir()`

La función **`os.listdir(path)`** devuelve una lista de cadenas de texto (`list[str]`) que contiene los nombres de todos los archivos y subdirectorios presentes en la ruta especificada. Si no se proporciona ningún argumento, lista el contenido del directorio de trabajo actual (`CWD`).

---

### 🗂️ Características Clave de `os.listdir()`

```text
[ Ruta Objetivo: '/ruta/al/directorio' ] ──( os.listdir )──> ['archivo.txt', 'subcarpeta', 'script.py']
```

| Aspecto | Descripción |
| :--- | :--- |
| 📄 Nombres Relativos | Devuelve únicamente los nombres de los elementos, no las rutas absolutas. Para operar sobre ellos se deben unir con `os.path.join()`. |
| 🔀 Orden Arbitrario | La lista no garantiza un orden alfabético o por fecha; depende del sistema de archivos subyacente. |
| 🚨 Manejo de Errores | Lanza `FileNotFoundError` si la ruta no existe, y `NotADirectoryError` si la ruta apunta a un archivo. |
```

````python
import os
from typing import Any

# 🛰️ [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 MÓDULO OS | Listado de Contenido con os.listdir()")
print("=" * 80)

# ================================================================================
# 🧪 1. DEMOSTRACIÓN BÁSICA DE OS.LISTDIR()
# ================================================================================
# Creación de estructura temporal de prueba y exploración de contenido.
# ================================================================================

print("\n--- 🚀 [PASO 1]: Inspección de un Directorio con os.listdir() ---")

target_directory: str = "demo_inspection_dir"

# Crear estructura de prueba
os.makedirs(os.path.join(target_directory, "subfolder_alpha"), exist_ok=True)
with open(os.path.join(target_directory, "notes.txt"), "w", encoding="utf-8") as file1:
    file1.write("Notas de prueba.")
with open(os.path.join(target_directory, "data.json"), "w", encoding="utf-8") as file2:
    file2.write("{}")

try:
    # 🔌 [1] Listar los archivos y directorios en la ruta especificada
    directory_contents: list[str] = os.listdir(target_directory)

    print(f"📋 Contenido hallado en '{target_directory}':")
    print(f"  🔹 Lista de elementos ({len(directory_contents)}): {directory_contents}")

finally:
    # Limpieza de archivos y carpetas temporales
    if os.path.exists(target_directory):
        for item in os.listdir(target_directory):
            item_path = os.path.join(target_directory, item)
            if os.path.isdir(item_path):
                os.rmdir(item_path)
            else:
                os.remove(item_path)
        os.rmdir(target_directory)
        print("  🔒 Entorno de prueba limpiado correctamente.")
````

## 🧱 Classificación y Filtrado de Contenido

### 💡 Discriminación entre Archivos y Directorios:
Puesto que `os.listdir()` retorna únicamente nombres simples, se debe combinar con `os.path.join()` junto con `os.path.isfile()` o `os.path.isdir()` para diferenciar el tipo de cada elemento.

````python
# ================================================================================
# 🧪 2. FILTRADO Y CATEGORIZACIÓN DE ELEMENTOS
# ================================================================================
print("\n--- 🚀 [PASO 2]: Clasificación de Archivos y Carpetas ---")

test_dir: str = "demo_classification_dir"
os.makedirs(os.path.join(test_dir, "logs"), exist_ok=True)
with open(os.path.join(test_dir, "config.env"), "w", encoding="utf-8") as temp_file:
    temp_file.write("ENV=DEV")

try:
    # ⚙️ [1] Obtener lista de nombres
    raw_items: list[str] = os.listdir(test_dir)

    files_found: list[str] = []
    folders_found: list[str] = []

    # ⚙️ [2] Clasificación iterativa construyendo rutas absolutas/relativas válidas
    for item_name in raw_items:
        full_element_path: str = os.path.join(test_dir, item_name)

        if os.path.isfile(full_element_path):
            files_found.append(item_name)
        elif os.path.isdir(full_element_path):
            folders_found.append(item_name)

    print(f"📋 Resumen de Clasificación para '{test_dir}':")
    print(f"  📄 Archivos ({len(files_found)}): {files_found}")
    print(f"  📁 Carpetas ({len(folders_found)}): {folders_found}")

finally:
    if os.path.exists(test_dir):
        for item in os.listdir(test_dir):
            path = os.path.join(test_dir, item)
            os.rmdir(path) if os.path.isdir(path) else os.remove(path)
        os.rmdir(test_dir)
````

## 🛒 Ejercicios Prácticos Avanzados Integrados

---

### 🧪 Ejercicio 1: Escáner y Clasificador de Directorios con Resumen de Extensión

**Escenario Empresarial:** Se requiere construir un auditor de sistema de archivos que inspeccione cualquier directorio provisto, contabilice de forma separada archivos y subdirectorios, y organice los archivos según su extensión (por ejemplo, `.txt`, `.json`, `.log`).

````python
# ================================================================================
# 🧪 3. EJERCICIO PRÁCTICO 1: ESCÁNER Y CLASIFICADOR DE CONTENIDO
# ================================================================================
print("\n--- 🧪 [PASO 3.1]: Ejercicio 1 - Inspector de Contenido por Extensión ---")


def scan_directory_structure(directory_path: str) -> dict[str, Any]:
    """Escanea un directorio y clasifica su contenido por tipo y extensión de archivo.

    Args:
        directory_path (str): Ruta del directorio a auditar.

    Returns:
        dict[str, Any]: Reporte detallado con carpetas, archivos y desglose por extensiones.
    """
    scan_report: dict[str, Any] = {
        "success": False,
        "total_items": 0,
        "directories": [],
        "files_by_extension": {},
        "error": None
    }

    try:
        # [1] Intentar listar el contenido del directorio
        items_list: list[str] = os.listdir(directory_path)
        scan_report["total_items"] = len(items_list)
        scan_report["success"] = True

        for item in items_list:
            full_path: str = os.path.join(directory_path, item)

            if os.path.isdir(full_path):
                scan_report["directories"].append(item)
            elif os.path.isfile(full_path):
                # Extraer la extensión del archivo (ejemplo: '.json')
                _, extension = os.path.splitext(item)
                ext_key: str = extension.lower() if extension else "sin_extension"

                if ext_key not in scan_report["files_by_extension"]:
                    scan_report["files_by_extension"][ext_key] = []
                
                scan_report["files_by_extension"][ext_key].append(item)

        print(f"  ✅ Escaneo de '{directory_path}' finalizado con éxito.")

    except FileNotFoundError:
        scan_report["error"] = f"El directorio '{directory_path}' no existe."
        print(f"  ❌ Error: {scan_report['error']}")
    except PermissionError:
        scan_report["error"] = f"Permisos insuficientes para acceder a '{directory_path}'."
        print(f"  ❌ Error: {scan_report['error']}")

    return scan_report


# Prueba del ejercicio
audit_dir: str = "workspace_audit_test"
os.makedirs(os.path.join(audit_dir, "backup"), exist_ok=True)
with open(os.path.join(audit_dir, "report.csv"), "w", encoding="utf-8") as f:
    f.write("id,status\n1,OK")
with open(os.path.join(audit_dir, "data.json"), "w", encoding="utf-8") as f:
    f.write("{}")

print(f"📋 Iniciando escaneo del directorio '{audit_dir}'...")
audit_results: dict[str, Any] = scan_directory_structure(audit_dir)

print("\n📋 Resumen de la Inspección:")
print(f"  🔹 Total de elementos:   {audit_results['total_items']}")
print(f"  📁 Subdirectorios ({len(audit_results['directories'])}): {audit_results['directories']}")
print("  📄 Archivos por extensión:")
for ext, files in audit_results["files_by_extension"].items():
    print(f"     • Extensión [{ext}]: {files}")

# Limpieza del entorno de prueba
if os.path.exists(audit_dir):
    for root, dirs, files in os.walk(audit_dir, topdown=False):
        for name in files:
            os.remove(os.path.join(root, name))
        for name in dirs:
            os.rmdir(os.path.join(root, name))
    os.rmdir(audit_dir)

# --------------------------------------------------------------------------------
# 🏁 [PASO 4]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Lección de os.listdir() finalizada.")
print("=" * 80)
````
