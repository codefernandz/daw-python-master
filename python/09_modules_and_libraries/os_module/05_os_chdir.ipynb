---

### 💡 El Método `os.chdir()`

La función **`os.chdir(path)`** (*Change Directory*) permite modificar la carpeta de trabajo actual desde la cual el proceso de Python resuelve todas las rutas relativas. 

---

### 🗂️ Comportamiento del Cambio de Directorio

```text
[ CWD Inicial: /app ] ──( os.chdir('/tmp') )──> [ Nuevo CWD: /tmp ] ──> Las rutas relativas ahora apuntan a /tmp
```

| Tipo de Error | Causa Principal | Prevención / Manejo |
| :--- | :--- | :--- |
| 🚨 `FileNotFoundError` | La ruta especificada no existe en el sistema de archivos. | Validar con `os.path.exists()` antes de cambiar. |
| 🚨 `NotADirectoryError` | La ruta provista existe, pero apunta a un archivo en lugar de una carpeta. | Validar con `os.path.isdir()`. |
```

````python
import os
from typing import Any

# 🛰️ [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 MÓDULO OS | Cambio del Directorio de Trabajo con os.chdir()")
print("=" * 80)

# ================================================================================
# 🧪 1. CAMBIO DE DIRECTORIO CON PREPARACIÓN Y RESTAURACIÓN DE CONTEXTO
# ================================================================================
# Creación de directorio temporal para demostrar os.chdir() de forma segura.
# ================================================================================

print("\n--- 🚀 [PASO 1]: Uso de os.chdir() y Confirmación con os.getcwd() ---")

# Guardar el directorio de origen para restaurarlo al finalizar
original_directory: str = os.getcwd()
temp_target_dir: str = "temp_working_dir"

# Crear directorio temporal para el ejemplo
if not os.path.exists(temp_target_dir):
    os.mkdir(temp_target_dir)

try:
    print(f"📋 Directorio de trabajo inicial: {original_directory}")

    # 🔌 [1] Cambiar al directorio especificado
    os.chdir(temp_target_dir)

    # 🔌 [2] Verificar el cambio de directorio
    new_directory: str = os.getcwd()
    print(f"  ✅ Nuevo directorio de trabajo: {new_directory}")

finally:
    # 🔌 [3] Restauración obligatoria del directorio original
    os.chdir(original_directory)
    print(f"  🔒 CWD restaurado al origen: {os.getcwd()}")

    # Limpieza del directorio temporal
    if os.path.exists(temp_target_dir):
        os.rmdir(temp_target_dir)
````

## 🧱 Manejo Seguro de Excepciones al Modificar el CWD

### 💡 Prevención de Errores al Navegar por el Sistema de Archivos:
Tratar de cambiar a una ruta inexistente o sin permisos detendrá el script. Es una buena práctica envolver llamadas a `os.chdir()` dentro de un bloque `try-except`.

````python
# ================================================================================
# 🧪 2. CAPTURA DE ERRORES AL INTENTAR CAMBIAR A RUTAS INVÁLIDAS
# ================================================================================
print("\n--- 🚀 [PASO 2]: Control de Excepciones con rutas inexistentes ---")

invalid_path: str = "/ruta/al/nuevo/directorio_inexistente"

# ⚙️ [1] Intento de cambio con manejo explícito de FileNotFoundError
try:
    print(f"📋 Intentando navegar a: '{invalid_path}'...")
    os.chdir(invalid_path)
except FileNotFoundError as error:
    print("  ❌ Excepción capturada: El directorio especificado no existe.")
    print(f"  🔍 Detalle técnico: {error}")
except NotADirectoryError as error:
    print("  ❌ Excepción capturada: La ruta no es un directorio válido.")
    print(f"  🔍 Detalle técnico: {error}")
````

## 🛒 Ejercicios Prácticos Avanzados Integrados

---

### 🧪 Ejercicio 1: Administrador de Contexto para Navegación Temporal de Directorios

**Escenario Empresarial:** Se necesita realizar operaciones temporales en una carpeta de trabajo específica y garantizar que, independientemente de si las tareas tienen éxito o fallan, el script regrese automáticamente al directorio original.

````python
# ================================================================================
# 🧪 3. EJERCICIO PRÁCTICO 1: NAVEGADOR TEMPORAL SEGURO DE DIRECTORIOS
# ================================================================================
print("\n--- 🧪 [PASO 3.1]: Ejercicio 1 - Gestor Temporal de CWD ---")


def execute_in_directory(target_path: str) -> dict[str, Any]:
    """Cambia temporalmente a un directorio objetivo, ejecuta una consulta y regresa al origen.

    Args:
        target_path (str): Ruta del directorio a inspeccionar.

    Returns:
        dict[str, Any]: Estado de la operación y rutas inspeccionadas.
    """
    execution_summary: dict[str, Any] = {
        "success": False,
        "previous_cwd": os.getcwd(),
        "active_cwd": None,
        "error": None
    }

    try:
        # [1] Intentar cambiar al directorio de destino
        os.chdir(target_path)
        execution_summary["active_cwd"] = os.getcwd()
        execution_summary["success"] = True
        print(f"  ✅ Navegación exitosa a: {execution_summary['active_cwd']}")

    except FileNotFoundError:
        execution_summary["error"] = f"La ruta '{target_path}' no fue encontrada."
        print(f"  ❌ Error: {execution_summary['error']}")
    finally:
        # [2] Garantizar el retorno al CWD original
        os.chdir(execution_summary["previous_cwd"])
        print(f"  🔒 Retorno garantizado a CWD inicial: {os.getcwd()}")

    return execution_summary


# Prueba del ejercicio
demo_dir: str = "workspace_test"
os.makedirs(demo_dir, exist_ok=True)

print(f"📋 Ejecutando cambio temporal hacia '{demo_dir}'...")
result: dict[str, Any] = execute_in_directory(demo_dir)

print("\n📋 Resultado de la Operación:")
print(f"  🔹 Operación completada: {result['success']}")
print(f"  🔹 CWD durante tareas:   {result['active_cwd']}")
print(f"  🔹 CWD final:            {result['previous_cwd']}")

# Limpieza del directorio de prueba
if os.path.exists(demo_dir):
    os.rmdir(demo_dir)

# --------------------------------------------------------------------------------
# 🏁 [PASO 4]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Lección de os.chdir() finalizada.")
print("=" * 80)
````
