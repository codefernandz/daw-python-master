| Propiedad | Detalle |
| :--- | :--- |
| 📍 **Módulo** | Módulos, Paquetes y Librerías / Módulo OS |
| 🎯 **Objetivo** | Comprender la creación de carpetas individuales mediante el método `os.mkdir()`, previniendo errores mediante validaciones con `os.path.exists()` y gestionando las excepciones `FileExistsError` y `FileNotFoundError`. |
| 🛠️ **Conceptos** | `os.mkdir()`, `Creación de Directorios`, `FileExistsError`, `FileNotFoundError`, `os.path.exists()` |

---

### 💡 El Método `os.mkdir()`

La función **`os.mkdir(path, mode=0o777)`** (*Make Directory*) crea un nuevo directorio en la ruta especificada. Es ideal para crear una **única carpeta** dentro de un directorio padre existente.

---

### 🗂️ Consideraciones Importantes de `os.mkdir()`

```text
[ Ruta Objetivo: './nueva_carpeta' ] ──( os.mkdir )──> ✅ Directorio Creado
[ Ruta Inexistente: './padre_no_existe/hijo' ] ──( os.mkdir )──> ❌ FileNotFoundError
```

| Situación | Comportamiento / Error | Solución Recomendada |
| :--- | :--- | :--- |
| 📁 La carpeta ya existe | Lanza `FileExistsError`. | Validar la existencia previa con `os.path.exists()`. |
| 🗂️ Directorio padre no existe | Lanza `FileNotFoundError` (no crea rutas anidadas automáticamente). | Usar `os.makedirs()` para estructuras multinivel. |
| 🔒 Sin permisos de escritura | Lanza `PermissionError`. | Verificar permisos del directorio base. |
```

````python
import os
from typing import Any

# 🛰️ [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 MÓDULO OS | Creación de Directorios con os.mkdir()")
print("=" * 80)

# ================================================================================
# 🧪 1. CREACIÓN BÁSICA Y VERIFICACIÓN CON OS.MKDIR()
# ================================================================================
# Intentos de creación de una sola carpeta controlando su existencia previa.
# ================================================================================

print("\n--- 🚀 [PASO 1]: Creación Controlada de Directorio Único ---")

new_directory_path: str = "nuevo_directorio"

try:
    # ⚙️ [1] Comprobar si la carpeta ya existe antes de intentar crearla
    if not os.path.exists(new_directory_path):
        # 🔌 [2] Crear el nuevo directorio
        os.mkdir(new_directory_path)
        print(f"  ✅ Directorio creado exitosamente: '{new_directory_path}'")
    else:
        print(f"  ⚠️ El directorio '{new_directory_path}' ya existe en el sistema.")

    # 🔌 [3] Confirmar que la carpeta existe y es un directorio
    is_created: bool = os.path.isdir(new_directory_path)
    print(f"📋 Verificación de tipo directorio: {is_created}")

finally:
    # Limpieza del directorio creado para mantener idempotencia
    if os.path.exists(new_directory_path):
        os.rmdir(new_directory_path)
        print("  🔒 Directorio de prueba eliminado correctamente.")
````

## 🧱 Control de Excepciones Comunes en `os.mkdir()`

### 💡 Manejo Seguro de FileExistsError y FileNotFoundError:
Intentar crear un directorio que ya existe o cuya ruta intermedia no existe causará una interrupción inmediata en la ejecución si no se captura la excepción adecuada.

````python
# ================================================================================
# 🧪 2. MANEJO DE EXCEPCIONES AL CREAR DIRECTORIOS
# ================================================================================
print("\n--- 🚀 [PASO 2]: Captura de Errores (FileExistsError y FileNotFoundError) ---")

existing_folder: str = "folder_existente_demo"
nested_invalid_path: str = os.path.join("padre_inexistente", "subcarpeta")

# Crear carpeta para forzar FileExistsError
if not os.path.exists(existing_folder):
    os.mkdir(existing_folder)

try:
    # ⚙️ [1] Intento de crear un directorio que ya existe
    print(f"📋 Intento 1: Creando carpeta existente '{existing_folder}'...")
    os.mkdir(existing_folder)

except FileExistsError as error:
    print("  ❌ Excepción capturada: El directorio ya existe.")
    print(f"  🔍 Detalle técnico: {error}")

try:
    # ⚙️ [2] Intento de crear ruta anidada donde el padre no existe
    print(f"\n📋 Intento 2: Creando ruta anidada sin carpeta padre '{nested_invalid_path}'...")
    os.mkdir(nested_invalid_path)

except FileNotFoundError as error:
    print("  ❌ Excepción capturada: La carpeta padre no existe.")
    print(f"  🔍 Detalle técnico: {error}")

finally:
    if os.path.exists(existing_folder):
        os.rmdir(existing_folder)
````

## 🛒 Ejercicios Prácticos Avanzados Integrados

---

### 🧪 Ejercicio 1: Creador Seguro de Estructura de Proyectos

**Escenario Empresarial:** Se necesita una función de inicialización de entornos que intente crear una lista de directorios de salida (`logs`, `exports`, `temp`) de forma idempotente, registrando el resultado individual de cada directorio sin detener el programa si alguno ya existe.

````python
# ================================================================================
# 🧪 3. EJERCICIO PRÁCTICO 1: GENERADOR SEGURA DE DIRECTORIOS
# ================================================================================
print("\n--- 🧪 [PASO 3.1]: Ejercicio 1 - Inicializador Idempotente de Carpetas ---")


def safe_create_directories(directory_list: list[str]) -> dict[str, Any]:
    """Crea múltiples directorios de forma individual y segura registrando su resultado.

    Args:
        directory_list (list[str]): Lista de rutas de carpetas a crear.

    Returns:
        dict[str, Any]: Resumen con carpetas creadas, preexistentes y con error.
    """
    summary: dict[str, Any] = {
        "created": [],
        "already_existed": [],
        "failed": []
    }

    for dir_path in directory_list:
        try:
            # [1] Intentar la creación directa
            os.mkdir(dir_path)
            summary["created"].append(dir_path)
            print(f"  ✅ Creado: '{dir_path}'")

        except FileExistsError:
            summary["already_existed"].append(dir_path)
            print(f"  ⚠️ Ya existía: '{dir_path}'")

        except FileNotFoundError:
            summary["failed"].append({"path": dir_path, "reason": "Ruta padre no encontrada"})
            print(f"  ❌ Falló: '{dir_path}' (Ruta padre no existe)")

        except PermissionError:
            summary["failed"].append({"path": dir_path, "reason": "Permisos insuficientes"})
            print(f"  ❌ Falló: '{dir_path}' (Sin permisos)")

    return summary


# Prueba del ejercicio
folders_to_initialize: list[str] = [
    "output_logs",
    "output_exports",
    "output_logs",  # Duplicado intencional para probar FileExistsError
    os.path.join("non_existing_parent", "child_dir")  # Para probar FileNotFoundError
]

print("📋 Inicializando estructura del proyecto...")
creation_report: dict[str, Any] = safe_create_directories(folders_to_initialize)

print("\n📋 Resumen del Estado de Carpetas:")
print(f"  🔹 Creadas con éxito ({len(creation_report['created'])}): {creation_report['created']}")
print(f"  🔹 Preexistentes ({len(creation_report['already_existed'])}): {creation_report['already_existed']}")
print(f"  🔹 Errores de creación ({len(creation_report['failed'])}): {creation_report['failed']}")

# Limpieza del entorno de prueba
for folder in creation_report["created"]:
    if os.path.exists(folder):
        os.rmdir(folder)

# --------------------------------------------------------------------------------
# 🏁 [PASO 4]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Lección de os.mkdir() finalizada.")
print("=" * 80)
````
