### 💡 El Módulo `datetime` en Python

El módulo estándar **`datetime`** proporciona clases orientadas a objetos para manipular fechas y horas de manera eficiente. Permite realizar operaciones aritméticas, extraer componentes de tiempo, formatear cadenas y calcular intervalos de tiempo.

---

### 🗂️ Principales Clases del Módulo

| Clase | Descripción | Ejemplo de Uso |
| :--- | :--- | :--- |
| 📅 **`date`** | Representa únicamente la fecha (Año, Mes, Día). Ideal para cumpleaños o fechas límite sin hora. | `datetime.date(2026, 8, 14)` |
| ⏰ **`time`** | Representa únicamente la hora (Hora, Minuto, Segundo, Microsegundo). Independiente de la fecha. | `datetime.time(14, 30, 0)` |
| 📆 **`datetime`** | Combina fecha y hora en una sola entidad. Representa un punto exacto en el tiempo. | `datetime.datetime.now()` |
| ⏱️ **`timedelta`** | Representa una duración o diferencia entre dos fechas/horas. Usado para sumas y restas. | `datetime.timedelta(days=7)` |

---

### 📐 Directivas Comunes de Formateo (`strftime` / `strptime`)

| Directiva | Significado | Ejemplo |
| :--- | :--- | :--- |
| `%Y` | Año con 4 dígitos | `2026` |
| `%m` | Mes con 2 dígitos rellenos con cero | `08` |
| `%d` | Día del mes con 2 dígitos | `14` |
| `%H` | Hora en formato 24 horas (00-23) | `15` |
| `%M` | Minutos con 2 dígitos (00-59) | `45` |
| `%S` | Segundos con 2 dígitos (00-59) | `09` |
| `%A` / `%B` | Nombre completo del día / mes (según *locale*) | `Friday` / `August` |

```python
from datetime import date, datetime, time, timedelta
from typing import Any

# 🛰️ [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 MÓDULO DATETIME | Fechas, Horas y Aritmética Temporal")
print("=" * 80)

# ================================================================================
# 🧪 1. CREACIÓN Y LECTURA DE FECHAS Y HORAS (DATETIME, DATE, TIME)
# ================================================================================
# Creación de instancias específicas y captura del tiempo actual del sistema.
# ================================================================================

print("\n--- 🚀 [PASO 1]: Creación de Objetos datetime, date y time ---")

# 🔌 [1] Creación de un objeto datetime específico (Año, Mes, Día, Hora, Minuto, Segundo)
specific_datetime: datetime = datetime(1495, 2, 14, 12, 30, 0)
print(f"📋 Fecha y hora histórica específica: {specific_datetime}")

# 🔌 [2] Obtención de la fecha y hora actual del sistema
current_datetime: datetime = datetime.now()
print(f"📋 Fecha y hora actual (now()):      {current_datetime}")

print("-" * 65)

# 🔌 [3] Trabajo exclusivo con fechas (date)
current_date: date = date.today()
specific_date: date = date(1495, 12, 31)

print(f"  🔹 Fecha actual (today()):          {current_date}")
print(f"  🔹 Fecha específica (date):         {specific_date}")

print("-" * 65)

# 🔌 [4] Trabajo exclusivo con horas (time)
current_time: time = current_datetime.time()
specific_time: time = time(23, 14, 56)

print(f"  🔹 Hora actual extraída (time()):   {current_time}")
print(f"  🔹 Hora específica definida (time): {specific_time}")
```

## 🧱 Formateo (`strftime`) y Parsing de Fechas (`strptime`)

### 💡 Conversión entre Cadenas y Objetos Datetime:
- **`strftime(format)` (*String Format Time*):** Convierte un objeto `datetime` a una cadena de texto formateada.
- **`strptime(date_string, format)` (*String Parse Time*):** Convierte una cadena de texto a un objeto `datetime` validando la estructura.

```python
# ================================================================================
# 🧪 2. FORMATEO (STRFTIME) Y PARSING (STRPTIME) DE CADENAS DE FECHA
# ================================================================================
print("\n--- 🚀 [PASO 2]: Formateo y Conversión de Fechas ---")

# ⚙️ [1] Formateo de objeto datetime a texto personalizado (strftime)
formatted_datetime_str: str = current_datetime.strftime("%Y-%m-%d %H:%M:%S")
formatted_date_str: str = current_datetime.strftime("%d/%m/%Y")
formatted_time_str: str = current_datetime.strftime("%H:%M:%S")

print(f"📋 Datetime formateado (%Y-%m-%d %H:%M:%S): {formatted_datetime_str}")
print(f"  🔹 Solo fecha formateada (%d/%m/%Y):      {formatted_date_str}")
print(f"  🔹 Solo hora formateada (%H:%M:%S):        {formatted_time_str}")

print("-" * 65)

# ⚙️ [2] Parseo de texto a objeto datetime (strptime)
raw_log_timestamp: str = "2026-08-14 18:45:30"
parsed_datetime: datetime = datetime.strptime(raw_log_timestamp, "%Y-%m-%d %H:%M:%S")

print(f"📋 Cadena de texto original: '{raw_log_timestamp}' (Tipo: {type(raw_log_timestamp).__name__})")
print(f"  ✅ Objeto datetime parseado:  {parsed_datetime} (Tipo: {type(parsed_datetime).__name__})")
print(f"  🔍 Componente Año extraído:  {parsed_datetime.year}")
```

## ⏱️ Operaciones Aritméticas de Tiempo con `timedelta`

### 💡 Suma y Resta de Intervalos Temporales:
El objeto `timedelta` permite sumar o restar días, segundos, microsegundos, milisegundos, minutos, horas o semanas a objetos `date` o `datetime`.

```python
# ================================================================================
# 📊 3. ARITMÉTICA TEMPORAL MEDIANTE TIMEDELTA
# ================================================================================
print("\n--- 📊 [PASO 3]: Cálculo de Intervalos Temporales con timedelta ---")

today: date = date.today()

# ⚙️ [1] Resta de un intervalo (7 días atrás)
one_week_ago: date = today - timedelta(days=7)

# ⚙️ [2] Suma de un intervalo (90 días futuros)
ninety_days_ahead: date = today + timedelta(days=90)

# ⚙️ [3] Resta entre dos objetos datetime para obtener una diferencia
event_start: datetime = datetime(2026, 8, 14, 9, 0, 0)
event_end: datetime = datetime(2026, 8, 14, 17, 30, 0)
duration: timedelta = event_end - event_start

print(f"📋 Fecha actual:                 {today}")
print(f"  🔹 Fecha de hace una semana:   {one_week_ago}")
print(f"  🔹 Fecha dentro de 90 días:    {ninety_days_ahead}")
print("-" * 65)
print(f"⏱️ Duración total del evento:    {duration} ({duration.total_seconds() / 3600:.2f} horas)")
```

### 🧪 Ejercicio 1: Analizador de Logs y Auditoría de Latencia del Servidor

**Escenario Empresarial:** Se requiere procesar una lista de registros de logs en formato de texto plano para identificar peticiones que superaron el umbral de latencia crítico de 2.5 segundos e identificar el tiempo total consumido por la transacción.

```python
# ================================================================================
# 🧪 4. EJERCICIO PRÁCTICO 1: AUDITORÍA DE AUDITORÍA DE LOGS Y LATENCIA
# ================================================================================
print("\n--- 🧪 [PASO 4.1]: Ejercicio 1 - Procesamiento y Auditoría de Logs ---")


def audit_server_logs(logs_data: list[dict[str, str]], latency_threshold_seconds: float) -> list[dict[str, Any]]:
    """Analiza registros de peticiones de red calculando la duración e identificando latencias críticas.

    Args:
        logs_data (list[dict]): Lista de logs con marcas de inicio y fin.
        latency_threshold_seconds (float): Umbral de alerta de latencia en segundos.

    Returns:
        list[dict]: Resumen auditado con objetos datetime y estado de alerta.
    """
    audit_summary: list[dict[str, Any]] = []

    for entry in logs_data:
        # Parseo de cadenas de texto a objetos datetime con microsegundos
        time_format: str = "%Y-%m-%d %H:%M:%S.%f"
        start_time: datetime = datetime.strptime(entry["start_time"], time_format)
        end_time: datetime = datetime.strptime(entry["end_time"], time_format)

        # Cálculo de diferencia de tiempo en segundos
        elapsed_timedelta: timedelta = end_time - start_time
        elapsed_seconds: float = elapsed_timedelta.total_seconds()

        is_critical: bool = elapsed_seconds > latency_threshold_seconds

        audit_summary.append({
            "request_id": entry["request_id"],
            "start_time": start_time,
            "elapsed_seconds": round(elapsed_seconds, 3),
            "is_critical": is_critical
        })

    return audit_summary


# Datos simulados de red
raw_logs: list[dict[str, str]] = [
    {"request_id": "REQ-001", "start_time": "2026-08-14 10:00:00.000000", "end_time": "2026-08-14 10:00:01.250000"},
    {"request_id": "REQ-002", "start_time": "2026-08-14 10:00:05.100000", "end_time": "2026-08-14 10:00:08.900000"},
    {"request_id": "REQ-003", "start_time": "2026-08-14 10:00:10.000000", "end_time": "2026-08-14 10:00:10.450000"}
]

audited_results: list[dict[str, Any]] = audit_server_logs(raw_logs, latency_threshold_seconds=2.5)

print("📋 Resultado de la Auditoría de Latencia:")
for record in audited_results:
    status_icon: str = "❌ [CRÍTICO]" if record["is_critical"] else "✅ [OK]"
    print(f"  {status_icon} Petición: {record['request_id']} | Duración: {record['elapsed_seconds']}s")
```

### 🧪 Ejercicio 2: Gestor Empresarial de SLA y Vencimiento de Contratos

**Escenario Empresarial:** Una plataforma SaaS debe calcular la fecha límite de resolución de un ticket de soporte basado en un Acuerdo de Nivel de Servicio (*SLA*). Si la fecha estimada cae en fin de semana (sábado o domingo), la fecha límite debe ajustarse automáticamente al siguiente lunes hábil a las 09:00 AM.

```python
# ================================================================================
# 🧪 5. EJERCICIO PRÁCTICO 2: CÁLCULO DE SLA DÍAS HÁBILES Y FINES DE SEMANA
# ================================================================================
print("\n--- 🧪 [PASO 4.2]: Ejercicio 2 - Calculadora de Vencimiento de SLA Business Days ---")


def calculate_sla_expiration(creation_dt: datetime, sla_hours: int) -> datetime:
    """Calcula la fecha de vencimiento de un ticket ajustando a días laborables si cae en fin de semana.

    Args:
        creation_dt (datetime): Momento exacto de apertura del ticket.
        sla_hours (int): Horas concedidas por el SLA de soporte.

    Returns:
        datetime: Fecha y hora límite ajustada.
    """
    # Suma de horas al momento inicial
    expiration_dt: datetime = creation_dt + timedelta(hours=sla_hours)

    # Verificación del día de la semana (0=Lunes, 5=Sábado, 6=Domingo)
    weekday_code: int = expiration_dt.weekday()

    if weekday_code == 5:  # Sábado -> Sumar 2 días para mover al Lunes
        expiration_dt = expiration_dt + timedelta(days=2)
        expiration_dt = expiration_dt.replace(hour=9, minute=0, second=0)
        print("  ⚠️ El vencimiento calculaba para un Sábado. Ajustado al Lunes a las 09:00 AM.")
    elif weekday_code == 6:  # Domingo -> Sumar 1 día para mover al Lunes
        expiration_dt = expiration_dt + timedelta(days=1)
        expiration_dt = expiration_dt.replace(hour=9, minute=0, second=0)
        print("  ⚠️ El vencimiento calculaba para un Domingo. Ajustado al Lunes a las 09:00 AM.")

    return expiration_dt


# Test Case: Ticket abierto un Viernes por la tarde con SLA de 24 horas
ticket_created: datetime = datetime(2026, 8, 14, 16, 0, 0)  # Viernes 14 de Agosto, 2026
ticket_sla_hours: int = 24

print(f"📋 Ticket creado el:  {ticket_created.strftime('%A, %Y-%m-%d %H:%M:%S')}")
sla_deadline: datetime = calculate_sla_expiration(ticket_created, ticket_sla_hours)
print(f"  ✅ Fecha Límite SLA: {sla_deadline.strftime('%A, %Y-%m-%d %H:%M:%S')}")

# --------------------------------------------------------------------------------
# 🏁 [PASO 6]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Cuaderno del Módulo Datetime finalizado con éxito.")
print("=" * 80)
```
