| 🛠️ **Conceptos** | `Transformation Methods`, `Sanitization (strip)`, `String Search (find vs index)`, `String Manipulation (replace, join, split)` |

### 💡 Explicación de la Arquitectura
* **Sintaxis General:** `string.metodo(argumentos)`
* **Transformación de Caso:** `lower()`, `upper()`, `title()`, `capitalize()`.
* **Limpieza y Reemplazo:** `strip()` (elimina espacios residuales), `replace(old, new)` (sustituye coincidencias).
* **Búsqueda y Pertenencia:**
  * `find(substring)`: Retorna la posición de inicio o **`-1`** si no encuentra la subcadena.
  * `index(substring)`: Retorna la posición de inicio pero arroja **`ValueError`** si no existe la subcadena.
* **Unión y División:**
  * `join(iterable)`: Une elementos de una lista/iterable mediante un delimitador.
  * `split(sep)`: Divide una cadena en una lista de subcadenas usando un delimitador (por defecto, cualquier espacio en blanco).

```python
# [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Métodos Nativos para Cadenas de Texto")
print("=" * 80)

# ================================================================================
# 🔠 1. MÉTODOS DE TRANSFORMACIÓN DE CASO (lower, upper, title, capitalize)
# ================================================================================
# Modificación de mayúsculas/minúsculas sobre instancias de str.
# ================================================================================

# [1] Variable de entrada base
raw_text: str = "hello python"

print("\n--- 🚀 [PASO 1]: Transformaciones de Formato y Capitalización ---")

# Aplicación de métodos de formato
lowercase_text: str = raw_text.lower()
uppercase_text: str = raw_text.upper()
titlecase_text: str = raw_text.title()
capitalized_text: str = raw_text.capitalize()

print(f"📋 Cadena de origen:      '{raw_text}'")
print("-" * 50)
print(f"✅ [1. lower()]:          '{lowercase_text}'")
print(f"✅ [2. upper()]:          '{uppercase_text}'")
print(f"✅ [3. title()]:          '{titlecase_text}'")
print(f"✅ [4. capitalize()]:     '{capitalized_text}'")
```

## 🧹 Limpieza (`strip`), Reemplazo (`replace`), Unión (`join`) y División (`split`)

### 💡 Explicación de los Casos de Uso:
1. **Limpieza (`strip()`):** Elimina espacios en blanco innecesarios al inicio y final del texto.
2. **Reemplazo (`replace()`):** Busca todas las apariciones de una subcadena y las sustituye por un nuevo valor.
3. **Concatenación Eficiente (`join()`):** Une listas de cadenas intercalando un separador (`" "`, `","`, `"-"`).
4. **División de Cadenas (`split()`):** Operación inversa a `join()`. Divide una cadena en una **lista de cadenas (`list[str]`)** usando un separador definido o espacios por defecto.

```python
# ================================================================================
# 🧹 2. LIMPIEZA, REEMPLAZO, UNIÓN Y DIVISIÓN (strip, replace, join, split)
# ================================================================================
print("\n--- 🧹 [PASO 2]: Limpieza, Reemplazo, Reconstrucción y División ---")

# [1] Caso de texto con espacios sobrantes
uncleaned_text: str = "   Hello Python   "
stripped_text: str = uncleaned_text.strip()

print(f"📋 Texto sin procesar:   '{uncleaned_text}'")
print(f"✅ [5. strip()]:         '{stripped_text}'")

# [2] Reemplazo de contenido dentro del texto
replacement_text: str = stripped_text.replace("Python", "World")
print(f"✅ [6. replace()]:       '{replacement_text}'")

# [3] Unión de elementos de una lista mediante delimitador
word_list: list[str] = ["Hello", "Python"]
joined_text: str = " ".join(word_list)
print(f"✅ [9. join()]:          '{joined_text}' (Unido con espacio)")

# [4] División de una cadena en lista (split)
csv_data: str = "Python,Data Science,Machine Learning"
split_words_default: list[str] = stripped_text.split()          # Por espacio por defecto
split_words_custom: list[str] = csv_data.split(",")             # Por coma explícita

print(f"✅ [10. split() por defecto]: '{stripped_text}' -> {split_words_default}")
print(f"✅ [10. split() delimitado]: '{csv_data}' -> {split_words_custom}")
```

## 🔍 Métodos de Búsqueda de Subcadenas (`find` vs `index`)

### 💡 Comparativa entre `find()` y `index()`:
* **`find(substring)`:** Es seguro de usar cuando no estamos seguros de si existe la subcadena, ya que si falla retorna **`-1`** sin detener el programa.
* **`index(substring)`:** Lanza una excepción de tipo **`ValueError`** si la subcadena no existe. Se suele usar dentro de bloques `try/except`.

```python
# ================================================================================
# 🔍 3. BÚSQUEDA Y LOCALIZACIÓN DE SUBCADENAS (find, index)
# ================================================================================
print("\n--- 🔍 [PASO 3]: Búsqueda de Posiciones e Índices ---")

sample_sentence: str = "Hello Python"
target_substring: str = "Python"
missing_substring: str = "Java"

# [1] Búsqueda con find()
found_index: int = sample_sentence.find(target_substring)
not_found_index: int = sample_sentence.find(missing_substring)

print(f"✅ [7. find() - Éxito]:     '{target_substring}' encontrado en índice {found_index}")
print(f"✅ [7. find() - Ausente]:   '{missing_substring}' no encontrado, retorna {not_found_index}")

# [2] Búsqueda con index() e integración con try/except
print("\n--- ⚙️ [Validación segura con index()]: ---")
try:
    matched_index: int = sample_sentence.index(target_substring)
    print(f"✅ [8. index() - Éxito]:    '{target_substring}' encontrado en índice {matched_index}")
    
    # Intentamos buscar una subcadena inexistente para evaluar la excepción
    sample_sentence.index(missing_substring)
except ValueError:
    print(f"❌ [8. index() - Excepción]: La subcadena '{missing_substring}' no fue encontrada.")

# --------------------------------------------------------------------------------
# 🏁 [PASO 4]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Módulo de métodos de cadenas finalizado con éxito.")
print("=" * 80)
```
