# ⚙️ Fundamentos de Python: Indexación y Slicing de Cadenas (`String Indexing & Slicing`)

| Propiedad | Detalle |
| :--- | :--- |
| 📍 **Módulo** | Fundamentos de Python |
| 🎯 **Objetivo** | Comprender la navegación por posiciones numéricas (índices positivos/negativos) y la extracción de subcadenas mediante la sintaxis de rangos (*slicing*: `[start:stop:step]`). |
| 🛠️ **Conceptos** | `Positive Indexing`, `Negative Indexing`, `Slicing [start:stop]`, `Step Parameter`, `String Reversal ([::-1])` |

---

### 💡 Explicación de la Arquitectura
* **Indexación Positiva (`0` a `N-1`):** Recorre el texto de izquierda a derecha comenzando en la posición `0`.
* **Indexación Negativa (`-1` a `-N`):** Recorre el texto de derecha a izquierda; el índice `-1` referencia siempre al último carácter.
* **Slicing (`[start:stop:step]`):**
  * **`start`:** Índice de inicio (**inclusive**).
  * **`stop`:** Índice de parada (**exclusivo**, no incluye este índice).
  * **`step`:** Tamaño del salto o paso (un paso negativo invierte el sentido).

```python
# [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Indexación Positiva y Negativa en Cadenas")
print("=" * 80)

# ================================================================================
# 📍 1. INDEXACIÓN BÁSICA (POSITIVA Y NEGATIVA)
# ================================================================================
# Acceso a caracteres individuales mediante corchetes [].
# ================================================================================

# [1] Cadenas de texto iniciales en inglés
sample_phrase: str = "Me encanta la programación"
target_word: str = "Python"

print("\n--- 🚀 [PASO 1]: Acceso Directo por Índice Positivo y Negativo ---")

# Acceso directo en frase corta
first_character: str = sample_phrase[0]
print(f"✅ Primer carácter de '{sample_phrase}': '{first_character}'")

print("-" * 50)
print(f"📋 Analizando la palabra clave: '{target_word}' (Longitud: {len(target_word)} caracteres)")

# Índices Positivos (0 -> 5)
first_char_positive: str = target_word[0]
last_char_positive: str = target_word[5]

# Índices Negativos (-1 -> -6)
last_char_negative: str = target_word[-1]
first_char_negative: str = target_word[-6]

print(f"✅ Indexación Positiva [0]:  '{first_char_positive}' | [5]:  '{last_char_positive}'")
print(f"✅ Indexación Negativa [-1]: '{last_char_negative}' | [-6]: '{first_char_negative}'")
```

## 📐 Extracción de Subcadenas (*Slicing*: `[start:stop:step]`)

### 💡 Explicación de las Reglas de Slicing:
1. **Rango Acotado (`[0:4]`):** Extrae desde el índice `0` hasta el `3` (el índice de parada `4` queda fuera).
2. **Omitir Límites (`[:3]` / `[3:]`):** Si omitimos `start`, asume el inicio (`0`). Si omitimos `stop`, lee hasta el final.
3. **Paso o Salto (`step`):** Define el incremento. Un salto de `2` toma caracteres salteados; un salto de `-1` recorre la cadena de derecha a izquierda, permitiendo **invertir el texto**.

```python
# ================================================================================
# ✂️ 2. RANGOS Y TROCEADO DE CADENAS (SLICING)
# ================================================================================
print("\n--- ✂️ [PASO 2]: Rangos de Indexación (Slicing) ---")

python_text: str = "Python"

# Extraer un rango de caracteres (start:0, stop:4) -> subcadena 'Pyth'
substring_range: str = python_text[0:4]

# Omitir inicio y final
prefix_substring: str = python_text[:3]  # De 0 a 2 -> 'Pyt'
suffix_substring: str = python_text[3:]  # De 3 al final -> 'hon'

print(f"✅ Subcadena [0:4]: '{substring_range}'")
print(f"✅ Omisión inicio [:3]: '{prefix_substring}'")
print(f"✅ Omisión final [3:]:  '{suffix_substring}'")

# ================================================================================
# 🔄 3. USO DEL PARÁMETRO STEP E INVERSIÓN
# ================================================================================
print("\n--- 🔄 [PASO 3]: Saltos (Step) e Inversión de Texto ---")

# Salto de 2 en 2
stepped_text: str = python_text[0:6:2]

# Inversión completa usando step negativo (-1)
reversed_word: str = python_text[::-1]

print(f"✅ Salto de 2 en 2 [0:6:2]: '{stepped_text}'")
print(f"✅ Texto invertido [::-1]:   '{reversed_word}'")

# --------------------------------------------------------------------------------
# 🏁 [PASO 4]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Módulo de indexación y slicing finalizado con éxito.")
print("=" * 80)
```
