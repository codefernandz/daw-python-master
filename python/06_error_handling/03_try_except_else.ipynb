### 💡 Explicación de la Arquitectura

La cláusula `else` en una estructura de manejo de excepciones se ejecuta **única y exclusivamente** si el bloque `try` finalizó con éxito sin lanzar ninguna excepción.

#### 📐 Principios de Diseño (*Best Practices*):
1. **Reducción del Alcance del `try`:** Solo debes colocar dentro del bloque `try` el código estrictamente propenso a fallar (p. ej., abrir un archivo o realizar la división).
2. **Ubicación de la Lógica de Éxito (`else`):** El código posterior a la operación riesgosa que depende de su éxito (p. ej., cerrar el archivo o procesar el resultado) se traslada al bloque `else`. Esto evita capturar accidentalmente excepciones no deseadas que ocurran en el procesamiento posterior.

```python
try:
    # Código propenso a excepciones (mínimo indispensable)
    ...
except SpecificException:
    # Gestión de errores
    ...
else:
    # Se ejecuta SOLO si el bloque 'try' no lanzó excepciones
    ...

````python
from typing import Any

# 🛰️ [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Manejo de Errores - Cláusula else")
print("=" * 80)

# ================================================================================
# 🧪 1. MANEJO DE ARCHIVOS CON TRY-EXCEPT-ELSE
# ================================================================================
# Intento de apertura de un archivo inexistente para evaluar el salto al except.
# ================================================================================

print("\n--- 🚀 [PASO 1]: Operación de Lectura de Archivo ---")

# 📋 Ruta de archivo objetivo con Type Hints
file_path: str = "non_existent_file.txt"

print(f"📋 Intentando acceder al archivo: '{file_path}'")
print("-" * 50)

# 🔌 [1] Apertura y lectura segura con try-except-else
try:
    # ⚙️ Operación propensa a FileNotFoundError
    file_stream = open(file_path, "r")
    file_content: str = file_stream.read()
except FileNotFoundError:
    print("❌ [ERROR DETECTADO]: FileNotFoundError ➔ El archivo especificado no existe.")
else:
    # ⚙️ Este bloque solo se ejecuta si el archivo abrió correctamente
    print("✅ [ÉXITO]: El archivo se abrió y se leyó correctamente.")
    print(f"📄 Contenido: {file_content}")
    file_stream.close()
    print("🔒 Archivo cerrado de forma segura.")
````

## 🧮 Separación de Lógica en Operaciones Aritméticas e Interacción

### 💡 Buenas Prácticas en el Uso de `else`:
Al solicitar datos al usuario para realizar cálculos, el cálculo o conversión se efectúa en el `try`, mientras que la presentación y formateo final del resultado se mueven al bloque `else`.

````python
# ================================================================================
# 📊 2. PROCESAMIENTO NUMÉRICO Y LÓGICA DE ÉXITO AISLADA
# ================================================================================
print("\n--- 📊 [PASO 2]: División de Constante por Entrada de Usuario ---")

# 📋 Constante para la operación
DIVIDEND_BASE: float = 100.0

print(f"📋 Operación: Divisió de {DIVIDEND_BASE} entre un valor del usuario")
print("-" * 50)

# 🔌 [1] Captura de excepciones con bloque else explicativo
try:
    # 📥 Lectura e intepretación numérica
    user_input: str = input("Introduce un número entero divisor: ")
    parsed_number: int = int(user_input)
    
    # ⚙️ Operación aritmética
    division_result: float = DIVIDEND_BASE / parsed_number

except ValueError:
    print("❌ [ERROR DETECTADO]: ValueError ➔ Debes introducir un número entero válido.")

except ZeroDivisionError:
    print("❌ [ERROR DETECTADO]: ZeroDivisionError ➔ No se puede dividir entre cero.")

else:
    # ⚙️ Lógica posterior a la operación exitosa
    print("✅ [ÉXITO]: Operación realizada sin excepciones.")
    print(f"📊 Resultado exacto: {division_result}")
    print(f"🔢 Resultado redondeado (2 decimales): {round(division_result, 2)}")

# --------------------------------------------------------------------------------
# 🏁 [PASO 3]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Módulo de la cláusula 'else' finalizado con éxito.")
print("=" * 80)
````
