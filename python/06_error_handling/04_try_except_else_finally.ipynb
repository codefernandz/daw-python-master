| Propiedad | Detalle |
| :--- | :--- |
| 📍 **Módulo** | Fundamentos de Python: Manejo de Errores |
| 🎯 **Objetivo** | Implementar la estructura completa de gestión defensiva de errores (`try-except-else-finally`) para garantizar la liberación de recursos (*cleanup*) independientemente de la ocurrencia de fallos. |
| 🛠️ **Conceptos** | `Try-Except-Else-Finally`, `Resource Cleanup`, `Guaranteed Execution`, `Catch-all Exception`, `Safe Resource Closing` |

---

### 💡 Explicación de la Arquitectura

La cláusula `finally` representa el bloque de **ejecución garantizada**. Se ejecuta **siempre**, sin importar si el bloque `try` finalizó exitosamente, si ocurrió una excepción capturada, o si se lanzó una excepción no gestionada.

#### 📊 Comparativa de Flujos (`else` vs `finally`):

| Cláusula | Condición de Ejecución | Caso de Uso Principal |
| :--- | :--- | :--- |
| `else` | **Únicamente** si el bloque `try` **NO** lanzó excepciones. | Procesamiento posterior de resultados / lectura de datos exitosa. |
| `finally` | **SIEMPRE**, exista o no una excepción previa. | Liberación de recursos (cerrar archivos, desconectar BD, limpiar sockets). |

#### 📐 Sintaxis Completa:

```python
try:
    # Código propenso a lanzar excepciones
    ...
except SpecificException:
    # Manejo de error específico
    ...
except Exception as err:
    # Captura genérica (Catch-all) para excepciones inesperadas
    ...
else:
    # Ejecución si NO hubo errores
    ...
finally:
    # Ejecución garantizada (Limpieza y liberación de recursos)
    ...

````python
from typing import Any, Optional

# 🛰️ [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Manejo de Errores - Estructura try-except-else-finally")
print("=" * 80)

# ================================================================================
# 🧪 1. MANEJO SEGURO DE RECURSOS (ARCHIVOS) CON FINALLY
# ================================================================================
# Intento de apertura de archivo con garantía de cierre e inspección de variables.
# ================================================================================

print("\n--- 🚀 [PASO 1]: Gestión de Archivos y Liberación Garantizada ---")

# 📋 Variables de control con Type Hints
target_filename: str = "non_existent_file.txt"
file_stream: Optional[Any] = None

print(f"📋 Intentando acceder al archivo: '{target_filename}'")
print("-" * 50)

# 🔌 [1] Estructura completa de manejo defensivo
try:
    # ⚙️ Operación propensa a FileNotFoundError
    file_stream = open(target_filename, "r")
    file_content: str = file_stream.read()

except FileNotFoundError:
    print("❌ [ERROR DETECTADO]: FileNotFoundError ➔ El archivo no se encontró.")

except Exception as unexpected_error:
    print(f"❌ [ERROR INESPERADO]: Ocurrió un error no contemplado: {unexpected_error}")

else:
    print("✅ [ÉXITO]: El archivo se abrió correctamente.")
    print(f"📄 Contenido: {file_content}")

finally:
    print("🧹 [CLEANUP]: El bloque finally se ejecuta siempre (con o sin excepciones).")
    
    # ⚙️ Verificación segura de liberación de recursos
    if file_stream is not None and not file_stream.closed:
        file_stream.close()
        print("🔒 [RECURSO]: Archivo cerrado exitosamente.")
    else:
        print("ℹ️ [RECURSO]: No hay stream de archivo abierto para cerrar.")
````

## 🧮 Entrada de Usuario y Aritmética Controlada

### 💡 Garantía de Cierre de Bloque en Aplicaciones Interactivas:
El bloque `finally` asegura que, independientemente de si la entrada del usuario es válida (`else`) o genera un error de conversión/división (`except`), la aplicación registra la trazabilidad del intento de operación antes de continuar el flujo.

````python
# ================================================================================
# 📊 2. PROCESAMIENTO INTERACTIVO CON TRAZABILIDAD GARANTIZADA
# ================================================================================
print("\n--- 📊 [PASO 2]: División Interactiva con try-except-else-finally ---")

# 📋 Constante para la operación
NUMERATOR_BASE: float = 10.0

print(f"📋 Operación: División de {NUMERATOR_BASE} entre entrada de usuario")
print("-" * 50)

# 🔌 [1] Captura selectiva con else y finally
try:
    user_input: str = input("Introduce un número entero divisor: ")
    parsed_number: int = int(user_input)
    division_result: float = NUMERATOR_BASE / parsed_number

except ValueError:
    print("❌ [ERROR DETECTADO]: ValueError ➔ Debes introducir un número entero válido.")

except ZeroDivisionError:
    print("❌ [ERROR DETECTADO]: ZeroDivisionError ➔ No se puede dividir entre cero.")

else:
    print(f"✅ [ÉXITO]: El resultado de la división es: {division_result:.2f}")

finally:
    print("🧹 [TRAZABILIDAD]: Finalizó el procesamiento de la entrada interactiva.")
````

## ⚠️ Evaluación Directa de Fallo por División Cero y Jerarquía de Captura

````python
# ================================================================================
# 🧪 3. DEMOSTRACIÓN DE CAPTURA ESPECÍFICA Y CATCH-ALL
# ================================================================================
print("\n--- 🧪 [PASO 3]: Simulación Directa de ZeroDivisionError ---")

# 📋 Declaración de variables para fallo forzado
forced_numerator: int = 10
forced_denominator: int = 0

print(f"📋 Evaluando: {forced_numerator} / {forced_denominator}")
print("-" * 50)

# 🔌 [1] Captura jerárquica con cláusula genérica
try:
    forced_result: float = forced_numerator / forced_denominator

except ZeroDivisionError:
    print("❌ [ERROR ESPECÍFICO]: ZeroDivisionError ➔ No se puede dividir entre cero.")

except TypeError:
    print("❌ [ERROR ESPECÍFICO]: TypeError ➔ Tipo de dato incorrecto para la operación.")

except Exception as generic_error:
    print(f"❌ [ERROR GENÉRICO]: Excepción no contemplada capturada: {generic_error}")

else:
    print(f"✅ [ÉXITO]: Resultado: {forced_result}")

finally:
    print("🧹 [CLEANUP]: Finalizó el bloque de prueba directa de errores.")

# --------------------------------------------------------------------------------
# 🏁 [PASO 4]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Módulo de la estructura try-except-else-finally finalizado con éxito.")
print("=" * 80)
````

## 📚 Resumen del Módulo: Manejo de Errores y Programación Defensiva

| Estructura | Propósito |
| :--- | :--- |
| `try ... except` | Intercepta excepciones específicas sin interrumpir el programa (*graceful degradation*). |
| `try ... except ... else` | Aisla la lógica que **solo debe ejecutarse si NO ocurrieron errores** en el `try`. |
| `try ... except ... else ... finally` | Garantiza la ejecución de código de limpieza (*resource cleanup*) exista o no una excepción. |
