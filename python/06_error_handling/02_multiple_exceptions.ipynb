# ⚙️ Fundamentos de Python: Captura Múltiple de Excepciones (`try-except-except`)

| Propiedad | Detalle |
| :--- | :--- |
| 📍 **Módulo** | Fundamentos de Python: Manejo de Errores |
| 🎯 **Objetivo** | Capturar y gestionar múltiples tipos de errores en tiempo de ejecución dentro de un mismo bloque de prueba mediante cláusulas `except` específicas. |
| 🛠️ **Conceptos** | `Multiple Exception Clauses`, `ValueError`, `ZeroDivisionError`, `Selective Error Handling`, `Input Casting Protection` |

---

### 💡 Explicación de la Arquitectura

En programas interactivos o con entrada de datos del usuario, una misma sección de código puede fallar por **múltiples razones distintas**:

1. **Fallo de Conversión (`ValueError`):** Ocurre si la entrada recibida (cadena de texto) no se puede castear/convertir a un número entero mediante `int()`.
2. **Fallo de Operación Arithmetic (`ZeroDivisionError`):** Ocurre si el divisor ingresado es exactamente `0`.

#### 📐 Sintaxis de Múltiples Cláusulas `except`:

```python
try:
    # Bloque de código con múltiples puntos potenciales de fallo
    ...
except ValueError:
    # Gestión específica para errores de tipo/valor de entrada
    ...
except ZeroDivisionError:
    # Gestión específica para operaciones matemáticas inválidas
    ...

````python
from typing import Any

# 🛰️ [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Manejo de Errores - Captura Múltiple")
print("=" * 80)

# ================================================================================
# 🧪 1. CAPTURA SELECTIVA CON MÚLTIPLES BLOQUES EXCEPT
# ================================================================================
# Evaluación de lectura de entrada, conversión int() y división entre 10.
# ================================================================================

print("\n--- 🚀 [PASO 1]: Procesamiento Defensivo con Múltiples Excepciones ---")

# 📋 Constante para la operación
DIVIDEND: int = 10

print("📋 [Ejemplo]: Captura selectiva de ValueError y ZeroDivisionError")
print("-" * 50)

# 🔌 [1] Estructura try con múltiples cláusulas except
try:
    # 📥 [1.1] Lectura de entrada (Simulada o interactiva)
    divisor_input: str = input("Introduce un número entero para dividir 10 entre él: ")
    
    # ⚙️ [1.2] Conversión de tipo (Punto potencial 1: ValueError)
    parsed_divisor: int = int(divisor_input)
    
    # ⚙️ [1.3] Operación matemática (Punto potencial 2: ZeroDivisionError)
    division_result: float = DIVIDEND / parsed_divisor
    
    print(f"✅ [ÉXITO]: El resultado de dividir {DIVIDEND} entre {parsed_divisor} es {division_result:.2f}")

except ValueError:
    print("❌ [ERROR DETECTADO]: ValueError ➔ Por favor, introduce un número entero válido.")

except ZeroDivisionError:
    print("❌ [ERROR DETECTADO]: ZeroDivisionError ➔ No es posible dividir un número entre cero.")

# --------------------------------------------------------------------------------
# 🏁 [PASO 2]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Módulo de captura múltiple de excepciones finalizado con éxito.")
print("=" * 80)
````
