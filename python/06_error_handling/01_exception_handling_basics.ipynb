### 💡 Explicación de la Arquitectura

La **programación defensiva** busca anticipar fallos durante la ejecución del programa para evitar interrupciones abruptas (*crashes*). En Python, las anomalías se representan como objetos de tipo **excepción** (*Exception*).

#### 📚 Principales Excepciones Incorporadas (*Built-in Exceptions*):

| Excepción | Descripción |
| :--- | :--- |
| `TypeError` | Se aplica una operación a un objeto de tipo incompatible. |
| `ValueError` | Un argumento recibe el tipo correcto pero un valor inapropiado. |
| `IndexError` | Acceso a un índice fuera del rango de una secuencia (`list`, `tuple`). |
| `KeyError` | Intento de acceder a una clave inexistente en un `dict`. |
| `AttributeError` | Intento de acceder a un atributo o método inexistente en un objeto. |
| `OSError` / `IOError` | Error en operaciones de entrada/salida (archivos, red). |
| `ZeroDivisionError` | Intento de realizar una división o módulo por cero ($x / 0$). |
| `ImportError` | Fallo al intentar importar un módulo o paquete. |
| `KeyboardInterrupt` | Interrupción manual de la ejecución por el usuario (`Ctrl + C`). |

#### 📐 Sintaxis Básica de Captura:
```python
try:
    # Código propenso a generar excepciones
    ...
except SpecificException:
    # Código de recuperación/gestión del error
    ...

````python
from typing import Any

# 🛰️ [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Manejo de Errores - Introducción a try-except")
print("=" * 80)

# ================================================================================
# 🧪 1. SIMULACIÓN DE ERROR SIN CONTROL DE EXCEPCIONES
# ================================================================================
# Demostración del comportamiento de una división por cero no interceptada.
# ================================================================================

print("\n--- 🚀 [PASO 1]: Escenario vulnerable a ZeroDivisionError ---")

# 📋 Declaración de operandos con Type Hints
numerator: int = 3
denominator: int = 0

print(f"📋 Numerador: {numerator} | Denominador: {denominator}")
print("-" * 50)

# ⚠️ La siguiente línea detendría la ejecución si no estuviera comentada:
# division_result: float = numerator / denominator  # ❌ Arroja ZeroDivisionError
print("⚠️ [ADVERTENCIA]: Operación '3 / 0' omitida sin bloque try para evitar la caída del programa.")
````

## 🛡️ Intercepción y Gestión Defensiva con `try...except`

### 💡 Mecánica de Captura:
Al envolver la operación riesgosa dentro de un bloque `try`, si se produce una excepción del tipo especificado (`ZeroDivisionError`), el intérprete detiene la ejecución del bloque `try` y salta **inmediatamente** al bloque `except`, evitando la interrupción del programa.

````python
# ================================================================================
# 📊 2. CAPTURA CONTROLADA CON BLOQUE TRY-EXCEPT
# ================================================================================
print("\n--- 📊 [PASO 2]: Ejecución Segura mediante try-except ---")

print("📋 [Ejemplo]: Intento de división por cero protegido")
print("-" * 50)

# 🔌 [1] Estructura de manejo defensivo de errores
try:
    print("⚙️ Intentando ejecutar la división dentro del bloque try...")
    division_result: float = numerator / denominator
    print(f"✅ Resultado: {division_result}")
except ZeroDivisionError:
    print("❌ [ERROR DETECTADO]: ZeroDivisionError ➔ No es posible dividir un número entre cero.")

# --------------------------------------------------------------------------------
# 🏁 [PASO 3]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Módulo básico de manejo de excepciones finalizado con éxito.")
print("=" * 80)
````
