### 💡 Explicación de la Arquitectura
* `print(*objects, sep=' ', end='\n')`: Imprime objetos en la salida estándar (soporta concatenación por comas y *f-strings*).
* `type(object)`: Retorna el tipo o clase del objeto pasado como parámetro.
* `len(sequence)`: Retorna la longitud (número de elementos/caracteres) de una secuencia o colección. Lanza `TypeError` si el objeto no es iterable (como `int` o `float`).
* `input(prompt)`: Captura una entrada de texto desde la consola/usuario (siempre retorna un valor de tipo `str`).

```python
# [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Funciones Nativas (Built-in Functions)")
print("=" * 80)

# ================================================================================
# 📢 1. SALIDA DE DATOS CON `print()` Y FORMATEO
# ================================================================================
# Formas de formatear texto: f-strings y separación por comas.
# ================================================================================

# [1] Declaración de variables en inglés
print_message: str = "Me encanta la programación en Python"
first_name: str = "Luis"
last_name: str = "Fernández"

print("\n--- 🚀 [PASO 1]: Impresión y Formateo de Texto con print() ---")

# Impresión directa
print(f"✅ [IMPRESIÓN DIRECTA]: {print_message}")

# Formateo 1: Método f-string (Recomendado)
print(f"✅ [F-STRING]: Hola, {first_name}!")

# Formateo 2: Separación por comas (agrega espacios automáticamente entre argumentos)
print("✅ [POR COMAS]:", "Hola,", last_name, "!")
```

## 🔍 Inspección de Tipos (`type()`) y Longitud (`len()`)

### 💡 Explicación de las Pruebas:
1. **Paso 2:** Inspección de clases con `type()`.
2. **Paso 3:** Obtención de longitud con `len()` en cadenas de texto y captura de un `TypeError` intencional al intentar evaluar un objeto de tipo `float`.

```python
# ================================================================================
# 📋 2. INSPECCIÓN DE TIPOS CON `type()`
# ================================================================================
print("\n--- 📋 [PASO 2]: Inspección de Tipos con type() ---")

sample_text: str = "Me encanta la programación"
sample_number: int = 3

print(f"✅ [TIPO CADENA]: El tipo de dato de 'sample_text' es:   {type(sample_text)}")
print(f"✅ [TIPO NUMÉRICO]: El tipo de dato de 'sample_number' es: {type(sample_number)}")

# ================================================================================
# 📏 3. LONGITUD DE SECUENCIAS CON `len()` Y CONTROL DE ERRORES
# ================================================================================
print("\n--- 📏 [PASO 3]: Medición de Longitud con len() y Manejo de TypeError ---")

text_length_target: str = "Me encanta la programación"
number_length_target: float = 4.2

# [1] Longitud válida en una cadena de texto
print(f"✅ [LONGITUD VÁLIDA]: La longitud de '{text_length_target}' es: {len(text_length_target)} caracteres")

# [2] Intento de obtener longitud en un flotante (Manejo defensivo con try-except)
try:
    # ⚠️ Esto lanzará TypeError intencional porque float no implementa __len__()
    invalid_len = len(number_length_target)  # type: ignore
except TypeError as error:
    print(f"❌ [ERROR DE TIPO] No se puede calcular len() en un objeto tipo 'float':")
    print(f"   ↳ Detalle técnico del sistema: {error}")
```

## ⌨️ Captura de Entrada de Usuario (`input()`)

### 💡 Explicación de la Captura:
La función `input()` pausa la ejecución del programa y espera la entrada del usuario. Todo dato ingresado a través de `input()` es evaluado **siempre** como una cadena de texto (`str`).

```python
# ================================================================================
# ⌨️ 4. ENTRADA DE DATOS POR CONSOLA CON `input()`
# ================================================================================
print("\n--- ⌨️ [PASO 4]: Captura de Entrada de Usuario con input() ---")

# [1] Solicitud de entrada (Si se ejecuta en entorno no interactivo, asigna valor por defecto)
try:
    user_input: str = input("Por favor indica tu nombre: ")
except EOFError:
    user_input: str = "Usuario Demo"

# [2] Inspección de valor y confirmación de tipo (siempre 'str')
print(f"✅ [ENTRADA USUARIO]: El valor ingresado ha sido: '{user_input}'")
print(f"📋 [CONFIRMACIÓN DE TIPO]: El tipo de dato ingresado es: {type(user_input)}")

# --------------------------------------------------------------------------------
# 🏁 [PASO 5]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Módulo de funciones incorporadas finalizado con éxito.")
print("=" * 80)
```
