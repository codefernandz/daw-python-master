### 💡 Ventajas de las Funciones con Parámetros

1. **Dinamismo y Flexibilidad:** A diferencia de las funciones sin parámetros, una función con parámetros ajusta su comportamiento y resultado según el argumento recibido.
2. **Aislamiento de Ámbito Local (*Scope Isolation*):** El parámetro actúa como una variable local dentro del cuerpo de la función. No interfiere con variables externas que tengan el mismo nombre.
3. **Reutilización Multiprocesamiento:** Una sola función puede procesar infinitos conjuntos de datos distintos sin necesidad de reescribir la lógica de filtrado o transformación.

## 🔤 Ejemplo 2: Procesamiento y Limpieza de Cadenas de Texto (`str`)

### 💡 Normalización de Entradas:
Procesar un parámetro de tipo texto para eliminar espacios sobrantes, convertir a minúsculas y contar palabras es un patrón habitual en saneamiento de datos.

````python
# ================================================================================
# 📊 2. CASO ADICIONAL 1: SANEAMIENTO Y ANÁLISIS DE TEXTO
# ================================================================================
print("\n--- 📊 [PASO 2]: Normalización y Métricas de Texto ---")


def analyze_text_payload(raw_text: str) -> dict[str, Any]:
    """Sanea una cadena de texto y calcula métricas básicas de contenido.

    Args:
        raw_text (str): La cadena de texto recibida para analizar.

    Returns:
        dict[str, Any]: Un diccionario con la versión limpia y sus métricas.
    """
    # ⚙️ Limpieza de espacios en bordes y conversión a minúsculas
    cleaned_text: str = raw_text.strip().lower()
    words_list: list[str] = cleaned_text.split()
    
    return {
        "cleaned_text": cleaned_text,
        "character_count": len(cleaned_text),
        "word_count": len(words_list)
    }


# 📋 Invocación con cadenas con ruido
dirty_input_1: str = "   Hola Mundo desde PYTHON  "
dirty_input_2: str = "Manejo de Funciones Modularizadas en Jupyter   "

print(f"📋 Entrada con espacio sobrante: '{dirty_input_1}'")
print("-" * 50)

metrics_result_1: dict[str, Any] = analyze_text_payload(dirty_input_1)
print(f"✅ [ANÁLISIS 1]: {metrics_result_1}")

metrics_result_2: dict[str, Any] = analyze_text_payload(dirty_input_2)
print(f"✅ [ANÁLISIS 2]: {metrics_result_2}")
````

## 🛒 Ejemplo 3: Cálculo Masivo en Estructuras de Diccionarios y Colecciones Complejas

````python
# ================================================================================
# 🧪 3. CASO ADICIONAL 2: PROCESAMIENTO DE CATÁLOGOS CON ESTRUCTURAS COMPLEJAS
# ================================================================================
print("\n--- 🧪 [PASO 3]: Filtrado de Productos por Stock Disponible ---")


def extract_in_stock_products(inventory: list[dict[str, Any]]) -> list[str]:
    """Filtra un catálogo de productos y devuelve los nombres de aquellos con stock > 0.

    Args:
        inventory (list[dict[str, Any]]): Lista de diccionarios con datos de productos.

    Returns:
        list[str]: Lista con los nombres de los productos disponibles.
    """
    available_items: list[str] = []
    
    for item in inventory:
        # ⚙️ Verificación de existencias
        if item.get("stock", 0) > 0:
            available_items.append(item["name"])
            
    return available_items


# 📋 Colección de productos
store_inventory: list[dict[str, Any]] = [
    {"name": "Monitor 4K", "stock": 12, "price": 350.00},
    {"name": "Teclado Mecánico", "stock": 0, "price": 80.00},
    {"name": "Mouse Inalámbrico", "stock": 5, "price": 25.00},
    {"name": "Auriculares", "stock": 0, "price": 45.00},
]

print("📋 Procesando inventario de la tienda:")
for prod in store_inventory:
    print(f"  🔹 {prod['name']:<20} | Stock: {prod['stock']}")

print("-" * 50)

# 🔌 Invocación de la función
available_products: list[str] = extract_in_stock_products(store_inventory)
print(f"✅ [PRODUCTOS DISPONIBLES]: {available_products}")

# --------------------------------------------------------------------------------
# 🏁 [PASO 4]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Módulo de funciones con un parámetro finalizado con éxito.")
print("=" * 80)
````
