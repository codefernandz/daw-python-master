### 💡 Motivación y Arquitectura de las Funciones

Una **función** es un bloque de código organizado y reutilizable diseñado para realizar una tarea específica. Las funciones solo se ejecutan cuando son invocadas (*called*).

#### 🚀 Ventajas Clave de la Modularización:
* **Reutilización de código (*DRY - Don't Repeat Yourself*):** Evita duplicar lógica idéntica en múltiples puntos del programa.
* **Mantenibilidad:** Si surge un error o cambia un requerimiento, la modificación se realiza únicamente dentro de la función.
* **Claridad y Organización:** Permite dividir problemas complejos en componentes más pequeños con nombres autodescriptivos.

---

### 📐 Componentes Principales de una Función:

1. `def`: Palabra clave obligatoria para declarar el inicio de la función.
2. **Nombre de la función:** Identificador en convención `snake_case` que debe describir claramente la acción que realiza.
3. **Parámetros (`parameters`):** Variables opcionales especificadas entre paréntesis `()` que la función acepta para operar.
4. **Docstring (`""" ... """`):** Documentación técnica que explica el propósito de la función, sus argumentos (`Args:`) y el valor retornado (`Returns:`).
5. **Cuerpo del Código:** Instrocciones indentadas que ejecutan la lógica interna.
6. `return`: Sentencia opcional que finaliza la ejecución y devuelve un resultado al punto de invocación.

```python
def function_name(parameter_1: int, parameter_2: int) -> int:
    """Resumen en una línea del comportamiento de la función.

    Args:
        parameter_1 (int): Descripción del primer parámetro.
        parameter_2 (int): Descripción del segundo parámetro.

    Returns:
        int: Descripción del valor devuelto.
    """
    # Lógica interna
    return parameter_1 + parameter_2

````python
from typing import Any

# 🛰️ [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Funciones - Declaración Básica y Documentación")
print("=" * 80)

# ================================================================================
# 🧪 1. DECLARACIÓN DE UNA FUNCIÓN SIN PARÁMETROS NI RETURN
# ================================================================================
# Definición de una función para imprimir banners de bienvenida.
# ================================================================================

print("\n--- 🚀 [PASO 1]: Función Simple de Saludo (Sin parámetros) ---")


def print_system_header() -> None:
    """Imprime un banner visual estándar en la consola para indicar el estado del sistema.

    Returns:
        None
    """
    print("-" * 50)
    print("🤖 SYSTEM_LOGGER: Módulo de funciones inicializado")
    print("-" * 50)


# 🔌 Invocación de la función
print("📋 Invocando la función 'print_system_header()':")
print_system_header()
````

## 📝 Documentación Formal con Docstrings y Retorno de Valores (`return`)

### 💡 Buenas Prácticas en la Documentación:
* **Uso de Type Hints:** Permite especificar los tipos de datos esperados en las entradas (`parameter: type`) y en la salida (`-> type`).
* **Estándar de Docstring:** Estructurar la documentación indicando explícitamente las secciones `Args:` y `Returns:`.

````python
# ================================================================================
# 📊 2. FUNCIÓN CON PARÁMETROS, TYPE HINTS Y RETORNO DE VALORES
# ================================================================================
print("\n--- 📊 [PASO 2]: Función Calculadora de Precio con Impuestos ---")


def calculate_total_price(base_price: float, tax_rate: float) -> float:
    """Calcula el precio total aplicando un porcentaje de impuesto sobre el precio base.

    Args:
        base_price (float): El precio base del producto o servicio.
        tax_rate (float): El porcentaje de impuesto a aplicar (ej. 0.21 para 21%).

    Returns:
        float: El monto total calculando base_price + (base_price * tax_rate).
    """
    # ⚙️ [1] Cálculo del impuesto y precio final
    tax_amount: float = base_price * tax_rate
    final_price: float = base_price + tax_amount
    return final_price


# 📋 Declaración de variables de prueba
item_price: float = 150.00
current_tax: float = 0.21

print(f"📋 Precio Base: ${item_price:.2f} | Tasa de Impuesto: {current_tax * 100}%")
print("-" * 50)

# 🔌 [2] Invocación de la función con retorno de valor
total_amount: float = calculate_total_price(base_price=item_price, tax_rate=current_tax)

print(f"✅ [ÉXITO]: El precio total calculado es: ${total_amount:.2f}")

# ℹ️ Inspección de la documentación de la función mediante __doc__
print("\n🔍 Inspección del Docstring registrado en la función:")
print("-" * 50)
print(calculate_total_price.__doc__)
````

## 🧮 Demostración de Reutilización de Código (*Code Reusability*)

````python
# ================================================================================
# 🧪 3. REUTILIZACIÓN DE FUNCIONES CON MÚLTIPLES ENTRADAS
# ================================================================================
print("\n--- 🧪 [PASO 3]: Invocación Iterativa de la Función en un Lote de Datos ---")

# 📋 Lista de precios a procesar
products_catalog: list[dict[str, Any]] = [
    {"name": "Laptop", "price": 1200.00},
    {"name": "Mouse", "price": 25.50},
    {"name": "Keyboard", "price": 75.00},
]

VAT_RATE: float = 0.21

print(f"📋 Procesando catálogo con tasa IVA del {VAT_RATE * 100}%:\n")

# 🔌 [1] Reutilización de la función dentro de un bucle for
for product in products_catalog:
    calculated_total: float = calculate_total_price(
        base_price=product["price"], 
        tax_rate=VAT_RATE
    )
    print(f"🛍️ Producto: {product['name']:<10} | Base: ${product['price']:>7.2f} | Total con IVA: ${calculated_total:>7.2f}")

# --------------------------------------------------------------------------------
# 🏁 [PASO 4]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Módulo básico de introducción a funciones finalizado con éxito.")
print("=" * 80)
````
