### 💡 Firma Fija vs. Firma Dinámica con Atributos Clave-Valor

Cuando una función declara parámetros nombrados específicos, Python valida que **únicamente** se envíen los argumentos definidos en la firma `def`.

* **Firma Fija:** Enviar una clave no declarada explícitamente (por ejemplo, `last_name="Garcia"`) provoca un `TypeError: got an unexpected keyword argument`.
* **Solución con `**kwargs`:** Al anteponer dos asteriscos `**` al parámetro (por convención `kwargs`), la función acepta **cualquier cantidad de pares clave-valor adicionales**, empaquetándolos automáticamente dentro de un **diccionario (`dict`)**.

---

### 📐 Comparativa de firmas:

```python
# ❌ Rígido: Lanza TypeError si se envía una clave no esperada
def display_profile_fixed(name: str, age: int, city: str, job: str) -> None:
    ...

# ✅ Dinámico: Acepta cualquier conjunto de argumentos clave-valor
def display_profile_dynamic(**kwargs: Any) -> None:
    ...

````python
from typing import Any

# 🛰️ [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Funciones - Argumentos Clave-Valor (**kwargs)")
print("=" * 80)

# ================================================================================
# 🧪 1. DEMOSTRACIÓN DE LA LIMITACIÓN DE FIRMA FIJA (TypeError)
# ================================================================================
# Definición de función con argumentos nombrados fijos e intercepción del error.
# ================================================================================

print("\n--- 🚀 [PASO 1]: Intento de Invocación con Clave Inesperada (TypeError) ---")


def print_user_info_fixed(name: str, age: int, city: str, profession: str) -> None:
    """Imprime información personal con una firma estricta de 4 parámetros.

    Args:
        name (str): Nombre del usuario.
        age (int): Edad del usuario.
        city (str): Ciudad de residencia.
        profession (str): Profesión u ocupación.
    """
    print(f"  🔹 Name: {name}")
    print(f"  🔹 Age: {age}")
    print(f"  🔹 City: {city}")
    print(f"  🔹 Profession: {profession}")


print("📋 Intentando enviar una clave adicional ('last_name') a la función fija:")
print("-" * 50)

# 🔌 Invocación con una clave no declarada para capturar el error
try:
    print_user_info_fixed(
        name="Juan",
        age=30,
        city="Madrid",
        profession="Ingeniero",
        last_name="Garcia"  # ❌ Clave no esperada
    )
except TypeError as keyword_error:
    print(f"❌ [ERROR DE TIPO CAPTURADO]: {keyword_error}")
````

## 📊 Ejemplo 2: Solución Dinámica mediante `**kwargs`

### 💡 Flexibilidad y Recorrido de Diccionarios:
Al declarar `**kwargs`, todos los pares clave-valor pasados a la función se empaquetan en un objeto `dict`. Esto nos permite usar métodos estándar como `.items()`, `.get()`, o `.keys()`.

````python
# ================================================================================
# 📊 2. SOLUCIÓN CON **kwargs E ITERACIÓN DE PARES CLAVE-VALOR
# ================================================================================
print("\n--- 📊 [PASO 2]: Función Dinámica con **kwargs ---")


def print_user_info(**kwargs: Any) -> None:
    """Imprime cualquier cantidad de datos personales recibidos como clave-valor.

    Args:
        **kwargs (Any): Diccionario de argumentos nombrados con la información del usuario.
    """
    # ⚙️ Muestra del diccionario empaquetado internamente
    print(f"  🔍 Diccionario **kwargs recibido (Tipo: {type(kwargs)}):")
    
    if not kwargs:
        print("  ⚠️ No se enviaron atributos para mostrar.")
        return

    # ⚙️ Iteración dinámica sobre las claves y valores
    for key, value in kwargs.items():
        print(f"  • {key.capitalize()}: {value}")


print("📋 Invocando la función 'print_user_info' con atributos flexibles:")
print("-" * 50)

# 🔌 Invocación con 5 argumentos clave-valor (incluyendo 'last_name')
print_user_info(
    name="Juan",
    age=30,
    city="Madrid",
    profession="Ingeniero",
    last_name="Garcia"
)
````

## 🛒 Ejemplo 3: Casos Prácticos de Aplicación Real para `**kwargs`

````python
# ================================================================================
# 🧪 3. CASOS PRÁCTICOS ADICIONALES: CONEXIÓN DB Y FILTRADO DINÁMICO
# ================================================================================
print("\n--- 🧪 [PASO 3]: Opciones de Configuración y Filtros Dinámicos ---")


# ⚙️ Caso A: Constructor de Conexiones a Bases de Datos con Parámetros Opcionales
def connect_database(db_name: str, host: str = "localhost", **options: Any) -> None:
    """Simula la conexión a una base de datos con opciones avanzadas dinámicas.

    Args:
        db_name (str): Nombre de la base de datos (obligatorio).
        host (str, optional): Host del servidor. Defaults to "localhost".
        **options (Any): Opciones de configuración adicionales (port, timeout, ssl, etc.).
    """
    print(f"🔌 Conectando a DB '{db_name}' en host '{host}'...")
    
    # ⚙️ Acceso seguro a valores opcionales mediante .get()
    port: int = options.get("port", 5432)
    ssl_enabled: bool = options.get("ssl", False)
    
    print(f"  🔹 Puerto asignado: {port}")
    print(f"  🔹 Configuración SSL: {ssl_enabled}")
    
    if "timeout" in options:
        print(f"  🔹 Timeout personalizado: {options['timeout']}s")


# ⚙️ Caso B: Filtrador Dinámico de Registros
def filter_records(records: list[dict[str, Any]], **filters: Any) -> list[dict[str, Any]]:
    """Filtra una lista de diccionarios según criterios arbitrarios enviados en kwargs.

    Args:
        records (list[dict[str, Any]]): Lista de diccionarios de datos.
        **filters (Any): Criterios de filtrado clave-valor.

    Returns:
        list[dict[str, Any]]: Lista con los registros que coinciden con todos los filtros.
    """
    matching_records: list[dict[str, Any]] = []
    
    for record in records:
        # Evalúa si el registro cumple con todas las parejas clave-valor requeridas
        if all(record.get(key) == value for key, value in filters.items()):
            matching_records.append(record)
            
    return matching_records


# 📋 Invocación de la conexión con opciones extra
print("📋 Probando configuración de Base de Datos:")
connect_database("production_db", host="db.internal.net", port=5433, ssl=True, timeout=15)

print("\n" + "-" * 50)

# 📋 Invocación del filtrado dinámico
dataset: list[dict[str, Any]] = [
    {"id": 1, "role": "admin", "status": "active"},
    {"id": 2, "role": "user", "status": "active"},
    {"id": 3, "role": "admin", "status": "inactive"}
]

print("📋 Filtrando dataset dinámicamente por role='admin' y status='active':")
filtered_results: list[dict[str, Any]] = filter_records(dataset, role="admin", status="active")
print(f"  ✅ Registros encontrados: {filtered_results}")

# --------------------------------------------------------------------------------
# 🏁 [PASO 4]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Módulo de argumentos clave-valor **kwargs finalizado con éxito.")
print("=" * 80)
````
