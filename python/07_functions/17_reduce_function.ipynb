### 💡 ¿Qué es la Función `reduce()`?

A diferencia de `map()` y `filter()`, que están integradas (*built-in*) y devuelven iteradores evaluados de forma perezosa (*Lazy Evaluation*), `reduce()` es una función de la librería estándar disponible en el módulo **`functools`**. 

`reduce()` procesa una secuencia aplicando de forma repetida una función binaria (que acepta dos argumentos: un **acumulador** y el **valor actual**) de izquierda a derecha. En cada iteración, toma el resultado devuelto y lo pasa como primer argumento de la siguiente iteración, colapsando o **reduciendo toda la colección a un único valor final**.

#### 📐 Sintaxis Básica:
```python
from functools import reduce

reduce(function, iterable[, initializer])
```

- **function**: Función binaria con la firma `f(accumulator, current_element) -> updated_accumulator`.
- **iterable**: Colección de elementos a procesar (`list`, `tuple`, `set`, etc.).
- **initializer** (Opcional): Valor inicial con el que se arranca la acumulación. Si se provee, se coloca antes de los elementos del iterable. Es vital para prevenir excepciones `TypeError` cuando la colección está vacía.

---

### 🔄 Flujo de Ejecución de `reduce()` (Paso a Paso)

Dada una lista `[A, B, C]` y una función `f(acc, item)`:

**Sin initializer:**

- Paso 1: `acc_1 = f(A, B)`
- Paso 2: `acc_2 = f(acc_1, C)`
- Resultado: `acc_2`

**Con initializer=0:**

- Paso 1: `acc_1 = f(0, A)`
- Paso 2: `acc_2 = f(acc_1, B)`
- Paso 3: `acc_3 = f(acc_2, C)`
- Resultado: `acc_3`

---

### ⚖️ Ventajas y Desventajas de `reduce()`

| Categoría | Aspecto | Descripción |
| :--- | :--- | :--- |
| 🟢 Ventaja | Elegancia Acumulativa | Permite expresar operaciones complejas de colapso de datos en una sola instrucción declarativa. |
| 🟢 Ventaja | Versatilidad de Salida | Puede transformar una lista en un escalar (float, int), un diccionario o un árbol de datos estructurado. |
| 🟢 Ventaja | Robustez con Initializer | Controla colecciones vacías asignando un estado inicial seguro. |
| 🔴 Desventaja | Curva de Legibilidad | Para agregaciones simples (sumas, máximos, mínimos), funciones integradas como `sum()`, `max()` y `min()` son sustancialmente más legibles y rápidas. |
| 🔴 Desventaja | Evaluación Inmediata | A diferencia de `map()` o `filter()`, procesa todos los elementos de forma inmediata (Eager Evaluation). |

````python
from functools import reduce
from typing import Callable, Any

# 🛰️ [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Funciones - Uso Detallado de reduce()")
print("=" * 80)

# ================================================================================
# 🧪 1. USO BÁSICO: BÚSQUEDA DEL MÁXIMO (FUNCIÓN DEF VS LAMBDA)
# ================================================================================
# Búsqueda del valor máximo en una lista usando reduce() y comparando sintaxis.
# ================================================================================

print("\n--- 🚀 [PASO 1]: Búsqueda de Valor Máximo con reduce() ---")


# ⚙️ [1] Definición de función binaria con Docstring norma Google y Type Hints
def get_higher_number(first_number: int, second_number: int) -> int:
    """Compara dos números enteros y retorna el mayor entre ellos.

    Args:
        first_number (int): Primer número a comparar (acumulador/anterior).
        second_number (int): Segundo número a comparar (elemento actual).

    Returns:
        int: El valor más alto entre ambos parámetros.
    """
    if int(first_number) > second_number:
        return first_number
    else:
        return second_number


numbers_list: list[int] = [23, 1, 45, 78, 5, 12, 98, 34, 67, 3, 56, 42, 89, 15, 7]

print(f"📋 Lista original de números: {numbers_list}")
print("-" * 50)

# 🔌 [2] Aplicación de reduce() con función nombrada 'def'
highest_number_def: int = reduce(get_higher_number, numbers_list)
print(f"  🔹 Resultado usando función 'def':    {highest_number_def}")

# 🔌 [3] Aplicación equivalente con expresión 'lambda'
highest_number_lambda: int = reduce(
    lambda acc, current: acc if acc > current else current, 
    numbers_list
)
print(f"  ✅ Resultado usando expresión 'lambda': {highest_number_lambda}")
````

## 📊 Ejemplo 2: El Parámetro `initializer` en `reduce()`

### 💡 Importancia de Asignar un Estado Inicial:
Proveer un valor inicial a `reduce()` garantiza:
1. Definir un valor por defecto si la lista está **vacía** (evita la excepción `TypeError: reduce() of empty sequence with no initial value`).
2. Permitir que el acumulador tenga un tipo de dato diferente al de los elementos de la lista.

````python
# ================================================================================
# 📊 2. REDUCE CON INITIALIZER Y PREVENCIÓN DE ERRORES
# ================================================================================
print("\n--- 📊 [PASO 2]: Uso del Parámetro 'initializer' ---")

empty_list: list[int] = []
valid_numbers: list[int] = [10, 20, 30]

print(f"📋 Lista vacía: {empty_list}")
print(f"📋 Lista válida: {valid_numbers}")
print("-" * 50)

# 🔌 [1] Suma acumulada con initializer sobre lista con datos
total_sum: int = reduce(
    lambda acc, val: acc + val, 
    valid_numbers, 
    0  # Initializer
)
print(f"  ✅ Suma acumulada (lista con datos, init=0): {total_sum}")

# 🔌 [2] Suma acumulada con initializer sobre lista vacía (Evita crash)
empty_sum: int = reduce(
    lambda acc, val: acc + val, 
    empty_list, 
    0  # Initializer
)
print(f"  ✅ Suma acumulada (lista vacía, init=0):     {empty_sum}")
````

## 🛒 Ejercicios Prácticos Adicionales

### 🧪 Ejercicio 1: Concatenación y Construcción de Cadenas
Uso de `reduce()` para acumular y dar formato a una secuencia de palabras formando una oración coherente con separadores específicos.

````python
# ================================================================================
# 🧪 3. EJERCICIO 1: CONCATENACIÓN DE TEXTO CON FORMATO
# ================================================================================
print("\n--- 🧪 [PASO 3]: Construcción de Cadenas de Texto con reduce() ---")

word_tokens: list[str] = ["Python", "permite", "escribir", "código", "limpio", "y", "expresivo"]

print(f"📋 Tokens de entrada: {word_tokens}")
print("-" * 50)

# ⚙️ Lambda que concatena agregando un espacio intermedio
formatted_sentence: str = reduce(
    lambda sentence_acc, word: f"{sentence_acc} {word}", 
    word_tokens
)

print(f"  ✅ Oración ensamblada: '{formatted_sentence}.'")
````

### 🧪 Ejercicio 2: Agregación de Métricas Financieras Complejas
Agregación de un historial de transacciones bancarias dividiendo el resultado final en un diccionario con el **balance total**, el **conteo total de operaciones** y la **transacción promedio**.

````python
# ================================================================================
# 🧪 4. EJERCICIO 2: AGREGACIÓN DE DICCIONARIOS DE MÉTRICAS FINANCIERAS
# ================================================================================
print("\n--- 🧪 [PASO 4]: Agregación de Métricas Financieras Complejas ---")

transactions: list[dict[str, Any]] = [
    {"id": "TX100", "type": "credit", "amount": 1200.50},
    {"id": "TX101", "type": "debit",  "amount": -300.00},
    {"id": "TX102", "type": "credit", "amount": 450.00},
    {"id": "TX103", "type": "debit",  "amount": -150.25},
    {"id": "TX104", "type": "credit", "amount": 800.00}
]

print("📋 Lista de transacciones a procesar:")
for tx in transactions:
    print(f"  • ID: {tx['id']} | Tipo: {tx['type']:6s} | Monto: ${tx['amount']:8.2f}")

print("-" * 50)

# ⚙️ Estado inicial del diccionario acumulador
initial_metrics: dict[str, Any] = {
    "total_balance": 0.0,
    "total_credits": 0.0,
    "total_debits": 0.0,
    "transaction_count": 0
}


# ⚙️ Función de reducción binaria
def aggregate_financials(acc: dict[str, Any], tx: dict[str, Any]) -> dict[str, Any]:
    """Acumula importes y conteos a partir de un objeto de transacción."""
    amount: float = tx["amount"]
    acc["total_balance"] += amount
    acc["transaction_count"] += 1

    if tx["type"] == "credit":
        acc["total_credits"] += amount
    elif tx["type"] == "debit":
        acc["total_debits"] += abs(amount)

    return acc


# 🔌 Procesamiento acumulativo con reduce()
financial_summary: dict[str, Any] = reduce(
    aggregate_financials, 
    transactions, 
    initial_metrics
)

# Cálculo de promedio final
avg_transaction: float = (
    financial_summary["total_balance"] / financial_summary["transaction_count"]
    if financial_summary["transaction_count"] > 0 else 0.0
)

print("✅ Resumen financiero acumulado:")
print(f"  🔹 Balance total:         ${financial_summary['total_balance']:.2f}")
print(f"  🔹 Total de ingresos:     ${financial_summary['total_credits']:.2f}")
print(f"  🔹 Total de egresos:      ${financial_summary['total_debits']:.2f}")
print(f"  🔹 Total de operaciones:  {financial_summary['transaction_count']}")
print(f"  🔹 Promedio por operación: ${avg_transaction:.2f}")

# --------------------------------------------------------------------------------
# 🏁 [PASO 5]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Cuaderno de la función reduce() finalizado con éxito.")
print("=" * 80)
````
