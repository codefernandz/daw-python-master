### 💡 Funcionamiento de los Parámetros por Defecto

Un **parámetro por defecto** es aquel que tiene un valor asignado en la declaración `def` mediante el operador de asignación `=`. 

#### 📐 Reglas de Sintaxis Obligatorias:
1. **Invocación Opcional:** Si quien llama a la función omite el argumento, Python utiliza automáticamente el valor predeterminado.
2. **Orden Estricto de Declaración:** Todos los parámetros **obligatorios** (sin valor por defecto) deben colocarse **ANTES** que cualquier parámetro por defecto. De lo contrario, Python lanzará un error de sintaxis (`SyntaxError: non-default argument follows default argument`).

```python
# ❌ INCORRECTO: Lanza SyntaxError
def invalid_function(tax_rate: float = 0.21, price: float) -> float:
    return price * tax_rate

# ✅ CORRECTO: Parámetro obligatorio primero, parámetro por defecto al final
def valid_function(price: float, tax_rate: float = 0.21) -> float:
    return price * tax_rate

````python
from typing import Any, Optional

# 🛰️ [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Funciones - Parámetros por Defecto")
print("=" * 80)

# ================================================================================
# 🧪 1. EJEMPLO PRINCIPAL: SALUDO PERSONALIZADO CON VALOR POR DEFECTO
# ================================================================================
# Saludo que adopta "Friend" cuando el cliente omite el argumento.
# ================================================================================

print("\n--- 🚀 [PASO 1]: Función de Saludo con Parámetro Opcional ---")


def greet_user(username: str = "Friend") -> str:
    """Devuelve un mensaje de saludo personalizado utilizando un valor predeterminado.

    Args:
        username (str, optional): El nombre de la persona a saludar. Defaults to "Friend".

    Returns:
        str: El mensaje de saludo formateado.
    """
    return f"Hola, {username}!"


print("📋 Probando invocaciones con y sin argumento:")
print("-" * 50)

# 🔌 [1] Invocación sin argumentos (usa el valor por defecto "Friend")
default_greeting: str = greet_user()
print(f"✅ [SIN ARGUMENTO]: {default_greeting}")

# 🔌 [2] Invocación con argumento explícito (sobrescribe el valor por defecto)
custom_greeting: str = greet_user("alumni")
print(f"✅ [CON ARGUMENTO 'alumni']: {custom_greeting}")
````

## 📊 Ejemplo 2: Formateador Financiero con Múltiples Parámetros por Defecto

### 💡 Combinando Parámetros Obligatorios y Opcionales:
En reportes bancarios o de comercio electrónico, los montos suelen requerir el código de moneda (`USD`, `EUR`) y la cantidad de decimales. Configurar estos valores por defecto simplifica las llamadas comunes.

````python
# ================================================================================
# 📊 2. CASO ADICIONAL 1: FORMATEADOR DE MONEDA MULTI-OPCIÓN
# ================================================================================
print("\n--- 📊 [PASO 2]: Formateador de Montos Monedas ---")


def format_currency(
    amount: float, 
    currency_symbol: str = "$", 
    decimals: int = 2, 
    show_code: bool = False
) -> str:
    """Formatea un número flotante como un monto de moneda configurable.

    Args:
        amount (float): El monto numérico obligatorio.
        currency_symbol (str, optional): Símbolo de la moneda. Defaults to "$".
        decimals (int, optional): Cantidad de cifras decimales. Defaults to 2.
        show_code (bool, optional): Si añade el sufijo 'USD'/'EUR'. Defaults to False.

    Returns:
        str: El texto formateado de la moneda.
    """
    formatted_number: str = f"{amount:,.{decimals}f}"
    base_format: str = f"{currency_symbol}{formatted_number}"
    
    if show_code:
        return f"{base_format} USD"
    return base_format


# 📋 Tests con combinaciones de argumentos
tx_amount: float = 1250.758

print(f"📋 Monto base de prueba: {tx_amount}")
print("-" * 50)

print(f"🔹 Por Defecto:        {format_currency(tx_amount)}")
print(f"🔹 Con Euro (€):       {format_currency(tx_amount, currency_symbol='€')}")
print(f"🔹 Sin Decimales:      {format_currency(tx_amount, decimals=0)}")
print(f"🔹 Con Código Visible: {format_currency(tx_amount, show_code=True)}")
````

## ⚠️ Caso Crítico: El Peligro de Argumentos Mutables por Defecto

### 💡 Anti-Patrón Común en Python:
Nunca debes usar objetos mutables (como `list` o `dict`) directamente como valores por defecto en la firma `def add_item(target_list: list = [])`. 

#### ❌ ¿Por qué es peligroso?
Python evalúa el valor predeterminado **una sola vez** al momento de definir la función, no en cada invocación. La misma lista persistirá en memoria compartida entre llamadas subsecuentes.

#### ✅ Solución Estándar con `Optional[T]` y `None`:
Usar `None` como valor por defecto e inicializar la nueva lista dentro del cuerpo de la función.

````python
# ================================================================================
# 🧪 3. CASO ADICIONAL 2: PATRÓN SEGURO PARA COLECCIONES MUTABLES
# ================================================================================
print("\n--- 🧪 [PASO 3]: Manejo Seguro de Listas como Parámetros Opcionales ---")


def append_log_entry(
    message: str, 
    log_history: Optional[list[str]] = None
) -> list[str]:
    """Añade una entrada de log a un historial. Previene el problema de mutable predeterminado.

    Args:
        message (str): El mensaje de log a registrar.
        log_history (Optional[list[str]], optional): Lista existente de logs. Defaults to None.

    Returns:
        list[str]: La lista de logs actualizada.
    """
    # ⚙️ Patrón seguro: Si es None, instanciar una nueva lista local
    if log_history is None:
        log_history = []
        
    log_history.append(message)
    return log_history


print("📋 Probando invocaciones independientes con lista mutable protegida:")
print("-" * 50)

# 🔌 Invocación 1 (lista nueva creada dinámicamente)
session_1_logs: list[str] = append_log_entry("Conexión establecida")
print(f"✅ [SESIÓN 1]: {session_1_logs}")

# 🔌 Invocación 2 (independiente, NO comparte la lista de la sesión 1)
session_2_logs: list[str] = append_log_entry("Error de autenticación")
print(f"✅ [SESIÓN 2]: {session_2_logs}")

# 🔌 Invocación 3 (pasando una lista existente para acumular)
append_log_entry("Reintentando conexión...", session_1_logs)
print(f"✅ [SESIÓN 1 REUTILIZADA]: {session_1_logs}")

# --------------------------------------------------------------------------------
# 🏁 [PASO 4]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Módulo de parámetros por defecto finalizado con éxito.")
print("=" * 80)
````
