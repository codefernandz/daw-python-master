# ⚙️ Fundamentos de Python: Argumentos Variables (`*args` y `**kwargs`)

| Propiedad | Detalle |
| :--- | :--- |
| 📍 **Módulo** | Fundamentos de Python: Funciones |
| 🎯 **Objetivo** | Dominar el manejo de funciones con un número indeterminado de argumentos posicionales (`*args`) y clave-valor (`**kwargs`), comprendiendo la estructura de datos interna que generan (`tuple` y `dict`). |
| 🛠️ **Conceptos** | `*args`, `**kwargs`, `Unpacking`, `Positional Arguments Tuple`, `Keyword Arguments Dict`, `Argument Ordering Rules` |

---

### 💡 Flexibilidad en la Firma de Funciones

En múltiples escenarios es imposible prever de antemano cuántos valores o configuraciones recibirá una función. Python resuelve esto mediante los operadores de desempaquetado `*` y `**`.

#### 📐 Convenciones Estándar:
1. `*args` (*Positional Arguments*): Captura cualquier cantidad de argumentos posicionales adicionales y los empaqueta en una **tupla (`tuple`)**.
2. `**kwargs` (*Keyword Arguments*): Captura cualquier cantidad de argumentos nombrados adicionales y los empaqueta en un **diccionario (`dict`)**.

> ⚠️ **Nota:** Las palabras `args` y `kwargs` son convenciones de la comunidad (*PEP 8*). Técnicamente es el asterisco (`*` o `**`) lo que otorga la funcionalidad, pero es **buena práctica obligatoria** respetar estos nombres.

---

### 📌 Orden Estricto de Parámetros en la Firma (`def`):

Si una función combina todos los tipos de parámetros, debe respetar **estrictamente** este orden para evitar errores de sintaxis (`SyntaxError`):

1. Parámetros posicionales estándar.
2. Parámetros por defecto (`default parameters`).
3. Argumentos posicionales variables (`*args`).
4. Argumentos clave-valor variables (`**kwargs`).

```python
def complex_signature(
    mandatory_param: str, 
    default_param: int = 10, 
    *args: float, 
    **kwargs: Any
) -> None:
    ...

````python
from typing import Any

# 🛰️ [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Funciones - Argumentos Variables (*args y **kwargs)")
print("=" * 80)

# ================================================================================
# 🧪 1. USO DE *args PARA ARGUMENTOS POSICIONALES DINÁMICOS
# ================================================================================
# Sumatoria y estadísticas sobre una cantidad indefinida de números flotantes.
# ================================================================================

print("\n--- 🚀 [PASO 1]: Procesamiento Dinámico de Números con *args ---")


def calculate_sum_and_average(*numbers: float) -> dict[str, float]:
    """Calcula la suma total y el promedio de una cantidad arbitraria de números.

    Args:
        *numbers (float): Cantidad indefinida de valores numéricos recibidos como tupla.

    Returns:
        dict[str, float]: Diccionario con el total, conteo y promedio calculado.
    """
    # ⚙️ Demostración de que *numbers es internamente una Tupla (tuple)
    print(f"  🔍 Tipo interno de *numbers: {type(numbers)} | Contenido: {numbers}")
    
    if not numbers:
        return {"total": 0.0, "count": 0, "average": 0.0}
        
    total_sum: float = sum(numbers)
    items_count: int = len(numbers)
    average_val: float = total_sum / items_count
    
    return {
        "total": total_sum,
        "count": float(items_count),
        "average": average_val
    }


print("📋 Probando invocaciones con diferente número de argumentos:")
print("-" * 50)

# 🔌 [1] Invocación con 3 argumentos
stats_1: dict[str, float] = calculate_sum_and_average(10.5, 20.0, 30.5)
print(f"✅ [RESULTADO 1]: {stats_1}\n")

# 🔌 [2] Invocación con 5 argumentos
stats_2: dict[str, float] = calculate_sum_and_average(5.0, 15.0, 25.0, 35.0, 45.0)
print(f"✅ [RESULTADO 2]: {stats_2}")
````

## 📊 Ejemplo 2: Uso de `**kwargs` para Argumentos Clave-Valor Dinámicos

### 💡 Procesamiento de Metadatos y Opciones Configurables:
`**kwargs` se utiliza habitualmente para recibir pares de clave-valor opcionales en sistemas de logs, constructores de objetos o consultas a bases de datos.

````python
# ================================================================================
# 📊 2. USO DE **kwargs PARA OPCIONES Y CONFIGURACIONES LIBRES
# ================================================================================
print("\n--- 📊 [PASO 2]: Registro de Eventos con Metadatos Variables via **kwargs ---")


def build_user_profile(user_id: int, username: str, **attributes: Any) -> dict[str, Any]:
    """Construye el perfil de un usuario aceptando atributos adicionales dinámicos.

    Args:
        user_id (int): ID único del usuario (parámetro obligatorio).
        username (str): Nombre de usuario (parámetro obligatorio).
        **attributes (Any): Atributos opcionales clave-valor recibidos como diccionario.

    Returns:
        dict[str, Any]: Perfil completo consolidado.
    """
    # ⚙️ Demostración de que **attributes es internamente un Diccionario (dict)
    print(f"  🔍 Tipo interno de **attributes: {type(attributes)} | Contenido: {attributes}")
    
    profile: dict[str, Any] = {
        "user_id": user_id,
        "username": username,
        "active": True
    }
    
    # ⚙️ Integración de los atributos adicionales recibidos
    profile.update(attributes)
    return profile


print("📋 Creando usuarios con diferentes metadatos de perfil:")
print("-" * 50)

# 🔌 Invocación 1 con metadatos de email y rol
user_1: dict[str, Any] = build_user_profile(
    user_id=101, 
    username="admin_user", 
    email="admin@company.com", 
    role="System Administrator"
)
print(f"✅ [USUARIO 1]: {user_1}\n")

# 🔌 Invocación 2 con metadatos de localización y preferencias
user_2: dict[str, Any] = build_user_profile(
    user_id=102, 
    username="dev_devon", 
    country="Spain", 
    theme="dark", 
    notifications=False
)
print(f"✅ [USUARIO 2]: {user_2}")
````

## ⚙️ Ejemplo 3: Combinación Completa (`Parámetros Estándar` + `*args` + `**kwargs`)

### 💡 Caso Real: Formateador Avanzado de Logs de Sistema
Combina un prefijo de log obligatorio, un mensaje principal, una lista indeterminada de etiquetas (*tags*) mediante `*args`, y parámetros de contexto mediante `**kwargs`.

````python
# ================================================================================
# 🧪 3. COMBINACIÓN DE PARÁMETROS ESTÁNDAR, *args Y **kwargs
# ================================================================================
print("\n--- 🧪 [PASO 3]: Sistema Integrado de Logs con Tags y Contexto ---")


def log_system_event(
    log_level: str, 
    message: str, 
    *tags: str, 
    **context: Any
) -> None:
    """Registra un evento formateado combinando nivel, mensaje, tags y metadatos.

    Args:
        log_level (str): Nivel del log ('INFO', 'WARNING', 'ERROR').
        message (str): Mensaje descriptivo del evento.
        *tags (str): Etiquetas o categorías del log (recibidas como tuple).
        **context (Any): Información contextual del evento (recibida como dict).

    Returns:
        None
    """
    formatted_tags: str = f"[{', '.join(tags)}]" if tags else "[NO_TAGS]"
    
    print(f"📌 [{log_level.upper()}] {message}")
    print(f"   🏷️ Tags: {formatted_tags}")
    
    if context:
        print("   🔍 Contexto Adicional:")
        for key, value in context.items():
            print(f"      • {key}: {value}")


print("📋 Ejecutando traza de logs con firma compleja:")
print("-" * 50)

# 🔌 Invocación 1 con Tags y Contexto
log_system_event(
    "ERROR", 
    "Fallo en la conexión a la Base de Datos", 
    "DB", "NETWORK", "CRITICAL",  # *args
    server_ip="192.168.1.50",    # **kwargs
    retry_attempts=3,            # **kwargs
    timeout_seconds=30.0         # **kwargs
)

print("-" * 50)

# 🔌 Invocación 2 solo con parámetros requeridos
log_system_event("INFO", "Proceso en segundo plano completado")

# --------------------------------------------------------------------------------
# 🏁 [PASO 4]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Módulo de argumentos variables finalizado con éxito.")
print("=" * 80)
````
