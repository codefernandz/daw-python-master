### 💡 Flexibilidad Múltiple en Funciones Anónimas

Las funciones lambda no están limitadas a un solo parámetro. Pueden recibir $N$ argumentos separándolos por comas en su declaración, manteniendo la restricción de ejecutar **una sola expresión**:

```python
# Sintaxis general con Múltiples Parámetros:
lambda arg_1, arg_2, ..., arg_n: expression

Estrategia,Ventajas,Sintaxis
Función def Tradicional,"Legibilidad alta, depuración paso a paso",Bucle for + if + .append()
Lambda + List Comprehension,"Compacta, altamente idiomática en Python",lambda lst: [x for x in lst if cond]
Lambda + filter(),"Evaluación perezosa (Lazy Evaluation) con iteradores","filter(lambda x: cond, lst)"
```

````python
from typing import Callable, Any

# 🛰️ [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Funciones - Lambdas Mutiparámetro y Listas")
print("=" * 80)

# ================================================================================
# 🧪 1. LAMBDAS CON MÚLTIPLES PARÁMETROS (SUMA DE DOS NÚMEROS)
# ================================================================================
# Comparación directa entre función tradicional 'def' y lambda binaria (2 args).
# ================================================================================

print("\n--- 🚀 [PASO 1]: Operaciones Binarias con Múltiples Parámetros ---")


# ⚙️ [1] Enfoque Tradicional con 'def'
def add_two_numbers(number1: float, number2: float) -> float:
    """Suma dos números recibidos como parámetros posicionales.

    Args:
        number1 (float): Primer operando.
        number2 (float): Segundo operando.

    Returns:
        float: Resultado de la suma.
    """
    return number1 + number2


# ⚙️ [2] Enfoque Anónimo con 'lambda'
add_two_numbers_lambda: Callable[[float, float], float] = lambda x, y: x + y

num_a: float = 12.0
num_b: float = 3.0

print(f"📋 Operandos de entrada: {num_a} y {num_b}")
print("-" * 50)

# 🔌 Invocación de ambas funciones
result_def: float = add_two_numbers(num_a, num_b)
result_lambda: float = add_two_numbers_lambda(num_a, num_b)

print(f"  🔹 Resultado con función estándar 'def': {result_def}")
print(f"  ✅ Resultado con expresión 'lambda':     {result_lambda}")
````

## 📊 Ejemplo 2: Filtrado de Listas (Divisibilidad entre 3)

### 💡 Comparativa entre Bucles Tradicionales y Lambda + List Comprehension:
Demostración de cómo reducir un bloque iterativo de 8 líneas a una sola expresión funcional.

````python
# ================================================================================
# 📊 2. FILTRADO DE ELEMENTOS DIVISIBLES ENTRE 3
# ================================================================================
print("\n--- 📊 [PASO 2]: Filtrado de Listas con List Comprehension y Lambda ---")


# ⚙️ [1] Método Tradicional con Bucles e Inspección Controlada
def filter_divisible_by_three_standard(numbers: list[int]) -> list[int]:
    """Filtra los elementos de una lista reteniendo solo los divisibles entre 3.

    Args:
        numbers (list[int]): Colección de enteros a evaluar.

    Returns:
        list[int]: Nueva lista con los números seleccionados.
    """
    divisible_elements: list[int] = []
    for element in numbers:
        if element % 3 == 0:
            divisible_elements.append(element)
        else:
            continue
    return divisible_elements


# ⚙️ [2] Método Compacto usando Lambda + List Comprehension
filter_divisible_by_three_lambda: Callable[[list[int]], list[int]] = (
    lambda number_list: [element for element in number_list if element % 3 == 0]
)

sample_numbers: list[int] = [34, 39, 12, 15, 16, 19]

print(f"📋 Lista original de prueba: {sample_numbers}")
print("-" * 50)

# 🔌 [1] Ejecución tradicional
filtered_standard: list[int] = filter_divisible_by_three_standard(sample_numbers)
print(f"  🔹 Resultado 'def' (Bucle for):          {filtered_standard}")

# 🔌 [2] Ejecución con Lambda
filtered_lambda: list[int] = filter_divisible_by_three_lambda(sample_numbers)
print(f"  ✅ Resultado 'lambda' (List Comp.):      {filtered_lambda}")
````

### 🧪 Ejercicio 1: Cálculo Geométrico (Distancia Euclidiana 2D)
Uso de lambdas con múltiples parámetros para calcular la distancia entre dos puntos $(x_1, y_1)$ y $(x_2, y_2)$ en un plano cartesiano:
$$d = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2}$$

````python
# ================================================================================
# 🧪 3. EJERCICIO 1: DISTANCIA EUCLIDIANA CON LAMBDA DE 4 PARÁMETROS
# ================================================================================
import math

print("\n--- 🧪 [PASO 3]: Geometría 2D con Lambdas Multiparámetro ---")

# ⚙️ Lambda que recibe 4 coordenadas float y retorna la distancia Euclidiana
calculate_distance_2d: Callable[[float, float, float, float], float] = (
    lambda x1, y1, x2, y2: math.sqrt((x2 - x1)**2 + (y2 - y1)**2)
)

point_a: tuple[float, float] = (1.0, 2.0)
point_b: tuple[float, float] = (4.0, 6.0)

print(f"📋 Punto A: {point_a} | Punto B: {point_b}")
print("-" * 50)

# 🔌 Unpacking de las tuplas al invocar la lambda multiparámetro
distance: float = calculate_distance_2d(*point_a, *point_b)
print(f"  ✅ Distancia cartesiana calculada: {distance:.4f}")
````

### 🧪 Ejercicio 2: Procesamiento Pareado de Listas (`zip` + `map` + `lambda`)
Demostración de cómo combinar múltiples listas elemento a elemento aplicando una lambda binaria de ponderación de notas.

````python
# ================================================================================
# 🧪 4. EJERCICIO 2: PONDERACIÓN DE EVALUACIONES (ZIP + MAP + LAMBDA)
# ================================================================================
print("\n--- 🧪 [PASO 4]: Ponderación Combinada de Listas de Datos ---")

exam_scores: list[float] = [80.0, 95.0, 70.0]
project_scores: list[float] = [90.0, 85.0, 100.0]

# ⚙️ Lambda que combina Nota Examen (60%) y Nota Proyecto (40%)
calculate_weighted_final: Callable[[float, float], float] = (
    lambda exam, project: round((exam * 0.60) + (project * 0.40), 2)
)

print(f"📋 Notas Exámenes (60%):  {exam_scores}")
print(f"📋 Notas Proyectos (40%): {project_scores}")
print("-" * 50)

# 🔌 Procesamiento simultáneo de ambas colecciones mediante map y zip
final_grades: list[float] = list(
    map(lambda pair: calculate_weighted_final(pair[0], pair[1]), zip(exam_scores, project_scores))
)

print(f"  ✅ Calificaciones finales ponderadas: {final_grades}")

# --------------------------------------------------------------------------------
# 🏁 [PASO 5]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Cuaderno de lambdas multiparámetro finalizado con éxito.")
print("=" * 80)
````
