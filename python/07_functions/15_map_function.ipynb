### 💡 ¿Qué es la Función `map()`?

La función integrada `map()` toma dos o más parámetros: una **función transformadora** como primer argumento y uno o varios **iterables** (listas, tuplas, conjuntos, etc.) como argumentos posteriores. Aplica la función a cada elemento de la secuencia y devuelve un **objeto iterador** (`map_iterator`) con los resultados.

#### 📐 Sintaxis Básica:
```python
map(func, iterable1, [iterable2, ...])
```

| Categoría | Aspecto | Descripción |
| :--- | :--- | :--- |
| 🟢 Ventaja | Código Declarativo | Promueve un estilo conciso y modular enfocándose en qué hacer y no en cómo iterar. |
| 🟢 Ventaja | Evaluación Perezosa (Lazy Evaluation) | Retorna un iterador que procesa elementos bajo demanda, optimizando drásticamente el consumo de memoria. |
| 🟢 Ventaja | Iteración Paralela | Acepta múltiples iterables y procesa sus elementos en paralelo (similar a zip()). |
| 🟢 Ventaja | Inmutabilidad | Genera nuevas colecciones sin modificar la estructura original (no side-effects). |
| 🔴 Desventaja | Legibilidad en Expresiones Complejas | Con lambdas intrincadas, un bucle for o una List Comprehension puede resultar más legible. |
| 🔴 Desventaja | Consumo Único del Iterador | Al ser un iterador, una vez recorrido el objeto map, sus elementos se agotan y no pueden re-evaluarse sin reinstanciarlo. |
```

````python
from typing import Callable, Any

# 🛰️ [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Funciones - Uso Detallado de map()")
print("=" * 80)

# ================================================================================
# 🧪 1. USO BÁSICO: FUNCIÓN NOMBRADA VS LAMBDA EN MAP()
# ================================================================================
# Cálculo del cuadrado de una lista de números demostrando el iterador devuelto.
# ================================================================================

print("\n--- 🚀 [PASO 1]: Transformación Básica con map() (def vs lambda) ---")


# ⚙️ [1] Definición de función nominada con Docstring norma Google y Type Hints
def calculate_square(number: float) -> float:
    """Calcula el cuadrado de un número proporcionado.

    Args:
        number (float): El número que se desea elevar al cuadrado.

    Returns:
        float: El cuadrado del número recibido.
    """
    return number ** 2


raw_numbers: list[int] = [1, 2, 3, 4, 5]

print(f"📋 Lista original de números: {raw_numbers}")
print("-" * 50)

# 🔌 [2] Aplicación de map() con función nominada 'def'
map_iterator: map[float] = map(calculate_square, raw_numbers)
print(f"🔍 Objeto retornado por map(): {map_iterator} (Tipo: {type(map_iterator)})")

# Conversión explícita del iterador a lista
squared_numbers_def: list[float] = list(map_iterator)
print(f"  🔹 Resultado con 'def' convertido a lista: {squared_numbers_def}")

# 🔌 [3] Aplicación equivalente directa con expresión 'lambda'
squared_numbers_lambda: list[float] = list(
    map(lambda element: element ** 2, raw_numbers)
)
print(f"  ✅ Resultado con 'lambda' convertido a lista: {squared_numbers_lambda}")
````

## 📊 Ejemplo 2: Procesamiento de Múltiples Iterables en Paralelo

### 💡 Múltiples Argumentos en `map()`:
Cuando se pasan varios iterables a `map()`, la función suministrada debe aceptar **tantos parámetros como iterables** se hayan provisto. La ejecución se detiene cuando el iterable más corto se agota.

````python
# ================================================================================
# 📊 2. PROCESAMIENTO MULTI-ITERABLE EN PARALELO
# ================================================================================
print("\n--- 📊 [PASO 2]: Iteración Paralela con Múltiples Secuencias ---")

base_values: list[float] = [2.0, 3.0, 5.0, 10.0]
exponents: list[float] = [2.0, 3.0, 2.0, 4.0]

print(f"📋 Colección de bases:      {base_values}")
print(f"📋 Colección de exponentes: {exponents}")
print("-" * 50)

# ⚙️ Lambda binaria que calcula la potencia de base^exponente recibiendo 2 colecciones
powered_results: list[float] = list(
    map(lambda base, exp: base ** exp, base_values, exponents)
)

print(f"  ✅ Resultado de potenciación elemento a elemento: {powered_results}")
````

## 🛒 Ejercicios Prácticos Adicionales

### 🧪 Ejercicio 1: Limpieza y Normalización de Datos de Texto
Uso de `map()` para sanitizar una lista de nombres con espacios sobrantes y capitalización inconsistente mediante métodos de cadena integrados.

````python
# ================================================================================
# 🧪 3. EJERCICIO 1: SANITIZACIÓN Y NORMALIZACIÓN DE CADENAS DE TEXTO
# ================================================================================
print("\n--- 🧪 [PASO 3]: Sanitización de Cadenas de Texto con map() ---")

raw_user_names: list[str] = [
    "  alBERTo  ", 
    "mARÍa   ", 
    "   jUAN pABLo ", 
    "  cARLOs "
]

print("📋 Nombres sin procesar (con espacios y mayúsculas/minúsculas irregulares):")
for name in raw_user_names:
    print(f"  • '{name}'")

print("-" * 50)

# ⚙️ Pipeline de transformación mediante lambda: eliminar espacios y aplicar title()
clean_names: list[str] = list(
    map(lambda name: name.strip().title(), raw_user_names)
)

print("✅ Nombres sanitizados y normalizados:")
for name in clean_names:
    print(f"  🔹 '{name}'")
````

### 🧪 Ejercicio 2: Cálculo de Precios con Impuestos y Descuentos
Aplicación de una transformación compleja con `map()` sobre una lista de diccionarios que representan artículos de una tienda.

````python
# ================================================================================
# 🧪 4. EJERCICIO 2: CÁLCULO FINANCIERO SOBRE ESTRUCTURAS COMPLEJAS
# ================================================================================
print("\n--- 🧪 [PASO 4]: Cálculo de Precios Netos sobre Diccionarios ---")

products: list[dict[str, Any]] = [
    {"name": "Smartphone", "price": 500.0, "discount_ratio": 0.10},
    {"name": "Headphones", "price": 80.0, "discount_ratio": 0.05},
    {"name": "Smartwatch", "price": 200.0, "discount_ratio": 0.15}
]

vat_rate: float = 0.21  # 21% IVA

print("📋 Productos originales:")
for prod in products:
    print(f"  • {prod['name']} | Precio base: ${prod['price']} | Desc: {prod['discount_ratio']*100}%")

print("-" * 50)

# ⚙️ Lambda que calcula: (Precio - Descuento) * (1 + IVA)
calculate_net_price: Callable[[dict[str, Any]], dict[str, Any]] = (
    lambda item: {
        "name": item["name"],
        "net_price": round(
            (item["price"] * (1 - item["discount_ratio"])) * (1 + vat_rate), 2
        )
    }
)

# 🔌 Transformación masiva de la lista de productos
processed_prices: list[dict[str, Any]] = list(map(calculate_net_price, products))

print("✅ Precios finales calculados (Con Descuento e IVA):")
for item in processed_prices:
    print(f"  🔹 {item['name']}: ${item['net_price']}")

# --------------------------------------------------------------------------------
# 🏁 [PASO 5]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Cuaderno de la función map() finalizado con éxito.")
print("=" * 80)
````
