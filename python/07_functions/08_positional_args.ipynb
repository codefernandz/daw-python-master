### 💡 Del Criterio Rígido a la Firma Dinámica

Cuando definimos una función con un número fijo de parámetros (por ejemplo, tres parámetros), Python exige pasar **exactamente** esa cantidad de argumentos posicionales.

* **Firma Fija:** Pasar un número menor o mayor de argumentos provoca una excepción inmediata de tipo `TypeError`.
* **Solución con `*args`:** Al anteponer un asterisco `*` al nombre del parámetro (por convención `args`), permitimos que la función acepte desde **0 hasta N argumentos posicionales**, empaquetándolos automáticamente en una **tupla (`tuple`)**.

---

### 📐 Comparativa de firmas:

```python
# ❌ Rígido: Lanza TypeError si no se envían exactamente 3 argumentos
def fixed_function(arg1: str, arg2: str, arg3: str) -> None:
    ...

# ✅ Dinámico: Permite 0, 1, 3, 10 o más argumentos sin romper la ejecución
def dynamic_function(*args: str) -> None:
    ...

````python
from typing import Any

# 🛰️ [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Funciones - Argumentos Posicionales (*args)")
print("=" * 80)

# ================================================================================
# 🧪 1. DEMOSTRACIÓN DE LA LIMITACIÓN DE FIRMA FIJA (TypeError)
# ================================================================================
# Definición de función rígida de 3 parámetros e intercepción del error esperado.
# ================================================================================

print("\n--- 🚀 [PASO 1]: Intento de Invocación con Exceso de Argumentos (TypeError) ---")


def print_classroom_students_fixed(student1: str, student2: str, student3: str) -> None:
    """Imprime los nombres de los estudiantes de una clase (requiere exactamente 3).

    Args:
        student1 (str): El nombre del primer estudiante.
        student2 (str): El nombre del segundo estudiante.
        student3 (str): El nombre del tercer estudiante.
    """
    print(f"  🔹 Estudiantes en clase: {student1}, {student2}, {student3}")


# 📋 Nombres de prueba (4 estudiantes)
name_1: str = "Lidia"
name_2: str = "Victor"
name_3: str = "Alberto"
name_4: str = "David"

print(f"📋 Intentando enviar 4 argumentos a una función que espera 3:")
print("-" * 50)

# 🔌 Intentamos llamar a la función fija con 4 argumentos para capturar el error
try:
    print_classroom_students_fixed(name_1, name_2, name_3, name_4)
except TypeError as error_message:
    print(f"❌ [ERROR DE TIPO CAPTURADO]: {error_message}")
````

## 📊 Ejemplo 2: Solución Dinámica mediante `*args`

### 💡 Flexibilidad y Recorredor Iterativo:
Al usar `*args`, la función acepta cualquier cantidad de nombres y podemos recorrer la tupla recibida o formatearla dinámicamente usando métodos de cadena como `join()`.

````python
# ================================================================================
# 📊 2. SOLUCIÓN CON *args E ITERACIÓN DE ELEMENTOS
# ================================================================================
print("\n--- 📊 [PASO 2]: Función Dinámica con *args ---")


def print_classroom_students(*students: str) -> None:
    """Imprime los nombres de cualquier cantidad de estudiantes recibidos.

    Args:
        *students (str): Tupla con los nombres de las personas de la clase.
    """
    # ⚙️ Muestra del contenido empaquetado en la Tupla
    print(f"  🔍 Tupla *students recibida: {students}")
    
    if not students:
        print("  ⚠️ No se han registrado estudiantes en esta sesión.")
        return

    # ⚙️ Formateo limpio para presentación
    formatted_names: str = ", ".join(students)
    print(f"  ✅ Estudiantes de la clase ({len(students)} registrados): {formatted_names}")


print("📋 Invocando la función dinámica con diferentes cantidades de argumentos:")
print("-" * 50)

# 🔌 Invocación 1 con 4 argumentos
print("👉 Invocación con 4 argumentos:")
print_classroom_students(name_1, name_2, name_3, name_4)

print("\n👉 Invocación con 2 argumentos:")
# 🔌 Invocación 2 con 2 argumentos
print_classroom_students("Elena", "Carlos")
````

## 🛒 Ejemplo 3: Casos Prácticos de Aplicación para `*args`

````python
# ================================================================================
# 🧪 3. CASOS PRÁCTICOS ADICIONALES: RUTAS DE ARCHIVOS Y PROMEDIOS
# ================================================================================
print("\n--- 🧪 [PASO 3]: Construcción Dinámica de Rutas y Métricas ---")


# ⚙️ Caso A: Constructor de Rutas de Sistema de Archivos
def build_file_path(*path_segments: str) -> str:
    """Une una cantidad variable de segmentos de carpeta en una ruta unificada.

    Args:
        *path_segments (str): Segmentos de la ruta de archivo.

    Returns:
        str: Ruta consolidada separada por '/'.
    """
    return "/".join(segment.strip("/") for segment in path_segments)


# ⚙️ Caso B: Promedio de Notas de Evaluaciones Semestrales
def calculate_grade_average(student_name: str, *grades: float) -> str:
    """Calcula el promedio de calificaciones arbitrarias para un estudiante.

    Args:
        student_name (str): Nombre del estudiante (parámetro posicional obligatorio).
        *grades (float): Serie variable de notas obtenidas.

    Returns:
        str: Mensaje con el desempeño general.
    """
    if not grades:
        return f"Estudiante {student_name}: Sin notas registradas."
        
    average: float = sum(grades) / len(grades)
    return f"Estudiante {student_name} | Evaluaciones: {len(grades)} | Promedio: {average:.2f}"


print("📋 Probando constructor de rutas dinámico:")
system_path: str = build_file_path("var", "log", "app", "errors.log")
print(f"  🔹 Ruta generada: /{system_path}")

print("\n📋 Probando cálculo de notas con cantidad variable de evaluaciones:")
report_1: str = calculate_grade_average("Ana Gomez", 8.5, 9.0, 7.5, 10.0)
report_2: str = calculate_grade_average("Luis Perez", 6.0, 7.5)

print(f"  ✅ {report_1}")
print(f"  ✅ {report_2}")

# --------------------------------------------------------------------------------
# 🏁 [PASO 4]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Módulo de argumentos posicionales *args finalizado con éxito.")
print("=" * 80)
````
