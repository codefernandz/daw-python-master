---

### 💡 Paradigma Funcional sobre Iterables

Las funciones `map()`, `filter()` y `reduce()` permiten procesar colecciones de datos (`list`, `tuple`, `set`, `dict`) sin necesidad de escribir bucles explícitos `for` o `while`.

#### 📐 Resumen de Funciones:

1. 🗺️ **`map(function, iterable)`:** Aplica una función transformadora a **cada elemento** del iterable. Devuelve un objeto iterador con el mismo número de elementos que el original.
2. 🔍 **`filter(function, iterable)`:** Evalúa cada elemento mediante una función con retorno booleano (*predicado*). Retorna un iterador únicamente con los elementos que evalúan como `True`.
3. 📉 **`reduce(function, iterable[, initializer])`:** Aplica repetidamente una función binaria (2 argumentos) de izquierda a derecha a los elementos del iterable para **reducir la colección a un único valor acumulado**.

> ⚠️ **Nota de Módulos:** `map()` y `filter()` son funciones integradas (*built-in*). `reduce()` debe importarse explícitamente desde el módulo estándar `functools`.

---

### 📊 Comparativa de Comportamiento:

| Función | Entrada | Salida | Evaluación | Importación |
| :--- | :--- | :--- | :--- | :--- |
| **`map()`** | $N$ elementos | $N$ elementos transformados | Perezosa (*Iterator*) | Built-in (no requiere) |
| **`filter()`** | $N$ elementos | $\le N$ elementos filtrados | Perezosa (*Iterator*) | Built-in (no requiere) |
| **`reduce()`** | $N$ elementos | **1 único valor acumulado** | Inmediata | `from functools import reduce` |

```python
from functools import reduce
from typing import Any, Callable

# 🛰️ [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Funciones - map(), filter() y reduce()")
print("=" * 80)

# ================================================================================
# 🧪 1. MAP(): TRANSFORMACIÓN ELEMENTO A ELEMENTO
# ================================================================================
# Demostración de conversión masiva de temperaturas de Celsius a Fahrenheit.
# ================================================================================

print("\n--- 🚀 [PASO 1]: Transformación de Colecciones con map() ---")

celsius_temperatures: list[float] = [0.0, 10.0, 20.0, 30.0, 40.0]

# ⚙️ Definición de la función de transformación
celsius_to_fahrenheit: Callable[[float], float] = lambda c: round((c * 9/5) + 32, 1)

print(f"📋 Lista original de temperaturas (°C): {celsius_temperatures}")
print("-" * 50)

# 🔌 [1] Aplicación de map() (Devuelve un objeto map_iterator)
fahrenheit_iterator: map[float] = map(celsius_to_fahrenheit, celsius_temperatures)

print(f"  🔍 Objeto retornado por map(): {fahrenheit_iterator} (Tipo: {type(fahrenheit_iterator)})")

# 🔌 [2] Conversión explícita del iterador a lista
fahrenheit_list: list[float] = list(fahrenheit_iterator)
print(f"  ✅ Resultado convertido a lista (°F): {fahrenheit_list}")
```

## 📊 Ejemplo 2: `filter()` — Selección de Datos mediante Predicados

### 💡 Evaluación Booleana:
`filter()` mantiene únicamente los registros donde la función de prueba retorne `True`.

```python
# ================================================================================
# 📊 2. FILTER(): SELECCIÓN SEGÚN CONDICIÓN BOOLEANA
# ================================================================================
print("\n--- 📊 [PASO 2]: Filtrado de Datos con filter() ---")

# 📋 Colección de puntuaciones de estudiantes
student_scores: list[int] = [45, 82, 91, 55, 68, 38, 100, 74]

# ⚙️ Predicado lambda que verifica si la nota es aprobatoria (>= 60)
is_passing_grade: Callable[[int], bool] = lambda score: score >= 60

print(f"📋 Puntuaciones originales: {student_scores}")
print("-" * 50)

# 🔌 Aplicación de filter() y conversión a lista
passing_scores: list[int] = list(filter(is_passing_grade, student_scores))

print(f"  ✅ Puntuaciones aprobatorias (>= 60): {passing_scores}")
print(f"  🔹 Total aprobados: {len(passing_scores)} de {len(student_scores)}")
```

## 📉 Ejemplo 3: `reduce()` — Acumulación e Integración Masiva

### 💡 Agregación Progresiva:
`reduce()` toma los dos primeros elementos del iterable, aplica la función, toma el resultado y el tercer elemento, y así sucesivamente hasta reducir la lista a un valor final.

```python
# ================================================================================
# 🧪 3. REDUCE(): ACUMULACIÓN Y CÁLCULO DEL MÁXIMO / MULTIPLICACIÓN
# ================================================================================
print("\n--- 🧪 [PASO 3]: Reducción de Colecciones a un Único Valor ---")

financial_transactions: list[float] = [150.50, 300.00, 45.25, 1200.00, 80.10]

print(f"📋 Transacciones financieras: {financial_transactions}")
print("-" * 50)

# ⚙️ [1] Suma acumulada de transacciones usando reduce()
total_revenue: float = reduce(
    lambda accumulator, current_val: accumulator + current_val, 
    financial_transactions
)

# ⚙️ [2] Búsqueda del valor máximo usando reduce()
max_transaction: float = reduce(
    lambda acc_max, current: current if current > acc_max else acc_max, 
    financial_transactions
)

print(f"  ✅ Suma total acumulada (reduce):   ${total_revenue:.2f}")
print(f"  ✅ Valor máximo encontrado (reduce): ${max_transaction:.2f}")
```

## 🛒 Ejercicios Prácticos Integrados

### 🧪 Pipeline Completo (`filter` ➡️ `map` ➡️ `reduce`)
Demostración de un flujo real de procesamiento de datos:
1. **Filtrar** productos en stock (`filter`).
2. **Transformar** calculando el valor total del inventario por producto (`map`).
3. **Acumular** el valor total del almacén (`reduce`).

```python
# ================================================================================
# 🧪 4. EJERCICIO INTEGRADO: PIPELINE DE PROCESAMIENTO EN CADENA
# ================================================================================
print("\n--- 🧪 [PASO 4]: Pipeline Combinado (Filter -> Map -> Reduce) ---")

# 📋 Inventario de productos (Nombre, Precio unitario, Stock disponible)
inventory: list[dict[str, Any]] = [
    {"name": "Laptop", "price": 1000.0, "stock": 5},
    {"name": "Mouse", "price": 25.0, "stock": 0},      # Sin stock
    {"name": "Keyboard", "price": 50.0, "stock": 10},
    {"name": "Monitor", "price": 300.0, "stock": 0},   # Sin stock
    {"name": "Webcam", "price": 80.0, "stock": 15}
]

print("📋 Inventario completo del almacén:")
for item in inventory:
    print(f"  • {item['name']}: ${item['price']} | Stock: {item['stock']}")

print("-" * 50)

# 🔌 [1] Paso 1 (Filter): Retener únicamente productos con stock > 0
in_stock_items: list[dict[str, Any]] = list(
    filter(lambda item: item["stock"] > 0, inventory)
)

# 🔌 [2] Paso 2 (Map): Calcular el valor total de inventario por producto (precio * stock)
item_values: list[float] = list(
    map(lambda item: item["price"] * item["stock"], in_stock_items)
)

# 🔌 [3] Paso 3 (Reduce): Sumar el valor total de todo el stock disponible
total_inventory_valuation: float = reduce(
    lambda total, val: total + val, 
    item_values, 
    0.0  # Inicializador
)

print(f"  🔹 Productos disponibles con stock: {len(in_stock_items)}")
print(f"  🔹 Valoración por producto en stock: {item_values}")
print(f"  ✅ Valor total del inventario activo: ${total_inventory_valuation:.2f}")

# --------------------------------------------------------------------------------
# 🏁 [PASO 5]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Cuaderno de map(), filter() y reduce() finalizado con éxito.")
print("=" * 80)
```
