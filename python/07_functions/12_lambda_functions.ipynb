### 💡 Anatomía y Sintaxis de las Funciones Lambda

Las **funciones lambda** son funciones pequeñas y anónimas (sin nombre explícito) que se definen en **una sola línea** utilizando la palabra clave `lambda`.

#### 📐 Sintaxis Básica:
```python
lambda argument_1, argument_2, ...: single_expression
```

#### 🏗️ Características Principales:

1. ⚡ **Expresión Única:** El cuerpo de una función lambda solo puede contener una única expresión. No se admiten bucles, declaraciones multilínea ni la palabra clave `return` (el retorno es implícito).
2. 🧹 **Uso Efímero / Anónimo:** Son ideales para pasarse como argumentos a otras funciones sin necesidad de contaminar el espacio de nombres global con funciones temporales de un solo uso.
3. 💾 **Eficiencia de Código:** Permiten escribir código más sucinto cuando se realizan transformaciones rápidas o filtrado de datos.

---

### ⚖️ Comparativa de Sintaxis:

```python
# 1. Definición Formal (def)
def divide_by_three(number: float) -> float:
    return number / 3.0

# 2. Expresión Anónima (lambda)
divide_by_three_lambda = lambda number: number / 3.0
```

````python
from typing import Callable, Any

# 🛰️ [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Funciones - Expresiones Anónimas (Lambda)")
print("=" * 80)

# ================================================================================
# 🧪 1. COMPARACIÓN DIRECTA: FUNCIÓN TRADICIONAL VS LAMBDA
# ================================================================================
# Demostración del cálculo de división entre 3 usando ambas aproximaciones.
# ================================================================================

print("\n--- 🚀 [PASO 1]: Comparación entre 'def' y 'lambda' ---")


# ⚙️ [1] Declaración tradicional con 'def'
def divide_by_three_standard(number: float) -> float:
    """Divide un número entre 3 utilizando una función estándar.

    Args:
        number (float): Número a dividir.

    Returns:
        float: Resultado de la división.
    """
    return number / 3.0


# ⚙️ [2] Declaración anónima con 'lambda' (Asignada a una variable con Type Hint Callable)
divide_by_three_lambda: Callable[[float], float] = lambda x: x / 3.0


input_value: float = 12.0

print("📋 Evaluando ambas funciones con el valor 12.0:")
print("-" * 50)

# 🔌 Invocación de la función estándar
result_standard: float = divide_by_three_standard(input_value)
print(f"  🔹 Resultado usando 'def':    {result_standard}")

# 🔌 Invocación de la función lambda
result_lambda: float = divide_by_three_lambda(input_value)
print(f"  ✅ Resultado usando 'lambda': {result_lambda}")
````

## 📊 Ejercicios Prácticos Adicionales

### 🧪 Ejercicio 1: Ordenamiento Personalizado de Colecciones con `sorted()`
Uno de los usos más habituales de `lambda` es servir como clave (`key`) de extracción en funciones de ordenación sobre estructuras de datos compuestas (listas de diccionarios o tuplas).

````python
# ================================================================================
# 📊 2. EJERCICIO 1: ORDENAMIENTO DE ESTRUCTURAS COMPLEJAS (sorted)
# ================================================================================
print("\n--- 📊 [PASO 2]: Ordenamiento de Diccionarios mediante Lambda Key ---")

# 📋 Lista de productos con nombre y precio
products_catalog: list[dict[str, Any]] = [
    {"name": "Laptop", "price": 1200.0, "stock": 5},
    {"name": "Mouse", "price": 25.5, "stock": 50},
    {"name": "Monitor", "price": 300.0, "stock": 12},
    {"name": "Keyboard", "price": 75.0, "stock": 25}
]

print("📋 Catálogo original de productos:")
for item in products_catalog:
    print(f"  • {item}")

print("-" * 50)

# ⚙️ Ordenar de menor a mayor precio utilizando lambda como extractor de clave
sorted_by_price: list[dict[str, Any]] = sorted(
    products_catalog, 
    key=lambda product: product["price"]
)

print("✅ Productos ordenados por precio (Ascendente):")
for item in sorted_by_price:
    print(f"  🔹 {item['name']}: ${item['price']}")
````

### 🧪 Ejercicio 2: Transformación y Filtrado de Datos con `map()` y `filter()`
Las funciones integradas `map()` (para transformar elementos) y `filter()` (para seleccionar elementos según una condición booleana) se combinan frecuentemente con expresiones `lambda`.

````python
# ================================================================================
# 🧪 3. EJERCICIO 2: TRANSFORMACIÓN CON map() Y FILTRADO CON filter()
# ================================================================================
print("\n--- 🧪 [PASO 3]: Uso de Lambda en Pipelines con map() y filter() ---")

raw_temperatures_celsius: list[float] = [0.0, 15.5, 22.0, 35.0, 40.2]

print(f"📋 Temperaturas originales (°C): {raw_temperatures_celsius}")
print("-" * 50)

# ⚙️ [1] Transformación con map(): Convertir Celsius a Fahrenheit [F = (C * 9/5) + 32]
temperatures_fahrenheit: list[float] = list(
    map(lambda temp_c: round((temp_c * 9/5) + 32, 1), raw_temperatures_celsius)
)
print(f"  🔹 Conversión a Fahrenheit con map(): {temperatures_fahrenheit}")

# ⚙️ [2] Filtrado con filter(): Seleccionar únicamente temperaturas consideradas "calurosas" (> 80 °F)
hot_temperatures: list[float] = list(
    filter(lambda temp_f: temp_f > 80.0, temperatures_fahrenheit)
)
print(f"  ✅ Temperaturas calurosas (> 80 °F) con filter(): {hot_temperatures}")

# --------------------------------------------------------------------------------
# 🏁 [PASO 4]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Cuaderno de funciones lambda finalizado con éxito.")
print("=" * 80)
````
