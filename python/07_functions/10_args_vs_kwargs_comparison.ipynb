---

### 💡 Análisis Comparativo: Ventajas de `**kwargs` frente a `*args`

Aunque ambos mecanismos aportan flexibilidad en el número de parámetros recibidos, existen diferencias clave en su usabilidad y mantenibilidad:

1. 📖 **Claridad y Legibilidad:** Al emplear `**kwargs`, cada valor está asociado a una clave explícita (`key=value`), lo que auto-documenta el propósito de cada dato en la llamada a la función.
2. 🔀 **Independencia de Posición:** Los argumentos en `*args` dependen **estrictamente de la posición/orden** en el que se envían. En `**kwargs`, el orden es completamente irrelevante ya que la extracción se realiza por clave.
3. 🛠️ **Evolución sin Romper Código (Backward Compatibility):** Permite añadir nuevos atributos opcionales a una función sin modificar la llamada de sistemas o código existente.

---

### 📋 Matriz de Decisión: ¿Cuándo usar cada uno?

| Criterio | Usar `*args` cuando... | Usar `**kwargs` cuando... |
| :--- | :--- | :--- |
| **Estructura Interna** | Se recibe como **Tupla (`tuple`)** | Se recibe como **Diccionario (`dict`)** |
| **Identificación** | Los elementos son anónimos o implícitos | Cada elemento requiere un **nombre/clave explícita** |
| **Importancia del Orden** | **Sí**, el orden secuencial importa | **No**, el acceso se realiza mediante claves |
| **Casos de Uso Típicos** | Listas de valores, operaciones matemáticas (`sum`), colecciones homogéneas | Opciones de configuración, filtros de búsqueda, metadatos, perfilado |

```python
from typing import Any

# 🛰️ [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Comparativa y Selección (*args vs **kwargs)")
print("=" * 80)

# ================================================================================
# 🧪 1. DEMOSTRACIÓN DE SENSIBILIDAD AL ORDEN (*args VS **kwargs)
# ================================================================================
# Muestra cómo un cambio de orden altera el resultado en *args, mientras que
# en **kwargs el comportamiento permanece inalterado.
# ================================================================================

print("\n--- 🚀 [PASO 1]: Impacto del Orden de Argumentos (*args vs **kwargs) ---")


# ⚙️ Caso A: Función basada en *args (Dependiente de Posición)
def format_user_address_positional(*address_parts: str) -> str:
    """Construye una dirección basada en el orden estricto de los argumentos recibidos.

    Args:
        *address_parts (str): Calle, Ciudad, Código Postal (en orden esperado).

    Returns:
        str: Dirección formateada.
    """
    return " | ".join(address_parts)


# ⚙️ Caso B: Función basada en **kwargs (Independiente del Orden)
def format_user_address_named(**address_details: str) -> str:
    """Construye una dirección mediante claves explícitas sin importar el orden.

    Args:
        **address_details (str): Claves esperadas como 'street', 'city', 'zip_code'.

    Returns:
        str: Dirección formateada.
    """
    street: str = address_details.get("street", "N/A")
    city: str = address_details.get("city", "N/A")
    zip_code: str = address_details.get("zip_code", "N/A")
    
    return f"Street: {street} | City: {city} | Zip: {zip_code}"


print("📋 Demostración con *args (Sensible a la posición):")
print("-" * 50)

# 🔌 [1] Orden correcto
addr_correct: str = format_user_address_positional("Main St 123", "Madrid", "28001")
print(f"✅ [ORDEN CORRECTO]: {addr_correct}")

# ❌ [2] Orden invertido (Provoca inconsistencia semántica)
addr_inverted: str = format_user_address_positional("Madrid", "28001", "Main St 123")
print(f"❌ [ORDEN INVERTIDO]:  {addr_inverted}\n")


print("📋 Demostración con **kwargs (Invariante al orden):")
print("-" * 50)

# 🔌 Invocación con orden de claves totalmente alternado
addr_kw_1: str = format_user_address_named(street="Main St 123", city="Madrid", zip_code="28001")
addr_kw_2: str = format_user_address_named(zip_code="28001", city="Madrid", street="Main St 123")

print(f"✅ [LLAMADA 1]: {addr_kw_1}")
print(f"✅ [LLAMADA 2]: {addr_kw_2}")
```

## 📊 Ejemplo 2: Integración Armónica de `*args` y `**kwargs`

### 💡 Caso Real: Procesador General de Transformación de Datos
Demostración de cómo combinar ambos tipos de parámetros en una misma función para procesar colecciones numéricas (`*args`) aplicando flags de configuración opcionales (`**kwargs`).

```python
# ================================================================================
# 📊 2. COMBINACIÓN DE *args (DATOS) Y **kwargs (OPCIONES DE CONFIGURACIÓN)
# ================================================================================
print("\n--- 📊 [PASO 2]: Procesamiento de Datos con Configuración Flexible ---")


def process_dataset(*data_points: float, **processing_options: Any) -> dict[str, Any]:
    """Procesa una serie de puntos de datos aplicando transformaciones configurables.

    Args:
        *data_points (float): Serie variable de valores numéricos a procesar.
        **processing_options (Any): Opciones como 'scale_factor', 'round_digits', 'label'.

    Returns:
        dict[str, Any]: Resultado consolidado del procesamiento.
    """
    if not data_points:
        return {"error": "No data provided", "results": []}

    # ⚙️ Extracción de opciones de configuración con valores por defecto con .get()
    scale_factor: float = processing_options.get("scale_factor", 1.0)
    round_digits: int = processing_options.get("round_digits", 2)
    dataset_label: str = processing_options.get("label", "DEFAULT_DATASET")

    # ⚙️ Transformación de los datos posicionales
    transformed_points: list[float] = [
        round(val * scale_factor, round_digits) for val in data_points
    ]

    return {
        "label": dataset_label,
        "count": len(transformed_points),
        "total_sum": round(sum(transformed_points), round_digits),
        "data": transformed_points
    }


print("📋 Procesando datos con configuraciones variables:")
print("-" * 50)

# 🔌 [1] Procesamiento simple usando valores por defecto
result_default: dict[str, Any] = process_dataset(10.255, 20.128, 30.891)
print(f"✅ [EJECUCIÓN POR DEFECTO]: {result_default}\n")

# 🔌 [2] Procesamiento personalizado pasando *args y **kwargs
result_custom: dict[str, Any] = process_dataset(
    10.255, 20.128, 30.891, 40.555,  # *args (Puntos de datos)
    scale_factor=1.5,                # **kwargs (Factor de escala)
    round_digits=1,                  # **kwargs (Redondeo)
    label="FINANCIAL_METRICS"        # **kwargs (Etiqueta)
)
print(f"✅ [EJECUCIÓN PERSONALIZADA]: {result_custom}")

# --------------------------------------------------------------------------------
# 🏁 [PASO 3]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Cuaderno comparativo finalizado con éxito.")
print("=" * 80)
```
