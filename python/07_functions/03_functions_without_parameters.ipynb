| 🛠️ **Conceptos** | `Parameterless Functions`, `Side Effects`, `Implicit Return`, `NoneType`, `Function Return Value` |

---

### 💡 Explicación del Comportamiento

1. **Funciones sin Parámetros:** Ocurre cuando los paréntesis `()` en la declaración `def` están vacíos. Indican que la función no requiere datos externos para ejecutar sus instrucciones.
2. **Efectos Secundarios (*Side Effects*):** Una función que solo ejecuta acciones (como imprimir en consola con `print()`, escribir en un archivo o modificar un estado global) pero no incluye un `return`, opera mediante efectos secundarios.
3. **Retorno Implícito (`None`):** En Python, si una función no tiene una sentencia `return` explícita (o ejecuta un `return` sin argumentos), **siempre devuelve `None`** (un objeto especial de tipo `NoneType`).

#### 📐 Esquema de Firma de Tipo:

```python
def execution_step() -> None:
    """La anotación -> None indica explícitamente que no hay valor de retorno."""
    ...

````python
from typing import Any

# 🛰️ [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Funciones - Sin Parámetros y Retorno None")
print("=" * 80)

# ================================================================================
# 🧪 1. DECLARACIÓN E INVOCACIÓN DE UNA FUNCIÓN SIN PARÁMETROS
# ================================================================================
# Definición de función con variable local e impresión directa en consola.
# ================================================================================

print("\n--- 🚀 [PASO 1]: Ejecución de Función con Efecto Secundario (print) ---")


def execute_parameterless_function() -> None:
    """Asigna un valor a una variable local e imprime un mensaje en consola.

    Returns:
        None: Esta función no devuelve ningún valor explícito.
    """
    # ⚙️ Asignación de variable local
    local_value: int = 2
    
    # ⚙️ Impresión directa en pantalla (Side Effect)
    print(f"  🔹 [DENTRO DE LA FUNCIÓN]: El valor de 'local_value' es: {local_value}")


print("📋 Invocando 'execute_parameterless_function()':")
print("-" * 50)

# 🔌 Invocación simple de la función
execute_parameterless_function()
````

## 🔍 Inspección del Valor Devuelto (`NoneType`)

### 💡 ¿Qué sucede si asignamos el resultado a una variable?
Al intentar almacenar el resultado de la llamada a una función sin `return`, la variable receptora capturará el objeto `None`.

````python
# ================================================================================
# 📊 2. EVALUACIÓN Y VERIFICACIÓN DEL RETORNO IMPLÍCITO (None)
# ================================================================================
print("\n--- 📊 [PASO 2]: Captura del Valor Devuelto e Inspección de Tipo ---")

print("📋 Capturando la salida de la función en la variable 'returned_value':")
print("-" * 50)

# 🔌 [1] Asignación de la invocación a una variable
returned_value: None = execute_parameterless_function()

print("\n📋 [RESULTADO DE LA ASIGNACIÓN]:")
print(f"  ✅ Valor capturado en 'returned_value': {returned_value}")
print(f"  🔍 Tipo de dato de 'returned_value': {type(returned_value)}")

# --------------------------------------------------------------------------------
# 🏁 [PASO 3]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Módulo de funciones sin parámetros finalizado con éxito.")
print("=" * 80)
````
