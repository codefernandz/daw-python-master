### 💡 Paso de Argumentos Múltiples

Cuando una función declara varios parámetros separados por comas `,` dentro de los paréntesis `()`, la invocación de la misma requiere enviar los argumentos correspondientes.

#### 📐 Tipos de Paso de Argumentos:
1. **Argumentos Posicionales:** Se asignan a los parámetros en el **mismo orden estricto** en el que fueron declarados.
2. **Argumentos Nombrados (*Keyword Arguments*):** Se especifican indicando explícitamente `nombre_parametro=valor`. Esto permite alterar el orden de envío y mejora la legibilidad.

```python
def calculate_metrics(base: float, multiplier: float) -> float:
    return base * multiplier

# Posicional (mismo orden)
result1 = calculate_metrics(10.0, 2.5)

# Nombrado / Keyword (orden explícito)
result2 = calculate_metrics(multiplier=2.5, base=10.0)

````python
from typing import Any

# 🛰️ [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Funciones - Múltiples Parámetros")
print("=" * 80)

# ================================================================================
# 🧪 1. EJEMPLO PRINCIPAL: CÁLCULO DE ÁREA GEOMÉTRICA (2 PARÁMETROS)
# ================================================================================
# Cálculo de la superficie de un rectángulo a partir de su base y su altura.
# ================================================================================

print("\n--- 🚀 [PASO 1]: Cálculo Geométrico con 2 Parámetros Numéricos ---")


def calculate_rectangle_area(base: float, height: float) -> float:
    """Calcula el área de un rectángulo dada su base y su altura.

    Args:
        base (float): La longitud de la base del rectángulo.
        height (float): La altura del rectángulo.

    Returns:
        float: El área total calculada (base * height).
    """
    return base * height


# 📋 Dimensiones de prueba
rect_base: float = 5.0
rect_height: float = 3.0

print(f"📋 Evaluando Rectángulo: Base = {rect_base}m | Altura = {rect_height}m")
print("-" * 50)

# 🔌 [1] Invocación Posicional
area_positional: float = calculate_rectangle_area(rect_base, rect_height)

# 🔌 [2] Invocación Nombrada (Keyword Arguments)
area_keyword: float = calculate_rectangle_area(height=rect_height, base=rect_base)

print(f"✅ [RESULTADO POSICIONAL]: Área = {area_positional} m²")
print(f"✅ [RESULTADO KEYWORD]:    Área = {area_keyword} m²")
````

## 📊 Ejemplo 2: Cálculo de Calificación Ponderada (3 Parámetros Numéricos)

### 💡 Ponderación de Evaluaciones:
Procesar las notas de un estudiante aplicando distintos porcentajes de peso sobre cada evaluación (Examen, Tareas, Proyecto).

````python
# ================================================================================
# 📊 2. CASO ADICIONAL 1: PROMEDIO PONDERADO DE ESTUDIANTES
# ================================================================================
print("\n--- 📊 [PASO 2]: Cálculo de Nota Final con Ponderaciones ---")


def calculate_weighted_grade(
    exam_score: float, 
    homework_score: float, 
    project_score: float
) -> float:
    """Calcula la nota final aplicando pesos del 50% al examen, 20% a tareas y 30% al proyecto.

    Args:
        exam_score (float): Nota del examen (0 a 100).
        homework_score (float): Nota de tareas (0 a 100).
        project_score (float): Nota del proyecto final (0 a 100).

    Returns:
        float: La nota promedio ponderada final.
    """
    # ⚙️ Aplicación de pesos
    weighted_exam: float = exam_score * 0.50
    weighted_homework: float = homework_score * 0.20
    weighted_project: float = project_score * 0.30
    
    final_grade: float = weighted_exam + weighted_homework + weighted_project
    return final_grade


# 📋 Notas del estudiante
student_exam: float = 85.0
student_homework: float = 90.0
student_project: float = 95.0

print(f"📋 Notas: Examen={student_exam} | Tareas={student_homework} | Proyecto={student_project}")
print("-" * 50)

# 🔌 Invocación con argumentos nombrados para máxima claridad
final_score: float = calculate_weighted_grade(
    exam_score=student_exam,
    homework_score=student_homework,
    project_score=student_project
)

print(f"✅ [NOTA FINAL PONDERADA]: {final_score:.2f} / 100.00")
````

## 🔤 Ejemplo 3: Generador de Correos Corporativos (3 Parámetros de Texto/Bool)

````python
# ================================================================================
# 🧪 3. CASO ADICIONAL 2: PROCESAMIENTO Y GENERACIÓN DE STRINGS
# ================================================================================
print("\n--- 🧪 [PASO 3]: Formateador de Correos de Empresa ---")


def generate_company_email(
    first_name: str, 
    last_name: str, 
    domain: str, 
    include_department: bool = False,
    department_code: str = ""
) -> str:
    """Genera una dirección de correo corporativo estandarizada en minúsculas.

    Args:
        first_name (str): Nombre del empleado.
        last_name (str): Apellido del empleado.
        domain (str): Dominio de la empresa (ej. 'techcorp.com').
        include_department (bool): Si es True, añade el prefijo del departamento.
        department_code (str): Código del departamento (ej. 'dev', 'hr').

    Returns:
        str: La dirección de correo electrónico formateada.
    """
    # ⚙️ Normalización a minúsculas y eliminación de espacios
    clean_first: str = first_name.strip().lower()
    clean_last: str = last_name.strip().lower()
    clean_domain: str = domain.strip().lower()
    
    if include_department and department_code:
        clean_dept: str = department_code.strip().lower()
        return f"{clean_first}.{clean_last}.{clean_dept}@{clean_domain}"
    
    return f"{clean_first}.{clean_last}@{clean_domain}"


# 📋 Datos de empleados
emp1_email: str = generate_company_email("Carlos", "Mendoza", "innovate.com")
emp2_email: str = generate_company_email(
    first_name="Ana", 
    last_name="Gomez", 
    domain="innovate.com", 
    include_department=True, 
    department_code="dev"
)

print("📋 Correos generados:")
print(f"  🔹 Empleado Estándar:  {emp1_email}")
print(f"  🔹 Empleado Depto:     {emp2_email}")
````

## 📈 Ejemplo 4: Calculadora de Interés Compuesto (4 Parámetros Financieros)

````python
# ================================================================================
# 📊 4. CASO ADICIONAL 3: FÓRMULA FINANCIERA MULTI-PARÁMETRO
# ================================================================================
print("\n--- 📊 [PASO 4]: Proyección Financiera de Interés Compuesto ---")


def calculate_compound_interest(
    principal: float, 
    annual_rate: float, 
    compounds_per_year: int, 
    years: int
) -> float:
    """Calcula el monto final acumulado usando la fórmula de interés compuesto.

    Fórmula: A = P * (1 + r/n)**(n*t)

    Args:
        principal (float): Capital inicial invertido.
        annual_rate (float): Tasa de interés anual expresada en decimales (ej. 0.05 para 5%).
        compounds_per_year (int): Frecuencia de capitalización al año (ej. 12 para mensual).
        years (int): Tiempo de la inversión expresado en años.

    Returns:
        float: El monto total acumulado (capital + intereses).
    """
    rate_per_period: float = annual_rate / compounds_per_year
    total_periods: int = compounds_per_year * years
    
    final_amount: float = principal * ((1 + rate_per_period) ** total_periods)
    return final_amount


# 📋 Parámetros de la inversión
initial_capital: float = 10000.00
interest_rate: float = 0.07  # 7%
monthly_compounds: int = 12   # Capitalización mensual
investment_years: int = 5

print(f"📋 Capital: ${initial_capital:,.2f} | Tasa Anual: {interest_rate * 100}% | Plazo: {investment_years} años")
print("-" * 50)

# 🔌 Invocación
total_investment: float = calculate_compound_interest(
    principal=initial_capital,
    annual_rate=interest_rate,
    compounds_per_year=monthly_compounds,
    years=investment_years
)

print(f"✅ [MONTO FINAL ACUMULADO]: ${total_investment:,.2f}")

# --------------------------------------------------------------------------------
# 🏁 [PASO 5]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Módulo de múltiples parámetros finalizado con éxito.")
print("=" * 80)
````
