### 💡 1. Ámbito de Variables (*Variable Scope*)

El **scope** o ámbito determina en qué partes del programa es accesible una variable:

* **Variable Global:** Se define fuera de cualquier función. Su alcance abarca todo el archivo y puede ser leída desde cualquier función o bloque.
* **Variable Local:** Se define dentro del cuerpo de una función. Solo existe durante la ejecución de dicha función y no es accesible desde el exterior.
* **Sombreado de Variables (*Shadowing*):** Si se define una variable local con el mismo nombre que una global, la variable local "oculta" a la global **únicamente dentro del cuerpo de la función**.

---

### 💡 2. Parámetros vs. Argumentos

Aunque suelen usarse como sinónimos, existe una diferencia conceptual clara:

| Concepto | Definición | Ejemplo |
| :--- | :--- | :--- |
| **Parámetro** | Variable declarada en la **definición** de la función (`def`). Actúa como variable local. | `first_number`, `second_number` en `def add_numbers(first_number, second_number)` |
| **Argumento** | Valor real pasado a la función en el momento de la **invocación** o llamada. | `3`, `5` en `add_numbers(3, 5)` |

---

### 🛠️ Herramientas de Documentación: Extensión `autoDocstring` en VS Code

Para mantener un estándar profesional sin escribir manualmente toda la estructura del Docstring:

1. **Instalación:** Busca la extensión `autoDocstring - Python Docstring Generator` en el Marketplace de Visual Studio Code.
2. **Uso:**
   * Declara tu función con sus Type Hints: `def add_numbers(first_number: float, second_number: float) -> float:`
   * En la línea inmediata inferior a `def`, escribe triple comilla doble `"""` y presiona `Enter`.
   * La extensión generará automáticamente la plantilla formateada en formato Google, NumPy o Sphinx (configurable).

```python
from typing import Any

# 🛰️ [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Funciones - Ámbitos, Parámetros y Docstrings")
print("=" * 80)

# ================================================================================
# 🧪 1. DEMOSTRACIÓN DE ÁMBITO GLOBAL VS. LOCAL (SCOPE)
# ================================================================================
# Variables globales accesibles globalmente y locales confinadas a la función.
# ================================================================================

print("\n--- 🚀 [PASO 1]: Ámbito Global vs. Ámbito Local ---")

# 📋 Variable Global
global_app_name: str = "Python Logger App"
global_counter: int = 100


def demonstrate_variable_scope() -> None:
    """Muestra la diferencia entre el ámbito global y el local,

    incluyendo el efecto de variable shadowing.

    Returns:
        None
    """
    # ⚙️ Variable Local (solo existe dentro de esta función)
    local_execution_id: str = "EXEC_001"
    
    # ⚙️ Sombreado de variable (Shadowing): reemplaza temporalmente a global_counter
    global_counter: int = 5
    
    print("  [DENTRO DE LA FUNCIÓN]:")
    print(f"  🔹 Variable Global (Lectura): '{global_app_name}'")
    print(f"  🔹 Variable Local: '{local_execution_id}'")
    print(f"  🔹 Variable Local (Shadowing de global_counter): {global_counter}")


# 🔌 Invocación de la función
print("📋 Ejecutando la función demostrativa de Scope:")
demonstrate_variable_scope()

print("\n📋 [FUERA DE LA FUNCIÓN - ÁMBITO GLOBAL]:")
print(f"  ✅ Variable Global intacta: {global_counter}")

# ⚠️ Intentar acceder a local_execution_id aquí lanzaría un NameError:
try:
    print(local_execution_id)  # type: ignore
except NameError as scope_error:
    print(f"  ❌ [ERROR ESPERADO DE SCOPE]: {scope_error}")
```

## 🧮 Parámetros vs. Argumentos con Docstrings Formales

### 💡 Aplicación de Type Hints y Convención Google:
La siguiente función ilustra la firma explícita con parámetros, el cuerpo con variables locales y el valor devuelto mediante `return`.

```python
# ================================================================================
# 📊 2. PARÁMETROS, ARGUMENTOS Y DOCUMENTACIÓN COMPLETA
# ================================================================================
print("\n--- 📊 [PASO 2]: Operación Aritmética con Parámetros Definitivos ---")

# 🔌 [1] Definición de la función (first_number y second_number son PARÁMETROS)
def add_two_numbers(first_number: float, second_number: float) -> float:
    """Suma dos números reales y devuelve el resultado de la operación.

    Args:
        first_number (float): El primer número operando.
        second_number (float): El segundo número operando.

    Returns:
        float: La suma resultante de first_number + second_number.
    """
    # ⚙️ Variable local que almacena la suma
    addition_result: float = first_number + second_number
    return addition_result


# 📋 Declaración de variables que serviran como argumentos de entrada
input_a: float = 3.0
input_b: float = 5.0

print(f"📋 Preparando invocación con Argumentos: {input_a} y {input_b}")
print("-" * 50)

# 🔌 [2] Invocación de la función (3.0 y 5.0 pasan como ARGUMENTOS)
calculation_test: float = add_two_numbers(input_a, input_b)

print(f"✅ [ÉXITO]: El resultado devuelto por 'add_two_numbers' es: {calculation_test}")

# --------------------------------------------------------------------------------
# 🏁 [PASO 3]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Módulo de ámbitos y parámetros finalizado con éxito.")
print("=" * 80)
```
