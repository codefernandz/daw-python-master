### 💡 Anatomía de la Recursividad

Una función recursiva es aquella que **se llama a sí misma** durante su ejecución para resolver un problema dividiéndolo en subproblemas idénticos pero de menor tamaño.

#### 🏗️ Pilares Imprescindibles:
1. 🛑 **Caso Base (*Base Case*):** Es la condición de parada. Define cuándo la función debe dejar de llamarse a sí misma y devolver un resultado concreto. Sin él, la función entra en un bucle infinito lanzando un `RecursionError`.
2. 🔄 **Caso Recursivo (*Recursive Step*):** Es la rama donde la función se invoca a sí misma reduciendo progresivamente el tamaño del problema para acercarse hacia el caso base.

---

### ⚠️ La Pila de Llamadas (*Call Stack*)

Cada llamada recursiva se apila en la memoria del intérprete de Python. Por defecto, Python tiene un límite máximo de profundidad de recursión (generalmente 1,000 llamadas) para proteger el sistema.

```python
def example_recursion(n: int) -> int:
    if n <= 0:          # 🛑 Caso Base
        return 0
    return n + example_recursion(n - 1)  # 🔄 Caso Recursivo

````python
from typing import Any

# 🛰️ [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Funciones - Recursividad y Pila de Llamadas")
print("=" * 80)

# ================================================================================
# 🧪 1. EJEMPLO BASE: CUENTA ATRÁS RECURSIVA (COUNTDOWN)
# ================================================================================
# Demostración del flujo recursivo básico con control de caso base.
# ================================================================================

print("\n--- 🚀 [PASO 1]: Ejecución de Cuenta Atrás Recursiva ---")


def countdown(start_number: int) -> None:
    """Realiza una cuenta atrás recursiva a partir del número especificado.

    Args:
        start_number (int): Número entero positivo desde el cual iniciar la cuenta.

    Returns:
        None
    """
    # ⚙️ [1] Imprime el número actual del paso recursivo
    print(f"  ⏳ Conteo: {start_number}")

    # 🛑 [2] Caso Base: Detener la recursión al llegar a 1 o menos
    if start_number <= 1:
        print("  ✅ [CASO BASE ALCANZADO]: ¡Tiempo agotado!")
        return

    # 🔄 [3] Caso Recursivo: Reducción del problema y llamada autorreferencial
    next_number: int = start_number - 1
    countdown(next_number)


print("📋 Iniciando cuenta atrás desde 5:")
print("-" * 50)

# 🔌 Invocación de prueba
countdown(5)
````

### 🧪 Ejercicio 1: Cálculo del Factorial ($n!$)
El factorial de un número $n$ se define como $n! = n \times (n-1)!$, teniendo como **Caso Base** que $0! = 1$ y $1! = 1$.

````python
# ================================================================================
# 📊 2. EJERCICIO 1: FACTORIAL DE UN NÚMERO
# ================================================================================
print("\n--- 📊 [PASO 2]: Cálculo Recursivo de Factorial ---")


def calculate_factorial(number: int) -> int:
    """Calcula el factorial de un número entero no negativo de forma recursiva.

    Args:
        number (int): Número entero del cual calcular el factorial.

    Returns:
        int: Resultado del factorial de 'number'.

    Raises:
        ValueError: Si el número recibido es negativo.
    """
    if number < 0:
        raise ValueError("❌ El factorial no está definido para números negativos.")

    # 🛑 Caso Base: 0! = 1 y 1! = 1
    if number <= 1:
        return 1

    # 🔄 Caso Recursivo: n * factorial(n - 1)
    return number * calculate_factorial(number - 1)


print("📋 Calculando factoriales de varios números:")
print("-" * 50)

# 🔌 Pruebas con diferentes valores
for val in [0, 1, 5, 7]:
    result: int = calculate_factorial(val)
    print(f"  ✅ Factorial de {val}! = {result}")
````

### 🧪 Ejercicio 2: Sucesión de Fibonacci ($F_n$)
En la sucesión de Fibonacci, cada término es la suma de los dos anteriores: $F_n = F_{n-1} + F_{n-2}$.
* **Casos Base:** $F_0 = 0$, $F_1 = 1$.

````python
# ================================================================================
# 🧪 3. EJERCICIO 2: SUCESIÓN DE FIBONACCI
# ================================================================================
print("\n--- 🧪 [PASO 3]: Generación de Fibonacci Recursivo ---")


def calculate_fibonacci(position: int) -> int:
    """Devuelve el enésimo término de la sucesión de Fibonacci de forma recursiva.

    Args:
        position (int): Posición deseada en la secuencia (0-indexed).

    Returns:
        int: Valor del término de Fibonacci en la posición indicada.
    """
    # 🛑 Caso Base: Las posiciones 0 y 1 devuelven su propio valor
    if position <= 0:
        return 0
    if position == 1:
        return 1

    # 🔄 Caso Recursivo: Suma de las dos posiciones previas
    return calculate_fibonacci(position - 1) + calculate_fibonacci(position - 2)


print("📋 Calculando los primeros 8 términos de la serie Fibonacci:")
print("-" * 50)

fib_sequence: list[int] = [calculate_fibonacci(i) for i in range(8)]
print(f"  ✅ Secuencia obtenida (posiciones 0 a 7): {fib_sequence}")
````

### 🧪 Ejercicio 3: Suma Recursiva de Elementos en una Lista
Demostración de cómo consumir una lista quitando su primer elemento (`head`) y sumándolo al resto de la lista (`tail`) mediante recursividad.

````python
# ================================================================================
# 🧪 4. EJERCICIO 3: SUMA RECURSIVA DE UNA LISTA
# ================================================================================
print("\n--- 🧪 [PASO 4]: Suma de Lista mediante Descomposición (Head + Tail) ---")


def recursive_sum(numbers_list: list[float]) -> float:
    """Calcula la suma total de una lista numérica procesándola de manera recursiva.

    Args:
        numbers_list (list[float]): Lista de números flotantes o enteros.

    Returns:
        float: Suma total de los elementos.
    """
    # 🛑 Caso Base: Una lista vacía suma 0
    if not numbers_list:
        return 0.0

    # 🔄 Caso Recursivo: Primer elemento + suma del resto de la lista (slicing)
    head: float = numbers_list[0]
    tail: list[float] = numbers_list[1:]
    
    return head + recursive_sum(tail)


print("📋 Sumando elementos de una lista sin usar bucles `for` ni `sum()`:")
print("-" * 50)

sample_data: list[float] = [10.5, 20.0, 5.25, 4.25]
total_result: float = recursive_sum(sample_data)

print(f"  🔹 Lista original: {sample_data}")
print(f"  ✅ Suma calculada de forma recursiva: {total_result}")

# --------------------------------------------------------------------------------
# 🏁 [PASO 5]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Cuaderno de funciones recursivas finalizado con éxito.")
print("=" * 80)
````
