---

### 💡 Explicación de la Arquitectura
* `user_name: str`: Declaración de una variable de texto con tipado explícito.
* `user_age: int`: Variable numérica entera.
* `account_balance: float`: Variable de punto flotante para valores decimales.
* `is_active_user: bool`: Variable booleana de estado (`True` / `False`).
* `type(var)` e `id(var)`: Funciones integradas (*built-in*) de Python para inspeccionar el tipo de dato y la dirección de memoria asignada a la variable.

```python
# [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Definición de Variables y Tipos Primitivos")
print("=" * 80)

# ================================================================================
# 📢 1. DECLARACIÓN DE VARIABLES Y TYPE HINTS
# ================================================================================
# Asignación de valores a variables en inglés siguiendo el formato 'snake_case'.
# ================================================================================

# [1] Variables primitivas con anotación de tipos explícita
user_name: str = "Alex Dev"
user_age: int = 28
account_balance: float = 1450.75
is_active_user: bool = True

# [2] Salida formateada de los datos registrados
print("\n--- 🚀 [PASO 1]: Impresión de datos almacenados en variables ---")
print(f"✅ [USUARIO]:  {user_name}")
print(f"✅ [EDAD]:     {user_age} años")
print(f"✅ [SALDO]:    ${account_balance:.2f}")
print(f"✅ [ESTADO]:   {'Activo' if is_active_user else 'Inactivo'}")
```

## 🧪 Inspección de Tipos de Datos y Direcciones de Memoria

### 💡 Explicación de las Pruebas:
1. **Paso 2:** Uso de `type()` para confirmar la clase del objeto asignado a cada variable.
2. **Paso 3:** Demostración de **Reasignación Dinámica** (*Dynamic Typing*), mostrando cómo Python permite cambiar el tipo de dato de una variable en tiempo de ejecución y cómo cambia su identidad en memoria mediante `id()`.

```python
# --------------------------------------------------------------------------------
# 🔍 [PASO 2]: Inspección de Tipos con type()
# --------------------------------------------------------------------------------
print("\n--- 🔍 [PASO 2]: Verificación del tipo de dato de cada variable ---")

print(f"📋 Tipo de 'user_name':       {type(user_name)}")
print(f"📋 Tipo de 'user_age':        {type(user_age)}")
print(f"📋 Tipo de 'account_balance': {type(account_balance)}")
print(f"📋 Tipo de 'is_active_user':  {type(is_active_user)}")

# --------------------------------------------------------------------------------
# 🔄 [PASO 3]: Reasignación Dinámica e Inspección de Memoria id()
# --------------------------------------------------------------------------------
print("\n--- 🔄 [PASO 3]: Reasignación dinámica de tipo e identidad en memoria ---")

# Memory ID inicial
initial_memory_id: int = id(user_age)
print(f"📍 Identificador de memoria inicial de 'user_age' ({user_age}): {initial_memory_id}")

# Reasignamos user_age con una cadena de texto (cambio de tipo dinámico)
user_age = "Veintiocho"  # type: ignore
updated_memory_id: int = id(user_age)

print(f"⚠️ 'user_age' reasignado a texto: '{user_age}'")
print(f"📋 Nuevo tipo de 'user_age': {type(user_age)}")
print(f"📍 Nuevo identificador de memoria: {updated_memory_id}")

# --------------------------------------------------------------------------------
# 🏁 [PASO 4]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Módulo de definición de variables completado.")
print("=" * 80)
```
