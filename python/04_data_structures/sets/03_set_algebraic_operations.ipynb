| Propiedad | Detalle |
| :--- | :--- |
| 📍 **Módulo** | Fundamentos de Python |
| 🎯 **Objetivo** | Dominar las operaciones álgebraicas de conjuntos (`union`, `intersection`, `difference`, `symmetric_difference`) y métodos de evaluación relacional (`isdisjoint`, `issubset`, `issuperset`), utilizando tanto métodos explícitos como operadores sobrecargados (`\|`, `&`, `-`, `^`, `<=`, `>=`). |
| 🛠️ **Conceptos** | `Set Union ()`, `Intersection (&)`, `Difference (-)`, `Symmetric Difference (^)`, `Disjoint Sets`, `Subset (<=)`, `Superset (>=)` |

---

### 💡 Explicación de la Arquitectura
* **Operaciones de Álgebra de Conjuntos:**
  * **Unión (`union` / `|`):** Combina todos los elementos no repetidos de ambos conjuntos.
  * **Intersección (`intersection` / `&`):** Retorna solo los elementos presentes en ambos conjuntos.
  * **Diferencia (`difference` / `-`):** Retorna los elementos presentes en el primer conjunto que no existen en el segundo ($A - B$).
  * **Diferencia Simétrica (`symmetric_difference` / `^`):** Retorna los elementos presentes en uno de los conjuntos pero no en ambos.
* **Evaluaciones Relacionales Booleans:**
  * **`isdisjoint()`:** Devuelve `True` si la intersección entre ambos conjuntos es vacía ($\emptyset$).
  * **`issubset()` / `<=`:** Devuelve `True` si todos los elementos del primer conjunto están incluidos en el segundo.
  * **`issuperset()` / `>=`:** Devuelve `True` si el conjunto actual contiene la totalidad de los elementos del otro conjunto.

```python
from typing import Any

# [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Operaciones Álgebraicas y Relacionales en Conjuntos")
print("=" * 80)

# ================================================================================
# 📐 1. OPERACIONES ÁLGEBRAICAS (Unión, Intersección, Diferencia y Dif. Simétrica)
# ================================================================================
# Combinación y filtrado de elementos mediante métodos nativos y operadores.
# ================================================================================

# [1] Declaración de conjuntos numéricos base con Type Hints
first_set: set[int] = {1, 2, 3, 4, 5}
second_set: set[int] = {4, 5, 6, 7, 8}

print("\n--- 🚀 [PASO 1]: Operaciones Álgebraicas Básicas ---")
print(f"📋 Conjunto A (first_set):  {first_set}")
print(f"📋 Conjunto B (second_set): {second_set}")

# [2] Unión: Métodos vs Operador (|)
union_set: set[int] = first_set.union(second_set)
union_operator: set[int] = first_set | second_set
print("-" * 50)
print(f"✅ [1. union()]:             {union_set}")
print(f"✅ [1. Operador '|']:        {union_operator}")

# [3] Intersección: Métodos vs Operador (&)
intersection_set: set[int] = first_set.intersection(second_set)
intersection_operator: set[int] = first_set & second_set
print("-" * 50)
print(f"✅ [2. intersection()]:      {intersection_set}")
print(f"✅ [2. Operador '&']:        {intersection_operator}")

# [4] Diferencia (A - B): Métodos vs Operador (-)
difference_set: set[int] = first_set.difference(second_set)
difference_operator: set[int] = first_set - second_set
print("-" * 50)
print(f"✅ [3. difference() (A - B)]: {difference_set}")
print(f"✅ [3. Operador '-']:        {difference_operator}")

# [5] Diferencia Simétrica: Métodos vs Operador (^)
sym_difference_set: set[int] = first_set.symmetric_difference(second_set)
sym_difference_operator: set[int] = first_set ^ second_set
print("-" * 50)
print(f"✅ [4. symmetric_difference()]: {sym_difference_set}")
print(f"✅ [4. Operador '^']:           {sym_difference_operator}")
```

## 🔍 Evaluaciones Relacionales (`isdisjoint`, `issubset`, `issuperset`)

### 💡 Métodos de Comprobación Booleana:
1. **`isdisjoint()`:** Útil para verificar la exclusión mutua de datos sin realizar un filtrado completo.
2. **`issubset()` / `<=`:** Determina relaciones de inclusión (e.g., validar que un conjunto de permisos requeridos esté dentro del perfil del usuario).
3. **`issuperset()` / `>=`:** Determina relaciones de contención.

```python
# ================================================================================
# 🔍 2. MÉTODOS Y OPERADORES RELACIONALES
# ================================================================================
print("\n--- 🔍 [PASO 2]: Evaluaciones Relacionales y Subconjuntos ---")

# [1] Verificación de conjuntos disjuntos (sin intersección)
are_disjoint: bool = first_set.isdisjoint(second_set)
print(f"🔍 [5. isdisjoint()]: ¿A y B no comparten elementos? -> {are_disjoint}")

# Creación de conjunto disjunto para prueba explícita
isolated_set: set[int] = {10, 20, 30}
are_isolated_disjoint: bool = first_set.isdisjoint(isolated_set)
print(f"🔍 [5. isdisjoint()]: ¿{first_set} y {isolated_set} son disjuntos? -> {are_isolated_disjoint}")

print("-" * 50)

# [2] Verificación de subconjuntos (issubset / <=)
is_subset_false: bool = first_set.issubset(second_set)
print(f"🔍 [6. issubset()]: ¿A es subconjunto de B? -> {is_subset_false}")

# Definición de subconjunto real
subset_a: set[int] = {1, 2}
superset_b: set[int] = {1, 2, 3, 4, 5}

is_subset_true: bool = subset_a.issubset(superset_b)
is_subset_operator: bool = subset_a <= superset_b

print(f"🔍 [6. issubset()]: ¿{subset_a} es subconjunto de {superset_b}? -> {is_subset_true}")
print(f"🔍 [6. Operador '<=']: Comprobación directa -> {is_subset_operator}")

print("-" * 50)

# [3] Verificación de superconjuntos (issuperset / >=)
is_superset_false: bool = first_set.issuperset(second_set)
print(f"🔍 [7. issuperset()]: ¿A es superconjunto de B? -> {is_superset_false}")

is_superset_true: bool = superset_b.issuperset(subset_a)
is_superset_operator: bool = superset_b >= subset_a

print(f"🔍 [7. issuperset()]: ¿{superset_b} es superconjunto de {subset_a}? -> {is_superset_true}")
print(f"🔍 [7. Operador '>=']: Comprobación directa -> {is_superset_operator}")

# --------------------------------------------------------------------------------
# 🏁 [PASO 3]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Módulo de operaciones álgebraicas en conjuntos finalizado con éxito.")
print("=" * 80)
```
