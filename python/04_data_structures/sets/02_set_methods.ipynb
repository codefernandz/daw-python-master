# ⚙️ Fundamentos de Python: Métodos Nativos de Conjuntos (`Set Built-in Methods`)

| Propiedad | Detalle |
| :--- | :--- |
| 📍 **Módulo** | Fundamentos de Python |
| 🎯 **Objetivo** | Dominar los métodos integrados de consulta (`len`), clonación (`copy`), inserción (`add`, `update`) y eliminación segura o estricta (`pop`, `remove`, `discard`, `clear`) en conjuntos (`set`). |
| 🛠️ **Conceptos** | `In-Place Mutation`, `Shallow Copy`, `Arbitrary Removal (.pop)`, `KeyError Exception (.remove)`, `Safe Removal (.discard)`, `Collection Clearing (.clear)` |

---

### 💡 Explicación de la Arquitectura
* **Inserción de Elementos:**
  * `add(element)`: Agrega un único elemento inmutable al conjunto. Si ya existe, no realiza cambios.
  * `update(iterable)`: Agrega múltiples elementos desde un iterable (`set`, `list`, `tuple`) de forma masiva.
* **Estrategias de Eliminación:**
  * `pop()`: Elimina y retorna un elemento arbitrario (ya que los conjuntos no tienen orden garantizado). Lanza `KeyError` si el conjunto está vacío.
  * `remove(element)`: Elimina un elemento específico. **Lanza `KeyError`** si el elemento no existe.
  * `discard(element)`: Elimina un elemento específico **sin lanzar error** si el elemento no se encuentra presente.

```python
from typing import Any

# [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Métodos Nativos de Conjuntos (Sets)")
print("=" * 80)

# ================================================================================
# 🔍 1. CONSULTA DE TAMAÑO, CLONACIÓN E INSERCIÓN
# ================================================================================
# Métodos para inspeccionar, duplicar e incorporar elementos al conjunto.
# ================================================================================

# [1] Declaración de conjunto base con Type Hints
user_names: set[str] = {"Juan", "María", "Pedro", "Ana"}

print("\n--- 🚀 [PASO 1]: Consulta, Clonación e Inserción ---")
print(f"📋 Set original inicial: {user_names}")

print("-" * 80)

# [2] Obtención de tamaño con len()
set_length: int = len(user_names)
print(f"📏 [1. len()]:  Cantidad de elementos -> {set_length}")

# [3] Clonación superficial con copy()
names_copy: set[str] = user_names.copy()
print(f"📋 [2. copy()]: Copia independiente creada -> {names_copy}")

# [4] Inserción de un solo elemento con add()
user_names.add("Luis")
print(f"✅ [3. add()]:  Tras agregar 'Luis' -> {user_names}")

# [5] Inserción masiva de elementos con update()
user_names.update({"Carlos", "Lucía"})
print(f"✅ [4. update()]: Tras agregar 'Carlos' y 'Lucía' -> {user_names}")
```

## 🗑️ Métodos de Extracción y Eliminación (`pop`, `remove`, `discard`, `clear`)

### 💡 Comparativa entre `remove()` y `discard()`:
* **`remove(x)`:** Uso adecuado cuando la ausencia del elemento representa una anomalía en la lógica del negocio que debe ser capturada mediante manejo de excepciones (`KeyError`).
* **`discard(x)`:** Uso idóneo para operaciones idempotentes donde solo se requiere garantizar que el elemento no permanezca en el conjunto, sin importar si existía previamente.

```python
# ================================================================================
# 🗑️ 2. MÉTODOS DE ELIMINACIÓN Y VACIADO
# ================================================================================
print("\n--- 🗑️ [PASO 2]: Extracción y Eliminación de Elementos ---")

# [1] Extracción arbitraria con pop()
popped_element: str = user_names.pop()
print(f"✅ [5. pop()]: Elemento arbitrario extraído: '{popped_element}'")
print(f"📋 Estado tras pop(): {user_names}")

# [2] Eliminación directa con remove() usando verificación previa
element_to_remove: str = "María"

print("-" * 80)
if element_to_remove in user_names:
    user_names.remove(element_to_remove)
    print(f"✅ [6. remove('{element_to_remove}')]: Estado tras eliminar -> {user_names}")
else:
    print(f"ℹ️ [6. remove('{element_to_remove}')]: '{element_to_remove}' no estaba en el conjunto (posiblemente fue extraído por pop(): '{popped_element}').")

# Captura de error intencionada al intentar eliminar un elemento que seguro no existe
try:
    user_names.remove("Victor")
except KeyError:
    print("❌ [6. remove('Victor')]: KeyError capturado correctamente (El elemento 'Victor' no existe).")

# [3] Eliminación segura con discard()
user_names.discard("Ana")
print("-" * 80)
print(f"✅ [7. discard('Ana')]: Estado tras descartar 'Ana' -> {user_names}")

# Intento de descartar elemento inexistente (no produce error)
user_names.discard("Victor")
print("✅ [7. discard('Victor')]: Ejecutado sobre elemento inexistente sin errores.")

# [4] Limpieza total con clear()
user_names.clear()
print("-" * 80)
print(f"✅ [8. clear()]: Conjunto vaciado completamente -> {user_names}")
print(f"📋 Verificación de la copia intacta:            {names_copy}")
```
