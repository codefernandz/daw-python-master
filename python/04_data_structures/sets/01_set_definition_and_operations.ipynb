### 💡 Explicación de la Arquitectura
* **Colección Desordenada de Elementos Únicos:** Un `set` no almacena índices ni mantiene un orden garantizado de acceso. No admite elementos duplicados.
* **Restricción de Elementos (`Hashable`):** Los elementos dentro de un `set` deben ser obligatoriamente **inmutables** (`str`, `int`, `float`, `tuple`). No puede contener colecciones mutables como listas (`list`), otros conjuntos (`set`) o diccionarios (`dict`).
* **Sintaxis de Creación:**
  * Literal con llaves: `{elem1, elem2, elem3}` (Nota: `{}` vacío crea un `dict`, no un `set`).
  * Constructor dinámico: `set(iterable)` útil para desduplicar tuplas o listas.

```python
from typing import Any

# [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Definición y Características de Conjuntos (Sets)")
print("=" * 80)

# ================================================================================
# 📦 1. CREACIÓN LITERAL DE CONJUNTOS ({})
# ================================================================================
# Declaración directa de un conjunto con elementos únicos e inmutables.
# ================================================================================

print("\n--- 🚀 [PASO 1]: Creación Literal con Llaves ---")

# [1] Definición usando la sintaxis literal con llaves {}
unique_names: set[str] = {"Alice", "Bob", "Charlie"}

print(f"📋 Set de nombres:  {unique_names}")
print(f"✅ Tipo de dato:     {type(unique_names)}")
print(f"📏 Longitud:         {len(unique_names)} elementos")
```

## 🔄 Conversión de Iterables a Conjuntos (`Set Casting`)

### 💡 Eliminación Automática de Duplicados:
Al pasar una estructura de datos (como una `tuple` o `list`) mediante el constructor `set()`, Python realiza dos acciones automáticas:
1. Recorre todos los elementos del iterable.
2. Descarta duplicados de forma eficiente utilizando sus valores hash.

```python
# ================================================================================
# 🔄 2. CONSTRUCCIÓN Y DESDUPLICACIÓN MEDIANTE set()
# ================================================================================
print("\n--- 🔄 [PASO 2]: Conversión desde Tupla (Set Casting) ---")

# [1] Definición de una tupla con un elemento duplicado ("Alice")
names_tuple: tuple[str, ...] = ("David", "Eva", "Frank", "Grace", "Alice", "Alice")

print(f"📦 Tupla original (con duplicados): {names_tuple}")
print(f"📏 Longitud tupla:                   {len(names_tuple)} elementos")

# [2] Conversión implícita a conjunto
set_from_tuple: set[str] = set(names_tuple)

print("-" * 50)
print(f"📋 Set resultante (desduplicado):    {set_from_tuple}")
print(f"✅ Tipo de dato:                     {type(set_from_tuple)}")
print(f"📏 Longitud set:                     {len(set_from_tuple)} elementos")

# --------------------------------------------------------------------------------
# 🏁 [PASO 3]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Módulo de definición de conjuntos finalizado con éxito.")
print("=" * 80)
```
