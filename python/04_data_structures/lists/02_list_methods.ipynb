### 💡 Explicación de la Arquitectura
* **Funciones Integradas (*Built-in Functions*):**
  * **`len(list)`:** Retorna la cantidad total de elementos en la lista.
  * **`max(list)` / `min(list)`:** Obtienen los valores máximo y mínimo respectivamente.
  * **`sorted(list)`:** Retorna una **nueva lista ordenada** sin alterar la lista original (*Inmutable*).
* **Métodos de Instancia (*In-place Methods*):**
  * **`list.sort()`:** Ordena los elementos directamente sobre la lista de origen (*Mutación en el sitio*).
  * **`list.copy()`:** Genera una copia superficial (*shallow copy*) independiente en memoria.
  * **`list.count(element)`:** Cuenta cuántas veces se repite un elemento dentro de la lista.
  * **`list.reverse()`:** Invierte el orden de los elementos in-place.

```python
# [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Métodos e Inspección de Listas")
print("=" * 80)

# ================================================================================
# 📊 1. FUNCIONES INTEGRADAS DE INSPECCIÓN (len, max, min)
# ================================================================================
# Obtención de métricas generales sobre colecciones numéricas.
# ================================================================================

# [1] Declaración de lista numérica inicial con Type Hints correctos
number_list: list[int] = [5, 2, 8, 1, 4]

print("\n--- 🚀 [PASO 1]: Inspección General con Funciones Integradas ---")
print(f"📋 Lista base inicial: {number_list}")

# Cálculo de métricas básicas
total_elements: int = len(number_list)
max_value: int = max(number_list)
min_value: int = min(number_list)

print(f"✅ [1. len()]: Total de elementos: {total_elements}")
print(f"✅ [2. max()]: Valor máximo:        {max_value}")
print(f"✅ [3. min()]: Valor mínimo:        {min_value}")
```

## 🔀 Ordenamiento (`sorted` vs `sort`), Clonación (`copy`), Conteo (`count`) e Inversión (`reverse`)

### 💡 Comparativa entre `sorted()` y `sort()`:
1. **`sorted(list)`:** Es una función global que devuelve una **nueva colección**, manteniendo intacta la variable de origen. Ideal cuando necesitas conservar los datos desordenados originales.
2. **`list.sort()`:** Es un método que **modifica la lista sobre la que se aplica** y retorna `None`. Es más eficiente en uso de memoria.

```python
# ================================================================================
# 🔀 2. ORDENAMIENTO (sorted vs sort)
# ================================================================================
print("\n--- 🔀 [PASO 2]: Comparativa de Ordenamiento (sorted vs sort) ---")

# [1] Uso de sorted() - No modifica la lista original
sorted_numbers: list[int] = sorted(number_list)

print(f"📋 Lista original intacta: {number_list}")
print(f"✅ [4. sorted()]: Nueva lista ordenada: {sorted_numbers}")

# [2] Uso de sort() - Modifica in-place
number_list.sort()
print(f"✅ [4. sort()]:   Lista modificada in-place: {number_list}")

# ================================================================================
# 🛠️ 3. MÉTODOS DE INSTANCIA (copy, count, reverse)
# ================================================================================
print("\n--- 🛠️ [PASO 3]: Copia, Conteo de Elementos e Inversión ---")

# [1] Clonación superficial de la lista
shallow_copy: list[int] = number_list.copy()
print(f"✅ [5. copy()]:    Copia independiente creada: {shallow_copy}")

# [2] Conteo de ocurrencias de un elemento
shallow_copy.append(2)  # Agregamos un segundo 2 para la prueba de conteo
occurrences: int = shallow_copy.count(2)
print(f"📋 Lista tras append(2): {shallow_copy}")
print(f"✅ [6. count()]:   El número 2 aparece {occurrences} veces")

# [3] Inversión de orden in-place
shallow_copy.reverse()
print(f"✅ [7. reverse()]: Lista invertida in-place: {shallow_copy}")

# --------------------------------------------------------------------------------
# 🏁 [PASO 4]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Módulo de métodos de listas finalizado con éxito.")
print("=" * 80)
```
