# ⚙️ Fundamentos de Python: Métodos de Adición e Inserción en Listas (`List Addition Methods`)

| Propiedad | Detalle |
| :--- | :--- |
| 📍 **Módulo** | Fundamentos de Python |
| 🎯 **Objetivo** | Dominar las operaciones de modificación de listas mediante adición de elementos individuales (`append`), fusión de colecciones (`extend`) e inserción por posición específica (`insert`). |
| 🛠️ **Conceptos** | `In-Place Mutation`, `Element Appending (append)`, `List Extension (extend)`, `Positional Insertion (insert)` |

---

### 💡 Explicación de la Arquitectura
* **Operaciones In-Place:** Todos estos métodos modifican la lista original de forma directa y retornan `None`.
* **`append(element)`:** Añade **un único elemento** al final de la lista (si se le pasa otra lista, la insertará como una sublista anidada).
* **`extend(iterable)`:** Desempaqueta y añade **cada uno de los elementos de un iterable** (lista, tupla, etc.) al final de la lista.
* **`insert(index, element)`:** Modifica la estructura desplazando los elementos existentes hacia la derecha para insertar el valor en la posición elegida.

```python
# [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Métodos Nativos de Adición e Inserción en Listas")
print("=" * 80)

# ================================================================================
# ➕ 1. ADICIÓN AL FINAL DE LA LISTA (append vs extend)
# ================================================================================
# Modificación de listas agregando elementos sueltos o colecciones completas.
# ================================================================================

# [1] Declaración de lista base
base_list: list[int] = [1, 2, 3]

print("\n--- 🚀 [PASO 1]: Adición de Elementos con append() y extend() ---")
print(f"📋 Lista base inicial: {base_list}")

# Uso de append() para insertar un elemento al final
base_list.append(4)
print(f"✅ [1. append(4)]:       Elemento añadido al final -> {base_list}")

# Uso de extend() para fusionar un iterable
extension_elements: list[int] = [5, 6]
base_list.extend(extension_elements)
print(f"✅ [2. extend([5, 6])]: Lista extendida con éxito  -> {base_list}")
```

## 📍 Inserción por Posición Específica (`insert`)

### 💡 Explicación de `insert(index, element)`:
* **`index`:** La posición numérica donde se colocará el nuevo valor (comenzando en `0`).
* **Comportamiento:** Los elementos presentes a partir de dicho índice son desplazados un lugar hacia la derecha sin sobrescribir ningún dato.

```python
# ================================================================================
# 📍 2. INSERCIÓN POSICIONAL (insert)
# ================================================================================
print("\n--- 📍 [PASO 2]: Inserción de Elementos en Posición Específica ---")

# [1] Reinstalación de lista de prueba
target_list: list[int] = [1, 2, 3]
inserted_value: int = 10
target_index: int = 1

print(f"📋 Lista original previa a la inserción: {target_list}")

# Uso de insert()
target_list.insert(target_index, inserted_value)
print(f"✅ [3. insert(1, 10)]:  Valor {inserted_value} insertado en índice {target_index} -> {target_list}")

# --------------------------------------------------------------------------------
# 🏁 [PASO 3]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Módulo de métodos de adición en listas finalizado con éxito.")
print("=" * 80)
```
