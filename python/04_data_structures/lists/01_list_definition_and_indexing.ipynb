### 💡 Explicación de la Arquitectura
* **Definición de Lista:** Colección ordenada y mutable de elementos encerrada entre corchetes `[]`.
* **Heterogeneidad:** Una misma lista puede almacenar tipos de datos mixtos (`int`, `str`, `bool`, `float`, etc.).
* **Listas Anidadas:** Estructuras multidimensionales donde un elemento de la lista es otra lista a la que se accede encadenando índices `list[i][j]`.
* **Conversión de Tipo (`list()`):** Transforma objetos iterables (como un `str`) en una lista de caracteres individuales.
* **Slicing de Listas (`[start:stop:step]`):** Extrae una sublista creando un nuevo objeto en memoria sin modificar la lista de origen.

```python
from typing import Any

# [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Creación, Estructura e Indexación de Listas")
print("=" * 80)

# ================================================================================
# 📦 1. CREACIÓN DE LISTAS, HETEROGENEIDAD Y ANIDAMIENTO
# ================================================================================
# Declaración de listas simples, vacías, anidadas y conversión desde cadenas.
# ================================================================================

# [1] Declaración de listas con Type Hints explícitos
sample_list: list[Any] = [1, 2, 3, "Python", True, 3.14]
empty_list: list[Any] = []
nested_list: list[Any] = [1, 2, [3, 4], ["a", "b"]]

# [2] Conversión de un string a lista de caracteres
source_string: str = "Hola"
string_char_list: list[str] = list(source_string)

print("\n--- 🚀 [PASO 1]: Declaración de Listas y Métricas de Longitud ---")
print(f"📋 Lista heterogénea inicial: {sample_list}")
print(f"📋 Lista vacía:               {empty_list}")
print(f"📋 Lista anidada:             {nested_list}")
print(f"📋 Lista desde string:        '{source_string}' -> {string_char_list}")

# Cálculo de longitud con len()
list_length: int = len(sample_list)
print(f"✅ Longitud de la lista base (len): {list_length} elementos")
```

## 📍 Indexación Básica y Acceso Multidimensional

### 💡 Explicación del Acceso por Índice:
1. **Acceso Unidimensional:** Utiliza un solo par de corchetes `sample_list[0]` (primer elemento) o `sample_list[-1]` (último elemento).
2. **Acceso Multidimensional (Listas Anidadas):** Para acceder a valores dentro de listas internas, se encadenan corchetes `nested_list[i][j]`, donde el primer índice selecciona la lista interna y el segundo el elemento dentro de esta.

```python
# ================================================================================
# 📍 2. INDEXACIÓN SIMPLE Y MULTIDIMENSIONAL
# ================================================================================
print("\n--- 📍 [PASO 2]: Acceso a Elementos por Índice ---")

# Indexación básica sobre lista simple
first_element: Any = sample_list[0]
last_element: Any = sample_list[-1]

print(f"✅ Primer elemento [0]:  {first_element}")
print(f"✅ Último elemento [-1]: {last_element}")

# Acceso a elementos en listas anidadas (multidimensional)
nested_element: int = nested_list[2][1]       # Accede al número 4 de la lista [3, 4]
nested_char: str = nested_list[3][1]          # Accede al carácter 'b' de la lista ["a", "b"]

print("-" * 50)
print(f"✅ Elemento anidado [2][1] de {nested_list[2]}: {nested_element}")
print(f"✅ Elemento anidado [3][1] de {nested_list[3]}: '{nested_char}'")
```

## ✂️ Troceado de Listas (*Slicing*: `[start:stop:step]`)

### 💡 Explicación de las Reglas de Slicing en Listas:
* **Rango Acotado (`[1:4]`):** Extrae desde el índice `1` hasta el `3` (el índice `4` queda fuera).
* **Intervalos / Saltos (`[::2]`):** Recorre toda la lista seleccionando un elemento cada dos posiciones.
* **Slicing Anidado (`nested_list[3][0:2]`):** Permite aplicar rangos directamente sobre la lista interna seleccionada.

```python
# ================================================================================
# ✂️ 3. SLICING Y EXTRACCIÓN DE SUBLISTAS
# ================================================================================
print("\n--- ✂️ [PASO 3]: Extracción de Sublistas (Slicing) ---")

# Slicing acotado [start:stop]
sliced_sublist: list[Any] = sample_list[1:4]

# Slicing con paso o salto [::step]
sublist_by_step: list[Any] = sample_list[::2]

# Copia completa de lista vía slicing [0:6]
full_copied_sublist: list[Any] = sample_list[0:6]

# Slicing sobre lista anidada
nested_sublist: list[str] = nested_list[3][0:2]

print(f"✅ Sublista acotada [1:4]:               {sliced_sublist}")
print(f"✅ Sublista con salto de 2 [::2]:        {sublist_by_step}")
print(f"✅ Sublista completa copia [0:6]:        {full_copied_sublist}")
print(f"✅ Slicing sobre lista anidada [3][0:2]: {nested_sublist}")

# --------------------------------------------------------------------------------
# 🏁 [PASO 4]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Módulo de definición e indexación de listas finalizado.")
print("=" * 80)
```
