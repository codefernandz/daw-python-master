# ⚙️ Fundamentos de Python: Métodos de Eliminación en Listas (`List Removal Methods`)

| Propiedad | Detalle |
| :--- | :--- |
| 📍 **Módulo** | Fundamentos de Python |
| 🎯 **Objetivo** | Dominar las técnicas de eliminación de elementos en listas por índice o extracción (`pop`), por valor específico (`remove`) y vaciado completo (`clear`). |
| 🛠️ **Conceptos** | `In-Place Mutation`, `Element Extraction (pop)`, `Value Removal (remove)`, `List Cleansing (clear)`, `Exception Handling (ValueError, IndexError)` |

---

### 💡 Explicación de la Arquitectura
* **`pop([index])`:** Elimina y **devuelve** el elemento ubicado en la posición especificada. Si no se pasa un índice, extrae el **último elemento** (comportamiento tipo Pila/Stack). Lanza `IndexError` si la lista está vacía o el índice está fuera de rango.
* **`remove(value)`:** Busca y elimina la **primera aparición** del valor indicado. Lanza `ValueError` si el elemento no existe en la lista.
* **`clear()`:** Vacía por completo la lista conservando su misma referencia en memoria.

```python
# [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Métodos Nativos de Eliminación en Listas")
print("=" * 80)

# ================================================================================
# 🗑️ 1. EXTRACCIÓN POR ÍNDICE Y POP (pop)
# ================================================================================
# Extracción de elementos modificando la lista e inspeccionando el valor extraído.
# ================================================================================

# [1] Declaración de lista base numérica
number_list: list[int] = [10, 20, 30, 40, 50]

print("\n--- 🚀 [PASO 1]: Extracción de Elementos con pop() ---")
print(f"📋 Lista base inicial: {number_list}")

# [2] Extracción por índice específico (índice 1 -> valor 20)
target_index: int = 1
popped_element: int = number_list.pop(target_index)

print(f"✅ [1. pop(1)]:  Valor extraído de índice {target_index}: {popped_element}")
print(f"📋 Lista tras pop(1): {number_list}")

# [3] Extracción por defecto (último elemento)
last_element: int = number_list.pop()
print(f"✅ [1. pop()]:   Valor extraído por defecto (último): {last_element}")
print(f"📋 Lista tras pop() por defecto: {number_list}")
```

## 🎯 Eliminación por Valor (`remove`) y Vaciado (`clear`)

### 💡 Explicación de Buenas Prácticas:
1. **`remove(value)`:** A diferencia de `pop()`, este método no requiere saber el índice del elemento, sino su contenido. Dado que lanza un `ValueError` si el elemento no existe, es recomendable usar bloques `try/except` o validar su presencia previa.
2. **`clear()`:** Es más eficiente para reiniciar una lista que reasignarla a `[]`, ya que mantiene la dirección de memoria original del objeto.

```python
# ================================================================================
# 🎯 2. ELIMINACIÓN POR VALOR (remove) Y VACIADO COMPLETO (clear)
# ================================================================================
print("\n--- 🎯 [PASO 2]: Eliminación por Valor (remove) y Limpieza (clear) ---")

value_to_remove: int = 30

# [1] Eliminación por valor con remove()
print(f"📋 Lista previa a remove({value_to_remove}): {number_list}")

try:
    number_list.remove(value_to_remove)
    print(f"✅ [2. remove({value_to_remove})]: Se eliminó la primera ocurrencia del valor {value_to_remove}")
    print(f"📋 Lista tras remove(): {number_list}")
except ValueError:
    print(f"❌ [Error]: El valor {value_to_remove} no existe en la lista.")

# [2] Vaciado total con clear()
number_list.clear()
print("-" * 50)
print(f"✅ [3. clear()]: Lista vaciada por completo -> {number_list}")
print(f"📋 Longitud final de la lista (len): {len(number_list)}")

# --------------------------------------------------------------------------------
# 🏁 [PASO 3]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Módulo de métodos de eliminación en listas finalizado con éxito.")
print("=" * 80)
```
