### 💡 Explicación de la Arquitectura
* **Definición de Tupla:** Colección ordenada e **inmutable** de elementos encerrada entre paréntesis `()`. Una vez creada, no es posible añadir, eliminar ni modificar sus elementos directamente.
* **Tupla de un Solo Elemento:** Requiere obligatoriamente una coma final `(42,)`. Sin la coma, Python lo evalúa como una expresión entre paréntesis entre enteros `(42) -> int`.
* **Indexación:** Acceso basado en posición mediante corchetes `tuple[i]`. Permite índices positivos `[0]` y negativos `[-1]`.
* **Modificación Indirecta:** Para alterar el contenido de una tupla se recurre al patrón de casteo: `tuple` $\rightarrow$ `list` $\rightarrow$ *mutación* $\rightarrow$ `tuple`.

```python
from typing import Any

# [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Creación, Inmutabilidad e Indexación de Tuplas")
print("=" * 80)

# ================================================================================
# 📦 1. CREACIÓN DE TUPLAS Y SINTAXIS DE ELEMENTO ÚNICO
# ================================================================================
# Declaración de tuplas heterogéneas y manejo del caso especial de elemento único.
# ================================================================================

# [1] Declaración de tupla heterogénea estándar
sample_tuple: tuple[Any, ...] = (1, 2, 3, "a", "b", "c")

# [2] Tupla de un solo elemento (notación obligatoria con coma final)
single_element_tuple: tuple[int] = (42,)

# [3] Demostración de la trampa sintáctica sin coma
not_a_tuple: int = (42)

print("\n--- 🚀 [PASO 1]: Declaración de Tuplas y Verificación de Tipos ---")
print(f"📋 Tupla heterogénea inicial:           {sample_tuple} | Tipo: {type(sample_tuple)}")
print(f"📋 Tupla de un solo elemento (42,):    {single_element_tuple} | Tipo: {type(single_element_tuple)}")
print(f"⚠️ Expresión sintáctica sin coma (42): {not_a_tuple} | Tipo: {type(not_a_tuple)}")
```

## 📍 Indexación Básica y Negativa

### 💡 Explicación del Acceso por Posición:
1. **Índice Positivo (`tuple[0]`):** Comienza en `0` para el primer elemento y avanza secuencialmente.
2. **Índice Negativo (`tuple[-1]`):** Comienza en `-1` para acceder al último elemento y retrocede secuencialmente hacia el inicio.

```python
# ================================================================================
# 📍 2. INDEXACIÓN SIMPLE Y REVERSA
# ================================================================================
print("\n--- 📍 [PASO 2]: Acceso a Elementos por Índice ---")

# Indexación positiva
first_element: Any = sample_tuple[0]
second_element: Any = sample_tuple[1]

# Indexación negativa
last_element: Any = sample_tuple[-1]
second_last_element: Any = sample_tuple[-2]

print(f"✅ Primer elemento [0]:     {first_element}")
print(f"✅ Segundo elemento [1]:    {second_element}")
print(f"✅ Último elemento [-1]:    '{last_element}'")
print(f"✅ Penúltimo elemento [-2]: '{second_last_element}'")
```

## 🔄 Concatenación y Modificación Indirecta

### 💡 Inmutabilidad y Patrón de Conversión:
* **Inmutabilidad:** Asignar un valor como `sample_tuple[0] = 99` lanza de forma directa un `TypeError`.
* **Concatenación:** Unir dos tuplas con `+` genera una **nueva tupla independiente en memoria**.
* **Patrón `tuple -> list -> tuple`:** Permite mutar indirectamente los datos cuando se requiere transformar la colección sin perder el contrato de inmutabilidad final.

```python
# ================================================================================
# 🔄 3. CONCATENACIÓN Y MODIFICACIÓN INDIRECTA (tuple <-> list)
# ================================================================================
print("\n--- 🔄 [PASO 3]: Concatenación y Modificación Indirecta ---")

# [1] Concatenación de tuplas (Genera un nuevo objeto)
extension_tuple: tuple[int, ...] = (4, 5, 6)
concatenated_tuple: tuple[Any, ...] = sample_tuple + extension_tuple

print(f"📋 Tupla base original:      {sample_tuple}")
print(f"✅ Tupla creada por suma (+): {concatenated_tuple}")

# [2] Demostración de mutación indirecta vía conversión a lista
print("-" * 50)
print("🛠️ Ejecutando patrón de mutación indirecta (tuple -> list -> tuple)...")

# Paso A: Conversión a lista mutable
mutable_list_conversion: list[Any] = list(sample_tuple)

# Paso B: Modificación de elemento en lista
mutable_list_conversion[0] = 10

# Paso C: Reconversión a tupla inmutable
modified_tuple: tuple[Any, ...] = tuple(mutable_list_conversion)

print(f"📋 Estado intermedio (lista): {mutable_list_conversion}")
print(f"✅ Resultado final (nueva tupla): {modified_tuple}")

# --------------------------------------------------------------------------------
# 🏁 [PASO 4]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Módulo de definición e indexación de tuplas finalizado con éxito.")
print("=" * 80)
```
