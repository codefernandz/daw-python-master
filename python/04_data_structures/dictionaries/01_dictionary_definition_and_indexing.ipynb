### 💡 Explicación de la Arquitectura
* **Definición de Diccionario:** Colección de pares clave-valor ordenada por orden de inserción (desde Python 3.7+).
* **Restricción de Claves:** Las claves deben ser obligatoriamente de **tipos inmutables y hashables** (`str`, `int`, `float`, `tuple`). Los valores pueden ser de cualquier tipo.
* **Acceso Directo vs Seguro:** 
  * `dict[key]`: Lanza `KeyError` si la clave no existe.
  * `dict.get(key, default)`: Devuelve `None` (o el valor por defecto configurado) sin interrumpir la ejecución.
* **Iteración Eficiente:** Permite recorrer llaves (`.keys()`), valores (`.values()`) o ambos simultáneamente mediante desempaquetado de tuplas (`.items()`).

```python
from typing import Any

# 🛰️ [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Creación, Indexación y Operaciones con Diccionarios")
print("=" * 80)

# ================================================================================
# 📦 1. CREACIÓN DE DICCIONARIOS
# ================================================================================
# Construcción mediante llaves {} y función constructora dict().
# ================================================================================

# [1] Declaración usando sintaxis literal con llaves {}
user_profile: dict[str, Any] = {
    "name": "Alice",
    "age": 30,
    "city": "Nueva York"
}

# [2] Declaración usando la función constructora dict()
secondary_profile: dict[str, Any] = dict(
    name="Bob",
    age=25,
    city="Los Ángeles"
)

print("\n--- 🚀 [PASO 1]: Creación de Diccionarios ---")
print(f"📋 Perfil 1 (llaves {{}}): {user_profile}")
print(f"📋 Perfil 2 (dict()):     {secondary_profile}")
```

## 📍 Indexación, Acceso Seguro e Inserción/Modificación

### 💡 Explicación del Acceso e Inserción:
1. **Acceso Directo (`dict[key]`):** Búsqueda directa por clave con complejidad $O(1)$.
2. **Acceso Seguro (`dict.get(key)`):** Previene caídas del sistema devolviendo `None` o un fallback opcional si la clave no existe.
3. **Asignación (`dict[key] = value`):** Si la clave existe, actualiza su valor. Si no existe, crea el par clave-valor en la colección.

```python
# ================================================================================
# 📍 2. INDEXACIÓN, BÚSQUEDA SEGURA Y MUTACIÓN
# ================================================================================
print("\n--- 📍 [PASO 2]: Acceso, Búsqueda Segura y Modificación ---")

# [1] Inicialización de diccionario de inventario
fruit_inventory: dict[str, Any] = {
    "fruit": "manzana",
    "color": "rojo",
    "quantity": 5
}

# Acceso directo mediante clave
print(f"✅ Fruta:    {fruit_inventory['fruit']}")
print(f"✅ Color:    {fruit_inventory['color']}")
print(f"✅ Cantidad: {fruit_inventory['quantity']}")

# Acceso seguro mediante get() para prevenir KeyError
flavor_access: Any = fruit_inventory.get("flavor")
fallback_access: Any = fruit_inventory.get("flavor", "Sin sabor especificado")

print("-" * 50)
print(f"🔍 Búsqueda de clave inexistente 'flavor' (.get()): {flavor_access}")
print(f"🔍 Búsqueda con valor de respaldo por defecto:    '{fallback_access}'")

# [2] Modificación de valor existente e inserción de nueva clave
fruit_inventory["quantity"] = 10         # Actualización
fruit_inventory["flavor"] = "dulce"       # Inserción

print("-" * 50)
print(f"✅ Inventario actualizado (quantity=10, flavor='dulce'):\n   {fruit_inventory}")
```

## 🗑️ Eliminación e Iteración de Diccionarios

### 💡 Estrategias de Eliminación y Recorrido:
* **`del dict[key]`:** Elimina el par clave-valor directamente. Lanza `KeyError` si la clave no existe.
* **`dict.pop(key)`:** Elimina la clave y **retorna el valor asociado** para su uso posterior.
* **Iteración:** 
  * Requerir solo claves: `for key in dict:` o `for key in dict.keys():`
  * Requerir solo valores: `for value in dict.values():`
  * Requerir ambos: `for key, value in dict.items():` *(Práctica recomendada)*

```python
# ================================================================================
# 🗑️ 3. ELIMINACIÓN DE CLAVES Y MÉTODOS DE ITERACIÓN
# ================================================================================
print("\n--- 🗑️ [PASO 3]: Eliminación de Elementos ---")

# [1] Eliminación con palabra clave del
del fruit_inventory["color"]
print(f"✅ Tras eliminar 'color' (del): {fruit_inventory}")

# [2] Extracción y eliminación con pop()
removed_value: Any = fruit_inventory.pop("flavor")
print(f"✅ Valor extraído con pop('flavor'): '{removed_value}'")
print(f"📋 Estado final del diccionario:    {fruit_inventory}")

print("\n--- 🔄 [PASO 4]: Métodos de Iteración ---")

# [3] Iteración sobre claves
print("⚙️ [Claves]:")
for key in fruit_inventory.keys():
    print(f"   • {key}")

# [4] Iteración sobre valores
print("⚙️ [Valores]:")
for value in fruit_inventory.values():
    print(f"   • {value}")

# [5] Iteración idiomática sobre pares clave-valor (.items)
print("⚙️ [Pares Clave-Valor (.items())]:")
for key, value in fruit_inventory.items():
    print(f"   • {key} -> {value}")

# --------------------------------------------------------------------------------
# 🏁 [PASO 5]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Módulo de definición e indexación de diccionarios finalizado.")
print("=" * 80)
```
