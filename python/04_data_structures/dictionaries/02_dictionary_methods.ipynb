| 🛠️ **Conceptos** | `Dictionary Views`, `Safe Lookup (.get)`, `In-Place Update (.update)`, `Default Assignment (.setdefault)`, `LIFO Removal (.popitem)`, `Shallow Copy` |

---

### 💡 Explicación de la Arquitectura
* **Vistas de Diccionario (*Dictionary Views*):** `keys()`, `values()` e `items()` devuelven objetos dinámicos que reflejan los cambios del diccionario en tiempo real sin crear copias en memoria.
* **Inserción y Actualización Eficiente:** 
  * `update()`: Fusiona diccionarios o pares clave-valor sobrescribiendo claves coincidentes.
  * `setdefault()`: Garantiza que una clave exista; si no está presente, la crea con el valor por defecto provisto.
* **Extracción Controlada:**
  * `pop(key)`: Extrae y retorna el valor de una clave dada (Lanza `KeyError` si no existe).
  * `popitem()`: Extrae y retorna el último par clave-valor insertado `(key, value)` siguiendo el orden LIFO (*Last-In, First-Out*).

```python
from typing import Any

# [1] Banner visual de inicio de ejecución
print("=" * 80)
print("🐍 FUNDAMENTOS DE PYTHON | Métodos Nativos de Diccionarios")
print("=" * 80)

# ================================================================================
# 🔍 1. VISTAS Y BÚSQUEDA SEGURA (keys, values, items, get)
# ================================================================================
# Consulta de estructuras clave-valor sin modificar la colección base.
# ================================================================================

# [1] Declaración de diccionario base con Type Hints
user_profile: dict[str, Any] = {
    "name": "Juan",
    "age": 30,
    "city": "Madrid"
}

print("\n--- 🚀 [PASO 1]: Vistas Dinámicas y Búsqueda Segura ---")
print(f"📋 Perfil base inicial: {user_profile}")

# [2] Obtención de vistas
dictionary_keys = user_profile.keys()
dictionary_values = user_profile.values()
dictionary_items = user_profile.items()

print(f"✅ [1. keys()]:   {dictionary_keys}")
print(f"✅ [2. values()]: {dictionary_values}")
print(f"✅ [3. items()]:  {dictionary_items}")

# [3] Búsqueda con get()
found_name: Any = user_profile.get("name")
missing_country: Any = user_profile.get("country")

print("-" * 50)
print(f"🔍 [4. get('name')]:    {found_name}")
print(f"🔍 [4. get('country')]: {missing_country} (Clave no existente)")
```

## 🛠️ Actualización, Asignación por Defecto y Extracción (`update`, `setdefault`, `pop`, `popitem`)

### 💡 Comportamiento de Mutación:
1. **`update()`:** Permite realizar actualizaciones masivas de datos en una sola línea de código.
2. **`setdefault()`:** Útil para evitar condicionales `if key not in dict:` al inicializar valores predeterminados.
3. **`popitem()`:** Elimina el par mas reciente en orden de inserción.

```python
# ================================================================================
# 🛠️ 2. MUTACIÓN, ASIGNACIÓN POR DEFECTO Y EXTRACCIÓN
# ================================================================================
print("\n--- 🛠️ [PASO 2]: Actualización y Extracción de Datos ---")

# [1] Fusionar y actualizar datos con update()
user_profile.update({"country": "España", "age": 31})
print(f"✅ [5. update()]:    Diccionario actualizado -> {user_profile}")

# [2] Asignación por defecto con setdefault()
profession_value: Any = user_profile.setdefault("profession", "Ingeniero")
print(f"✅ [6. setdefault()]: Valor obtenido: '{profession_value}'")
print(f"📋 Estado tras setdefault(): {user_profile}")

# [3] Extracción por clave con pop()
popped_age: Any = user_profile.pop("age")
print("-" * 50)
print(f"✅ [7. pop('age')]:   Valor extraído: {popped_age}")
print(f"📋 Estado tras pop('age'):   {user_profile}")

# [4] Extracción del último elemento con popitem()
last_item: tuple[str, Any] = user_profile.popitem()
print(f"✅ [7. popitem()]:    Último par extraído: {last_item}")
print(f"📋 Estado tras popitem():    {user_profile}")
```

## 📋 Clonación Superficial (`copy`) y Vaciado (`clear`)

### 💡 Gestión de Memoria:
* **`copy()`:** Crea una **copia superficial** (*shallow copy*). Modificar la copia no altera el diccionario de origen.
* **`clear()`:** Elimina todas las entradas del diccionario directamente en memoria, dejándolo vacío `{}`.

```python
# ================================================================================
# 📋 3. CLONACIÓN SUPERFICIAL Y LIMPIEZA TOTAL
# ================================================================================
print("\n--- 📋 [PASO 3]: Copia Independiente y Vaciado de Diccionario ---")

# [1] Copia independiente con copy()
shallow_copy: dict[str, Any] = user_profile.copy()
print(f"✅ [8. copy()]:   Copia superficial creada -> {shallow_copy}")

# [2] Limpieza de la colección base con clear()
user_profile.clear()
print(f"✅ [9. clear()]:  Diccionario base vaciado -> {user_profile}")
print(f"📋 Verificación de la copia intacta:      {shallow_copy}")

# --------------------------------------------------------------------------------
# 🏁 [PASO 4]: Cierre de Ejecución
# --------------------------------------------------------------------------------
print("\n" + "=" * 80)
print("✅ [FIN DE EJECUCIÓN] Módulo de métodos de diccionarios finalizado con éxito.")
print("=" * 80)
```
